## Dynamic Programming 

### What is Dynamic Programming?
Dynamic Programming (DP) is a technique to solve problems by:

    - Breaking them into overlapping subproblems

    - Solving each subproblem once

    - Storing results (memoization / tabulation)

    - Reusing them to build the final answer

It is fundamentally about:

    “Turn exponential recursion into polynomial time by caching repeated work.”


Note: DP is not just as a coding trick, but as a state-transition optimization framework.

🔎 When Should You Think of DP?

A problem is likely DP if:

    - It asks for min / max / count / number of ways

    - There are choices at each step

    - There is an optimal substructure

    - Brute force recursion is exponential

    - You see overlapping subproblems



### Core Concepts

1. Optimal Substructure

If optimal solution to problem depends on optimal solutions to subproblems.

Example:
Shortest path in DAG
Knapsack
LIS

2. Overlapping Subproblems

Same subproblem computed multiple times in recursion.

Example:
Fibonacci
fib(5)
 ├── fib(4)
 │    ├── fib(3)
 │    └── fib(2)
 └── fib(3)   ← repeated

#### Two Main DP Styles

🔹 1. Top-Down (Memoization)

    - Recursive + cache.


def fib(n, memo={}):
    if n <= 1:
        return n
    if n in memo:
        return memo[n]

    memo[n] = fib(n-1, memo) + fib(n-2, memo)
    return memo[n]

Time: O(n)
Space: O(n)


🔹 2. Bottom-Up (Tabulation)

Iterative, build from smallest subproblem.

def fib(n):
    dp = [0]*(n+1)
    dp[1] = 1

    for i in range(2, n+1):
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n]


4️⃣ The 5-Step DP Thinking Framework (Critical for E6)

When solving DP in interviews:

Step 1: Define the State

What uniquely defines a subproblem?

Examples:

    dp[i] → best solution up to index i

    dp[i][j] → best using first i items with capacity j

    dp[i][k] → best with i elements and k transactions

Step 2: Define the Transition

How does current state depend on previous states?

dp[i] = max(
    take current,
    skip current
)


Step 3: Base Cases

    What are smallest valid states?

Step 4: Compute Order

    Top-down or bottom-up?

Step 5: Optimize Space (E6-level expectation)
    Can you reduce 2D → 1D?
    Can you remove recursion?
    Can you compress states?

🔵 3. Knapsack-Type DP
