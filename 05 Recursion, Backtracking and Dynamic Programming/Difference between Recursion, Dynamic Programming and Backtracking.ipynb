## **Recursion vs Dynamic Programming vs Backtracking**

**Recursion** is the foundational mechanism where a function calls itself to solve smaller sub-problems. 

**Dynamic Programming (DP)** is an optimization of recursion that saves sub-problem results to avoid repeating work. 

**Backtracking** uses recursion to explore all possible paths, undoing choices when hitting dead ends. 

### **How are they related**

* Recursion as the Foundation: Both Dynamic Programming (in its top-down memoization form) and Backtracking rely heavily on recurison to break a large problem down into manageable states. 

* The progression of Problem-Solving: May algorithmic problems start as a basic recursive idea, which then splits based on characteristics:

    - If different paths reach the same repeated sub-problmes, it becomes a **Dynamic Programming** problem. 

    - If you need to try different choices, check constraints, and undo (or prune) invalid paths to find valid configurations, it becomes a **Backtracking** problem

#### Key Differences at a Glance:


| Concept | Primary Purpose | Core Mechanism | Overlapping Subproblems? |
| :--- | :--- | :--- | :--- |
| **Recursion** | Decomposing a problem into smaller instances of itself | Function calls itself with a base case | Yes (often leads to redundant work without optimization) |
| **Dynamic Programming** | Finding optimal or counted solutions efficiently | Memoization (top-down) or Tabulation (bottom-up) to store results | Yes (specifically designed to exploit this) |
| **Backtracking** | Exploring combinatorial search and constraint spaces | Choice, recursive exploration, validation, and undo/prune | No (explores distinct paths or configurations) |
