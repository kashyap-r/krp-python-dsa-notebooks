### Applications of Backtracking 

1. Combinatorial Problems
    * Subsets 
    * Permutations 
    * Combinations 

```python
# Generate all subsets 
def subsets(nums):
    res = []

    def backtrack(start, path):
        res.append(path[:])
        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i+1, path)
            path.pop()
    backtrack(0, [])
    return res

arr = [1, 2, 3, 4]
result = subsets(arr)

for rows in result: 
    print (rows)
```

🚦 When to Use Backtracking?
Use backtracking when:
✅ Strong signals:
- “Find all possible solutions”
- “Generate all combinations/permutations”
- “Check constraints at each step”
- Problem size is small (n ≤ 20 typically)
🚫 Avoid when:
- Input size is large (n > 25+)
- No pruning possible → brute-force explosion
- Dynamic Programming fits better

⚖️ Pros and Cons
👍 Pros
- Guarantees correct and complete solution
- Elegant recursive structure
- Easy to implement for many problems

👎 Cons
- Exponential time complexity
- Can be slow without pruning
- High recursion depth → stack issues

🧠 Backtracking vs Other Techniques

| Technique        | Use Case |
|------------------|---------|
| Backtracking     | All solutions, constraints |
| DFS              | Traverse graph |
| Dynamic Programming | Overlapping subproblems |
| Greedy           | Local optimal choices |

🧭 How to Recognize Backtracking in Interviews
If you see:
- “Return all…”
- “Try all possibilities”
- “Constraints at each step”
- “Grid + path + choices”
👉 Think Backtracking + DFS

🚀 Pro Tips (Important for E6)
- Prune early → biggest optimization lever
- Use sets/bitmasks for faster constraint checks
- Convert recursion → iterative if needed for scale
- Combine with:
  - Memoization (Backtracking + DP)
  - Heuristics (ordering choices)
