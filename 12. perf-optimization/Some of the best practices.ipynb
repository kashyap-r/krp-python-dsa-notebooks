### Raising Exceptions in Python to enforce business rules

Beyond catching exceptions, you can deliberately raise them using the raise keyword. This is useful for enforcing business rules or input validation. It provides a signal that something is wrong before it causes a harder-to-debug failure somewhere further downstream.

```python
# Raising value error example

value = 2_000
if value > 1_000:   
    # raise the ValueError
    raise ValueError("Please add a value lower than 1,000")
else:
    print("Congratulations! You are the winner!!")
```

```python
value = 2_000
try:
    if value > 1_000:
          
        # raise the ValueError
        raise ValueError("Please add a value lower than 1,000")
    else:
        print("Congratulations! You are the winner!!")
              
# if false then raise the value error
except ValueError as e:
        print(e)
```

### Exception chaining with raise ... from

Beyond raising standalone exceptions, you can chain exceptions using raise ... from. This preserves the original exception as context, making tracebacks much clearer when one error causes another.

```python
try:
    open("config.json")
except FileNotFoundError as e:
    raise RuntimeError("App failed to start: config file missing") from e
```

### Structural Pattern Matching - useful when dealing with multiple exception types in a more readable way.

```python
try:
    result = 1 / 0
except Exception as e:
    match e:
        case ZeroDivisionError():
            print("You cannot divide by zero.")
        case NameError():
            print("Variable not defined.")
        case _:
            print("An unexpected error occurred.")
```

Exception groups and except* 

Python 3.11 introduced exception groups to handle multiple exceptions in a single block using except*. This is useful for managing asynchronous code or scenarios where multiple exceptions might be raised. Example:


```python
try:
    raise ExceptionGroup("Multiple errors", [ValueError("Bad value"), TypeError("Bad type")])
except* ValueError as ve:
    print(f"Handling ValueError: {ve}")
except* TypeError as te:
    print(f"Handling TypeError: {te}")
```

add_note method for exceptions (Python 3.11)

Exceptions now have an add_note() method, which allows developers to add custom notes to exceptions for better debugging. Example:


```python
try:
    raise ValueError("Invalid input")
except ValueError as e:
    e.add_note("This happened while processing user input.")
    e.add_note("Consider validating input before processing.")
    raise
```

Can you catch multiple exceptions in a single try-except block?

Yes, Python allows you to catch multiple exceptions in a single try-except block by using a tuple of exception types. For example: except (TypeError, ValueError):. This will handle either a TypeError or a ValueError in the same block.

Creating Custom Exception in Python 

You can create a custom exception by defining a new class that inherits from the built-in Exception class. For example:

class MyCustomError(Exception):
    pass

Suppressing exceptions in Python 

You can suppress exceptions by using the contextlib.suppress context manager. It allows you to specify exception types that should be ignored. For example:

from contextlib import suppress
with suppress(FileNotFoundError):
    open('non_existent_file.txt')
