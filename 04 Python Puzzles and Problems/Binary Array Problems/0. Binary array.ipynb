### Binary Array 

A binary array simply means an array whose elements are retricted to two possible values 

Example: arr = [0, 1, 0, 1, 1]
or 
arr = [1, 1, 1, 0, 0]

Uses: 
- Rperesenting boolean states
- Encoding sets, falgs, or masks 
- Used in prblems like "maximum consecutive 1s, "minimum flips", or binary subarray sum 



### classic binary array problem

Given a binary array, find he length of the longest contiguous sequesnce of 1s

Input: [1, 1, 1, 0, 0, 1, 1, 1, 1, 0, 0, 1, 1, 0, 1]
Output: 4 

```python
# naive method
arr = [1, 1, 1, 0, 0, 1, 1, 1, 1, 0, 0, 1, 1, 0, 1]

counta = 0
countb = 0
 
for i in arr:
    if i == 1:
        counta += 1  
    else:
        countb = max(counta, countb)
        counta = 0
# bugfix # Final check for trailing 1s
# need to do this to avoid 
countb = max(counta, countb)

print (countb)
```

```python
## Corrected naive version

def longest_run_of_ones(arr):
    max_count = 0
    current_count = 0
    
    for num in arr:
        if num == 1:
            current_count += 1
            max_count = max(max_count, current_count)
        else:
            current_count = 0   # reset when we hit a 0
    
    return max_count

# Example
arr = [1, 1, 0, 1, 1, 1, 0, 1]
print(longest_run_of_ones(arr))  # Output: 3

```

## Problem: Longest Run of 1s if You Can Flip at Most One 0

Statement: Given a binary array, find the length of the longest contiguous sequence of 1s if you are allowed to flip at most one 0 into a 1

Example: 
Input : [1, 1, 0, 1, 1, 1, 0, 1]
Output: 6 [filp the first 0 -> the longest run is [1, 1, 1, 1, 1, 1]]

**Step 1:** Intuition 
* Without flips, it's just the longest run of 1s 
* with one flip we can tolerate one 0 inside the window

* This is a sliding window problem
    - Expand the window while the number of zeros <= 1 
    - Shrink the window when zeroes > 1
    - Track the maximum window length 

**Step 2:** Implementation
 

```python
def longest_run_of_ones(arr):
    left = 0 
    zeros = 0 
    max_len = 0 

    for right in range(len(arr)):
        if arr[right] == 0:
            zeros += 1 
        
        # if more than 1 zero, shrink window 
        while zeros > 1:
            if arr[left] == 0:
                zeros -= 1
            left += 1 

        # Update max length 
        max_len = max(max_len, right - left + 1)
    return max_len 

# arr = [1,1,0,1,1,1,0,1]
# output = 6 
arr = [1, 1, 1, 0, 0, 1, 1, 1, 1, 0, 0, 1, 1, 0, 1]
print(longest_run_of_ones(arr))  
# Output: 6
```
