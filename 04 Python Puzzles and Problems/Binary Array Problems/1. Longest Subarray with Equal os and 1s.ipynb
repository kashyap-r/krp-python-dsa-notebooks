Problem Statement: Longest Subarray with Equal 0s and 1s 
Difficulty: Medium 

Description: Given a binary array, find the length of the longest contiguous subarray with equal number of 0s and 1s.

Example: 
Input: [0, 1, 0, 1, 1, 0, 0]
Output: 6 ( i.e. subarray [0, 1, 0, 1, 1, 0] has 3 zeros and 3 ones)


### Approach 
1. Replace 0 and -1. Now the problem reduces to finding the longest subarray with sum = 0 
2. Use a prefix sum + hashmap 
    - Keep track of cumulative sum 
    - If the same sum appears again at index j and earlier at index i, then the subarray (i+1... j) has equal 0s and 1s.
    - Store the first occurance of each sum in a dictionary

Evolution of the approach: 
### Step: Replace 0 with -1 
Why ?
- If we treat 0 as -1, then every subarray with equal numbers of 0s and 1s will have a sum 0.
- That's because each 1 contributes +1, each 0 contributes -1. 
- Equal counts -> total sum cancels out to 0 

Example Walkthrough 
Original array:
[0, 1, 0, 1, 1, 0, 0]

Replace 0 -> -1:
[-1, 1, -1, 1, 1, -1, -1]

Now compute prefix sums:
Index:   0   1   2   3   4   5   6
Value:  -1   1  -1   1   1  -1  -1
Prefix: -1   0  -1   0   1   0  -1




```python
def longest_equal_zeros_ones(arr):
    prefix_sum = 0
    seen = {0: -1}   # sum -> first index
    max_len = 0
    
    for i, num in enumerate(arr):
        prefix_sum += 1 if num == 1 else -1
        
        if prefix_sum in seen:
            max_len = max(max_len, i - seen[prefix_sum])
        else:
            seen[prefix_sum] = i
    
    return max_len

# Example
arr = [0,1,0,1,1,0,0]
print(longest_equal_zeros_ones(arr))  # Output: 6
```
