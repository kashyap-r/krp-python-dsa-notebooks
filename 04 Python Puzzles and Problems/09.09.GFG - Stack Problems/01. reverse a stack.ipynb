### Reverse a Stack without using extra space in O(n)

Problem Statement: reverse a Stack without using recursion and extra space. 

Example:
Input 1-2-3-4
Output: 4-3-2-1




### 1. Traditional Solution 

Given a stack st[], reverse the stack so that the top element becomes the top while preserving the order of the elements accordingly. 

Example 
Input: st[] = [1, 2, 3, 4]
Output: [1, 2, 3, 4]
Explanation:





[Approach #1] Recursive Approach

The idea is to use the recursion to reverse the given stack 

- First, we keep removing the elements from the stack until stack becomes empty 
- Once the stack is empty, start going back in recursion. At each step insted of placing the element back on the top , we insert it into the bottom
- To insert an element at the bottom, we recursively pop all elements, push the current element, and then put the popped elements back.



```python
def insertAtBottom(st, x):
    if not st:
        st.append(x)
        return
    top = st.pop()
    insertAtBottom(st, x)
    st.append(top)

def reverseStack(st):
    if not st:
        return 
    
    top = st.pop()
    reverseStack(st)
    insertAtBottom(st, top)

if __name__ == "__main__":
    st = [1, 2, 3, 4]
    reverseStack(st)

    while st:
        print(st.pop, end=" ")
```

[Approach 2] Iterative Approach - O(n) Time and O(n) Space

We use an auxiliary stack to hold elements while popping them from the original stack. Since stacks are LIFO, pushing all elements into the auxiliary stack will naturally reverse their order. Finally, we replace the original stack with the auxiliary stack.
