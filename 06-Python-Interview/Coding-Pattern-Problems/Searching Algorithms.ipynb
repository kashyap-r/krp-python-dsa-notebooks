## 1. Linear Search Algorithm

Given an array arr[] of n integers and an interger element x, find whether element x is present in the array. Return the index of the first occurance of x in the array, or -1 if it doesnt exist.

Input: arr = [1, 2, 3, 4, 5, -2], x = 3 
Output: 2


```python
def linear_search(arr, key):
    n = len(arr)
    for i in range(n):
        if arr[i] == key:
            return i
    return -1

## usage
arr = [1, 2, 3, 4, 5, -2]
key = -4
position = linear_search(arr, key)
if position == -1:
    print(f"The integer {key} not found int array!")
else: 
    print (f"The integer {key} was found at position {position}")
```


## 2. Binary Search 

Binary Search is a searching algorithm that operates on a sorted or monotonic search space, repeatedly dividing it into halves to find a target value or optimal answer in logarithmic time O(log N).

Conditions to apply Binary Search:
    - The input list must be sorted 
    - Access to any element of data structure should take a constant time.

Algorithm 

Step 1: Divide the search space into two halves by finding the middle index "mid"

Step 2: Compare the middle of the search space with the search "key"

Step 3: If the key is found in the middle, the process is terminated 

Step 4: If the key is not found in the middle, choose which half will be used as the next search space. 

        -> If the key is smaller than the middle, then the left side is used for next search 
        
        -> if the key is larger than the middle, then the right side is used as the next search space.

Step 5: Continue this process until the key is found or the total search space is exhausted. 




Pro-Tip: The Two Python Patterns

When writing binary search, pick one of these two standard styles to avoid index errors:

• Closed Window (Most Common): Use right = len(arr) - 1 and while left <= right:. Update pointers with mid + 1 or mid - 1.

• Half-Open Window: Use right = len(arr) and while left < right:. Update pointers with left = mid + 1 or right = mid.



Pattern A: The Half-Open Window [left, right)

• Setup: right = len(arr), loop runs while left < right:, right moves to mid.
• The Magic: When arr[mid] == target, it treats it like the element is too large (right = mid). This forces the window to squeeze leftward, trapping the first occurrence.


```python
def half_open_binary_search(arr, target):
    left = 0
    right = len(arr)  # Right is outside valid indices

    while left < right:
        mid = left + (right - left) // 2
        if arr[mid] >= target:
            right = mid  # Squeezes search to the left half
        else:
            left = mid + 1
            
    # Verification step
    if left < len(arr) and arr[left] == target:
        return left
    return -1
```


Pattern B: The Closed Window [left, right]

• Setup: right = len(arr) - 1, loop runs while left <= right:, right moves to mid - 1.

• The Magic: When arr[mid] == target, we save mid as a temporary best answer (ans = mid) and continue searching left (right = mid - 1) to see if an even earlier copy exists.


```python
def closed_window_binary_search(arr, target):
    left = 0
    right = len(arr) - 1
    ans = -1  # Variable to lock down our best candidate

    while left <= right:
        mid = left + (right - left) // 2
        
        if arr[mid] == target:
            ans = mid        # Found a match! Save it...
            right = mid - 1  # ...but keep looking left for duplicates!
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
            
    return ans
```

Recursive Approach in implmenting the Binary Search 

Create a recursive function and compare the mid of search space with the key. 

Based on the result either return the index where the key is found or call the recursive function for the next search space.


```python
def binary_search_rec(arr, low, high, key):
    if high >= low:
        mid = low + (high - low) // 2
        if arr[mid] == key:
            return mid 
        elif arr[mid] > key:
            return binary_search_rec(arr, low, mid - 1, key)
        else: 
            return binary_search_rec(arr, mid + 1, high, key)
    else: 
        return -1 

if __name__ == '__main__':
    arr = [2, 3, 4, 10, 40]
    x = 10    
    result = binary_search_rec(arr, 0, len(arr)-1, x)    
    if result != -1:
        print("Element is present at index", result)
    else:
        print("Element is not present in array")
```

## 3. Boundary Search 

The boundary search algorithm inDSA most commonly refers to a specialized application of Binary Search used to find the exact index where a specific condition (or predicate) transitions from false to true. 

Instead of searching for a single exactmatch, it is designed to find the transition point or "boundary" within a dataset. 

**1. The Core Concept: Finding the Transition Point**

Imagine an array where elements up to a certain point fail a condition (false), and all elements after it pass the condition (true). This property is called monotonicity. 

Index:      0       1       2       3       4       5       6 
Array:  [flase,     false,  false,  true,   true,   true,   true]
                                    ^
                            The boundary index (3)

The boundary search algorithm modifies standard binary search to trap the pointers at this exact transition zone rather than stopping early upon finding a generic match. 

**2. Common Applications in DSA**

This pattern is frequently seen in competetive programming and technical interviews to solve:

*1.* First and Last Occurance: In a sorted array with duplicates (e.g., [1, 2, 2, 2, 3]), find the first time 2 appears (left boundary) or the last time it appears (right boundary).

*2.* Lower Bound & Upper Bound: Finding the first element that is >= target or strictly > target 

*3.* Binary Search on Answer Space: Solving Optimization problems where you search for minimum or maximum possible valid answer (e.g., the Book Allocation Problem or Aggressive Cows)

**3. Implementation Blueprint (The Shrinking Window)**

TO find a boundary, developers typically use a half-open range [left, right]or a shrinking window template. When a 
matching element is foundm, the algorithm does not return immediately, it narraows the search space to find the absolute edge.



**NOTE : Remember this crucial rule - Binary Search (and the standaard boundary search) only works if the data is sorted or follows a strict predicatble property (monotonicity)**

For Ex. in the below array arr = [1, 2, 3, 3, 4, 5, 6, 3, 3, 3, 8, 9, 0], the data is unsorted. The number 3 appears in scattered clusters. In an unsorted array loke this, binary boundary search fails, and you must use a linear search O(N) or a Hash Map.

Given below is the impmentation in python for finding the first occurance (the left boundary) of the target:

```python
def find_left_boundary(arr, target):
    left = 0 
    right = len(arr) # half-open interval [left, right]

    while left < right:
        mid = left + (right - left) // 2

        if arr[mid] >= target:
            # If mid is a candidate, the true boundary must be at the mid or further to the left
            right = mid
        else: 
            # If arr[mid] is too small, the boundary must be to the right of the mid
            left = mid + 1
    return left # left and right converge precisely at the boundary
```

Use Case 1: Count Occurances in the Sorted array i.e. you can find the start poistion, end position, and total count in O(log N) time by using two boundary searches:

    1. Left Boundary: Find the first occurance of the target 
    2. Right Boundary: Find the first eleemnt greater thant he target (upper bound).
    3. Count: Right Boundary - Left Boundary 

Use Case 2: Scattered Clusters (Unsorted array)

If the array is unsorted, we use a HashMap or linear tracking to find all start / end positions of the repetitions.


```python
def find_left_boundary(arr, target):
    """Finds the first occurance index of target in SORTED array."""
    left, right = 0, len(arr)
    while left < right:
        mid = left + (right -left) // 2
        if arr[mid] >= target:
            right = mid 
        else:
            left = mid + 1
    return left 

def find_right_boundary(arr, target):
    """Finds the idnex of the first element GREATER than target in a sorted array"""
    left, right = 0, len(arr)
    while left < right:
        mid = left + (right - left) // 2
        if arr[mid] > target:
            right = mid
        else: 
            left = mid + 1
    return left          

def analyze_sorted_occurances(arr, target):
    """Calculates start, end, and count in O(log N) time for sorted array"""
    start = find_left_boundary(arr, target)

    # check if the target actually exists in the array 
    if start >= len(arr) or arr[start] != target:
        return {"start_pos": -1, "end_pos": -1, "total_count": 0}
    end = find_right_boundary(arr, target) - 1
    count = (end - start) + 1
    return {"start_pos": start, "end_pos": end, "total_count": count}

def find_scattered_clusters(arr, target):
    """Finds intervals of repetitions in an UNSORTED array in O(N) time"""
    intervals = [] 
    in_cluster = False 
    start_idx = -1 

    for i, val in enumerate(arr):
        if val == target:
            if not in_cluster:
                start_idx = i 
                in_cluster = True 
            else:
                if in_cluster:
                    intervals.append((start_idx, i - 1))
                    in_cluster = False
    # Catch a cluster if it extends to the very end of the array
    if in_cluster:
        intervals.append((start_idx, len(arr) - 1))
        
    return intervals

## 
arr = [1, 2, 3, 3, 3, 3, 3, 4, 5, 6, 8, 9]
target_val = 3
left_b = find_left_boundary(arr, target_val)
print (f"The left boundary of {target_val} in the array is at index {left_b}")

right_b = find_right_boundary(arr, target_val)
print (f"The right boundary of {target_val} in the array is at index {right_b}")

result = analyze_sorted_occurances(arr, target_val)
print(f"Array: {arr}")
print(f"Target: {target_val}")
if result["total_count"] > 0:
    print(f"Positions: From index {result['start_pos']} to {result['end_pos']}")
else:
    print(f"Target {target_val} not found!")
print(f"Total Repetitions: {result['total_count']}\n")


# --- Test Case 2: Your Unsorted Array with Scattered Clusters ---
unsorted_arr = [1, 2, 3, 3, 4, 5, 6, 3, 3, 3, 8, 9, 0]
print("--- Use Case 2: Unsorted Array Scattered Clusters ---")
cluster_intervals = find_scattered_clusters(unsorted_arr, target_val)
print(f"Array: {unsorted_arr}")
print(f"Target: {target_val}")
print("Occurs in the following [start_index, end_index] intervals:")
for interval in cluster_intervals:
    length = interval[1] - interval[0] + 1
    print(f"  - Position {interval[0]} to {interval[1]} (Repeated {length} times)")

print () 
print (cluster_intervals)
```

Rotated Sorted Array (Boundary Search Variant )

A rorated sorted array is an array that was originally sorted in ascending order but then shifted (rotated) to the left or right by some pivot order 

For example, [1, 2, 3, 4, 5, 6, 7] rotated becomes [4, 5, 6, 7, 1, 2, 3]

Even though, the array as a whole is unsorted, it is partially sorted in to two monotonic segments [4, 5, 6, 7] and [1, 2, 3]. We can use a modified binary search to find the inflection point (the boundary where it frops from a high number to a low number) or a target element in O(logN) time. 

The rule of Thumb

If you plsit a sorted array in half, at least one side is always perfetly sorted

    If arr[left] <= arr[mid], the left side is sorted.

    If arr[mid] <= arr[right], the right side is sorted 
    

```python
def search_rotated_array(arr, target):
    left, right = 0, len(arr) -1 
    while left <= right:
        mid = left + (right - left) // 2 

        if arr[mid] == target:
            return mid 

        # Check if the left half is normally sorted 
        if arr[left] <= arr[mid]:
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else: 
                left = mid + 1

        #otherwise the right half must be sorted 
        else: 
            # target lies within the sorted half
            if arr[mid] < target <= arr[right]:
                left = mid + 1 
            else: 
                right = mid + 1 

        return -1

# --- Quick Test ---
rotated_arr = [4, 5, 6, 7, 1, 2, 3]
print(f"Index of 2: {search_rotated_array(rotated_arr, 2)}") # Output: 5
print(f"Index of 6: {search_rotated_array(rotated_arr, 6)}") # Output: 2
```

4. Worked example: rank retrieved documents

Problem: Sort documents by descending relevance score, then ascending latency. Preserve input order when both fields tie.

```python
documents = [
    {"id": "A", "score": 0.91, "latency": 120},
    {"id": "B", "score": 0.95, "latency": 180},
    {"id": "C", "score": 0.95, "latency": 90},
    {"id": "D", "score": 0.91, "latency": 120},
]

ranked = sorted(
    documents,
    key=lambda doc: (-doc["score"], doc["latency"])
)

print([doc["id"] for doc in ranked])
# ['C', 'B', 'A', 'D']
```

Python compares tuple keys from left to right:
1. Negative score puts larger scores first.
2. Latency breaks score ties.
3. Stability preserves A before D.

```python
# Creates a new list; the original order is retained in documents.
ranked = sorted(documents, key=lambda doc: doc["score"])

# Modifies the original list and returns None.
documents.sort(key=lambda doc: doc["score"])
```

Complexity: O(n log n) time in the worst case. sorted() also creates a new list.
