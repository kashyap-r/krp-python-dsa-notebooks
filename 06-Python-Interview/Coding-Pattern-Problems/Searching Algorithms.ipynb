Linear Search Algorithm

Given an array arr[] of n integers and an interger element x, find whether element x is present in the array. Return the index of the first occurance of x in the array, or -1 if it doesnt exist.

Input: arr = [1, 2, 3, 4, 5, -2], x = 3 
Output: 2


```python
def linear_search(arr, key):
    n = len(arr)
    for i in range(n):
        if arr[i] == key:
            return i
    return -1

## usage
arr = [1, 2, 3, 4, 5, -2]
key = -4
position = linear_search(arr, key)
if position == -1:
    print(f"The integer {key} not found int array!")
else: 
    print (f"The integer {key} was found at position {position}")
```


Binary Search 

Binary Search is a searching algorithm that operates on a sorted or monotonic search space, repeatedly dividing it into halves to find a target value or optimal answer in logarithmic time O(log N).

Conditions to apply Binary Search:
    - The input list must be sorted 
    - Access to any element of data structure should take a constant time.

Algorithm 
Step 1: Divide the search space into two halves by finding the middle index "mid"
Step 2: Compare the middle of the search space with the search "key"
Step 3: If the key is found in the middle, the process is terminated 
Step 4: If the key is not found in the middle, choose which half will be used as the next search space. 
    -> If the key is smaller than the middle, then the left side is used for next search 
    -> if the key is larger than the middle, then the right side is used as the next search space.

Step 5: Continue this process until the key is found or the total search space is exhausted. 



```python
# Iterative algorithm 

def binary_search(arr, key):
    low = 0 
    high = len(arr) -1 
    
    """   
    The middle element mid is calculated as 
    mid = ( low + high ) // 2
    Alternatively, for very large datasets, mid can also be calculated as..
    mid = low + (high - low) // 2
    """
    while low < high:
        mid = ( low + high ) // 2
        print (low, mid, high, arr[mid], key)

        if arr[mid] == key:
            return mid 
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1 
    return -1 

## Usage
arr = [11, 12, 13, 14, 15]
key = 4
print (f"low, mid, high, arr[mid], key")
bs = binary_search(arr, key)

print ("Position:", bs)
```

Complexity Analysis
Time: O(log n) i.e time is the function of the length of the array, since with every iteration the search space i,e, the length of the array reduces by half

Space: O(1)

Recursive Approach in implmenting the Binary Search 

Create a recursive function and compare the mid of search space with the key. 

Based on the result either return the index where the key is found or call the recursive function for the next search space.


```python
def binary_search_rec(arr, low, high, key):
    # if high >= low:
    # if low <= high:

        mid = low + (high - low) // 2

        if arr[mid] == key:
            return mid 

        elif arr[mid] > key:
            return binary_search_rec(arr, low, mid - 1, key)
        else: 
            return binary_search_rec(arr, mid + 1, high, key)
    # else: 
        return -1 

if __name__ == '__main__':
    arr = [2, 3, 4, 10, 40]
    x = 10
    
    result = binary_search_rec(arr, 0, len(arr)-1, x)
    
    if result != -1:
        print("Element is present at index", result)
    else:
        print("Element is not present in array")
```

4. Worked example: rank retrieved documents

Problem: Sort documents by descending relevance score, then ascending latency. Preserve input order when both fields tie.

```python
documents = [
    {"id": "A", "score": 0.91, "latency": 120},
    {"id": "B", "score": 0.95, "latency": 180},
    {"id": "C", "score": 0.95, "latency": 90},
    {"id": "D", "score": 0.91, "latency": 120},
]

ranked = sorted(
    documents,
    key=lambda doc: (-doc["score"], doc["latency"])
)

print([doc["id"] for doc in ranked])
# ['C', 'B', 'A', 'D']
```

Python compares tuple keys from left to right:
1. Negative score puts larger scores first.
2. Latency breaks score ties.
3. Stability preserves A before D.

```python
# Creates a new list; the original order is retained in documents.
ranked = sorted(documents, key=lambda doc: doc["score"])

# Modifies the original list and returns None.
documents.sort(key=lambda doc: doc["score"])
```
