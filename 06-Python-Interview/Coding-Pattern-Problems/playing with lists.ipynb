**The difference: sort() VS sorted()**

The main difference is that .sort() modifies the original list, while sorted() leaves the original list alone and returns a brand new list.

```python
numbers = [3, 7, 1, 9, 5]
print (numbers)
result = numbers.sort()
print (numbers)
print (result)

arr = [6, 2, 8, 4, 0]
print (arr)
new_list = sorted(arr)
print (arr)
print (new_list)
```

**The Difference: reverse() VS reversed()**

Similar to sorting, .reverse() flips the original list in-place, while reversed() leaves the original list alone and returns an iterator object.


```python
letters = ['a', 'b', 'c']
print (letters)
result = letters.reverse() 
print (letters)
print (result)

letters = ['a', 'b', 'c','d']
res= reversed(letters)
print (letters)
print (res)
print (list(res))
```

5. Display only the diagonal elements from L-R and similarly from R-L -> Done

```python
# print only the diagonal elements 
mat = [[0 for _ in range(3)] for _ in range(3)]
print (mat)

mat = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
# from left to R 
for i in range(len(mat)):
    for j in range(len(mat[i])):
        if i == j:
            print (mat[i][j])
# Time O(n^2)
print () 
# from R to L 
for i in range(len(mat)):
    print (mat[i][-1-i])

# Calculate the width of the matrix
n = len(mat)
# Extract into a list
right_to_left_diagonal = [mat[i][n - 1 - i] for i in range(n)]
print (right_to_left_diagonal)
left_to_right_diagonal = [mat[i][i] for i in range(n)]
print(left_to_right_diagonal)
# Time : O(2n) = O(n) 
```
