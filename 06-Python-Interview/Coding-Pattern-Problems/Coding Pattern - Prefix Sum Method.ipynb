## Prefix Sum 

Core Idea 

A Prefix Sum array stores cumulative sums so that range queries can be answered in O(1) time. 

Given: 
    arr= [a0, a1, a2, a3, ... , an-1]

We build: 
    prefix[i] = a0 + a1 +a2 + a3 + ... + ai 

So: 
    prefix[i] = prefix[i-1] + arr[i]

Why is this powerful?

Without prefix sum:
    * Range Sum query -> O(n)

With prefix sum:
    * Range Sum query -> O(1) 

sum(L, R) = prefix[R] - prefix[L-1]

This is a preprocessing for fast queries pattern.


# Basic Example 
# Input

arr = [2, 4, 1, 6, 3]

# Build Prefix
prefix[0] = 2

prefix[1] = 2 + 4 = 6

prefix[2] = 6 + 1 = 7

prefix[3] = 7 + 6 = 13

prefix[4] = 13 + 3 = 16

Query Example:
Sum from index 1 to 3:
= prefix[3] - prefix[0]
= 13 - 2 
= 11 

Correct value: 4 + 1 + 6


**When Should You Use Prefix Sum?**

Use it when:

Multiple range queries

    - Sum of subarrays 

    - Count of elements in range 

    - Frequency in intervals

Subarray sum problems

    - Count subarrays = K

    - Longest subarray sum = K

    - Subarray divisible by K 

2D grid problems 

    - Matrix Sum Problems 

    - Image Smoothing 

    - Heatmap aggregation 

Difference Array (range updates)

Probability & Ml Use cases

Worked Example #1: Range Sum Query 

Problem
Given array and multiple queries (L, R), return sum.

Real-world Context:
Log volume per minute. Query total logs below time windows.

Approach
1. Build prefix array 
2. For each query:
    sum = prefix[R] - prefix[L-1]

```python
# Naive solution 

nums = [2, 4, 1, 7, 3]
queries = [(0, 2), (1, 3), (2, 4)]
# Answers: [7, 12, 11]

def range_sums (nums, queries):
    return [sum(nums[L:R+1]) for L, R in queries]

# Each query takes O(R − L + 1) time. For q queries, worst-case time is O(qn). 
# Slicing also creates a temporary list.
print (range_sums(nums, queries))
```

```python
# Precompute cumulative sums once

nums = [2, 4, 1, 7, 3]
queries = [(0, 2), (1, 3), (2, 4)]

def range_sums(nums, queries):
    prefix = [0]

    for value in nums:
        prefix.append(prefix[-1] + value)
    return [prefix[R + 1] - prefix[L] for L, R in queries]

# Preprocessing: O(n). Each query: O(1).
# Total time: O(n + q). Extra space: O(n), plus O(q) for the returned answers.
# Negative values and single-element ranges work too. Queries must satisfy 0 <= L <= R < len(nums).
```

```python
# Prefix-sum solution using explicit indices, without slicing or comprehensions

def range_sums(nums, queries):
    prefix = [0] * (len(nums) + 1)

    # for i in range(len(nums)):
    #     prefix[i+1] = prefix[i] + nums[i]

    for i in range(len(nums)):
        prefix[i] = prefix[i-1] + nums[i]
    # print (prefix)
    # print () 
    # print (ps)

    result = []
    for L,R in queries:
        result.append(prefix[R] - prefix[L-1])

    return result

# usage 
nums = [2, 4, 1, 7, 3]
queries = [(0, 2), (1, 3), (2, 4)]

print (range_sums(nums, queries))
```

```python
# Another way 
def build_prefix(nums):
    prefix = [0] * (len(nums) + 1)

    for i, num in enumerate(nums):
        prefix[i + 1] = prefix[i] + num

    return prefix


def range_sum(prefix, left, right):
    return prefix[right + 1] - prefix[left]


nums = [3, 2, 5, 1, 6]

prefix = build_prefix(nums)

print(prefix)
print(range_sum(prefix, 1, 3))
```

| Operation | Time | Space |
|---|---|---|
| Construct prefix | O(n) | O(n) |
| Range query | O(1) | O(1) auxiliary |
| Q range queries | O(n + Q) | O(n) |

```python
class NumArray:
    def __init__(self, nums):
        self.prefix = [0] * len(nums)
        self.prefix[0] = nums[0]

        for i in range(1, len(nums)):
            self.prefix[i] = self.prefix[i-1] + nums[i]

    def sumRange(self, left, right):
        if left == 0:
            return self.prefix[right]
        return self.prefix[right] - self.prefix[left-1]


"""
Build: O(n)
Query: O(1)
"""

# arr = [1, 2, 3, 4, 5]
arr = [-1]
soln = NumArray(arr)
# print (soln.sumRange(1, 3))
print (soln.sumRange(0, 0))

# print (sum(arr[1:4]))
    
```

Worked Example #2: Find Pivot Index 
Leetcode #724

Problem
Find index where left sum = right sum 

Aproach
1. Compute total sum 
2. Maintan running left sum 
3. Check 

    left_sum == total - left_sum - arr[i]


```python
def pivot_index(nums):
    total = sum(nums)
    left_sum = 0 

    for i in range(len(nums)):
        if left_sum == total - left_sum - nums[i]:
            return i 
        left_sum += nums[i]
    return -1 

```
