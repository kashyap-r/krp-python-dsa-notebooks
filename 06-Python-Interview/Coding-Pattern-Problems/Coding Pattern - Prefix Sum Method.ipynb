## Prefix Sum 

Core Idea 

A Prefix Sum array stores cumulative sums so that range queries can be answered in O(1) time. 

Given: 
    arr= [a0, a1, a2, a3, ... , an-1]

We build: 
    prefix[i] = a0 + a1 +a2 + a3 + ... + ai 

So: 
    prefix[i] = prefix[i-1] + arr[i]

Why is this powerful?

Without prefix sum:
    * Range Sum query -> O(n)

With prefix sum:
    * Range Sum query -> O(1) 

sum(L, R) = prefix[R] - prefix[L-1]

This is a preprocessing for fast queries pattern.


# Basic Example 
# Input
arr = [2, 4, 1, 6, 3]

# Build Prefix
prefix[0] = 2
prefix[1] = 2 + 4 = 6
prefix[2] = 6 + 1 = 7
prefix[3] = 7 + 6 = 13
prefix[4] = 13 + 3 = 16

Query Example:
Sum from index 1 to 3:
= prefix[3] - prefix[0]
= 13 - 2 
= 11 

Correct value: 4 + 1 + 6


**When Should You Use Prefix Sum?**

Use it when:

Multiple range queries
    - Sum of subarrays 
    - Count of elements in range 
    - Frequency in intervals

Subarray sum problems
    - Count subarrays = K
    - Longest subarray sum = K
    - Subarray divisible by K 

2D grid problems 
    - Matrix Sym Problems 
    - Image Smoorthing 
    - Heatmap aggregation 

Difference Array (range updates)

Probability & Ml Use cases 
    - Cumulative ditribution 
    - Histogram prefixing 
    - Rolling wondow metrics 
    - Feature aggregation by time bucket 


Variants of Prefix Sum

2D Prefix Sum

Worked Example #1: Range Sum Query 

Problem
Given array and multiple queries (L, R), return sum.

Real-world Context:
Log volume per minute. Query total logs below time windows.

Approach
1. Build prefix array 
2. For each query:
    sum = prefix[R] - prefix[L-1]

```python
class NumArray:
    def __init__(self, nums):
        self.prefix = [0] * len(nums)
        self.prefix[0] = nums[0]

        for i in range(1, len(nums)):
            self.prefix[i] = self.prefix[i-1] + nums[i]

    def sumRange(self, left, right):
        if left == 0:
            return self.prefix[right]
        return self.prefix[right] - self.prefix[left-1]


"""
Build: O(n)
Query: O(1)
"""

arr = [1, 2, 3, 4, 5]
soln = NumArray(arr)

print (soln.sumRange(1, 3))

print (sum(arr[1:4]))
    
```
