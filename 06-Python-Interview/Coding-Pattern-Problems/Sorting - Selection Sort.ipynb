1. Selection Sort 

- Comparison based sorting algorithm 
- sorts by repeatedly selecting the smallest or largest element from the unsorted portion andswapping it with the first unsorted element

Steps: 
1. Find the smallest element and swap it with the first element. This way we get the smallest element as its correct position. 

2. Then find the smallest among the remaining elements (or second smallest) and swap it with second element

3. We keep doingthis until we get all the element moved to the correct position. 


```python
def selection_sort(arr):
    n = len(arr)

    for i in range(n-1):
        min_idx = i 

        for j in range(i+1, n):

            # to sort in ascending order 
            if arr[j] < arr[min_idx]:
            # to sort in descending order 
            # if arr[j] > arr[min_idx]:
                min_idx = j 

        arr[i], arr[min_idx] = arr[min_idx], arr[i]


## Usage
if __name__ == "__main__":
    arr = [64, 25, -12, 22, -11, 0]

    selection_sort(arr)
    print (arr)
```

Complexities

Time- O(n^2)
    - Outer loop to select an element of the array one by one - O(n)
    - Inner loop to compare the element with every other lement of the array - O(n)
    - Overall complexity in a for loop = O(n) * O(n) = O(n^2)

Auxiliary Space: O(1) as the only extra memory used for temporary variable.
