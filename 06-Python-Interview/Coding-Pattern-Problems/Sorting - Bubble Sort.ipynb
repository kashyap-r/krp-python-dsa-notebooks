2. Bubble Sort 

- again, a comparison based sort 
- just repeatedly swap the adjacent element if they're in the wron order. 

- Sorts the array using multiple passes. 
- After the first pass, the maximum goes to end (its correct position). 
- After the second pass, the second largest goes to second last position 
- In every pass, process only those that have already not moved to the correct position


```python
def bubble_sort(arr):
    n = len (arr)
    swapped = False
    for i in range(n-1):
        # Why n - 1 - i? 
        # After i passes, the last i elements are already in their final positions.
        for j in range(n - 1 - i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if swapped == False:            
            break

# usage
arr = [5, 3, 1, 4, 6, 7, 8, 9]
bubble_sort(arr)
print(arr)  # [1, 3, 4, 5]    
```
