Exchange Sort or Interchange sort

```python
def interchange_sort(arr):
    n = len(arr)
    for i in range(n-1):
        for j in range(i+1, n):
            if arr[i] > arr[j]:
                arr[i], arr[j] = arr[j], arr[i]

# usage 
arr = [5, 6, 1, 3]

interchange_sort(arr)
print (arr)
```
