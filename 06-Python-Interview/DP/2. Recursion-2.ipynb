#### String Reversal 

```python
str = 'Kash'

print (str[1:])

str[::-1]
```

```python
def reverseString(str):
    if len(str) < 1:
        return str
    else:
        return reverseString(str[1:]) + str[0]


print (reverseString('Kash'))
```

```python
dtr = 'Kash'

for i in range(len(dtr)-1, -1, -1):
    print(dtr[i], end=' ')
```
