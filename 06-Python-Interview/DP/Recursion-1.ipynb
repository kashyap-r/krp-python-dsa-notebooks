#### Recursion 

Recursion is a programming technique where a function calls itself either directly or indirectly to solve a problem by breaking it into smaller, simpler subproblems

Recursive Function 

def recrusive_function(parameters):
    if base_case_condition :
        return base_result 
    else:
        return recursive_function(modified_parameters)

Recursive function 
 - Base Case: The stopping condition that prevents infinite recursion 
 - Recursive Case: The part of the function where it calls itself with modified parameters

 

```python
# Example: find the factorial of a number 

def factorial(n):
    if n == 0:
        return 1 
    
    else:
        return n * factorial(n-1)


print (factorial(5))
```

```python
# Calculate the nth fibonacci number 
# fibonacci number: where each number is the sum of two preceeding numbers, starting from 0 and 1 #  
# ex. 0, 1, 1, 2, 3, 5, 8

def fibo(n):
    if n == 0:
        return 0 
    elif n == 1:
        return 1 
    else:
        return fibo(n-1) + fibo(n-2)
    
print (fibo(10))
```

Explanation: 

Base Cases: If n == 0, the function returns 0. If n == 1, the function returns 1. These two cases are necessary to stop the recursion.

Recursive Case: function calls itself twice with decrements of n (i.e., fibonacci(n-1) and fibonacci(n-2)), summing results of these calls.

Types of Recursion:

Recursion can be broadly classified into two types: tail recursion and non-tail recursion. The main difference between them is related to what happens after the recursive call. 

    Tail Recursion: 
        The recursive call is the last thing the function does, so nothing happens after it returns. 

    Non-Tail Recursion: 
        The function does more work after the recursive call returns, so it cant be optimized into a loop. 



```python
def tail_factorial(n, acc=1):
    if n == 0:
        return acc 
    else:
        return tail_factorial(n-1, acc * n)
    

def nontail_factorial(n):
    if n == 0:
        return 1 
    else:
        return n * nontail_factorial(n-1)


print (tail_factorial(5))
print (nontail_factorial(5))
```

Explanation:

def tail_fact(n, acc=1): - Defines a tail-recursive factorial function with an accumulator acc to store intermediate results.

if n == 0: return acc - Base case: when n reaches 0, return the accumulated result.

return tail_fact(n-1, acc * n) - Tail-recursive call: multiplies acc by n before the call, so no extra work is left after recursion.

def nontail_fact(n): - Defines a non-tail-recursive factorial function.

if n == 0: return 1 - Base case: when n == 0, return 1 (factorial of 0 is 1).

return n * nontail_fact(n-1) - Non-tail call: multiplication happens after the recursive call returns, so more work remains after recursion.

```python
def function(n):
    print(f"Entering function with n={n}")
    if n == 4:
        print(f"Base case reached with n={n}, returning {n}")
        return n
    else:
        result = 2 * function(n+1)
        print(f"Returning {result} for n={n}")
        return result

print("Final result:", function(2))
```
