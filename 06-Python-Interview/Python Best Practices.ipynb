

### Arrays

1. In python, slicing creates new lists, so a more efficient recursion style for arrays uses indices

2. Avoid using insert(0, x) and pop(0) for a list-based stack: they shift other elements and cost O(n).

3. Reversing an array 
Better to use reversed(array_name)

fruits = ["apple", "banana", "cherry", "dates"]
for fruit in reversed(fruits):
    print (fruit, end=' ')
print ()
print (fruits)

#### Using fruits[::-1] will create a brand new array/ list in memory. 
#### If the list contains millions of items, slicig will consume significant memory and time

reversed() avoids this completely by simply changing the direction the index pointer reads the existing data.


### Recursion: When to decide to go for recursion 

1. The problem is natually defined in terms of smaller subproblems of the same type
    - "answer for n = combone answers for n-1 and something"
    - "answer for this tree = combine answers for left subtree and right subtree"
    - "answer for this range [l, r] = combine asnwers for [l, m] and [m+1, r]

2. You're working with recursive data structures:
    * Trees (binary tree, n-ary tree, segment tree)
    * Graphs (DFS, Connected components)
    * Linked Lists (often ok but python depth can be an issue for very long ones)

3. You need to explore multiple ppossibilities / paths (Backtracking)
    - All subsets, permutations, combinations
    - paths in a maze 
    - N-Queens, sudoku, et

4. Divide-and-conquer Algorithms: 
    * Merge sort, quick sort
    * Binary search

5. The resursive solution is signifcantly clearer than the iterative one 
    If the iterative solution requires manually simulating a call stack with a list, the recursive version is often clearer

###. Recrusion is not ideal when:

1. The logic is just a simple loop over a range or list.
    Example: summing an array, finding max, counting elements: a for loop is simpler and avoids overhead.

2. The recursion depth will be very large (≫ 1000) in Python.  
    - Python has a default recursion depth limit (~1000).
    - Deep recursion can hit RecursionError.
    - For very deep problems, you usually want an iterative approach.

3. Performance is critical and function call overhead matters.  
    - Tight loops with millions of iterations are often faster as plain loops than recursive calls.

4. Tail recursion:  
    - Some languages (like Haskell, Scala) optimize tail recursion.
    - Python does not do tail-call optimization, so “tail-recursive style” doesn’t save stack frames here.

### Trade-offs: Recursion vs iteration 

Pros of Recursion 
    * Expressiveness / clarity for certain patterns:
        - Tree Traversals 
        - DFS / Backtracking
        - Divide and Conquer 

    * Closer to mathematical definitions (e.g. factorial, fibonacci, recurrences)
    * Less boilerplate for managing your owm stack / state:
        Python call stack is effectively the stack data structure

Cons of Recursion: 
1. Stack Depth Limit
    - Risk of "RecursionError: maximum recursion depth exceeded" for deep recursions. 
    - You can tweak the limit using sys.setrecursionlimit, but that’s usually a code smell unless you know what you’re doing and control the environment.

2. Overhead 
    - Function calls are more expensive than simple loop iterations.
    - Each call adds a new frame to the call stack (memory usage).

3. Harder to debug for beginners  
    - If the base case or recursive step is slightly wrong, you can get infinite recursion or weird behavior.
    - Need practice reading stack traces and tracing calls.

4. Complexity doesn’t magically improve  
    - Recursion doesn’t change big-O by itself.
    - E.g., recursive Fibonacci naïvely is worse (exponential) than iterative Fibonacci.
    - For overlapping subproblems, you need memoization or DP to avoid recomputation.

A practical “decision rule” you can use
When you see a problem:
    1. Can I express it as:
        - result for input = combine results of the same problem on smaller inputs?
        - Is there an obvious base case?
        If no → probably iterative or some other paradigm.

    2. Does it involve:
        - Trees / graphs / nested structures?
        - Backtracking / exploring all possibilities?
        - Divide-and-conquer?
        If yes → recursion is usually a strong candidate.

    3. Will the recursion depth stay reasonable (≲ 1000 in Python)?
        - If yes → recursion is fine.
        - If no → either:
            - change to an iterative + stack approach, or
            - use a different algorithm.
            
    4. Is the recursive solution simpler to read and maintain than the iterative version?
        - If yes → go recursive.
        - If no → stick to loops.
