### Add a timeout to a retrying tool call

Multi-file code review. An agent calls a read-only search service. The service sometimes raises TimeoutError. An AI assistant drafted a retry helper, but a reviewer reports that it retries some failures it should surface immediately and ignores the configured timeout.

Contract
- max_attempts includes the first call. A value of 2 permits two calls total. A value below 1 must raise ValueError before the service is called.
- Pass timeout_seconds to the service on every attempt. It is a per-attempt timeout.
- Retry only TimeoutError. If the final attempt times out, propagate that exception. Any other exception must propagate immediately.
- Return the first successful result, including an empty list.


```python
def run_with_retry(operation, *, max_attempts):
    for _ in range(max_attempts + 1):
        try:
            return operation()
        except Exception:
            continue
    return None
```

Assume the client accepts client.search(query, timeout=...). You can test with a fake client.

Your task: Patch both files and write one focused test. A good starting case is a client that times out on its first call and returns [] on its second. With max_attempts=2 and timeout_seconds=5, the result should be [], and the client should have received timeout=5 exactly twice.
