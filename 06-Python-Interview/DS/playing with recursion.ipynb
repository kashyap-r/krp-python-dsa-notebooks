How you can practice (concrete path) Recursion

Implement both iterative and recursive versions for:
    - sum of array
    - max of array
    - reverse array
    - factorial, Fibonacci (with and without memoization)

Move to index-based recursion:
    - sum using i
    - print / process subarrays using (start, end) indices
    (this connects directly to your earlier “all subarrays” triple-loop snippet).

Do branching recursion:
    - subsets / power set
    - permutations
    - all binary strings of length n

Do tree problems:
    - depth / height of tree
    - sum of nodes
    - count leaves
    - inorder / preorder / postorder traversals

For each problem, ask:
    - Is recursion clearer?
    - How deep can the recursion go?
    - What’s the time / space complexity?
