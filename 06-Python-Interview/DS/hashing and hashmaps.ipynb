### Hashing 

    = converting data -> Fixed size numeric value (hash code) 

i.e. it is a technique where ... 
input (key) -> hash funtion -> hash value (index)

A hash function:
    - Takes input of ay size
    - Produces a fixed size integer
    - ideallyy distributes values uniformly 

Real-world uses of Hashing 
1. Password storage
2. Caching Systems (Redis, Memcached)
3. Databases & Indexing 
4. Load Balancing 
    Ensures:
        - Same user goes to same server
        - Even traffic distribution
    This technique is called Consistent Hashing.



```python
hash('apple')
```

### Hash Map 

    - is simply a data structure that uses Hashing. 
    - dict in python is a hash map, is a data structure that stores key-value pairs using hashing. 

What Kind of Coding problems do hash maps solve ?

Pattern 1: Frequency Counting 
            Ex. Find Duplicates in an array 
Pattern 2: Two Sum Problem 

Pattern 3: Grouping Problems 
            Ex. Group Anagrams 

Pattern 4: Sub array problems 
            Ex. Find if subarray sum equals k

Pattern 5: Caching / Memoization 
            Ex. DynamiC Programming 






```python
print (hash('apple'))
print (hash('pplea'))
```

"What is hashing?"
    Hashing is a technique that maps keys to fixed-size values using a hash function, enabling constant-time average complexity lookups.

"What is a hash map?"
    A hash map is a key-value data structure built on top of hashing that provides O(1) average insert, delete, and search operations.


```python
print(hash("apple"))
print(hash("banana"))
print(hash(100))
print(hash((1, 2, 3)))
```
