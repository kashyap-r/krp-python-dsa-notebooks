### Graphs

1. #### What is a Graph?
- is a non-linear data structure.
- is a collection of nodes connected by edges used to represent relationships between entities

The limitation of tree is, it can only represent hierarchical data. For situations where nodes or vertices are randomly connected with each other other, we use Graph.

Example situations where we use graph data structure are, a social network, a computer network, a network of locations used in GPS and many more examples where different nodes or vertices are connected without any hierarchic or constraint on structure.

**Components of Graph Data Structure**

A Graph consists of:

- Vertices (Nodes) -> Entities 
    - Can be labelled or unlabelled 
- Edges (Connections) -> Relationships between entities
    - Used to connect any two nodes in any possible way
    - Can be ordered pair of nodes in a directed graph 
    - every edge can be labelled / unlabelled

Mathematically: 

    G = (V, E)

Where 
- V = set of vertices
- E = set of edges 

Example:

A ---- B
|      |
C ---- D

Here:
- Nodes -> A, B, C, D
- Edges -> (A, B), (A, C), (B, D), (C, D)

#### Representation of Graph Data Structure:

There are multiple ways to store a graph: The following are the most common representations.

    Adjacency Matrix
    Adjacency List

Adjacency Matrix Representation of Graph Data Structure:

here, the graph is stored in a 2D matrix where rows and columns denote vertices. Each entry in the matrix represents the weight of the edge between those vertices. 

matrix[i][j] = 1 if there is an edge between vertex i and vertex j.
matrix[i][j] = 0 if there is no edge.


```python
# matrix representation for undirected graph

def createUndiGraph_matrix(V, edges):
    mat = [[0 for _ in range (V)] for _ in range(V)]

    for it in edges:
        u = it[0]
        v = it[1]
        mat[u][v] = 1
        mat[v][u] = 1
    return mat

if __name__ == "__main__":
    V = 3

    # List of edges (u, v)
    edges = [[0, 1], [0, 2], [1, 2]]

    # Build the graph using edges
    mat = createUndiGraph_matrix(V, edges)

    print("Adjacency Matrix Representation:")
    for i in range(V):
        for j in range(V):
            print(mat[i][j], end=" ")
        print()
```

```python
# Matrix representation of directed graph 
def create_diGraph_matrix(V, edges):
    mat = [[0 for _ in range(V)] for _ in range(V)]

    for item in edges:
        u = item[0]
        v = item[1]
        mat[u][v] = 1

    return mat

if __name__ == "__main__":
    V = 3

    # List of edges (u, v)
    edges = [[1, 0], [2, 0], [1, 2]]

    # Build the graph using edges
    mat = create_diGraph_matrix(V, edges)

    print("Adjacency Matrix Representation:")
    for i in range(V):
        for j in range(V):
            print(mat[i][j], end=" ")
        print()
    
```

Adjacency List Representation of Graph:

This graph is represented as a collection of array lists. There is an array of pointer which points to the edges connected to that vertex. 


```python
# Representation of an undirected graph 

def createGraph(V, edges):
    adj = [[] for _ in range(V)]

    # add each edge to the adjacency list 
    for it in edges: 
        u = it[0]
        v = it[1]
        adj[u].append(v)
        # for an undirected graph
        adj[v].append(u)
    return adj

if __name__ == "__main__":
    V = 3 

    # List of edges
    edges = [[0, 1], [0, 2], [1, 2]]

    # Build the graph using edges
    adj = createGraph(V, edges)

    print ("Adjacency list representation of the graph:")
    for i in range(V):
        print (f"{i}: ", end=" ")
        for j in adj[i]:
            print (j, end=" ")
        print ()
```

```python
# representation of a directed graph 

def createGraph(V, edges):
    adj = [[] for _ in range(V)]

    for it in edges:
        u = it[0]
        v = it[1]
        adj[u].append(v)

    return adj

if __name__ == "__main__":
    V = 3

    # List of edges (u, v)
    edges = [[1, 0], [1, 2], [2, 0]]

    # Build the graph using edges
    adj = createGraph(V, edges)

    print("Adjacency List Representation:")
    for i in range(V):
        
        # Print the vertex
        print(f"{i}:", end=" ")
        for j in adj[i]:
            
            # Print its adjacent
            print(j, end=" ")
        print()
```

### Core Graph Operations

1. Graph Traversal 

Tewo fundamental algorithms.

##### BFS (Breadth First Search) / Level Order Traversal

- Uses queue

Good For:
- Shortest path (unweighted)
- level traversal 

Note: Traversal Technique of Graph are used to visit all the vertices of a graph systematically. These methods help to explore the graph completely and are useful for solving many graph-related problems.


```python
from collections import deque 

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            print (node)
            visited.add(node)
            queue.extend(graph[node])

# Complexity: O(V + E)
```

##### Depth First Search (DFS)

Uses Stack or Recursion


```python
def dfs(graph, node, visited):
    if node in visited:
        return 

    visited.add(node)
    print (node)

    for neighbor in graph[node]:
        dfs(graph, neighbor, visited)
```
