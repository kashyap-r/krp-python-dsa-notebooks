Singly Linked List
    - is a chain of nodes
    - Each node has 
        - data 
        - pointer to the next node 

Advantages of a SLL 
1. Dynamic size (no pre-allocation)
    - Grows and shrinks at runtime 
    - No ersizing cost like python lists 

Why this matters
    - When data volume is unknown or highly variable 
    - Example
        Streaming logs, incremental ingestion pipelines 

2. O(1) insertion / deletion (when pointer is known)
    - Insert at head -> O(1)
    - Delete after a given node -> O(1)

📌 Example
Maintaining a queue of tasks where items are constantly added/removed

3. No shifting cost 
    - Arrays reqire shifting elements ->. O(n)
    - Linked LIst just relinks pointers 

    Example
        - Undo stacks, operation histories

Disadvantages of a Singly Linked List
    ❌ 1️⃣ No random access
    - list[i] → ❌ not possible
    - Must traverse from head → O(n)
    📌 Pain point
    Anything that smells like indexing or slicing

    ❌ 2️⃣ Poor cache locality
    - Nodes scattered in memory
    - CPU cache misses → slower in practice
    📌 Why Python lists often win despite worse big-O

    ❌ 3️⃣ Reverse traversal is impossible
    - No prev pointer
    - Reversing list costs O(n)

    ❌ 4️⃣ Extra memory per element
    - Pointer overhead per node
    - Python object overhead makes this worse

    ❌ 5️⃣ Complex to debug and maintain
    - Pointer bugs = nightmare
    - One bad link → memory leak / infinite loop

Real-Life Applications (Where SLLs Shine)
✅ 1️⃣ Implementing stacks and queues
    - Stack → push/pop at head
    - Queue → enqueue at tail, dequeue at head
    📌 Common in interview questions and low-level systems

✅ 2️⃣ Undo / Redo functionality (basic)
    - Each operation points to the next
    - Lightweight history tracking
    📌 Text editors, simple command history

✅ 3️⃣ Graph adjacency lists
    - Each node maintains a linked list of neighbors
    📌 Especially in C/C++ graph implementations

✅ 4️⃣ Memory-constrained embedded systems
    - Predictable pointer operations
    - No costly reallocation
    📌 Firmware, RTOS task lists

✅ 5️⃣ Streaming / sequential data processing
    - Process nodes one by one
    - No need for random access
    📌 Log processing, event streams

Real-Life Use-Cases Where Singly Linked Lists Fail Hard
❌ 1️⃣ Search-heavy systems
    - Databases
    - Search engines
    - Analytics queries
    Why?
    O(n) lookup kills performance.
📌 Use instead: hash tables, trees, indexes

❌ 2️⃣ ML / Data Engineering workloads
    - Feature vectors
    - Tensor operations
    - Batch processing
    📌 Reality check
    NumPy, Pandas, PyTorch → contiguous memory is king

❌ 3️⃣ UI rendering and timelines
    - Need fast back/forward traversal
    - Random jumps (scrolling)
    📌 Doubly linked lists or arrays win

❌ 4️⃣ High-frequency trading / low-latency systems
    - Pointer chasing = cache misses
    - Unpredictable memory access
    📌 Arrays + pre-allocation dominate here

❌ 5️⃣ Python production code (most of the time)
    - Python’s list is:
    - Cache-friendly
    - Amortized O(1) append
    - Faster in practice
    📌 That’s why Python doesn’t expose a native linked list

Interview Gold: When to Use vs Avoid
    ✔ Use Singly Linked List when:
    - Insert/delete is frequent
    - Access is sequential
    - Size is unknown
    - Memory fragmentation is acceptable
    ✘ Avoid when:
    - Random access is required
    - Performance is critical
    - Data is numeric or vectorized
    - You’re in Python and list works
    

```python
class Node:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next


class SinglyLinkedList:
    def __init__(self, values=()):
        self.head = None
        curr = None

        for value in values:
            if self.head is None:
                self.head = Node(value)
                curr = self.head
            else:
                curr.next = Node(value)
                curr = curr.next

    # ---------- Traversal and display ----------

    def __iter__(self):
        curr = self.head
        while curr:
            yield curr.value
            curr = curr.next

    def display(self):
        print(" -> ".join(map(str, self)) + " -> None"
              if self.head else "None")

    # ---------- Search ----------

    def search(self, value):
        """Return the first matching position, or -1."""
        for position, item in enumerate(self):
            if item == value:
                return position
        return -1

    def node_at(self, position):
        """Return the node at a position; also used by insert/delete."""
        if position < 0:
            raise IndexError("Position out of bounds")

        curr = self.head
        for _ in range(position):
            if curr is None:
                raise IndexError("Position out of bounds")
            curr = curr.next

        if curr is None:
            raise IndexError("Position out of bounds")

        return curr

    # ---------- Insertion ----------

    def prepend(self, value):
        """Insert at the beginning."""
        self.head = Node(value, self.head)

    def append(self, value):
        """Insert at the end."""
        if self.head is None:
            self.prepend(value)
            return

        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = Node(value)

    def insert(self, position, value):
        """Insert at position, shifting existing elements right."""
        if position == 0:
            self.prepend(value)
            return

        prev = self.node_at(position - 1)
        prev.next = Node(value, prev.next)

    def insert_after(self, position, value):
        """Insert after the node at position."""
        curr = self.node_at(position)
        curr.next = Node(value, curr.next)

    # ---------- Deletion ----------

    def delete_first(self):
        """Remove and return the first value."""
        if self.head is None:
            raise IndexError("List is empty")

        value = self.head.value
        self.head = self.head.next
        return value

    def delete_last(self):
        """Remove and return the last value."""
        if self.head is None or self.head.next is None:
            return self.delete_first()

        curr = self.head
        while curr.next.next:
            curr = curr.next

        value = curr.next.value
        curr.next = None
        return value

    def delete_at(self, position):
        """Remove and return the value at position."""
        if position == 0:
            return self.delete_first()

        prev = self.node_at(position - 1)
        if prev.next is None:
            raise IndexError("Position out of bounds")

        value = prev.next.value
        prev.next = prev.next.next
        return value

    def remove(self, value):
        """Delete the first occurrence; return whether it was found."""
        if self.head is None:
            return False

        if self.head.value == value:
            self.head = self.head.next
            return True

        curr = self.head
        while curr.next:
            if curr.next.value == value:
                curr.next = curr.next.next
                return True
            curr = curr.next

        return False

## Usage

ll = SinglyLinkedList([10, 20, 30])
ll.display()                  # 10 -> 20 -> 30 -> None

ll.prepend(5)
ll.append(40)
ll.insert(2, 15)
ll.insert_after(3, 25)
ll.display()                  # 5 -> 10 -> 15 -> 20 -> 25 -> 30 -> 40 -> None

print(ll.search(25))           # 4
print(ll.search(99))           # -1
print(ll.node_at(3).value)     # 20

# Traversal: __iter__ lets us use an ordinary for loop.
for value in ll:
    print(value, end=" ")      # 5 10 15 20 25 30 40
print()

print(ll.delete_first())       # 5
print(ll.delete_last())        # 40
print(ll.delete_at(1))         # 15
print(ll.remove(25))           # True
print(ll.remove(99))           # False
ll.display()                  # 10 -> 20 -> 30 -> None

print(list(ll))                # [10, 20, 30]

empty = SinglyLinkedList()
empty.display()               # None
empty.append(7)
print(empty.delete_last())     # 7
empty.display()               # None
```

```python
class Node: 
    def __init__(self, data, next=None):
        self.data = data 
        self.next = next 

class LinkedList:
    def __init__(self):
        self.head = None 

    # insert at beginning 
    def insert_at_beginning(self, data):
        node = Node(data)
        node.next = self.head
        self.head = node 

    def insert_at_end(self, data):
        node= Node(data)
        if self.head is None:
            self.head = node 
            return 
        curr = self.head 
        while curr.next:
            curr = curr.next
        curr.next = node

    def insert_at_position(self, position, data):
        if position < 0:
            return None
        if position == 0:
            self.insert_at_beginning(data)
            return 
        curr = self.head 
        for _ in range(position-1):
            if curr is None:
                raise IndexError("Position Out of range")
            curr = curr.next
        if curr is None:
            raise IndexError("Position Out of range")
        node = Node(data) 
        node.next = curr.next
        curr.next = node 

    def insert_after_value(self, target, data):
        curr = self.head 
        while curr and curr.data != target:
            curr = curr.next 
        if curr is None:
            raise IndexError("Target Not Found")
        node = Node(data)
        node.next = curr.next
        curr.next = node 

    # ---------- Deletion ----------
    def delete_at_beginning(self):
        if self.head is None:
            raise IndexError("List is empty")
        self.head = self.head.next
 
    def delete_at_end(self):
        if self.head is None:
            raise IndexError("List is empty")
        if self.head.next is None:
            self.head = None
            return
        cur = self.head
        while cur.next.next:
            cur = cur.next
        cur.next = None
 
    def delete_at_position(self, pos):
        if self.head is None:
            raise IndexError("List is empty")
        if pos == 0:
            self.head = self.head.next
            return
        cur = self.head
        for _ in range(pos - 1):
            if cur.next is None:
                raise IndexError("Position out of range")
            cur = cur.next
        if cur.next is None:
            raise IndexError("Position out of range")
        cur.next = cur.next.next
 
    def delete_by_value(self, value):
        if self.head is None:
            return False
        if self.head.data == value:
            self.head = self.head.next
            return True
        cur = self.head
        while cur.next and cur.next.data != value:
            cur = cur.next
        if cur.next is None:
            return False
        cur.next = cur.next.next
        return True
 
    # ---------- Search ----------
    def search(self, value):
        """Return index of first match, or -1."""
        cur, i = self.head, 0
        while cur:
            if cur.data == value:
                return i
            cur = cur.next
            i += 1
        return -1
 
    # ---------- Traversal / Display ----------
    def length(self):
        count, cur = 0, self.head
        while cur:
            count += 1
            cur = cur.next
        return count
 
    def display(self):
        parts, cur = [], self.head
        while cur:
            parts.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(parts) + " -> None")
 
 
if __name__ == "__main__":
    ll = LinkedList()
    ll.insert_at_end(10)
    ll.insert_at_end(20)
    ll.insert_at_beginning(5)
    ll.insert_at_position(2, 15)
    ll.insert_after_value(20, 25)
    ll.display()                    # 5 -> 10 -> 15 -> 20 -> 25 -> None
 
    print(ll.search(20))            # 3
    print(ll.search(99))            # -1
 
    ll.delete_at_beginning()
    ll.delete_at_end()
    ll.delete_at_position(1)
    ll.delete_by_value(10)
    ll.display()                    # 20 -> None
    print(ll.length())              # 1
```
