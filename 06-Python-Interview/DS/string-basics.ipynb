Textual data in Python is handled with str objects, or strings. 

Strings are immuatble sequences of unicode code points. String literals are written in variety of ways: 

Single quotes: 'allows embedded "double" quotes'
Double quotes: "allows embedded 'single' quotes"
Triple quotes: '''Three single quotes''', """Three double quotes"""

Note: Triple quoted strings may span multiple lines 

Reference: 

https://docs.python.org/3/builtins/stdtypes.html#string-methods
