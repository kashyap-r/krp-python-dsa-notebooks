### 1. enumerate() - Keep Track of Indexes 

Used to iterate over an iterable while keeping track of both the index and the value. 
It returns pairs in the form (index, element)

```python
languages = ["Python", "TypeScript", "ShellScript", "Java"]

fruits = ['apple', 'banana', 'cherry', 'date']
for language in languages:
    print(language, end=' ')
print ()
print ()

for idx, value in enumerate(fruits):
    print(idx, value)
```

### zip (*iterables, strict=False)

* Used for dot product, matrix transpositions, or aligning layers and weights. 
* Add strict=True arguemment to prevent silent debugging nightmares by raising a ValueError if the iterable are of unequal length

Zip is used to combinetwo or more iterables like lists, tuples, strings, dictionaries into a single iterator of tuples. 


```python
fruits = ['apple', 'banana', 'cherry', 'dates']
slno = ['A1', 'A2', 'A3', 'A4']
qty = [100, 200, 300, 250]

prod = zip(slno, fruits)
for i, x in prod:
    print (i, x)
print () 

prod_qty = zip(fruits, qty)
for prod, qty in prod_qty:
    print (prod, qty)
```

### Lambda Functions in Python

A lambda function is a small, anonymous function that is defined without a name. 

Unlike standard functions created using the def keyword, lambda functions are written in a single line of code using the lambda keyword.

They are typically used for short, simple operations that are only needed for a brief moment, often as arguments passed into higher-order functions.

Syntax:

    lambda arguements: expression

• lambda: The keyword that tells Python you are creating an anonymous function.
• arguments: A comma-separated list of inputs (similar to arguments in a standard function).
• expression: A single piece of code that is executed and automatically returned. You cannot include complex logic like loops or multiple lines here.

Example: 

A regular Function 
def square(x):
    return x ** 2

A Lambda Function 

square = lambda x: x ** 2


Key Limitations

While convenient, lambda functions have structural boundaries:

• Single Expression Only: They cannot contain branches like if/elif/else blocks (though they can use simple conditional ternary operators like x if x > 0 else y).

• No Side Effects: They cannot contain loops (for or while) or statements like pass, assert, or raise.

• Readability: If the logic gets too complex, using a lambda function makes code harder to read. In those cases, a standard def function is always preferred.



Common Use Cases

Lambda functions are incredibly useful when paired with built-in Python functions like map(), filter(), and sorted().

1. Sorting with a Custom Key

You can use a lambda function to tell the sorted() function exactly how to sort complex data, such as a list of tuples.


```python
pairs = [('Jackfruit', 1.5), ("Apple", 1.2), ("Banana", 0.5), ("Cherry", 2.5)]
print (pairs)
print()
# Sort the items by price (the second element in the tuple)
sorted_pairs = sorted (pairs, key=lambda item: item[1])
print (sorted_pairs)
print ()
sorted_pairs = sorted (pairs, key=lambda item: item[0])
print (sorted_pairs)
```


2. Transforming Data with map()

The map() function applies a function to all items in an input list.

```python
numbers = [1, 2, 3, 4]

# Double every number in the list 
doubled = list(map(lambda x: x**2, numbers))

print (doubled)
```


3. Filtering Data with filter()

The filter() function extracts items from a list that meet a specific condition.

```python
numbers = [1, 2, 3, 4, 5, 6]

# Keep only the even numbers
evens = list(filter(lambda x: x%2 == 0, numbers))
print (evens)
```

4. Conditional Logic (Ternary Operator)

You can use a simple if-else structure inside a lambda function, provided it stays on a single line.

```python
check_num = lambda x: "positive" if x >= 0 else "negative"

print (check_num(5))
print (check_num(-4))
```

5. Sorting Dictionaries by a Specific Key

Lambda functions are highly effective when you need to sort a list of dictionaries by a specific internal value.

```python
users = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age":24},
    {"name": "Charlie", "age": 33}
]

sorted_users = sorted(users, key=lambda user: user["age"])
print (sorted_users)
```


6. String Manipulation with map()

You can clean or format data lists instantly using lambda functions.

```python
names = ["  alice ", "BOB", " chArlie  "]

# Strip whitespace and capitalize the first letter
clean_names = list(map(lambda name: name.strip().capitalize(), names))
print(clean_names)
# Output: ['Alice', 'Bob', 'Charlie']
```

### Shallow Copy vs deep Copy 

In Python, the difference between shallow copy and deep copy only becomes relevant when you are dealing with compound objects (objects that contain other objects, like nested lists or diectionaries)

copy module handles these operations:

    copy.copy(x) creates a shallow copy

    copy.deepcopy(x) creates deep copy 


The 3 levels of copying 

Assingment (=): Creates new reference to the exact same object. Modifying to level element afffects both. Modifying nexted element affects both.

Shallow Copy: Copies the outer container, but references the original inner objects. MOdifying top level element affects only the copy. Modifying nested element affects both. 

Deep Copy: Recursively copies the container and everything inside it. Modifying the top level element affects only the copy. Modifying nested element affects only the copy. 


```python
# Shallow Copy (copy.copy)
"""
A shallow copy constructs a new collection object and then populates it with references to the child objects found in the original.
• Analogy: Imagine copying a folder full of shortcuts. You have a brand new folder, but clicking the shortcuts still opens the exact same original files.
• Other ways to shallow copy: 
    Slicing a list (list[:]), 
    calling list.copy(), dict.copy(), or 
    using type constructors like list(original_list)
"""

import copy 
original = [[1, 2, 3], [4, 5, 6]]
shallow = copy.copy(original)

# 1. Modifying the top level structure 
shallow.append([7, 8, 9])
print ("Originl after append: ", original)
print ("Shallow after append: ", shallow)

# 2. Modifying a nested mutable object 
shallow[0][0] = 99
# shallow[1][1] = 'x'
print ("Originl after nested change: ", original)
print ("Shallow after nested change: ", shallow)
```

```python
# Deep Copy 

""" A deep copy constructs a new collection object and then recursively inserts copies of the child objects found in the original. The two objects become entirely independent in memory.
    • Analogy: Imagine copying a folder and making actual duplicates of every document inside it. Changing a document in the new folder has zero impact on the old folder.
"""

import copy

original = [[1, 2, 3], [4, 5, 6]]
deep = copy.deepcopy(original)

# Modifying a nested mutable object
deep[0][0] = 100
print("Original:", original)  
print("Deep Copy:", deep)    
```


Key Exceptions & Behaviors

1. Immutable Objects: If a container holds only immutable objects (like integers, strings, or tuples), a shallow copy is completely sufficient. Python optimization prevents duplicating immutable structures because they cannot be changed anyway.

2. Performance Trade-off: Deep copies are significantly slower and consume more memory because they must crawl through the entire object graph to replicate every element. Use them only when completely necessary.

3. Circular References: copy.deepcopy() maintains a memo dictionary during its execution. This prevents infinite loops if your custom classes contain circular references (e.g., Object A references Object B, which references Object A).
