### Enumerate 

used to iterate over an iterable while keeping track of both the index and the value. 

It returns pairs int eh form (index, element)

```python
a = ["Python", "TypeScript", "ShellScript", "Java"]

# for i in a:
#     print(i)

for i, v in enumerate(a):
    print(i, v)
```

The below code conversta the enumerate object into a list of tuples with each tuple containing the index and coresponding element from the list. 


```python
a = ["A", "B", "C"]
r = list(enumerate(a))
print (r)
```

```python
a = ["Python", "TypeScript", "ShellScript", "Java"]
e = enumerate(a)
print (next(e)) 
print (next(e))
```

#### Enumerating dictionary items 


```python
d = {"a": 10, "b": 20}

for i, (k, v) in enumerate(d.items()):
    print(i, k, v)
```
