```python
# Python dictionary comprehension 

sq = {x: x**2 for x in range(1, 6)}
print (sq)

# Creating a dictionary from two lists 

keys = ['a', 'b', 'c','d', 'e']
values = [10, 20, 30, 40, 50]

d = {k:v for (k, v) in zip(keys, values)}
print (d)


# Using fromkeys() Method 

d = dict.fromkeys(range(5), True)
print (d)

b = {x.upper(): x*3 for x in 'coding'}
print (d)

for x in 'coding':
    print (x, x*3)

c = {fruit: len(fruit) for fruit in ['apple', 'banana', 'cherry']}
print (c)

#  dictionary comprehension with conditional statements
d = {x: x**3 for x in range(10) if x**3 % 4 == 0}
print (d)

# Nested Dictionary Comprehension 
s = "GF"
res = {x: {y: x + y for y in s} for x in s}
print(res)
```

### 1. Word frequency 

Input: 
words = ["apple", "banana", "apple", "orange", "cherry", "apple", "orange"]

Output:


Assumptions: 
1. Fruit names are only letters (do I need to check)
2. convert everything to lowercase to ensure accuracy in comparison
3. Any limit on the size of the dataset


```python
words = ["apple", "banana", "apple", "orange", "cherry", "apple", "orange"]

# naive solution 
freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print (freq)
print ()
# optimized solution, not using counter
new_freq = {}
for word in words:
    new_freq[word] = new_freq.get(word, 0) + 1
print (new_freq)
```
