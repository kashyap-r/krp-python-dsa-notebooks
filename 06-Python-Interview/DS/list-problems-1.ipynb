Generating every subarray 

```python
arr = [1, 2, 3, 4]
n = len(arr)
count = 0
for i in range(n):
    for j in range(i+1, n+1):
        print (arr[i:j])
        count += 1
print (count)
```

```python
# Another approach to generate all the subarrays - using List Comprehension
arr = [1, 2, 3, 4]
n = len(arr)

all_subarrays = [arr[i:j] for i in range(n) for j in range(i+1, n+1)]
print (all_subarrays)
len(all_subarrays)
```

```python
# Usign a generator function (memory efficient)

def generate_subarrays_lazily(arr):
    n = len(arr)
    for start in range(n):
        for end in range(start, n):
            yield arr[start:end+1]

# Usage 
arr= [1, 2, 3, 4]
for sub in generate_subarrays_lazily(arr):
    print (sub)
```


How Generators Work: "Lazy Evaluation"

A generator function (using the yield keyword) does not create a list. It does not calculate anything in advance. Instead, it returns a tiny iterator object that sits completely empty.
When you loop over a generator:
1. It runs the loops until it hits the first yield statement.
2. It pauses its execution and hands one single subarray out to your loop.
3. Your loop processes it (e.g., prints it, sums it, checks a condition).
4. The old subarray is instantly thrown away by Python's garbage collector.
5. The generator wakes back up, calculates the next single subarray, and repeats.


| Metric | List Comprehension / Master List | Generator Function (`yield`) |
| :--- | :--- | :--- |
| **Memory Used (RAM)** | **Explodes exponentially** ($O(N^3)$) | **Stays perfectly flat** ($O(N)$ or $O(1)$) |
| **Initial Start Speed** | **Slow** (Must wait for all subarrays to be created) | **Instant** (Calculates the first item immediately) |
| **Risk of Crashing** | **High** (Triggers `MemoryError` on larger lists) | **Zero** (Can handle arrays of infinite length safely) |

Sum of a subarray


```python
# Brute force method

arr = [1, 2, 3, 4]
n = len(arr)
for start in range(n):
    for end in range(start+1, n+1):
        subarray = arr[start:end]
        print (subarray, sum(subarray))
```
