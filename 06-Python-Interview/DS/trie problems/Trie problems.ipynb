## Tries 

Hash tables can do: 
    search(word) -> O(1)

But cannot efficiently support prefix queries.

Example problem: 
    Find all words starting with "app"

Hash table approach: Scan the entire dictionary - O(N) 

Trie approach: Traverse Prefix -> O(prefix length)



### When will a problem require a Trie?
In interviews, keywords in problem statements are huge hints.

| Problem Hint | Likely Data Structure |
|---|---|
Prefix search | Trie |
Dictionary lookup | Trie |
Autocomplete system | Trie |
Word suggestions | Trie |
Prefix matching | Trie |
Spell checker | Trie |
Word search in board | Trie |
Many strings with common prefix | Trie |

### Problem Type 1 - Implement Trie 

Problem Statement: 
Design a Trie with functions: 
    insert(word)
    search(word)
    startsWith(prefix)

Expected Behaviour:

    Trie trie = new Trie()

    trie.insert("apple")
    trie.search("apple") → True
    trie.search("app") → False
    trie.startsWith("app") → True

This tests:
    - Trie structure
    - Insert logic
    - Prefix traversal
