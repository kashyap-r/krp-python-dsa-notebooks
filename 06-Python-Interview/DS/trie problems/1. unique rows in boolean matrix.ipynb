## Trie based Problems

Problem 1: Unique Rows in Boolean Matrix

Given a binary marix your task is to find all unique rows of the given matrix in order of their appearance in the matrix

Example:

Input: mat[][] = [  [1, 1, 0, 1], 
                    [1, 0, 0, 1], 
                    [1, 1, 0, 1]]

Output: [   [1, 1, 0, 1], 
            [1, 0, 0, 1]]

Explanation: 

The two unique rows are R1: [1, 1, 0, 1] and R2: [1, 0, 0, 1]
As R1 appeared at row-0 ad R2 appeared at row-1 on the resulting list R1 is kept before R2


### Approach 1
### The Hashing Approach (Using Python set)

The most straight forward alternative is to use a "set" to track row uniqueness. 

Since lists are mutable and cannot be hashed, we convert each row into a tuple nefore adding it to the set. 


```python
def unique_rows_set(matrix):
    seen = set() 
    unique_result = []

    for row in matrix: 
        # Convert to tuple so it can be hashed and stored in a set
        row_tuple = tuple(row)
        if row_tuple not in seen:
            seen.add(row_tuple)
            unique_result.append(row)

    return unique_result
""" 
Time: O(R * C)
Auxiliary Space: R * C
"""


```

```python
class TrieNode: 
    def __init__(self):
        self.children: list[TrieNode | None] = [None, None]

class BinaryTrie: 
    def __init__(self):
        self.root = TrieNode() 

    def insertRow(self, row):
        """
         
        """
        node = self.root
        is_new_row = False

        for bit in row:
            if node.children[bit] is None:
                node.children[bit] = TrieNode()
                is_new_row = True
            node = node.children[bit]
        return is_new_row

def find_unique_rows(matrix):    
    trie = BinaryTrie()
    unique_rows = []

    for row in matrix:
        if trie.insertRow(row):
            unique_rows.append(row)
    return unique_rows

if __name__ == "__main__":
    matrix = [[1, 1, 0, 1],
              [1, 0, 0, 1], 
              [1, 1, 0, 1]]

    print (find_unique_rows(matrix))
    print ()
    matrix = [[1, 1, 0, 1]]
    print (find_unique_rows(matrix))
    print ()
    matrix = [[1, 1, 0, 1], [1, 1, 0, 1]]
    print (find_unique_rows(matrix))
    print ()
    matrix = [[-1, -1, -0, 1]]
    print (find_unique_rows(matrix))
    print ()
```
