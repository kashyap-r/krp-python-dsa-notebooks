Problem Statement:
    A workflow begins with a document awaiting processing. Write a function called process_document(state) that returns an updated state without modifying the original dictionary.

Facts/ Given: 
Input is a document 
initial state:  

To-Do: Once the document is done processing, return the updated state without modifying the original dictionary

Input: 
state = {
    "document": "employee_policy.pdf",
    "status": "pending",
    "attempts": 0,
    "processed": False
}

Requirement: 

    1. increment attempts by 1 
    2. change status to 'completed'
    3. change the processed to "True"
    4. preserve the document field
    5. Return a new dictionary 

Expected result: 
state = {
    "document": "employee_policy.pdf",
    "status": "completed",
    "attempts": 1,
    "processed": True
}

```python
state = {
    "document": "employee_policy.pdf",
    "status": "pending",
    "attempts": 0,
    "processed": False
}

def process_documents(status):
    return {
            **state,             
            "status": "completed",
            "attempts": state["attempts"] + 1,
            "processed": True}

new_state = process_documents(state)

print ("Original state:", state)
print ("Updated state: ", new_state)
```
