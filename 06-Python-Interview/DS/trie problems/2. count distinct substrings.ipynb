Count Distinct Substrings

Given a string s consisting of lowercase English letters, find the total number of distinct non-empty substrings of s.

Note: A substring is a contiguous sequence of characters within a string. Two substrings are considered distinct if they have different character sequences, regardless of their positions in the string.

Input: s = "ababa"
Output: 9
Explanation: 

All distinct substrings of "ababa" are: 

"a", "b", "ab", "ba", "aba", "bab", "abab", "baba", "ababa".


```python
# Naive method

s = "ababa"
n = len(s)
result = set()

for i in range(n): 
    for j in range (i+1, n+1):
        result.add(s[i:j])

print (result)
```

```python
## Using Trie 

class TrieNode: 
    def __init__(self):
        self.children = {} 

def count_distinct_substrings(s):
    root = TrieNode() 
    distinct_count = 0 
    n = len(s) 
    result = []

    for i in range(n):
        node = root 
        for j in range(i, n): 
            ch = s[j]
            if ch not in node.children: 
                node.children[ch] = TrieNode()                 
                distinct_count += 1
            node = node.children[ch] 
            result.append(node.children)
            
    return distinct_count

input = "ababa"

print (count_distinct_substrings(input))
print (result)
```
