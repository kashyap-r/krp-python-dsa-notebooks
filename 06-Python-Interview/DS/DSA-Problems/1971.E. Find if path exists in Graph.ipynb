1971. Find if Path Exists in Graph
Easy

There is a bi-directional graph with n vertices, where each vertex is labeled from 0 to n - 1 (inclusive). The edges in the graph are represented as a 2D integer array edges, where each edges[i] = [ui, vi] denotes a bi-directional edge between vertex ui and vertex vi. Every vertex pair is connected by at most one edge, and no vertex has an edge to itself.

You want to determine if there is a valid path that exists from vertex source to vertex destination.

Given edges and the integers n, source, and destination, return true if there is a valid path from source to destination, or false otherwise.

Input: n = 3, edges = [[0,1],[1,2],[2,0]], source = 0, destination = 2
Output: true
Explanation: There are two paths from vertex 0 to vertex 2:
- 0 → 1 → 2
- 0 → 2

Input: n = 6, edges = [[0,1],[0,2],[3,5],[5,4],[4,3]], source = 0, destination = 5
Output: false
Explanation: There is no path from vertex 0 to vertex 5.

Analysis:
We need to determine whether source and destination are connected in how many ways.

Naive Approach
- Would be to traverse by repeatedly scanning the edge list 

Start at source vertex 0. For each vertex we visist scan the entire edge list to find its neighbours. Use a visited set to avoid cycles. 

This solution works but finding nieghbors repeatedly is expensive 

Upto V vertices visted
scan all E edges for each vertex 
means time: O(V*E + V) 
Extra space O(V)

We can improve this by building an adjacency list, so each vertex's neighbours are directly accessible. 

2. Optimal Approach: Adjacency list + Iterative DFS 

For the first example:
edges = [[0, 1], [1, 2], [2, 0]]

graph = [
    [1, 2],  # Neighbors of vertex 0
    [0, 2],  # Neighbors of vertex 1
    [1, 0]   # Neighbors of vertex 2
]

Use these data structures:
| Structure | Purpose |
|---|---|
| List of lists | Store neighbors of each vertex |
| Stack (`list`) | Track vertices to explore using DFS |
| Set | Track discovered vertices and prevent revisiting |

The algorithm:
1. Build the adjacency list, adding each edge in both directions.
2. Put source on the stack and mark it visited.
3. Pop a vertex. If it equals destination, return True.
4. Add its unvisited neighbors to the stack, marking them immediately.
5. If the stack becomes empty, return False.


```python
# Optimal Solution

def valid_path(n, edges, source, destination):
    if source == destination:
        return True 

    graph = [[] for _ in range(n)]

    # Build the undirected graph 
    for u, v in edges:
        graph[u].append(v)
        graph[v].append(u)

    stack = [source]
    visited = {source}

    while stack:
        node = stack.pop() 

        if node == destination:
            return True 

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)
    return False

    
n = 3
edges = [[0, 1], [1, 2], [2, 0]]
source = 0
destination = 2

print (valid_path(n, edges, source, 3))
```
