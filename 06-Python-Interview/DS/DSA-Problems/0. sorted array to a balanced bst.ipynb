### Sorted Array to a Balanced BST 

Given a sorted array arr[]. convert it into a Balanced BST. Return the root of the BST 

Note: A Balanced Binary Search Tree is a type of binary tree in which the difference between the heights of the left and rght sobtrees of every node is at most 1

Example 1: 

Input:  arr[] = [1, 5, 9, 14, 23, 27]
Output: [[9], [1, 23], [N, 5, 14, 27]]

Example 3: 
Input: arr[] = [10, 20, 30]
Output: [[20], [10, 30]]

The Approach: 

Given arr = [1, 5, 9, 14, 23, 27]

1. Find the middle element. The middle element is 14 (index 3). This becomes the root. 
2. Left Subtree: Repeat the process for the left half [1, 5, 9]. The middle is 5 which becomes the left child of 14 
3. Right Subtree: Rpeat the process for right half [23, 27]. The middle can be chosen as 27 (0r 23) which becomes the right child of 14

The optimal iterative soltion to building a BST from a Sorted Array 

```python

# Validting BST and Balance 
# Two Checks 
#   1. Is it a valid BST? (left values < root < right values)
#   2. Is it height balanced? (the height difference between 
#      left and right subtrees never exceeds 1)

def is_bst_valid_and_balanced(root):
    # helper the checkthe bst properties and calculate the height in one pass
    def check(node, min_val=float('-inf'), max_val=float('inf')):
        if not node:
            return True, 0 # is balanced, height

        # 1. validate BST value range constraint
        if not min_val < node.value < max_val:
            return False, 0 

        # 2. check the left and right subtrees recursively 
        left_valid, left_height = check(node.left, min_val, node.value)
        right_valid, right_height = check(node.right, node.value, max_val)

        # 3. Calculate heights and balance factor
        is_balanced = left_valid and right_valid and abs(left_height - right_height) <= 1
        current_height = max(left_height, right_height) + 1

        return is_balanced, current_height

    balanced, _ = check(root)
    return balanced

"""
Complexity Check for Both Variations:
• Time Complexity: O(N) because every value in the array is processed exactly once.
• Space Complexity: O(log N) as the maximum depth of the manual stack or recursive 
stack stays bounded by the height of the balanced tree.
"""

# --- Execution ---
arr = [1, 5, 9, 14, 23, 27]
root = sorted_array_to_bst_iterative(arr)

print(f"Is the constructed tree a perfectly balanced BST? -> {is_bst_valid_and_balanced(root)}")        
print () 
print ()

nums = [1, 12, 3, 4, 5, 6, 7, 8, 9]
tree1 = sorted_array_to_bst_iterative(nums)
print(f"Is the constructed tree a perfectly balanced BST? -> {is_bst_valid_and_balanced(tree1)}")        
print() 

nums = []
tree1 = sorted_array_to_bst_iterative(nums)
print(f"Is the constructed tree a perfectly balanced BST? -> {is_bst_valid_and_balanced(tree1)}") 

nums = [1, 1, 1, 1, 1]
tree1 = sorted_array_to_bst_iterative(nums)
print(f"Is the constructed tree a perfectly balanced BST? -> {is_bst_valid_and_balanced(tree1)}") 
```
