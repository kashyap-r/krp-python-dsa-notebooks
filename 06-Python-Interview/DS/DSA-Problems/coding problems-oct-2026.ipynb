**Problem #2.**

1.Easy. Two Sum
Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

Example 1:
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

Example 2:
Input: nums = [3,2,4], target = 6
Output: [1,2]

Example 3:
Input: nums = [3,3], target = 6
Output: [0,1]


```python
class Solution: 
    def twoSum(self, nums, target):
        # Dictionary to map: { number_value : its_index }
        seen = {}
        
        for i, num in enumerate(nums):
            complement = target - num
            
            # Check if the needed matching number has already been seen
            if complement in seen:
                return [seen[complement], i]
            
            # Otherwise, record the current number and its index
            seen[num] = i


if __name__=="__main__":
    nums = [2,7,11,15]
    target = 9
    twoSum = Solution() 
    d = twoSum.twoSum(nums, target)
    print (d)
```

"""
15.M. 3Sum
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] 
such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

"""

```python

# Using two pointer method
def find3Sum(arr):
    arr.sort()

    result = []

    for i in range(len(arr)-2):
        if i > 0 and arr[i] == nums[i-1]:
            continue
        # fix the first element, point the left pointer to the 2nd element
        left = i+1
        # second pointer top point to the last element
        right = len(arr)-1

        while left < right:
            total = arr[i] + arr[left] + arr[right]

            if total == 0:
                result.append([arr[i], arr[left], arr[right]])

                while left < right and arr[left] == arr[left+1]:
                    left+= 1 
                while left < right and arr[right] == arr[right-1]:
                    right -= 1
                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

if __name__ == "__main__":
    nums = [-1, 0, 1, 2, -1, -4]
    print (find3Sum(nums))
```

```python
# <!-- Using hashmap -->

def threeSum(nums):
    result = set()
    n = len(nums)

    for i in range(n):
        seen = {}
        for j in range(i + 1, n):
            complement = - (nums[i] + nums[j])
            if complement in seen:
                triplet = tuple(sorted([nums[i], nums[j], complement]))
                result.add(triplet)
            seen[nums[j]] = j

    return [list(triplet) for triplet in result]

if __name__ == "__main__":
    nums = [-1, 0, 1, 2, -1, -4]
    print (find3Sum(nums))
```

Input: 
    arr = [4, 2, 4, 1, 2, 7, 1]

Expected Output: [4, 2, 1,7]

Note: 
Retain the first occurance of each element


```python
arr = [4, 2, 4, 1, 2, 7, 1]

def remove_duplicates(arr):
    seen = set() 
    result = []

    for ele in arr:
        if not ele in seen:
            result.append(ele)
        seen.add(ele)
    return result

result = remove_duplicates(arr)
print (result)
```

Nested List Sum — Easy

Given [[1,2], [3,4], [5,6]], calculate the sum of all values using 
nested loops.

```python
def nested_sum(arr):
    total = 0 
    for i in arr:
        for j in i:
            total += j
    return total
arr = [[1,2], [3,4], [5,6]]
print (nested_sum(arr))

matrix = [
    [1, 2, 3],
    [4, 5],
    [6, 7, 8, 9]
]
print (nested_sum(matrix))

"""
Time: O(N) generally 
O(rows * cols) - when both are equal
"""
```

Given: 
matrix = [
    [3, 7, 1],

Return the largest value across all inner lists.

Expected Result: 
10 

Hints: 
Track one variable as the largest 
Think about negative only inputs 
do not initialize largest = 0 


```python
def largest_in_matrix(arr):
    # Find the maximum element of each row, then find the max of those maximums
    return max(max(row) for row in arr)

matrix = [
    [3, 7, 1],
    [8, 2],
    [4, 10, 6]
]

# print(largest_in_matrix(matrix)) # Output: 10
```

```python
def largest_in_matrix(arr):
    largest = arr[0][0]
    for row in arr:
        for element in row:
            if element > largest:
                largest = element
    return largest

matrix = [
    [3, 7, 1],
    [8, 2],
    [4, 10, 6]
]
print (largest_in_matrix(matrix))
```

Problem #18

Best Model from List of Tuples
Given: 

models = [
    ("model_a", 0.91),
    ("model_b", 0.95),
    ("model_c", 0.89)
]

Return: 
("model_b", 0.95)

Do not sort the whole list.
Hints
- Scan once.
- Maintain the best tuple or best score.
- Target complexity: O(n).


```python
models = [
    ("model_a", 0.91),
    ("model_b", 0.95),
    ("model_c", 0.89)
]

def get_finest_model(models):
    finest = models[0]
    for model, score in models:
        if score > finest[1]:
            finest = (model, score)
    return finest

# def get_bestest_model(models):
#     # Lambda tells max() to compare using the score (item[1])
#     return max(models, key=lambda item: item[1])

print (get_finest_model(models))
# print (get_bestest_model(models))

""" 

Complexity

• Time Complexity: O(N) because it scans the list exactly once.
• Space Complexity: O(1) constant space.

"""
```
