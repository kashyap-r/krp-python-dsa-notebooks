217. Contains Duplicate

Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

Input: nums = [1,2,3,1]
Output: true
Explanation:
    The element 1 occurs at the indices 0 and 3.

    You need the element and its indices, when there is a duplicate
"""

Constraints / Boundary Conditions/ (Edge) Test Cases:
1. Size of the array 

2. Size of the array elements

3. Whether the array elements can be positive/ negative integers, decimals, fractions, etc in which case 

4. whether to consider asbolute value for comparison?

Assumptions: 
1. That the array will have only numbers and not special characters or empty spaces

```python
from collections import defaultdict

class Solution:
    def contains_duplicates(self, arr): 
        n = len(arr)
        seen = set()

        for i in range (n):
            if arr[i] in seen:
                return True
            seen.add(arr[i])
        return False

    def get_duplicates(self, arr):
        n = len(arr)
        d = dict() 

        for i in range(n):
            if arr[i] not in d:
                d[arr[i]] = []
            d[arr[i]].append(i)
        return d
    
    # Awesoem solution
    def get_duplicates_2(self, arr):
        counts = {}
        for value in arr:
            counts[value] = counts.get(value, 0) + 1
        return {value: count for value, count in counts.items() if count > 1}

    def fetch_duplicates(self, arr):
        d = defaultdict(list)
        for i, element in enumerate(arr):
            d[element].append(i)
        return d


if __name__ == "__main__":
    arr = [1, 2, 2, 2.0, 3, 4, 5, 5]

    s = Solution()
    print (s.contains_duplicates(arr))

    # d = s.get_duplicates(arr)
    # for k, v in d.items():
    #     print (f"{k} is repeated {v} times")

    d = s.get_duplicates_2(arr)
    for k, v in d.items():
        print(f"{k} is repeated {v} times")

    
    # d = s.fetch_duplicates(arr)
    # print (type(d))
    # for k, v in d.items():
    #     print(f"{k} is repeated in indices: {v}")
```

**Problem #2.**

1.Easy. Two Sum
Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

Example 1:
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

Example 2:
Input: nums = [3,2,4], target = 6
Output: [1,2]

Example 3:
Input: nums = [3,3], target = 6
Output: [0,1]


```python
class Solution: 
    def twoSum(self, nums, target):
        # Dictionary to map: { number_value : its_index }
        seen = {}
        
        for i, num in enumerate(nums):
            complement = target - num
            
            # Check if the needed matching number has already been seen
            if complement in seen:
                return [seen[complement], i]
            
            # Otherwise, record the current number and its index
            seen[num] = i


if __name__=="__main__":
    nums = [2,7,11,15]
    target = 9
    twoSum = Solution() 
    d = twoSum.twoSum(nums, target)
    print (d)
```

```python
"""
15.M. 3Sum
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] 
such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

"""
# Using two pointer method
def find3Sum(arr):
    arr.sort()

    result = []

    for i in range(len(arr)-2):
        if i > 0 and arr[i] == nums[i-1]:
            continue
        # fix the first element, point the left pointer to the 2nd element
        left = i+1
        # second pointer top point to the last element
        right = len(arr)-1

        while left < right:
            total = arr[i] + arr[left] + arr[right]

            if total == 0:
                result.append([arr[i], arr[left], arr[right]])

                while left < right and arr[left] == arr[left+1]:
                    left+= 1 
                while left < right and arr[right] == arr[right-1]:
                    right -= 1
                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

if __name__ == "__main__":
    nums = [-1, 0, 1, 2, -1, -4]
    print (find3Sum(nums))
```

```python
# <!-- Using hashmap -->

def threeSum(nums):
    result = set()
    n = len(nums)

    for i in range(n):
        seen = {}
        for j in range(i + 1, n):
            complement = - (nums[i] + nums[j])
            if complement in seen:
                triplet = tuple(sorted([nums[i], nums[j], complement]))
                result.add(triplet)
            seen[nums[j]] = j

    return [list(triplet) for triplet in result]

if __name__ == "__main__":
    nums = [-1, 0, 1, 2, -1, -4]
    print (find3Sum(nums))
```
