207. Course Schedule
Medium


There are a total of numCourses courses you have to take, labeled from 0 to numCourses - 1. You are given an array prerequisites where prerequisites[i] = [ai, bi] indicates that you must take course bi first if you want to take course ai.

    For example, the pair [0, 1], indicates that to take course 0 you have to first take course 1.

Return true if you can finish all courses. Otherwise, return false.

 

Example 1:

Input: numCourses = 2, prerequisites = [[1,0]]
Output: true
Explanation: There are a total of 2 courses to take. 
To take course 1 you should have finished course 0. So it is possible.

Example 2:

Input: numCourses = 2, prerequisites = [[1,0],[0,1]]
Output: false
Explanation: There are a total of 2 courses to take. 
To take course 1 you should have finished course 0, and to take course 0 you should also have finished course 1. So it is impossible.

Hint 1
This problem is equivalent to finding if a cycle exists in a directed graph. If a cycle exists, no topological ordering exists and therefore it will be impossible to take all courses.

Hint 2
Topological Sort via DFS - A great tutorial explaining the basic concepts of Topological Sort.

Hint 3
Topological sort could also be done via BFS.


#### Anaylsis:

Given [course, prerequisite], determine whether every course can be completed.

numCourses = 2
prerequisites = [[1, 0]]

Complete 0 before 1.
Answer: True

Model this as a directed graph:
- Vertex: a course.
- Edge: prerequisite → course.
- A directed cycle makes it impossible to complete all courses.
Naive approach

Try every permutation of the courses. For each ordering, check whether every prerequisite appears before its dependent course.

Return True if any ordering works; otherwise return False.

For V courses and E dependencies:
- There are V! permutations.
- Building a position map and checking one ordering takes O(V + E).
- Time: O(V! × (V + E)).
This generates many invalid orders. Instead, we can select only courses that are currently ready.

Optimal approach: Kahn’s topological sort
| Structure | Purpose |
|---|---|
| Adjacency list | Courses that depend on each course |
| Indegree array | Number of outstanding prerequisites |
| Queue | Courses with zero outstanding prerequisites |
| Counter | Number of courses completed |


```python
from collections import deque

class Solution:
    def canFinish(
        self,
        numCourses: int,
        prerequisites: list[list[int]]
    ) -> bool:

        graph = [[] for _ in range(numCourses)]
        indegree = [0] * numCourses

        # Build edges: prerequisite → course.
        for course, prerequisite in prerequisites:
            graph[prerequisite].append(course)
            indegree[course] += 1

        # Start with every course that has no prerequisites.
        queue = deque(
            course
            for course in range(numCourses)
            if indegree[course] == 0
        )

        completed = 0

        while queue:
            course = queue.popleft()
            completed += 1

            for dependent in graph[course]:
                # One prerequisite has now been completed.
                indegree[dependent] -= 1

                # All prerequisites are now satisfied.
                if indegree[dependent] == 0:
                    queue.append(dependent)

        return completed == numCourses

```

```python
# naive Solution 
from itertools import permutations

def canFinishPermutations(numCourses, prerequisites):
    # 1. Generate all possible orderings (permutations) of the courses
    # There are V! total permutations.
    all_possible_orders = permutations(range(numCourses))
    
    for order in all_possible_orders:
        # 2. Build a position map for the current ordering: O(V)
        # e.g., if order is (1, 0, 2), position map is {1: 0, 0: 1, 2: 2}
        position = {course: idx for idx, course in enumerate(order)}
        
        # 3. Check whether every prerequisite appears before its dependent course: O(E)
        is_valid_order = True
        for dest, src in prerequisites:
            # If the prerequisite (src) appears AFTER the dependent course (dest),
            # this entire ordering is invalid.
            if position[src] >= position[dest]:
                is_valid_order = False
                break  # Stop checking this permutation early
        
        # 4. Return True if even a single ordering works
        if is_valid_order:
            print(f"Valid schedule found: {list(order)}")
            return True
            
    # If we tried all V! permutations and none worked, return False
    return False

# --- Example Usage ---
# Example 1: No cycles (Possible to finish)
# Course 1 depends on Course 0 (0 -> 1)
print("Example 1:")
print("Result:", canFinishPermutations(2, [[1, 0]])) 

print("\n" + "="*20 + "\n")

# Example 2: Has a cycle (Impossible to finish)
# Course 1 depends on 0, and Course 0 depends on 1 (0 <-> 1)
print("Example 2:")
print("Result:", canFinishPermutations(2, [[1, 0], [0, 1]]))
```

```python
# Naive Solution 

def canFinishNaive(numCourses, prerequisites):
    completed = set()
    
    # We loop up to numCourses times. 
    # In the worst case, we can only unlock exactly 1 course per full pass.
    for _ in range(numCourses):
        course_found_this_pass = False
        
        for course in range(numCourses):
            # If we've already completed this course, skip it
            if course in completed:
                continue
                
            # Naively scan the entire prerequisites list to see if
            # this specific course has any incomplete prerequisites
            has_unmet_prereq = False
            for dest, src in prerequisites:
                if dest == course and src not in completed:
                    has_unmet_prereq = True
                    break  # We cannot take this course yet
            
            # If all prerequisites for this course are met, we complete it!
            if not has_unmet_prereq:
                completed.add(course)
                course_found_this_pass = True
                break  # Break out to start a fresh pass with our updated completed set
                
        # If a full scan of all courses yields no progress, a cycle exists
        if not course_found_this_pass:
            break
            
    # If the total completed courses equals numCourses, it's possible to finish
    return len(completed) == numCourses

# --- Example Usage ---
# Example 1: No cycles (Possible to finish)
# Course 1 depends on Course 0 (0 -> 1)
print("Example 1:")
print("Result:", canFinishNaive(2, [[1, 0]])) 

print("\n" + "="*20 + "\n")

# Example 2: Has a cycle (Impossible to finish)
# Course 1 depends on 0, and Course 0 depends on 1 (0 <-> 1)
print("Example 2:")
print("Result:", canFinishNaive(2, [[1, 0], [0, 1]]))    
```
