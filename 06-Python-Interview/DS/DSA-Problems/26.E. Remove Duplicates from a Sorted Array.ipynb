Problem: Modify the array in place so that each distinct value appears once in the first k positions. Return k.


Method 1: The Naive Solution (Extra Memory + Copying)

nums = [1, 1, 2, 2, 2, 4]

# Return: 3
# Required prefix: nums[:3] == [1, 2, 4]
# Values beyond that prefix do not matter.

```python
nums = [1, 1, 2, 2, 2, 4]

def removeDuplicates_naive(nums):
    if not nums:
        return 0 

    # 1. COllect unique elements in order 
    unique_elements = []
    for num in nums:
        if not unique_elements or num != unique_elements[-1]:
            unique_elements.append(num)
    
    # 2. Overwrite the first k positions of the original array 
    k = len(unique_elements)
    nums[:k] = unique_elements

    return k

# --- Verification ---
nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = removeDuplicates_naive(nums)
print(f"k = {k}")           # Output: 5
print(f"Array: {nums[:k]}") # Output: [0, 1, 2, 3, 4]
nums
```


Why Interviewers Reject This

• Time Complexity: O(N) because it loops through the array of length N linearly.
• Space Complexity: O(N) because it allocates a new unique_elements array structure in memory. The problem explicitly states we must modify the array in-place with O(1) auxiliary memory space.


Optimal Solution: Two Pointer In-Place Approach 

Because the array is already sorted, all duplicate values are guaranteed to sit adjacent to each other. We can solve this with a single pass without using any extra memory by maintaining two pointers:
1. write_index: Points to where the next unique element should be written.
2. read_index: Scans through the list to find new distinct numbers.

The Logic

We initialize write_index = 1 because the first element (nums[0]) is always unique relative to itself. 

We start scanning with read_index at position 1. 

Every time nums[read_index] is different from nums[write_index - 1], we have discovered a brand new distinct number! We write it at write_index and bump that pointer forward.

```python
from typing import List

class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        i = 0 
        j = 1 
        while i < len(nums)-1:
            if nums[i] != nums[i+1]:
                nums[j] = nums[i+1]
                j += 1 
            i += 1
        return j
#
nums = [1, 1, 2]
soln = Solution()
result = soln.removeDuplicates(nums)
print (result)

nums = [0,0,1,1,1,2,2,3,3,4]
result = soln.removeDuplicates(nums)
print (result)
```
