# Insertion Sort Referesher

Insertion Sort: grow a sorted prefix 

For each element, shift larger elements to the right and insert in into the resulting gap

```python
def insertion_sort(nums):
    for i in range(1, len(nums)):
        current = nums[i]
        j = i - 1
        while j >= 0 and nums[j] > current:
            nums[j+1] = nums[j]
            j -= 1
        nums[j+1] = current 
    return nums

# 
arr = [22, 11, -12, 3, 0, 99]
print (insertion_sort(arr))
```

# Merge Sort refresher  

```python
# Merge sort using recursion 
def merge_sort(nums):
    if len(nums) <= 1:
        return nums[:]  # Return a new list even for the base case.

    mid = len(nums) // 2

    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])

    merged = []
    i = j = 0

    while i < len(left) and j < len(right):
        # Choosing left on equality preserves stability.
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1

    merged.extend(left[i:])
    merged.extend(right[j:])

    return merged

# --- Quick Test ---


rotated_arr = [4, 5, 6, 0, -7, 1, 2, 3]
print(f"Sorted array:\n{merge_sort(rotated_arr)}") 
```

For [7, 2, 6, 3]:
- Split into [7, 2] and [6, 3].
- Sort each half into [2, 7] and [3, 6].
- Repeatedly take the smaller front element: 2, 3, 6, 7.
Why O(n log n)? There are approximately log n splitting levels, and each level performs O(n) total merging work.

This implementation uses O(n) peak auxiliary space, plus O(log n) recursion depth. Slicing also creates additional allocation work.

Role connection: The merge operation is useful when combining already sorted results from multiple sources. For data larger than memory, external sorting sorts chunks and then merges them.

#### Quick Sort 

Partition, then sort the partitions 

Algorithm:

    1. Choose a pivot 
    2. Partition values into those smaller than, equal to, and larger than the pivot 
    3. Recursively sort the smaller and larger groups 
    4. Combine them

For [7, 2, 6, 3, 6], choosing pivot gives:

smaller = [2, 3]
equal = [6, 6]
larger = [7]


Balanced partitions give O(n log n) work. Repeatedly producing extremely unbalanced partitions can give O(n²).

Quickselect uses the same partition idea but continues only into the partition containing the desired rank. Its expected time is O(n), with O(n²) worst case.

Problem: Return the index of the target in an ascending sorted list, or -1 if absent

Solution: Binary Search

```python
def binary_search(arr, target):
    lo = 0 
    hi = len(arr)-1

    while lo <= hi:
        mid = lo + (hi - lo) // 2

        if arr[mid] == target: 
            return mid

        if arr[mid] < target:
            lo = mid + 1
        else: 
            hi = mid -1
    return -1

## Usage

# arr = [7, 2, 16, 3, 6] - unsorted array
arr = sorted([7, 6, 2, 16, 3, 2, 6])
target = 6 
print (arr)
result = binary_search(arr, target)
if result == -1:
    print ("Not Found")
else: 
    print ("Found at position ", result)
```

Find the absolute last (rightmost) occurence of a duplicate element in a sorted array

Two Patterns to the solution

Pattern A: The closed window [left, right] (Most intutive)

Here, when arr[mid] == target, we save mid as a temporary best answer (ans = mid) and continue searching to the right side (left = mid + 1) to see if a later copy exists. 

```python
def find_last_occurence(arr, target):
    left = 0 
    right = len(arr) -1
    ans = -1  

    while left <= right:
        mid = left + (right - left) // 2

        if arr[mid] == target:
            ans = mid # found a match
            left = mid + 1 # but keep looking RIGHT for later duplicates in the array
        elif arr[mid] < target:
            left = mid + 1 
        else: 
            right = mid - 1
    return ans

# 
arr = [2, 4, 5, 6, 6, 7, 9]
ans = find_last_occurence(arr, 6)
print (ans)
```

Worked Example #3

Problem: Return the first index whose value is greater than or equal to target. Return len(nums) if no such value exists.

Assume the array to be sorted, then, Binary search or its variant 

Else, it has to be linear search with time complexity as O(N)

```python
def lower_bound(nums, target):
    lo, hi = 0, len(nums)  # Search interval is [lo, hi).

    while lo < hi:
        mid = (lo + hi) // 2

        if nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid  # mid could be the first valid position.

    return lo

#
nums = [2, 4, 4, 4, 7, 9]

print(lower_bound(nums, 4))   # 1
print(lower_bound(nums, 5))   # 4
print(lower_bound(nums, 100))  # 6
```
