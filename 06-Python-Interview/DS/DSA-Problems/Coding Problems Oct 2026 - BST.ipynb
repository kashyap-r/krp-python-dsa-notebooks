```python
# Binary Search Tree 
from collections import deque 

class Node:
    def __init__(self, value):
        self.data = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

def insert_recursively(root, key):
    new_node = Node(key)
    if root is None:
        return new_node

    if key == root.data:
        return root
    
    if key < root.data:
        root.left = insert_recursively(root.left, key)
    else:
        root.right = insert_recursively(root.right, key)
    return root 

def insert_iteratively(root, key):
    new_node = Node(key)
    if root is None:
        return new_node 
    
    current = root 
    while current is not None: 
        if current.data == key:
            return root 
        elif current.data > key and current.left is not None:
            current = current.left 
        elif current.data < key and current.right is not None:
            current = current.right
        else:
            break 
    
    if current.data > key:
        current.left = new_node
    else:
        current.right = new_node 
    
    return root

# print / display BST level order 
def display_BST_level_order(root):
    q = deque([root])

    res = []
    first = True 

    while q:
        current = q.popleft() 

        if not first:
            res.append(", ")
        first = False 

        if current is not None: 
            res.append(str(current.data))
            q.append(current.left)
            q.append(current.right)
        else: 
            res.append("N")
    return res

# Binary Search Tree - Search (Recursively)
def recursive_search(root, key):
    if root is None:
        return False 

    if root.data == key:
        return True 

    if key > root.data:
        return recursive_search(root.right, key)
    
    return recursive_search(root.left, key)    

# Complexity: Time O(h), Space O(h) - recursion stack

# Binary Search Tree - Search Iterative method 
def interative_search(root, key):
    present = False 

    while root is not None: 
        if root.data == key:
            present = True 
            break 
        elif root.data > key: 
            root = root.left 
        else: 
            root = root.right

    return present
# Complexity: Time O(h), Space O(1)

# Deletion from a Binary Search tree
# Refer: https://www.geeksforgeeks.org/dsa/deletion-in-binary-search-tree/
from collections import deque

# Node structure
class Node:
    def __init__(self, x):
        self.data = x
        self.left = None
        self.right = None

# Calculate Height
def getHeight(root, h):
    if root is None:
        return h - 1
    return max(getHeight(root.left, h + 1),
               getHeight(root.right, h + 1))

# Print Level Order
def levelOrder(root):
    q = deque()
    q.append((root, 0))
    lastLevel = 0
    
    # function to get the height of tree
    height = getHeight(root, 0)

    #  printing the level order of tree
    while q:
        node, lvl = q.popleft()

        if lvl > lastLevel:
            print()
            lastLevel = lvl
        
        #  all levels are printed
        if lvl > height:
            break
         
        #  printing null nodes
        if node.data != -1:
            print(node.data, end=" ")
        else:
            print("N", end=" ")
        
        # null node has no children
        if node.data == -1:
            continue

        if node.left is None:
            q.append((Node(-1), lvl + 1))
        else:
            q.append((node.left, lvl + 1))

        if node.right is None:
            q.append((Node(-1), lvl + 1))
        else:
            q.append((node.right, lvl + 1))

# Get inorder successor (smallest in right subtree)
def getSuccessor(curr):
    curr = curr.right
    while curr is not None and curr.left is not None:
        curr = curr.left
    return curr

# Delete a node with value x from BST
def delNode(root, x):
    if root is None:
        return root

    if root.data > x:
        root.left = delNode(root.left, x)
    elif root.data < x:
        root.right = delNode(root.right, x)
    else:
        
        # node with 0 or 1  children
        if root.left is None:
            return root.right
        if root.right is None:
            return root.left
        
        #  Node with 2 children
        succ = getSuccessor(root)
        root.data = succ.data
        root.right = delNode(root.right, succ.data)

    return root
# Complesity: Time O(h), Space O(h)

if  __name__ == "__main__":
    root = Node(10)
    root.left = Node(5)
    root.right = Node(15)
    root.right.left = Node(12)
    root.right.right = Node(18)
    
    x = 15
    root = delNode(root, x)
    levelOrder(root)



if __name__=="__main__":
    root = Node(2) 
    root.left = Node(1) 
    root.right = Node(3)
    root.right.right = Node(6)

    key = 4 
    root = insert_recursively(root, key)

    root = insert_iteratively(root, 9)
    root = insert_iteratively(root, -1)
    res = display_BST_level_order(root)
    print ("[" + "".join(res) + "]")

    
    print ("Searching for 9 ... ",recursive_search(root, 9) )

    print ("Searching for -1 ... ",interative_search(root, -1) )

    print ("Searching for -11 ... ",interative_search(root, -11) )

    print ("Deleting the node 9.. ", delNode(root, 9))

    res = display_BST_level_order(root)
    print ("[" + "".join(res) + "]")
```

Problem #1 

Sorted array to a Balanced Binary Search Tree 

Given a sorted array. convert it into a Balanced Binary Search Tree (BST)
Return the root of the BST

A Balanced BST is a Binary Tree in which the difference between the hieghts of the left and right subtrees of every node is at most one 

Input: arr = [10, 20, 30]

Expected Output: [[20], [10, 30]]


Input: arr = [1, 5, 9, 14, 23, 27]
Output: [[9], [1, 23], [N, 5, 14, 27]]

Approach #1 - Using recursion 

The idea is to use recursion to build the tree from a sorted array. 

We first find the middle element of the array and make it the root of the tree. 

Then recursively, repeat the same process for the eleft subarray (to form the left subtree) and the right subarray (to form the right subtree)


```python
class Node:
    def __init__(self, val):
        self.data = val 
        self.left = None 
        self.right = None 

def sortedArraytoBST_recursive(arr, start, end):
    if start > end: 
        return None 

    mid = start + (end - start) // 2 
    root = Node(arr[mid])

    # Divide from the middle element
    root.left = sortedArraytoBST_recursive(arr, start, mid - 1)
    root.right = sortedArraytoBST_recursive(arr, mid + 1, end)

    return root

def sortedArrayToBST(arr):
    return sortedArraytoBST_recursive(arr, 0, len(arr) - 1)




```

Problem #1 

Valid BST 

Medium LC 98 

Problem statement 
Given the root of a binary tree, determine whether it is a valid BST. A valid BST requires all values in the left subtree to be strictly smaller and all values in the right subtree to be strictly larger.

Example:

        10
       /  \
      5    15
          /  \
         6    20

Expected Output: False 

Intuition: 
By defition... 
root.left.val < root.val < root.right.val 

However, additionally, every node must satisfy bounds inherited from all its ancestors. 

Algorithm:

Start with no bounds at the root

For each node: 

    1. Check that lower < node.value < upper 
    2. When moving left, set the upper bound to the current value 
    3. when moving right, set the lower bound to the current value 
    4. Both subtrees must be valid

Assumption: 
- The node values are integers


```python
def is_valid_bst(root):
    def validate(node, low, high):
        if node is None:
            return True 

        if not (low < node.val < high):
            return False 

        return (
            validate(node.left, low, node.val)
            and 
            validate(node.right, node.val, high)
        )
    return validate(root, float("-inf"), float("-inf"))

# Complexity: O(n) time, O(h) recursion space.

## Usage 

root = build_binary_tree([10, 5, 15, None, None, 6, 20])
print(is_valid_bst(root))  # False

root = build_binary_tree([10, 5, 15, 2, 7, 12, 20])
print(is_valid_bst(root))  # True

```

```python
# Alternative 

def is_valid_bst_inorder(root) -> bool:
    stack = []
    current = root
    previous = None

    while current is not None or stack:
        while current is not None:
            stack.append(current)
            current = current.left

        current = stack.pop()

        if previous is not None and current.value <= previous:
            return False

        previous = current.value
        current = current.right

    return True
```

Problem #2: 

Kth Smallest element in BST 
Given the root of a BST and an integer k, return the kth smallest value. Assume 1 <= k <= number of nodes.
Example:
        5
       / \
      3   7
     / \   \
    2   4   9

Input: K = 3 

Solution Evolution: 
Approach #1: Brute-Force

Traverse the entire tree, collect values, sort them, return the index k-1 

Time: O(n log n)
Space: O(n) 

Approach #2: Explot the bST property 
InOrder traversal of BST gives sorted values 

We can aviod sorting 

Approach #3: Iterative Inorder with early stopping 

Complexity: O(h + k) time for this early-stopping traversal and O(h) stack space. Worst-case time O(n).
Interview follow-up: What if the interviewer asks for thousands of kth-smallest queries?
Augment each BST node with its subtree size. With maintained subtree counts, each query takes O(h), or O(log n) on a balanced BST.


```python
# Approach #3

def kth_smallest(root, k):
    stack = []
    current = root

    while stack or current:

        while current:
            stack.append(current)
            current = current.left

        current = stack.pop()
        k -= 1

        if k == 0:
            return current.val

        current = current.right

    return None
```
