
String Pattern Matching (KMP Algorithm)

When searching for a pattern inside a text (like searching for 3, 3, 4 inside your scattered array or ABC inside a string), a naive search takes O(N * M) time because it backtracks to the beginning of the pattern every time a mismatch occurs.
The Knuth-Morris-Pratt (KMP) algorithm optimizes this to O(N + M) time by avoiding backtracking. It uses a Longest Prefix Suffix (LPS) table (also called a π table or failure function). The LPS array stores how much of the pattern's prefix matches its own suffix.
If a mismatch happens midway through a match, KMP uses the LPS table to skip ahead to the next viable starting point without re-checking characters it has already seen.

```python
def build_lps_table(pattern):
    """Precomputes the Longest Prefix Suffix array for the pattern."""
    m = len(pattern)
    lps = [0] * m
    length = 0  # length of the previous longest prefix suffix
    i = 1
    
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def kmp_search(text, pattern):
    """Searches for all starting indices where the pattern occurs in the text."""
    n, m = len(text), len(pattern)
    lps = build_lps_table(pattern)
    matches = []
    
    i = 0  # index for text
    j = 0  # index for pattern
    
    while i < n:
        if pattern[j] == text[i]:
            i += 1
            j += 1
            
        if j == m:
            # Pattern found! Save the starting index
            matches.append(i - j)
            j = lps[j - 1]  # Reset j using the LPS table to find next matches
            
        elif i < n and pattern[j] != text[i]:
            # Mismatch occurred after j matches
            if j != 0:
                j = lps[j - 1]  # Smart skip using LPS
            else:
                i += 1
                
    return matches

# --- Quick Test ---
text_sequence = [1, 2, 3, 3, 4, 5, 6, 3, 3, 4, 8, 9, 0]
pattern_sequence = [3, 3, 4]

found_at = kmp_search(text_sequence, pattern_sequence)
print(f"Pattern found at starting indices: {found_at}") 
# Output: [2, 7] (Matches starting at index 2 and index 7)
```
