34. M Find First and Last Position of Element in Sorted Array

Given an array of integers nums sorted in non-decreasing order, find the starting and ending position of a given target value.

If target is NOT found in the array return [-1, -1]

You must run an algorithm with O(log N) runtime complexity

Example 1: 

Input

nums = [5, 7, 7, 8, 8, 10, 12]
target 8 

Output: [3, 4]



Given that the array is sorted, the best way to search for an element would be Binary Search. 

However, this solution needs a little tweaking with the standard algorithm of Binary search 

Lets just start with BS 
On finding a match, save it, then keep searching:
- First occurrence: continue to the left.
- Last occurrence: continue to the right

```python
# Leet code submitted ... :)
def search_range(arr, target):
    def find_boundary(find_first):
        lo, hi = 0, len(arr) - 1
        ans = -1 

        while lo <= hi:
            mid = lo + (hi - lo) // 2

            if arr[mid] < target:
                lo = mid + 1
            elif arr[mid] > target:
                hi = mid - 1
            else: 
                ans = mid

                if find_first:
                    hi = mid - 1 # look for an earlier match
                else: 
                    lo = mid + 1 # look for a later match
        
        return ans

    return [find_boundary(True), find_boundary(False)]

## 
arr = [5, 7, 7, 8, 8, 10]
target = 15 
result = search_range(arr, target)
print (result)
```

```python
# Another Approach 
def extreme_insertion_index(arr, target, find_first):
    left, right = 0, len(arr) - 1
    ans = -1
    
    while left <= right:
        mid = left + (right - left) // 2
        
        if arr[mid] == target:
            ans = mid
            if find_first:
                right = mid - 1  # Keep squeezing left
            else:
                left = mid + 1   # Keep squeezing right
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
            
    return ans

def search_range(arr, target):
    """Returns a list of [first_idx, last_idx] or [-1, -1] if not found."""
    first = extreme_insertion_index(arr, target, find_first=True)
    if first == -1:
        return [-1, -1]
    
    last = extreme_insertion_index(arr, target, find_first=False)
    return [first, last]

# --- Test ---
sorted_arr = [2, 5, 5, 5, 8]
print(search_range(sorted_arr, 5))  # Output: [1, 3]

```
