Given an Array bui;d a binary tree 

arr =  [3, 9, 20, None, None, 15, 7]

The array represents a tree in level order: read nodes from top to bottom, left to right. None means a child is missing. 

Usually it is level order unless specified otherwise. 



```python

arr =  [3, 9, 20, None, None, 15, 7]

class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None 

# build the tree manually 
#         
root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)

# root.left.left = None
# root.left.right = None 

root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

```

We now want sa function that builds the tree from an array. 

We will use a queue.. why ?

We need to assign the children to the nodes in the order we created them:
1. Assign children to 3
2. Assign children to 9 
3. Assign children to 20. 

That is first in, first out (FIFO) - exactly what a queue provides. 

This implementation assumes the input is a valid level-order representation; trailing missing children can be omitted.

```python
from collections import deque

class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None

def build_binary_tree(values):
    if not values or values[0] is None:
        return None

    root = TreeNode(values[0])
    queue = deque([root])

    i = 1 # index of the next child entry 
    while queue and i < len(values):
        parent = queue.popleft()

        # next entry represents this parent's left child
        if values[i] is not None:
            parent.left = TreeNode(values[i])
            queue.append(parent.left)

        i += 1 

        # Following entry represents this parent's right child
        if i < len(values) and values[i] is not None:
            parent.right = TreeNode(values[i])
            queue.append(parent.right)

        i += 1 
    return root

# verify the tree by printing each levels
def print_levels(root):
    if root is None:
        return 

    queue = deque([root])

    while queue:
        level = []
        for _ in range(len(queue)):
            node = queue.popleft()
            level.append(node.data)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        print (level)

# Usage
values = [3, 9, 20, None, None, 15, 7]

root = build_binary_tree(values)

print(root.data)              # 3
print(root.left.data)         # 9
print(root.right.data)        # 20
print(root.right.left.data)   # 15
print(root.right.right.data)  # 7
print(root.left.left)        # None

print_levels(root)
```

Complexity of buildign the Binary Tree

Time: O(n) - process each array entry at most once. 
Auxiliary Space: O(W) for the queue, where w is the maximum width of the tree; O(n) in the worst case

Tree Storage: O(m) for the m non-None nodes. 

```python
root = build_binary_tree([3, 9, 20, None, None, 15, 7])
print_levels(root)
```

Build a binary tree using a stack 

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


def build_tree_preorder(values):
    if not values:
        return None

    if values[0] is None:
        if len(values) != 1:
            raise ValueError("Unexpected values after an empty root")
        return None

    root = Node(values[0])
    # Each tuple represents a child position waiting to be filled.
    # Push right first so left is popped first.
    stack = [(root, "right"), (root, "left")]

    for value in values[1:]:
        if not stack:
            raise ValueError("Extra values after the tree is complete")
        parent, side = stack.pop()
        if value is None:
            continue  # This child remains None.
        child = Node(value)
        if side == "left":
            parent.left = child
        else:
            parent.right = child
        stack.append((child, "right"))
        stack.append((child, "left"))
    if stack:
        raise ValueError("Missing child entries: include all None markers")
    return root

# verify the tree by printing each levels
def print_levels(root):
    if root is None:
        return 
    queue = deque([root])
    while queue:
        level = []
        for _ in range(len(queue)):
            node = queue.popleft()
            level.append(node.val)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        print (level)

## Usage 
values = [
    3,
    9, None, None,
    20,
    15, None, None,
    7, None, None
]

root = build_tree_preorder(values)
print_levels(root)  # Our earlier BFS printing function
```

```python
def build_tree(values):
    if not values or values[0] is None:
        return None

    root = Node(values[0])
    parents = [root]
    front = 0
    i = 1

    while front < len(parents) and i < len(values):
        parent = parents[front]
        front += 1

        if values[i] is not None:
            parent.left = Node(values[i])
            parents.append(parent.left)
        i += 1

        if i < len(values) and values[i] is not None:
            parent.right = Node(values[i])
            parents.append(parent.right)
        i += 1

    return root
```

```python
class QueueUsingStacks:
    def __init__(self):
        self.incoming = []
        self.outgoing = []

    def enqueue(self, item):
        self.incoming.append(item)

    def dequeue(self):
        if not self.outgoing:
            while self.incoming:
                self.outgoing.append(self.incoming.pop())

        return self.outgoing.pop()

    def __bool__(self):
        return bool(self.incoming or self.outgoing)
```

```python
def build_tree_with_stacks(values):
    if not values or values[0] is None:
        return None

    root = Node(values[0])
    queue = QueueUsingStacks()
    queue.enqueue(root)
    i = 1

    while queue and i < len(values):
        parent = queue.dequeue()

        if values[i] is not None:
            parent.left = Node(values[i])
            queue.enqueue(parent.left)
        i += 1

        if i < len(values) and values[i] is not None:
            parent.right = Node(values[i])
            queue.enqueue(parent.right)
        i += 1

    return root
```
