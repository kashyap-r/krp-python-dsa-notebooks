**Memorise iterative preorder DFS first. Add iterative inorder for BST problems. Understand postorder as “children must finish before the parent.”**

About the choice of traversal - Think about when the node's work can be done

| Question to ask | Traversal | Examples |
|---|---|---|
| Can I process the node immediately? | **Preorder** | Search, count nodes, invert tree, root-to-leaf path sum |
| Do I need values in sorted order in a BST? | **Inorder** | Kth smallest, BST iterator, validate BST |
| Do I need answers from both children first? | **Postorder** | Height, balance, diameter, subtree sums |


```python
# Preorder: you default iterative DFS 

def preorder(root):
    if root is None:
        return []

    stack = [root]
    result = []

    while stack:
        node = stack.pop()
        result.append(node.value)

        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)

    return result
```

Remember: pop → process → push children. Push right first to process left first.

Often, the only adaptation is adding information to each stack entry.


For example, finding maximum depth:
Here, depth is passed downward from parent to child, so preorder works. A path sum can similarly carry a running sum.

```python
# For Example..
def max_depth(root):
    if root is None:
        return 0

    stack = [(root, 1)]
    deepest = 0

    while stack:
        node, depth = stack.pop()
        deepest = max(deepest, depth)

        if node.right:
            stack.append((node.right, depth + 1))
        if node.left:
            stack.append((node.left, depth + 1))

    return deepest
```

2. Inorder: the essential BST template

```python
def inorder(root):
    stack = []
    result = []
    curr = root

    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left

        curr = stack.pop()
        result.append(curr.value)
        curr = curr.right

    return result
```

Remember: go left → pop and process → go right.

In a BST with distinct keys, this produces strictly increasing values:
- Kth smallest: stop after processing k nodes.
- Validate BST: check each value is greater than the previously processed value.
- BST iterator: preserve the stack between calls.
However, a BST problem doesn’t automatically require inorder traversal:

| BST task | Better starting point |
|---|---|
| Search for a value | Compare and follow one branch |
| Successor/predecessor | Follow one branch while keeping a candidate |
| Lowest common ancestor | Follow the shared direction until the targets split |
| Visit values in sorted order | Inorder |

3. Postorder: when results flow upward

For diameter, a node needs the heights of both subtrees before calculating the longest path through itself. That dependency suggests postorder.

Use the flag template you’ve already seen:

```python
def postorder(root):
    if root is None:
        return []

    stack = [(root, False)]
    result = []

    while stack:
        node, ready = stack.pop()

        if ready:
            result.append(node.value)
        else:
            stack.append((node, True))

            if node.right:
                stack.append((node.right, False))
            if node.left:
                stack.append((node.left, False))

    return result
```

Remember: schedule the parent for later → process children → return to parent.

For height, diameter, or balance calculations, replace the append with the parent’s calculation and store the child results where the parent can retrieve them. The flag controls processing order; it doesn’t itself store those results.

Why can height use both preorder and postorder?

It depends on how you formulate the solution:
- Preorder: carry depth downward and track the maximum.
- Postorder: calculate height = 1 + max(left_height, right_height) upward.
A problem may allow several traversals. The dependency in your chosen solution determines the order.

4. Graph DFS: use the stack pattern plus a visited set

Graphs can have cycles and multiple routes to the same node. Unlike an ordinary tree, you must usually track visited nodes.

```python
def dfs(graph, start):
    stack = [start]
    seen = {start}
    result = []

    while stack:
        node = stack.pop()
        result.append(node)

        for neighbour in graph.get(node, []):
            if neighbour not in seen:
                seen.add(neighbour)
                stack.append(neighbour)

    return result
```

Mark nodes when pushing them to avoid adding the same node repeatedly.

For graphs:
- Reachability / connected components: this basic template works.
- Inorder: has no standard meaning because there are no designated left/right children.
- Topological sorting / directed cycle detection: need additional state to track active and finished nodes; the basic visited-set template alone is insufficient.
- Shortest path with equal-cost edges: use the BFS you already know.
This simple graph template is suitable for reachability; it need not reproduce the exact visiting order of recursive DFS.

Your recognition checklist

| What the problem needs | Start with |
|---|---|
| Visit/search all tree nodes | Preorder stack |
| Carry depth, running sum, or path state downward | Preorder with `(node, state)` |
| Sorted processing of BST values | Inorder stack |
| Combine completed child results | Postorder |
| Explore arbitrary connections | Graph DFS + visited |
| Levels or fewest equal-cost moves | BFS |
| Locate a key within a BST | Direct comparisons, usually no stack |
