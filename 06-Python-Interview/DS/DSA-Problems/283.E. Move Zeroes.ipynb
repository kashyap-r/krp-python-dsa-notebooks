283.E. Move Zeroes 

Given an interger array nums, move all zeroes to the end while maintaining the relative order of nonzero elements. 

Note that you must do this in-place without making a copy of the array.

Example 1:

Input: nums = [0,1,0,3,12]
Output: [1,3,12,0,0]

Example 2:

Input: nums = [0]
Output: [0]

Hint 1
In-place means we should not be allocating any space for extra array. But we are allowed to modify the existing array. However, as a first step, try coming up with a solution that makes use of additional space. For this problem as well, first apply the idea discussed using an additional array and the in-place solution will pop up eventually.

Hint 2
A two-pointer approach could be helpful here. The idea would be to have one pointer for iterating the array and another pointer that just works on the non-zero elements of the array.

Analysis: 

Pattern Recognition 
    * In-place modification
    * Preserve relative order 
    * No requirement for a contiguous window 
    * Two indices progressing forward

Same-direction read/write pointers





```python
# Leet code Submission
def move_zeroes(arr):
    slow = 0
    for fast in range(len(arr)):
        if arr[fast] != 0:
            arr[slow], arr[fast] = arr[fast], arr[slow]
            slow += 1

arr = [0,1,0,0,3,12]
print (arr)
move_zeroes(arr)
print(arr)


# Time O(n)
# Space O(1)
```
