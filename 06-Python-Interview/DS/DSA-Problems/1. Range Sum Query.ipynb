Given an qrray an dmultiple queries (L, R), return sum 

Context: Log volume per minute. Query total logs between time windows. 

Approach:
1. Build prefix array
2. For each query, sum = prefix[R] - prefix[L-1]

```python
class NumArray:
    def __init__(self, nums: list[int]):
        self.prefix = [0] * len(nums)
        self.prefix[0] = nums[0]

        for i in range(1, len(nums)):
            self.prefix[i] = self.prefix[i - 1] + nums[i]

    def sumRange(self, left, right):
        if left == 0:
            return self.prefix[right]
        else:
            return self.prefix[right] - self.prefix[left - 1]

"""
Here:
Time Complexity: O(n) for initialization, O(1) for sumRange queries
Query Complexity: O(1) for sumRange queries
Space Complexity: O(n) for storing the prefix sum array
"""

if __name__ == "__main__":
    nums = [-2, 0, 3, -5, 2, -1]
    num_array = NumArray(nums)
    print(num_array.sumRange(0, 2))  # Output: 1
    print(num_array.sumRange(2, 5))  # Output: -1
    print(num_array.sumRange(0, 5))  # Output: -3

# Edge cases:
# 1. Empty array: The class should handle the case where the input array is empty
# 2. Single element array: The class should handle the case where the input array has only one element
# Boundary cases:
# 1. Querying the entire array: The class should handle the case where the sum  is requested for the entire array   
# 2. Querying a single element: The class should handle the case where the sum is requested for a single element in the array   
# 3. Querying the first or last element: The class should handle the case where the sum is requested for the first or last element in the array     
# 4. Querying with left and right indices being the same: The class should handle the case where the left and right indices are the same, which means the sum of a single element is requested  
# 5. Querying with left index greater than right index: The class should handle the case where the left index is greater than the right index, which is an invalid query and should return 0 or raise an exception.
# 6. Querying with negative indices: The class should handle the case where the left or right index is negative, which is an invalid query and should return 0 or raise an exception.   
# 7. Querying with indices out of bounds: The class should handle the case where the left or right index is out of bounds, which is an invalid query and should return 0 or raise an exception. 
# 8. Querying with non-integer indices: The class should handle the case where the left or right index is not an integer, which is an invalid query and should return 0 or raise an exception.  
# 9. Querying with non-list input: The class should handle the case where the input is not a list, which is an invalid input and should raise an exception.
# 10. Querying with non-integer elements in the list: The class should handle the case where the input list contains non-integer elements, which is an invalid input and should raise an exception.
# 11. Querying with very large input: The class should handle the case where the input list is very large, which may cause performance issues or memory errors.
# 12. Querying with very small input: The class should handle the case where the input list is very small, which may cause performance issues or memory errors.
# 13. Querying with very large indices: The class should handle the case where the left or right index is very large, which may cause performance issues or memory errors.
# 14. Querying with very small indices: The class should handle the case where the left or right index is very small, which may cause performance issues or memory errors.
# 15. Querying with very large sums: The class should handle the case where the sum of the elements in the array is very large, which may cause performance issues or memory errors.
# 16. Querying with very small sums: The class should handle the case where the sum of the elements in the array is very small, which may cause performance issues or memory errors.
# 17. Querying with very large differences between left and right indices: The class should handle the case where the difference between the left and right indices is very large, which may cause performance issues or memory errors.
# 18. Querying with very small differences between left and right indices: The class should handle the case where the difference between the left and right indices is very small, which may cause performance issues or memory errors.
# test harness:
```
