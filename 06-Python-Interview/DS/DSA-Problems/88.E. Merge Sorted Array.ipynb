88. Merge Sorted Array

You are given two interger arrays nums1 and nums2, sorted in non-decreasing order, and two integers m and n, representing the number of elements in nums1 and nums2 respectively. 

Merge nums1 and nums2 into a single array sorted in non-decreasing order. 

The final sorted array should not be returned by the function, but instead be stored inside the array nums1. To accommodate this, nums1 has a length of m + n, where the first m elements denote the elements that should be merged, and the last elements are set to 0 and should be ignored. nums2 has a length of n. 

Example 1:

Input: nums1 = [1,2,3,0,0,0], m = 3, nums2 = [2,5,6], n = 3
Output: [1,2,2,3,5,6]
Explanation: The arrays we are merging are [1,2,3] and [2,5,6].
The result of the merge is [1,2,2,3,5,6] with the underlined elements coming from nums1.

Example 2:

Input: nums1 = [1], m = 1, nums2 = [], n = 0
Output: [1]
Explanation: The arrays we are merging are [1] and [].
The result of the merge is [1].

Example 3:

Input: nums1 = [0], m = 0, nums2 = [1], n = 1
Output: [1]
Explanation: The arrays we are merging are [] and [1].
The result of the merge is [1].
Note that because m = 0, there are no elements in nums1. The 0 is only there to ensure the merge result can fit in nums1.


Constraints: 

1. nums1.length == m + n

2. nums2.length = n 

3. 0 <= m, n <= 200 

4. 1 <= m + n <= 200 

5. nums1[i] and nums2[j] can be both negative and positive integers. 

Question: 
1. What to do with the duplicate items - retain one of them or include them? 

Follow up: Can you come up with an algorithm that runs in m + n time ?

```python

nums1 = [1,2,3,0,0,0]
m = 3
nums2 = [2,5,6]
n = 3

# Output: [1,2,2,3,5,6]
```

The Naive Approach 

The most straightforward way to solve this is to take advantage of the trailing zeroes at the end of nums1. 

We can overwrite those zeroes with the elements of nums2, and then call 

Python's built-in .sort() method on the entire combined array.

```python
def merge_naive(nums1, m, nums2, n):
    nums1[m:] = nums2
    nums1.sort()

# Usage 
nums1 = [1,2,3,0,0,0]
m = 3
nums2 = [2,5,6]
n = 3

print (f"First array: {nums1}")
print (f"Second array: {nums2}")
merge_naive(nums1, m, nums2, n)
print (f"First array after merge: {nums1}")
```

Complexity Analysis 

Time: O((m+n) log(m+n))
Merging the lists takes linear time, but calling .sort() forces python to run Timsort, which takes O(NlogN) operations 

Space: O(1) if you do not count the internal sorting stack space used by Pythons sort() algorithm. 



Method 2: The Optimal Solution (Three-Pointer Approach)

The naive method ignores a massive hint: both arrays are already sorted.

If we try to fill nums1 starting from the front (index 0), we run into a major issue—we risk overwriting the valid elements of nums1 before we've had a chance to compare them. 

To avoid this without using extra memory, we fill the array from the back to the front.

The Logic

    We place three pointers:
    1. p1: Points to the last valid element in nums1 (m - 1).
    2. p2: Points to the last element in nums2 (n - 1).
    3. p: Points to the absolute end of nums1's memory slot (m + n - 1).

We compare the elements at p1 and p2, write the larger value at index p, and move the pointers backward.

```python
def merge_sorted_array_optimal(nums1, m, nums2, n):
    p1 = m - 1 # pointer for valid elements in nums1 
    p2 = n - 1 # pointer for the last element in nums2
    p = m + n - 1 # pointer for the write position at the end of nums1

    # Compare the elements from the back while both arrays have elements
    # remaining 
    while p1 >= 0 and p2 >= 0:
        if nums1[p1] > nums2[p2]:
            nums1[p] = nums1[p1]
            p1 -= 1
        else: 
            nums1[p] = nums2[p2]
            p2 -= 1
        p -= 1 

    # If there are any remaining elements left in nums2, copy them over
    # If the elemnts remain in nums1, they are already in their correct 
    # order 
    while p2 >= 0:
        nums1[p] = nums2[p2]
        p2 -= 1 
        p -= 1

# Usage 
nums1 = [1,2,3,0,0,0]
m = 3
nums2 = [2,5,6]
n = 3

print (f"First array: {nums1}")
print (f"Second array: {nums2}")
merge_sorted_array_optimal(nums1, m, nums2, n)
print (f"First array after merge: {nums1}")

```


Complexity Analysis

• Time Complexity: O(m + n) because we process each slot of the combined array exactly once in a single reverse sweep.

• Space Complexity: O(1) absolute constant memory space.

Why and, rather than or, in the first loop?

Inside it, you access both a[i] and b[j]. Both positions must be valid.

Once either array is exhausted, comparisons are no longer possible. The leftover loops process the other array.

Why is it safe to append the remaining elements directly?

They are already sorted, and the merge has already emitted all elements that precede them.

Invariant: result is the correctly sorted merge of the processed prefixes a[:i] and b[:j]. All unprocessed values are at least as large as the last emitted value.
