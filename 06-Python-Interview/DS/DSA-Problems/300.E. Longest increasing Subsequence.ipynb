The Problem (LeetCode 300): Given an integer array, return the length of the longest strictly increasing subsequence. Remember, a subsequence can skip elements!

The Intuition (Dynamic Programming Approach)

We create a dp array where dp[i] stores the length of the longest increasing subsequence that ends at index i.
• We initialize every slot to 1 (since a single number is an increasing subsequence of length 1).
• For every element, we look backward at all previous elements. If the current element is larger than a previous element, we try to extend that previous subsequence chain.


```python
def lengthOfLIS(nums):
    if not nums:
        return 0
        
    # Every element is an increasing subsequence of length 1 by default
    dp = [1] * len(nums)
    
    for i in range(len(nums)):
        for j in range(i):
            # If the current number is bigger than the past number
            if nums[i] > nums[j]:
                # Can we make a longer chain by appending nums[i] to nums[j]'s chain?
                dp[i] = max(dp[i], dp[j] + 1)
                
    return max(dp)

# Example test
nums = [10, 9, 2, 5, 3, 7, 101, 18]
print(lengthOfLIS(nums))  
# Output: 4  (The subsequence is [2, 3, 7, 101] or)
```
