560. Subarray Sum Equals K

Given an array of integers nums and an integer k, return the total number of subarrays whose sum equals to k.

A subarray is a contiguous non-empty sequence of elements within an array.

Example 1:

Input: nums = [1,1,1], k = 2
Output: 2

Example 2:

Input: nums = [1,2,3], k = 3
Output: 2

```python
# optimal solution 

def subarray_sum(nums, k):
    frequencies = {0: 1}
    prefix = 0 
    count = 0 
    for num in nums:
        prefix += num 
        # each matching earlier prefix gives one subarray ending here
        count += frequencies.get(prefix - k, 0)
        # record this prefix for subsequent positions 
        frequencies[prefix] = frequencies.get(prefix, 0) + 1 

    return count

# Usage
nums = [1, 2, 1, 2, 1]
k = 2 

result = subarray_sum(nums, k)
print (result)
```

```python
def subarray_sum(nums, k):
    prefix_indices = {0: [-1]}  # Zero sum before index 0.
    prefix = 0
    result = []

    for end, num in enumerate(nums):
        prefix += num

        # Each matching earlier prefix identifies a valid start.
        for previous in prefix_indices.get(prefix - k, []):
            result.append(nums[previous + 1:end + 1])

        # Store this index after checking for matches.
        prefix_indices.setdefault(prefix, []).append(end)

    return result
```
