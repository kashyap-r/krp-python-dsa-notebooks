
Maximum Subarray (Kadane's Algorithm)

The Problem (LeetCode 53): Given an integer array, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.

The Intuition

At any given index, you have two choices:
1. Add the current number to the running subarray sum you've been building.
2. Throw away the past and start a brand new subarray beginning exactly at the current number.


```python
def maxSubarray(arr):
    max_global = arr[0]
    max_current = arr[0]

    for i in range(1, len(arr)):
        # Decide: Add tot he current streak, or start fresh?
        max_current = max(arr[i], max_current + arr[i])
        

        # track the best sum we have see so far
        if max_current > max_global:
            max_global = max_current
    return max_global

# Usage 
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print (maxSubarray(nums))
```

```python
def maxSubArrayWithIndices(arr):
    if not nums:
        return 0, []
    max_global = nums[0]
    max_current = nums[0]
    start = 0
    best_start = 0
    best_end = 0

    for i in range(1, len(arr)):
        if arr[i] > max_current + arr[i]:
            max_current = nums[i]
            start = i
        else:
            max_current = max_current + nums[i]

        # Update the global maximum window if we find a better sum
        if max_current > max_global:
            max_global = max_current
            best_start = start
            best_end = i
        
    # Extract the exact continuous slice from the array
    max_subarray = nums[best_start : best_end + 1]

    return max_global, max_subarray


# --- Verification ---
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
max_sum, sub_arr = maxSubArrayWithIndices(nums)

print(f"Maximum Sum:      {max_sum}")       # Output: 6
print(f"Winning Subarray: {sub_arr}")       # Output: [4, -1, 2, 1]
```
