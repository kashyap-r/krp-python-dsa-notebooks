```python
stack = []
# only way to insert at the front 
stack.insert(0, 2)
print (stack)
stack.insert(0, 1)
print (stack)

stack.append(3)
print (stack)

print(stack.pop())
print (stack)

stack.extend([4, 5, 6])
print (stack)
print (stack.pop())
print (stack)
```

```python
from collections import deque 

dq = deque([])

dq.append(1)
dq.append(2)
print (dq)
dq.append(3)
print (dq)
dq.append(4)
print (dq)
dq.pop()
print(dq)
dq.appendleft(0)
print(dq)
dq.appendleft(0)
print(dq)
dq.popleft()
print(dq)

```
