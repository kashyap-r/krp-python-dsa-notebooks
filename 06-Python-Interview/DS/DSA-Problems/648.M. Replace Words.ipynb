In English, we have a concept called root, which can be followed by some other word to form another longer word - let's call this word derivative. For example, when the root "help" is followed by the word "ful", we can form a derivative "helpful".

Given a dictionary consisting of many roots and a sentence consisting of words separated by spaces, replace all the derivatives in the sentence with the root forming it. If a derivative can be replaced by more than one root, replace it with the root that has the shortest length.

Return the sentence after the replacement.

Example 1:

Input: dictionary = ["cat","bat","rat"], sentence = "the cattle was rattled by the battery"
Output: "the cat was rat by the bat"

Example 2:

Input: dictionary = ["a","b","c"], sentence = "aadsfasf absbs bbab cadsfafs"
Output: "a a b c"


Analysis: 

The key requirement is: **Replace each word with its shortest dictionary prefix**


```python
dictionary = ["cat", "bat", "rat"]
sentence = "the cattle was rattled by the battery"

# Output: "the cat was rat by the bat"
```

```python
# Naive Solution: Check eveery root for every word 
# Sort the roots by length, then the first matching prefix is guaranteed to be the shortest

def replace_words(dictionary, sentence):
    roots = sorted(dictionary, key=len)
    words = sentence.split()
    print (roots)
    print (words)

    for i, word in enumerate(words):
        for root in roots:
            if word.startswith(root):
                words[i] = root
                break
    return " ".join(words)

print (replace_words(dictionary, sentence))
```

Complexity Analysis:
For D roots, W words in the sentence and maximum root length L
- Sorting O(D log D)
- Prefix comparisons: O(W * D * L) worst case

What can you avoid? Checking roots separately 
Suppose the dictionary contains: ["cat", "car", "care"]
The naive approach repeatedly checks the shared prefix "ca"

**Optimal Approach**

A trie stores that shared prefix once. For each sentence word, walkthrough its characters 
- if a character has no mactching child, no dictionary root matches - skip the loop, n keep the word!
- If you reach is_end, you've found a root - return it immediately 
- Because you walk from the shortest prefix to the longest, the first root found is the shortest. 

Useful edge cases: no matching root, multiple matching roots, a word exactly equal to a root, and repeated words in the sentence.

```python
class Trie:
    def __init__(self):
        self.children = {}
        self.is_end = False

    def insert(self, word):
        node = self

        for ch in word:
            if ch not in node.children:
                node.children[ch] = Trie()
            node = node.children[ch]

        node.is_end = True

    def shortest_root(self, word):
        node = self

        for i, ch in enumerate(word):
            if ch not in node.children:
                return word

            node = node.children[ch]

            if node.is_end:
                return word[:i + 1]

        return word


def replace_words(dictionary, sentence):
    trie = Trie()

    for root in dictionary:
        trie.insert(root)

    return " ".join(
        trie.shortest_root(word)
        for word in sentence.split()
    )
### Usage
print(replace_words(
    ["cat", "bat", "rat"],
    "the cattle was rattled by the battery"
))
# the cat was rat by the bat

print(replace_words(
    ["a", "ab", "abc"],
    "abc abacus apple xyz"
))
# a a a xyz
```

| Approach | Time | Space |
|---|---|---|
| Naive | O(D log D + W × D × L) | O(D + S), including working lists/output |
| Trie | **O(R + S) expected** | **O(R + S)**, including trie/output |
