Implement a Trie 

words = ["cat", "car", "care", "dog"]


```python
# Represents one node in the Trie
class TrieNode:                                 
    # Constructor called whenever a new node is created
    def __init__(self):                        
        # Dictionary mapping each character to its child TrieNode 
        self.children = {}                      
        # True if this node marks the end of a complete word
        self.is_end = False                     

# Represents the entire Trie data structure
class Trie:                                      
    # Constructor called when creating a Trie
    def __init__(self):    
        # Create an empty root node shared by all words                     
        self.root = TrieNode()                   

    # Insert a word into the Trie
    def insert(self, word):      
        # Begin traversal from the root node               
        node = self.root                        

        # Process each character of the word, left to right
        for ch in word:                         
            # Check whether the character ch already exists in 
            # the current node's children ditionary
            if ch not in node.children:
                # If absent, create a new TrieNode() and store it in
                # node.children[ch] (the ch is the dictionary key)         
                node.children[ch] = TrieNode()  
            # Move to the child node representing this character
            node = node.children[ch]            
        # Mark the final node as the end of a complete word
        node.is_end = True                      

    # Search for an exact word in the Trie
    def search(self, word):
        # Start searching from the root node
        node = self.root                        
        # Examine each character in the search word
        for ch in word:                         
            # If the required character path doesn't exist
            if ch not in node.children:        
                # Word cannot exist, so return False immediately 
                return False                   
            # Otherwise, move to the matching child node
            node = node.children[ch]            
        # True only if the final node marks a complete word
        return node.is_end                      

    # Check whether any stored word starts with prefix
    def starts_with(self, prefix):              
        # Begin traversal from the root node
        node = self.root                        
        # Examine each character in the prefix
        for ch in prefix:                       
            # If any character is missing from the path
            if ch not in node.children:        
                # No stored word can have this prefix 
                return False                   
            # Move to the next node along the prefix path
            node = node.children[ch]            
        # Entire prefix path exists in the Trie
        return True        
                         
    # Print every complete word stored in the Trie
    def print_words(self):                     
        # Recursive DFS; prefix stores characters visited so far
        def dfs(node, prefix):                  
            # Check whether the current node ends a valid word
            if node.is_end:                     
                # Print the word constructed along the current path
                print(prefix)                   
            # Iterate over all children of the current node
            for ch, child in node.children.items():
                # Recursively visit each child, extending the prefix  
                dfs(child, prefix + ch)         
        # Start DFS from the root with an empty prefix
        dfs(self.root, "")                      
    
    # Print the Trie as an indented tree
    def print_structure(self):                  
        # DFS where level represents indentation depth
        def dfs(node, level):                   
            # Visit each character and its corresponding child
            for ch, child in node.children.items():  
                # Mark complete words with an asterisk
                marker = "*" if child.is_end else ""
                # Print character with indentation and marker  
                print(" " * level + ch + marker)
                # Recursively print descendants with greater indentation      
                dfs(child, level + 2)          
        # Display the root of the Trie
        print("(root)")                         
        # Begin traversal, indenting root's children by 2 spaces
        dfs(self.root, 2)                       


# ---------------- EXAMPLE USAGE ----------------
trie = Trie()                                   # Create an empty Trie

trie.insert("cat")                              # Insert the word "cat"
trie.insert("car")                              # Insert the word "car", sharing prefix "ca"
trie.insert("care")                             # Insert "care", extending the existing word "car"
trie.insert("dog")                              # Insert "dog", creating a separate branch

print(trie.search("cat"))                       # True: "cat" is a complete stored word
print(trie.search("ca"))                        # False: "ca" is only a prefix, not a complete word
print(trie.search("care"))                      # True: "care" is stored

print(trie.starts_with("ca"))                   # True: multiple words begin with "ca"
print(trie.starts_with("do"))                   # True: "dog" begins with "do"
print(trie.starts_with("xy"))                   # False: no stored word begins with "xy"

trie.print_words()                              # Print all complete words stored in the Trie
trie.print_structure()                          # Print the hierarchical structure of the Trie
```

Problem #1 

Implement a Trie, with prefix search nd auto complete 

words = ["cat", "car", "care", "dog"]

```python
# Implement a trie 

class TrieNode:
    def __init__(self):
        self.children = {} 
        self.is_end = False 

class Trie:
    def __init__(self):
        # create root with empty node
        self.root = TrieNode() 

    def insert(self, word):
        node = self.root 

        for ch in word:
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
        node.is_end = True 

    def search(self, word):
        node = self.root 

        for ch in word: 
            if ch not in node.children: 
                return False 
            node = node.children[ch]
        return node.is_end

    def startsWith(self, prefix):
        node = self.root 
        for ch in prefix:
            if ch not in node.children:
                return False
            node = node.children[ch]
            
        return True

    def autoComplete(self, prefix):
        node = self.root 

        for ch in prefix: 
            if ch not in node.children:
                return [] 
            node = node.children[ch]

        results = []
        self._find_all_words(node, prefix, results)
        return results

    def _find_all_words(self, current_node, current_word, results):
        if current_node.is_end: 
            results.append(current_word)

        for ch, next_node in current_node.children.items():
            self._find_all_words(next_node, current_word+ch, results)

    # Print the Trie as an indented tree
    def print_structure(self):                  
        # DFS where level represents indentation depth
        def dfs(node, level):                   
            # Visit each character and its corresponding child
            for ch, child in node.children.items():  
                # Mark complete words with an asterisk
                marker = "*" if child.is_end else ""
                # Print character with indentation and marker  
                print(" " * level + ch + marker)
                # Recursively print descendants with greater indentation      
                dfs(child, level + 2)          
        # Display the root of the Trie
        print("(root)")                         
        # Begin traversal, indenting root's children by 2 spaces
        dfs(self.root, 2)     

    # Print the word path 
    def print_word_path(self, word):
        # Begin at the root
        node = self.root
        # Print root node 
        print("(root)")                         
        # Traverse characters in the requested word
        for i, ch in enumerate(word):            
            # Check if the character path exists
            if ch not in node.children:          
                print(f"Word '{word}' not found")
                return
            # Move to the child TrieNode
            node = node.children[ch]             
            # Mark nodes representing complete words
            marker = "*" if node.is_end else ""  
            # Print character at its tree depth
            print("  " * (i + 1) + ch + marker)  
        # Check whether the full word is stored
        if not node.is_end:                      
            print(f"'{word}' is only a prefix, not a complete word")                  

if __name__ == "__main__":

    myTrie = Trie()

    words = ['cat', 'car', 'care', 'do', 'dog', 'don', 'dirty']

    for word in words:
        myTrie.insert(word)

    myTrie.print_structure()
    print (f"Found 'cat' in the Trie .. {myTrie.search('cat')}!")
    print (f"Are there any words starting with 'ca' in the Trie.. {myTrie.startsWith('ca')}!\n")
    print (f"These are the words starting with 'ca' in the Trie.. \n{myTrie.autoComplete('ca')}")
```

```python
# Auto complete with a twist 

def autocomplete(self, prefix, limit=3):
    if len(prefix) < 2:
        return []

    if limit <= 0:
        return []

    node = self.root

    # Navigate to the prefix node
    for ch in prefix:
        if ch not in node.children:
            return []

        node = node.children[ch]

    suggestions = []

    def dfs(current, word):
        # Stop once enough suggestions are found
        if len(suggestions) >= limit:
            return

        if current.is_end:
            suggestions.append(word)

        # Sorted children for lexicographic ordering
        for ch in sorted(current.children):
            dfs(current.children[ch], word + ch)

            if len(suggestions) >= limit:
                return

    dfs(node, prefix)

    return suggestions

## 
trie.autocomplete("ca", limit=3)
# ["car", "carbon", "care"]
```

1. Count the number of words matching a prefix

Input: The trie contains 
    words = ["car", "cat", "care", "carbon", "do", "dog"]

Expected Results:

Prefix "ca"  → 4 words
Prefix "car" → 3 words
Prefix "do"  → 2 words
Prefix "xyz" → 0 words

Approach #1 (Best Approch)

Store prefix_count in every node (recommended)

Rather than repeatedly traversing descendants, maintain a counter during insertion.

Important edge case: The above implementation counts repeated insertions. If you want unique words, check whether the complete word already exists before incrementing counters.

Approach #2 
Extend startsWith() using DFS
Your existing startsWith() already navigates to the node representing the last character of the prefix.

From there, we simply traverse its subtree and count nodes where is_end == True.

```python
class TrieNode:
    def __init__(self):
        self.children = {} 
        self.is_end = False 
        self.prefix_count = 0 

class Trie: 
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root 

        for ch in word: 
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
        node.is_end = True
        node.prefix_count += 1

    def count_prefix(self, prefix):
        node = self.root 

        for ch in prefix: 
            if ch not in node.children:
                return 0 
            node = node.children[ch]

        return node.prefix_count

    def starts_with(self, prefix):
        return self.count_prefix(prefix) > 0

    def countPrefix(self, prefix):
        node = self.root

        # Step 1: Navigate to the prefix
        for ch in prefix:
            if ch not in node.children:
                return 0

            node = node.children[ch]

        # Step 2: Count complete words below this node
        def dfs(current):
            count = 1 if current.is_end else 0

            for child in current.children.values():
                count += dfs(child)

            return count

        return dfs(node)

if __name__ == "__main__":

    myTrie = Trie()

    words = ['cat', 'car', 'care', 'do', 'dog', 'don', 'dirty']

    for word in words:
        myTrie.insert(word)

    
    # print (f"Found 'cat' in the Trie .. {myTrie.search('cat')}!")
    print (f"Are there any words starting with 'ca' in the Trie.. {myTrie.starts_with('ca')}!\n")
    

```

Problem #3 

Find the longest word in the Trie 

Input: 
words = ["car", "cat", "care", "carbon", "do", "dog"]

Output: words = ["car", "cat", "care", "carbon", "do", "dog"]

The longest word is "carbon" with length 6.

Approach A — DFS to find the longest word

How it works:
    1. Start at the root with an empty string.
    2. Traverse each child, appending its character.
    3. Whenever a complete word is encountered, compare its length against the longest found.
    4. Return the longest after exploring all branches.
    Time complexity is O(S + C) for traversal and string construction, where S is the number of nodes and C is the total cost of building prefixes. With Python's prefix + ch, this can be O(S × H) in the worst case, where H is maximum depth.


```python
def longest_word(self):
    longest = "" 

    def dfs(node, prefix):
        nonlocal longest 

        # update longst only for complete words 
        if node.is_end and len(prefix) > len(longest):
            longest = prefix 

        # explore every child 
        for ch, child in node.children.items():
            dfs(child, prefix+ch)
    dfs(self.root, "")

    return longest
```

Approach B 

Maintain the longest word during insertion
If longest-word queries are frequent, you can avoid DFS entirely.


```python
class Trie:
    def __init__(self):
        self.root = TrieNode()
        self.max_word = ""

    def insert(self, word):
        node = self.root

        for ch in word:
            if ch not in node.children:
                node.children[ch] = TrieNode()

            node = node.children[ch]

        node.is_end = True

        # Update the longest word
        if len(word) > len(self.max_word):
            self.max_word = word

    def longest_word(self):
        return self.max_word
```

Now longest_word() is O(1) to retrieve the stored string reference, with O(L) cost if the caller copies or prints the string.

This approach assumes insertions only. Supporting deletion requires additional bookkeeping or recomputation when the longest word is removed.

Problem #4 

Validate or restrict prefix length for auto complete 

This is actually a different requirement from finding the longest word.


Consider these three constraints:
Constraint	            Example	Purpose
Minimum prefix length	At least 2 characters	Avoid overly broad autocomplete queries
Maximum prefix length	At most 20 characters	Enforce API input limits

Longest stored word	"carbon" (6 chars)	Know the maximum word length currently indexed

You can enforce prefix length directly, without traversing the Trie.

Should maximum prefix length equal the longest stored word? Not necessarily.

For example, if your longest stored word is six characters, a prefix of length eight cannot match any stored word. You can reject it early as an optimization.

But an application-level maximum prefix length should usually be a fixed, documented limit, independent of the current dictionary contents.

You can also maintain self.max_word_length during insertion and use:

if len(prefix) > self.max_word_length:
    return False


```python
def validate_prefix(prefix, min_length=2, max_length=20):
    if not isinstance(prefix, str):
        raise TypeError("Prefix must be a string")

    if not min_length <= len(prefix) <= max_length:
        raise ValueError(
            f"Prefix must contain between "
            f"{min_length} and {max_length} characters"
        )

    return True

def starts_with(self, prefix):
    validate_prefix(prefix)

    node = self.root

    for ch in prefix:
        if ch not in node.children:
            return False

        node = node.children[ch]

    return True
```

Auto-complete feature using Trie



```python


class TrieNode():
    def __init__(self):
        # Initialising one node for trie
        self.children = {}
        self.last = False


class Trie():
    def __init__(self):

        # Initialising the trie structure.
        self.root = TrieNode()

    def formTrie(self, keys):

        # Forms a trie structure with the given set of strings
        # if it does not exists already else it merges the key
        # into it by extending the structure as required
        for key in keys:
            self.insert(key)  # inserting one key to the trie.

    def insert(self, key):

        # Inserts a key into trie if it does not exist already.
        # And if the key is a prefix of the trie node, just
        # marks it as leaf node.
        node = self.root

        for a in key:
            if not node.children.get(a):
                node.children[a] = TrieNode()

            node = node.children[a]

        node.last = True

    def suggestionsRec(self, node, word):

        # Method to recursively traverse the trie
        # and return a whole word.
        if node.last:
            print(word)

        for a, n in node.children.items():
            self.suggestionsRec(n, word + a)

    def printAutoSuggestions(self, key):

        # Returns all the words in the trie whose common
        # prefix is the given key thus listing out all
        # the suggestions for autocomplete.
        node = self.root

        for a in key:
            # no string in the Trie has this prefix
            if not node.children.get(a):
                return 0
            node = node.children[a]

        # If prefix is present as a word, but
        # there is no subtree below the last
        # matching node.
        if not node.children:
            return -1

        self.suggestionsRec(node, key)
        return 1


# Driver Code
keys = ["hello", "dog", "hell", "cat", "a",
        "hel", "help", "helps", "helping"]  # keys to form the trie structure.
key = "h"  # key for autocomplete suggestions.

# creating trie object
t = Trie()

# creating the trie structure with the
# given set of strings.
t.formTrie(keys)

# autocompleting the given key using
# our trie structure.
comp = t.printAutoSuggestions(key)

if comp == -1:
    print("No other strings found with this prefix\n")
elif comp == 0:
    print("No string found with this prefix\n")

# This code is contributed by amurdia and muhammedrijnas
```
