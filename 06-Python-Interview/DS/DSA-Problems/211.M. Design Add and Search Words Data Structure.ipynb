

Problem Description

Design a data structure that supports adding new words and finding if a string matches any previously added string.
Implement the WordDictionary class:
• WordDictionary() Initializes the object.
• void addWord(word) Adds word to the data structure, making it available for future matching.
• boolean search(word) Returns true if there is any string in the data structure that matches word or false otherwise. word may contain dots '.' where dots can be matched with any letter (acting as a wildcard character).

Example

Input:

["WordDictionary","addWord","addWord","addWord","search","search","search","search"]
[[],["bad"],["dad"],["mad"],["pad"],["bad"],[".ad"],["b.."]]

Ouput: 
[null,null,null,null,false,true,true,true]
