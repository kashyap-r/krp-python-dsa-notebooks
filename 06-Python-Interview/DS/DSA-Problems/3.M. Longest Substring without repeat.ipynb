Difficulty Level: Medium

Given a string s, find the length of the longest substring without repeating characters.

Example #1
Input: s = "abcabcbb"
Output: 3 
Explanation: The answer is "abc", with the length of 3.

Example #2
Input: s = "bbbbb"
Output: 1 
Explanation: The answer is "b", witht he length of 1

Example #3
Input: s = "pwwkew"
Output: 3
Explanation: The answer is "wke", with the length of 3. 
Note that the answer must be a substring, "pwke" is a subsequence and not a substring.



Constraints & Edge Cases

• Constraints:
	• 0 ≤ s.length ≤ 5 × 10⁴
	• s consists of English letters, digits, symbols, and spaces.
	
• Edge Cases to Consider:
	• Empty String (s = ""): The output must be 0.
	• All Unique Characters (s = "abcdef"): The output should equal the length of the string itself (6).
	• All Identical Characters (s = "nnnnnn"): The output must be 1.
	• String with Spaces/Symbols (s = "a b c!"): Spaces and symbols are valid characters and count toward duplicates.


```python
# generate all possibel substrings  
# then compare set(input_string) == output(substring)
def longest_substring(s):

    longest = 0
    for i in range(len(s)):
        seen = set()  
        for j in range(i, len(s)):
            if s[j] in seen:
                break 
            seen.add(s[j])            
            longest = max(longest, j- i + 1)
    return longest

s = "abcabcbb"

result = longest_substring(s)
print (result)

# Complexity: Time O(n^2), Space O(1)
```

```python
# The optimal soltuion - leet code submission 
# Sliding windpw + set 

def longest_substring_optimal(s):
    seen = set() 
    left = longest = 0 

    for right, char in enumerate(s):
        while char in seen:
            seen.remove(s[left])
            left += 1
        seen.add(char)
        longest = max(longest, right - left + 1)
    return longest

s = "abcabcbb"

result = longest_substring_optimal(s)
print (result)

# Complexity: Time - O(n), Space(n)
```

Time: O(n) expected. Each character enters the window once and leaves at most once, despite the nested while.

Extra space: O(min(n, number of possible characters)).

To avoid repeatedly remocing characters .. store the last seen index in a dictionary. and jump left directly past its previous occurance

```python
def longest_substring_optimal2(s):
    last_seen = {}
    left = longest = 0 
    for right, char in enumerate(s):
        if char in last_seen:
            left = max(left, last_seen[char] + 1)

        last_seen[char] = right
        longest = max(longest, right - left + 1)

    return longest 
```

Time: O(n) expected. Space: O(min(n, alphabet size)).

Both versions are linear-time; this version replaces repeated removals with a direct boundary jump.

```python
# Another optimal solution 

def length_of_longest_substring(s):
    seen = set()
    left = 0
    best = 0

    for right in range(len(s)):

        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])

        best = max(best, right - left + 1)

    return best

```

```python
# The Naive Solution

# Test every single possible substring within the string
# For each substring we check if all its characters are completely unique. 
# If the are, we calculate its length and track the maximum length seen. 

def longest_substring_naive(s):
    def all_unique(substring):
        # A set drops duplicates; if lengths match, all characters are unique
        return len(substring) == len(set(substring))

    max_len = 0 
    n = len(s)

    for i in range(n):
        for j in range(i+1, n+1):
            current_substring = s[i:j]
            # print (len(current_substring), set(current_substring))
            if all_unique(current_substring):
                max_len = max(max_len, len(current_substring))

    return max_len

# --- Verification ---
print(longest_substring_naive("abcabcbb"))  # Output: 3
# print(longest_substring_naive("pwwkew"))    # Output: 3
```


Complexity Analysis

• Time Complexity: O(N³) where N is the length of the string.

	• Generating all possible substring windows takes O(N²) loops.

	• Inside the loops, converting the substring into a set to check uniqueness takes linear time proportional to the substring length, adding another O(N) cost layer. This approach will quickly crash or time out on larger platforms.

• Space Complexity: O(min(N, M)) where M is the size of the alphabet set, to store characters in the helper unique uniqueness set verification map.


The Optimal Approach 

The Sliding Window (Two-Pointer) technique with a Hash Map tracker is the gold standard approach for string boundary problems. 

It completely eliminates the nested lookup loops, bringing the time complexity down from a sluggish O(N^3) to a lightning-fast O(N).


The Sliding Window Strategy

Instead of resetting our search from scratch whenever we encounter a duplicate character, we use a dynamic window that expands and shrinks as it moves across the string:

1. Two Pointers: We use left to track the start of the current unique window and right to scan forward character-by-character.

2. The Hash Map Tracker: We maintain a dictionary to remember the most recent index position where we saw each character: { character: index }.

3. Handling Duplicates: As right moves forward, if it encounters a character that is already in our Hash Map, it means we found a duplicate!
	• The Catch: We only care if the duplicate is inside our current window (to the right of our left pointer).
	• If it is, we instantly jump our left pointer forward to map[char] + 1 (just past the last known position of that duplicate).

4. Updating Max: At each step,the current window length is right - left + 1. 
    We update our running maximum.
