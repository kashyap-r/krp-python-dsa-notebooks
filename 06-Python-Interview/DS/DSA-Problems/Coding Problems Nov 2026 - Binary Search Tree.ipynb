### Binary Search Tree

```python
from collections import deque

class TreeNode:
    def __init__(self, value):
        self.value = value 
        self.left = None 
        self.right = None 

def insert_bst_rec(root, key):
    if not root:
        return TreeNode(key)
    if key == root.value:
        return root 
    if key < root.value:
        root.left = insert_bst_rec(root.left, key)    
    else: 
        root.right = insert_bst_rec(root.right, key)
    return root

def insert_bst_iterative(root, key):
    new_node = TreeNode(key)
    if not root:
        return new_node     
    
    current = root
    parent = None
    
    # 1. Navigate to the insertion spot while keeping track of the parent
    while current:
        parent = current
        if current.value == key:
            return root
        elif key < current.value:
            current = current.left 
        else:
            current = current.right 
            
    # 2. Attach the new node to the parent
    if key < parent.value:
        parent.left = new_node
    else:
        parent.right = new_node 
    return root

def display_tree(root):
    if root is None:
        return
    
    # Using the imported deque for a proper level-order traversal
    queue = deque([root])
    while queue:
        current = queue.popleft()
        print(f"{current.value}", end=" ")
        
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)
    print()

# Usage 
root = TreeNode(2)
root = insert_bst_iterative(root, 1)
root = insert_bst_rec(root, 3)
root = insert_bst_iterative(root, 6)
root = insert_bst_rec(root, 4)

display_tree(root)
```
