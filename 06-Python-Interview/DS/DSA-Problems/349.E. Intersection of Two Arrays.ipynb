349.E. Intersection of Two Arrays

Given two integer arrays nums1 and nums2, return an array of their intersection. Each element in the result must be unique and you may return the result in any order.

Example 1:

Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2]

Example 2:

Input: nums1 = [4,9,5], nums2 = [9,4,9,8,4]
Output: [9,4]
Explanation: [4,9] is also accepted.



```python
# Naive Approach 

def arr_intersection_naive(arr1, arr2):
    result = set() 

    for num in arr1:
        if num in arr2:
            result.add(num)
    return list(result)

# Usage
nums1 = [4,9,5]
nums2 = [9,4,9,8,4]

print (arr_intersection_naive(nums1, nums2))
```

Complexity: 
Time: O(N * M)
Space: O(min(N, M))

Another naive solution (Sorting + Binary Search)

If we sort nums2 first, we can replace the slow linear lookup with a highly optimized Binary Search 

Complexity: 
Time: O(MlogM + NlogM). Sorting nums2 takes O(MlogM), and performing a binary search for each of the N elements takes O(NlogM)

Space: O(1) if we ignore the space used by the sorting algorithm


The Optimal Solution (Hash Set Lookup)

We can optimize the runtime down to absolute linear time by trading a tiny amount of memory. 

Instead of looking up items inside a raw list, we convert nums1 into a Hash Set. Checking if an item exists inside a Python set runs in (O(1) constant time.
