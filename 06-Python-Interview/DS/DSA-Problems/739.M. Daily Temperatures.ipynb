739. Daily Temperatures
Medium

Given an array of integers temperatures represents the daily temperatures, return an array answer such that answer[i] is the number of days you have to wait after the ith day to get a warmer temperature. If there is no future day for which this is possible, keep answer[i] == 0 instead.

Example 1:

Input: temperatures = [73,74,75,71,69,72,76,73]
Output: [1,1,4,2,1,1,0,0]

Example 2:

Input: temperatures = [30,40,50,60]
Output: [1,1,1,0]

Example 3:

Input: temperatures = [30,60,90]
Output: [1,1,0]

Constraints:

    1 <= temperatures.length <= 105
    30 <= temperatures[i] <= 100

Hint 1
If the temperature is say, 70 today, then in the future a warmer temperature must be either 71, 72, 73, ..., 99, or 100. We could remember when all of them occur next.

This is a “next greater element” problem: for each day, find the first future day with a strictly warmer temperature, then return the distance between their indices.

1. Analysis and Constraints 
    - The array is non-empty 
    - Ewual temperatures do not count was warmer 
    - We need the first warmer future day, not the hottest future day 
    - The final day's answer is always 0 
    - Preserve the input and return a separate asnwer array 

The main challenge is avoiding repeated scans of future days.

2. Naive Solution: search forward from every day
For each index i, examine subsequent days until finding a temperature greater than temperatures[i]


```python
def daily_temperatures_naive(temperatures):
    n = len(temperatures)
    answer = [0] * n 

    for i in range(n):
        for j in range(i+1, n):
            if temperatures[j] > temperatures[i]:
                answer[i] = j - 1
                break 
    return answer 

```

3. Optimal approach: monotonic stack

Instead of searching forward separately for each day:
Keep unresolved days on a stack. When a warmer day arrives, resolve every colder day at the top.

Store indices, because we need to calculate the number of days between temperatures.

The temperatures at those indices remain non-increasing from bottom to top. Equal temperatures may coexist.

For each day:
1. While today is warmer than the day at the stack’s top:
   - Pop that earlier day.
   - Set its answer to today_index - earlier_index.
2. Push today’s index onto the stack.
3. Any indices left at the end keep their answer of 0.

```python
class Solution:
    def dailyTemperatures(self, temperatures: list[int]) -> list[int]:
        answer = [0] * len(temperatures)
        stack = []  # Indices of days still waiting for a warmer day.

        for today, temperature in enumerate(temperatures):

            # Today resolves all colder days at the top.
            while stack and temperature > temperatures[stack[-1]]:
                previous_day = stack.pop()
                answer[previous_day] = today - previous_day

            # Today now waits for its own warmer future day.
            stack.append(today)

        return answer
```
