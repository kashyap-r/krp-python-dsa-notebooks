
Problem Title: Remove Duplicates from Sorted Array II

• LeetCode Problem Number: 80
• Difficulty Level: Medium

Problem Description

Given an integer array nums sorted in non-decreasing order, remove some duplicates in-place such that each unique element appears at most twice. The relative order of the elements should be kept the same.

Since it is impossible to change the length of the array in some languages, you must instead have the result be placed in the first part of the array nums. 

More formally, if there are k elements after removing the duplicates, then the first k elements of nums should hold the final result. It does not matter what you leave beyond the first k elements.

Return k after placing the final result in the first k slots of nums.
Do not allocate extra space for another array. You must do this by modifying the input array in-place with O(1) extra memory.


Constraints

• 1 ≤ nums.length ≤ 3 × 10⁴
• -10^4 <= nums[i] <= 10^4
• nums is sorted in non-decreasing order.

Example 1: 
Input: nums = [1, 1, 1, 2, 2, 3]
Output: 5, nums = [1, 1, 2, 2, 3, _]
Explanation: Your function should return k = 5, with the first five elements of nums being 1, 1, 2, 2 and 3 respectively.

Example 2:
Input: nums = [0, 0, 1, 1, 1, 1, 2, 3, 3]
Output: 7, nums = [0, 0, 1, 1, 2, 3, 3, _, _]
Explanation: Your function should return k = 7, with the first seven elements of nums being 0, 0, 1, 1, 2, 3 and 3 respectively.


The Naive Solution 

The intuitive approach is to count the frequency of each number using a tracking dictionary or hash map, and then rebuild the array using only the elements that have a count capped at 2. 


```python
def removeDuplicates2_naive(arr):
    from collections import Counter

    #1. Count the frequencies of each element 
    counts = Counter(arr)

    #2. Rebuild the unique elements list, allowing at most 2 of each 
    rebuilt = []
    for num in arr:
        # we process in order, but only add if we haven't hit the cap 
        # in the rebuilt list
        if rebuilt.count(num) < 2:
            rebuilt.append(num)

    # 3. Overwrite the original array in-place 
    k = len(rebuilt)
    arr[:k] = rebuilt 
    return k


```

• Time Complexity: O(N²) or O(N) depending on how cleanly you track the insertion count, but rebuilt.count(num) makes this specific snippet O(N²).

• Space Complexity: O(N) because it allocates separate storage blocks (counts and rebuilt), which violates the O(1) in-place interview constraint.



Method 2: The Optimal Solution (The Two-Pointer Look-Back Strategy)

Because the array is sorted, duplicates are still grouped tightly together. We can use a fast two-pointer technique.

Instead of looking at the immediate previous element, we look two positions behind the current write_index. 

If our scanning element (nums[read_index]) is different from nums[write_index - 2], it means we haven't recorded two copies of this number yet. It is safe to write it down!

```python
def removeDuplicates_optimal(nums):
    # If the array has 2 or fewer elements, it naturally meets the constraint
    if len(nums) <= 2:
        return len(nums)
        
    # Start writing at index 2, since the first two elements are always safe
    write_index = 2
    
    for read_index in range(2, len(nums)):
        # Compare the scanned element with the element 2 positions behind the write pointer
        if nums[read_index] != nums[write_index - 2]:
            nums[write_index] = nums[read_index]
            write_index += 1
            
    return write_index

# --- Verification ---
nums = [0, 0, 1, 1, 1, 1, 2, 3, 3]
k = removeDuplicates_optimal(nums)
print(f"k = {k}")                # Output: 7
print(f"Modified Array: {nums[:k]}") # Output: [0, 0, 1, 1, 2, 3, 3]
```


Complexity Analysis

• Time Complexity: O(N) because it completes a single linear sweep across the array.
• Space Complexity: O(1) absolute constant memory.
