217.E.Contains Duplicate 

Given an interger array nums, return True if any value appears twice in the array, and return false if every element is distinct.

Example 1:
nums = [1, 2, 3, 1]

Output = True 

Input nums = [1, 2, 3, 4]
Output False 


Constraints or Clarifications: 

1. Whether the array has only positive, negative or zero values 
2. Any limits on the size of the array 
3. CanI sort the input 

Edge Cases: 
1. Empty array 
2. Array with one element 
3. Array with two elements 
4. Array with all distinct elements 

```python
# latest Submission

def containsDuplicate(nums):    
    return len(nums) != len(set(nums))

nums = [1, 2, 3, 4]
containsDuplicate(nums)
```

```python
## Old Submission 

from typing import List

class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        seen = set()
        for num in nums:
            if num in seen:
                return True
            seen.add(num)
        return False
    
""" 
Time Complexity: O(n)
Space Complexity: O(n)
"""

Soln = Solution()
arr = [-1, -1]
print (f"Duplicate pressent in array -> {Soln.containsDuplicate(arr)}")

arr = [0, 1, 2, 3, 4, 5, 5.2]
print (f"Duplicate pressent in array -> {Soln.containsDuplicate(arr)}")

arr = [1, 2, 3, 4]
print (f"Duplicate pressent in array -> {Soln.containsDuplicate(arr)}")

arr = []
print (f"Duplicate pressent in array -> {Soln.containsDuplicate(arr)}")

arr = [-1]
print (f"Duplicate pressent in array -> {Soln.containsDuplicate(arr)}")
```
