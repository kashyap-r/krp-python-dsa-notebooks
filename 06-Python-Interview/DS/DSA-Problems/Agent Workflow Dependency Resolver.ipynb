Problem #1 - Agent workflow Dependency Resolver
Medium Graph - HashMap - Queue - Topological Sort

An AI workflow platform executes multiple tasks. Some tasks depend on other tasks completing successfully.

You are given:
- tasks: a list of unique task identifiers.
- dependencies: a list of pairs [A, B], meaning task A must complete before task B can execute.

Implement:
def resolve_workflow(tasks, dependencies):
    pass

Return a valid execution order containing every task exactly once. If the dependencies contain a cycle, return an empty list.

When multiple tasks are ready, any valid order is acceptable.

Input 
tasks = ["retrieve", "embed", "rerank", "generate", "evaluate"]
dependencies = [
    ["retrieve", "embed"],
    ["embed", "rerank"],
    ["rerank", "generate"],
    ["generate", "evaluate"]
]

Expected Output
["retrieve", "embed", "rerank", "generate", "evaluate"]

Two Common choices:
    - DFS with cycle detection 
    - Kahn algorithm using indegree counts and a queue

For this interview, I'd start with Kahn's algorithm because it's iterative, relatively straightforward in Python, and naturally models tasks becoming ready.
