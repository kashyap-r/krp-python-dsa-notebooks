20. Valid Parentheses
Easy

Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

An input string is valid if:
    Open brackets must be closed by the same type of brackets.
    Open brackets must be closed in the correct order.
    Every close bracket has a corresponding open bracket of the same type.

Example 1:
Input: s = "()"
Output: true

Example 2:
Input: s = "()[]{}"
Output: true

Example 3:
Input: s = "(]"
Output: false

Example 4:
Input: s = "([])"
Output: true

Example 5:
Input: s = "([)]"
Output: false

Constraints:

    1 <= s.length <= 104
    s consists of parentheses only '()[]{}'.

Hint 1: Use a stack of characters.
Hint 2: When you encounter an opening bracket, push it to the top of the stack.
Hint 3: When you encounter a closing bracket, check if the top of the stack was the opening for it. If yes, pop it from the stack. Otherwise, return false.


#### Analysis 

This is a Stack problem because the most recently opened bracket must be closed first

1. Constraints and Clarifications 
- length of the string 
- characters are limited to just the paranthesis (){}[]
- Return Boolean 
- DO NOT NEED to handle letters, spaces, or other characters 

2. Solution Approach
Counting opening and closing brackets is not enough..
    - the counts should not only match bu the nesting should be correct 

We need to remember which bracket was opened most recently. A stack provides this last-in, first out behaviour 

1. OPening bracket -> push onto the stack 
2. Closing bracket -> it must match the top of the stack 
3. If the stack is empty or the type is wrong -> return False 
4. After procesing everything stack must be empty

Useful early check: an odd-length string cannot be valid because every bracket needs a partner.


#### Naive Solution 

4. Naive solution: repeatedly remove matching pairs

A valid nonempty bracket string always contains an innermost adjacent pair: (), [], or {}.

Repeatedly remove those pairs:
- If everything disappears, the string was valid.
- If a pass removes nothing but characters remain, it was invalid.

Time: O(n²) worst case.
Auxiliary space: O(n) for temporary strings.

```python
def is_valid_naive(s: str) -> bool:
    while s:
        reduced = s.replace("()", "").replace("[]", "").replace("{}", "")
        print (reduced)
        # No matching adjacent pair could be removed.
        if reduced == s:
            return False

        s = reduced

    return True

print (is_valid_naive("{}"))
print (is_valid_naive("({)}"))
```

```python
# Optimal Solution 
class Solution:
    def isValid(self, s: str) -> bool:
        if len(s) % 2 != 0:
            return False

        matching = {
            ")": "(",
            "]": "[",
            "}": "{"
        }

        stack = []

        for char in s:
            if char in matching:
                # A closing bracket needs a matching opening
                # at the top of the stack.
                if not stack or stack.pop() != matching[char]:
                    return False
            else:
                # The input contains only bracket characters,
                # so this must be an opening bracket.
                stack.append(char)

        # Any remaining openings were never closed.
        return not stack

# Usage
solution = Solution()  # Use the Solution class from above.

print(solution.isValid("()[]{}"))  # True
print(solution.isValid("{[]}"))    # True
print(solution.isValid("([)]"))    # False
print(solution.isValid("(("))      # False
```

Mandatory edge cases

| Case | Input | Expected | What it catches |
|---|---|---|---|
| One character | `"("` | `False` | Odd length |
| Simple pair | `"()"` | `True` | Basic matching |
| Adjacent pairs | `"()[]{}"` | `True` | Independent valid pairs |
| Nested pairs | `"{[]}"` | `True` | Correct nesting |
| Wrong bracket type | `"(]"` | `False` | Type mismatch |
| Incorrect nesting | `"([)]"` | `False` | Counts match, order fails |
| Closing bracket first | `")("` | `False` | Empty-stack handling |
| Unclosed brackets | `"(("` | `False` | Nonempty stack at the end |

```python
# Naive Solution 

def is_valid_naive(s: str) -> bool:
    while s:
        # Find one adjacent matching pair.
        for i in range(len(s) - 1):
            pair = s[i:i + 2]
            print (pair)
            if pair in ("()", "[]", "{}"):
                # Remove these two characters.
                s = s[:i] + s[i + 2:]
                break  # Restart scanning the shortened string.
        else:
            # The for-loop finished without finding a pair.
            return False

    # Every bracket was successfully removed.
    return True

# Usage

print(is_valid_naive("{[()]}"))  # True
print(is_valid_naive("()[]{}"))  # True
print(is_valid_naive("([)]"))    # False
print(is_valid_naive(")("))      # False
print(is_valid_naive("(("))      # False
```
