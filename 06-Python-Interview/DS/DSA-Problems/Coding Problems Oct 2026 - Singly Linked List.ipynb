Problem #1 

Reverse a Singly LInked List (Iterative) 

Intuition
    - Flip arrows one by one 
    - Maintain prev, curr, next 
    

```python
class Node:
    def __init__(self, data):
        self.data = data 
        self.next = None 
    
class SinglyLinkedList:
    def __init__(self):
        self.head = None 

    def reverse_list_iterative(self):
        prev = None 
        curr = self.head 

        while curr:
            # Store next
            nxt = curr.next 
            # reverse the current node's next pointer
            curr.next = prev 
            # move pointers one position ahead
            prev = curr 
            curr = nxt 
        self.head = prev 

    def printList(self):
        curr = self.head 
        while curr:
            print (f"{curr.data}", end="")
            if curr.next:
                print (" -> ", end="")
            curr = curr.next
        print ()
```

Leetcode #206 (Difficulty: Easy)
Given the head of a singly linked list, reverse the list, and return the reversed list.

Input: head = [1,2,3,4,5]
Output: [5,4,3,2,1]

Follow up: A linked list can be reversed either iteratively or recursively. Could you implement both?

```python
# Approach #3
# 1. define the structure of a node
class Node:
    def __init__(self, data=0, next=None):
        self.data = data
        self.next = next 

# 2. helper function to construct a singly linked list from an array
def array_to_linked_list(arr):
    if not arr:
        return None 
    head = Node(arr[0])
    curr = head
    for i in range(1, len(arr)):
        curr.next = Node(arr[i])
        curr = curr.next
    return head

# 3. helper function to print out a linked list 
def print_list(head):
    if not head:
        return "list empty"
    curr = head 
    while curr:
        print (f"{curr.data}", end="")
        if curr.next:
            print (f" -> ", end="")
        curr = curr.next
    print()

# 4. The core in-place reversal function leetcode #206
def reverse_linked_list(head):
    prev = None 
    curr = head 
    while curr:
        next_node = curr.next # 1. temporarily save the next node's address
        curr.next = prev # 2. FLIP the pointer backwards to point to prev
        prev = curr # 3. Move prev forward to the current node
        curr = next_node # 4. Move curr forward to the saved node

    # prev ends up sitting exactly on the new head node of the reversed list
    return prev 
"""
Time Complexity: O(N) because the loop processes / touches each of the nodes exactly once

Space Complexity: O(1) we usedjust the one variable and manipulated existing ponter references 
in place without instantiating new nodes.
"""

# 5. Follow up - implement the linked list reversal recursively
def reverse_list_recursive(head):
    # base case: empty list or single element list
    if not head or not head.next: 
        return head

    # recursive step: reverse the entire remaining rest of the chain 
    new_head = reverse_list_recursive(head.next)

    # The flip: make the next node point back to the current node
    head.next.next = head 

    # disconnect: cut the current node'd old forward pointer arrow
    head.next = None

    # return the new head node that was bubbled up from the base case
    return new_head

# --- Execution Framework Verification ---
array_input = [10, 20, 30, 40, 50]

# Construct the linked list
list_head = array_to_linked_list(array_input)
print("Original Singly Linked List:")
print_list(list_head)

# Reverse the linked list in-place
reversed_head = reverse_linked_list(list_head)
print("\nReversed Singly Linked List:")
print_list(reversed_head)

print("\nReversed the reversed Singly Linked List recursively:")
straightened_head = reverse_list_recursive(reversed_head)
print_list(straightened_head)
```

The Recursive Strategy

1. Base Case: If the list is empty (head is None) or contains only a single item (head.next is None), it is already reversed. Return it immediately as the new head.

2. The Leap of Faith: Assume the rest of the list starting from head.next is already successfully reversed by a recursive call. This call returns the new_head (the original tail node).

3. The Flip: The node immediately following our current node (head.next) needs to point its next pointer back at our current node (head.next.next = head).

4. Disconnect: Break the current node's old forward link by setting head.next = None so we don't accidentally create an infinite loop.


Visual Step-by-Step Call Stack Trace

Let's look at a short 3-node list: 1 -> 2 -> 3 -> None.

1. reverse_list_recursive(Node 1): head.next is Node 2. Calls reverse_list_recursive(Node 2).
2. reverse_list_recursive(Node 2): head.next is Node 3. Calls reverse_list_recursive(Node 3).
3. reverse_list_recursive(Node 3): [BASE CASE] Node 3 has no next item. Returns Node 3.

Now the call stack begins to unwind and execute the flip steps:
• Unwinding back to Node 2 (head = Node 2):
	• new_head is currently tracking Node 3.
	• head.next.next = head translates to: Node 2.next.next = Node 2 → Node 3.next = Node 2. (Arrow flipped!)
	• head.next = None translates to: Node 2.next = None.
	• Current structure layout: 1 -> 2 <- 3 and 2 -> None. Returns Node 3.
• Unwinding back to Node 1 (head = Node 1):
	• new_head is still tracking Node 3.
	• head.next.next = head translates to: Node 1.next.next = Node 1 → Node 2.next = Node 1. (Arrow flipped!)
	• head.next = None translates to: Node 1.next = None.
	• Current structure layout: None <- 1 <- 2 <- 3. Returns Node 3.
Final Resulting Output: The function returns Node 3, which serves as the new entry point head for the completely inverted string layout: 3 -> 2 -> 1 -> None.


Complexity Verification

• Time Complexity: O(N) because the execution frames travel down to the tail node and back up to the head node, performing fixed O(1) pointer operations at each node exactly once.
• Space Complexity: O(N) extra space allocation needed strictly to support the call stack frames. If a linked list has 50,000 nodes, it will create 50,000 recursive execution frames in memory, making the iterative three-pointer framework more memory-efficient for scaling real-world system loads.
