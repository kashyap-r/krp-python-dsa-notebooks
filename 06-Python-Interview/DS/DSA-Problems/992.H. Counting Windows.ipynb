992. Subarrays with K Different Integers

Given an integer array nums and an integer k, return the number of good subarrays of nums.

A good array is an array where the number of different integers in that array is exactly k.

    For example, [1,2,3,1,2] has 3 different integers: 1, 2, and 3.

A subarray is a contiguous part of an array.

Input: nums = [1,2,1,2,3], k = 2
Output: 7
Explanation: Subarrays formed with exactly 2 different integers: 
[1,2], [2,1], [1,2], [2,3], [1,2,1], [2,1,2], [1,2,1,2]


Analysis: 
What is needed: 
    Count subarray 
    exactly k distinct (elements)
    contiguous 

The key relatinship: 
    Exactly(K)= AtMost(K) - AtMost(K-1)

Illustration: 
nums = [1 ,2 ,1]
k = 2 

All non-empty subarrays 

[1]          1 distinct
[1,2]        2 distinct
[1,2,1]      2 distinct
[2]          1 distinct
[2,1]        2 distinct
[1]          1 distinct
Therefore 

AtMost(2) = 6
AtMost(1) = 3

Exactly(2) = 6 - 3 = 3





Illustration: 

Look at the wording k = 2, means array haveing 2 distinct values, NOT length of 2.

For nums = [1, 2, 1, 2, 3] and k = 2, there are 7 qualifying subarrays:

| Indices (inclusive) | Subarray | Distinct values |
|---|---|---|
| `0–1` | `[1, 2]` | `{1, 2}` |
| `1–2` | `[2, 1]` | `{1, 2}` |
| `2–3` | `[1, 2]` | `{1, 2}` |
| `3–4` | `[2, 3]` | `{2, 3}` |
| `0–2` | `[1, 2, 1]` | `{1, 2}` |
| `1–3` | `[2, 1, 2]` | `{1, 2}` |
| `0–3` | `[1, 2, 1, 2]` | `{1, 2}` |

The two occurrences of [1, 2] count separately because they occupy different positions.

[1, 2, 3] at indices 2–4 does not qualify—it has three distinct values.


```python
from collections import defaultdict

def at_most_k(nums, k):
    if k < 0:
        return 0

    freq = defaultdict(int)
    left = 0
    count = 0

    for right in range(len(nums)):
        freq[nums[right]] += 1

        while len(freq) > k:
            freq[nums[left]] -= 1

            if freq[nums[left]] == 0:
                del freq[nums[left]]

            left += 1

        count += right - left + 1

    return count


def exactly_k(nums, k):
    return at_most_k(nums, k) - at_most_k(nums, k - 1)
```

```python
def subarray(nums, k):
    n = len(nums)
    if not 1 <= k<= n:
        return []
    
    result = []
    for i in range(n - k + 1):
        curr = []
        for j in range(i, i+k):
            curr.append(nums[j])
        result.append(curr)
    return result

# usage
nums = [1,2,1,2,3]
k = 2

result = subarray(nums, k)
print (result)
```
