1.Easy. Two Sum
Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

Example 1:
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

Example 2:
Input: nums = [3,2,4], target = 6
Output: [1,2]

Example 3:
Input: nums = [3,3], target = 6
Output: [0,1]

Hints
- Brute force is two nested loops.
- Can previous values be remembered?
- Dictionary lookup can reduce O(n²) to O(n).

```python
class Solution: 
    def twoSum_basic(self, nums, target):
        n = len(nums)

        for i in range(len(nums)):
            for j in range(i+1, n):
                if nums[i] + nums[j] == target:
                    return (i, j)
        return None

    def twoSum(self, nums, target):
        # Dictionary to map: { number_value : its_index }
        seen = {}
        
        for i, num in enumerate(nums):
            complement = target - num
            
            # Check if the needed matching number has already been seen
            if complement in seen:
                return [seen[complement], i]
            
            # Otherwise, record the current number and its index
            seen[num] = i


if __name__=="__main__":
    # nums = [2, 7, 11, 15]
    nums = [2, 1, 11, 15, -11, 7]
    target = 9 
    twoSum = Solution() 
    result = twoSum.twoSum_basic(nums, target)
    print(result)
    d = twoSum.twoSum(nums, target)
    print (d)
```

"""
15.M. 3Sum
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] 
such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

"""

```python

# Using two pointer method
def find3Sum(arr):
    arr.sort()

    result = []

    for i in range(len(arr)-2):
        if i > 0 and arr[i] == nums[i-1]:
            continue
        # fix the first element, point the left pointer to the 2nd element
        left = i+1
        # second pointer top point to the last element
        right = len(arr)-1

        while left < right:
            total = arr[i] + arr[left] + arr[right]

            if total == 0:
                result.append([arr[i], arr[left], arr[right]])

                while left < right and arr[left] == arr[left+1]:
                    left+= 1 
                while left < right and arr[right] == arr[right-1]:
                    right -= 1
                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

if __name__ == "__main__":
    nums = [-1, 0, 1, 2, -1, -4]
    print (find3Sum(nums))
```

### Constructing a Binary Tree from a given array / list 

**Note: Use the Level-Order (Breadth-First Construction) Strategy**

Steps: 
1. Create a TreeNode class to hold the value, left and the right attributes 

2. Turn the first element (3) into root node and place it inside the  queue

3. Use a pointer or index tracking variable (1) to sweep through the rest of the array elements 

4. While the queue is not empty and you still have elements in the array:
    - Pop the fron parent node off the queue

    - Grab the next array elelemtn for the left child. If it isn't None, build a node, attach it, and push into the queue. 

    - Grab the next array element for the right child. If it isn't None, build a node attach it, and push it into the queue.
    

```python
from collections import deque

class TreeNode: 
    def __init__(self, val):
        self.value = value
        self.left = None 
        self.right = None 

# build the tree using the lever order search algo
def build_binary_tree(values):
    if not values or values[0] is None:
        return None 
    
    # the first value of the array being a node
    root = TreeNode(values[0])
    queue = deque([root])

    i = 1 # index of the next child entry on the array

    while queue and i < len(values):
        parent = queue.popleft()

        # next entry represents this parent's left child
        if values[i] is not None: 
            parent.left = TreeNode(values[i])
            queue.append(parent.left)
        
        i += 1 

        # Following entry represents this parent's right child 
        if i< len(values) and values[i] is not None: 
            parent.right = TreeNode(values[i])
            queue.append(parent.right)

        i += 1
    return root

# now print the tree and check 
def display_binary_tree(root):
    if root is None:
        return 

    queue = deque([root])

    while queue:
        level = []
        for _ in range(len(queue)):
            node = queue.popleft()
            level.append(node.value)

            if node.left:
                queue.append(node.left)

            if node.right:
                queue.append(node.right)
        print (level)
    
values = [3, 9, 20, None, None, 15, 7]

root = build_binary_tree(values)
display_binary_tree(root)
```

Given a binary tree and a target sum, return all root-to-leaf paths whose node values sum to the target.
Input: [5,4,8,11,None,13,4,7,2,None,None,5,1]
Target: 22
Output: [[5,4,11,2],[5,8,4,5]]

Function: path_sum(root, target) -> list[list[int]]
Hints
Maintain a current path and remaining sum. 
Append a node before exploring its children, record a copy when a leaf satisfies the sum, and pop() the node when returning. 

This is classic backtracking.

Approach: 

Use DFS with Backtracking 

Maintain the current root-to-node path and the sum still needed. 

Save a path only when you reach a leaf with the required sum.


```python
class Node: 
    def __init__(self, value):
        self.value = value 
        self.left = None 
        self.right = None 

def path_sum(root, target):
    result = []
    path = [] 

    def dfs(node, remaining):
        if node is None:
            return 

        # Choose: include this mode
        path.append(node.value)
        remaining -= node.value

        # check if it is a leaf.
        is_leaf = node.left is None and node.right is None

        # this is the place (leaf node) where the path adds up to the sum 
        if is_leaf and remaining == 0:
            result.append(path.copy())

        dfs(node.left, remaining)
        dfs(node.right, remaining)

        # undo remove this node before explaoring another branch
        path.pop() 

    dfs(root, target)
    return result

# Usage 
arr = [5, 4, 8, 11, None, 13, 4, 7, 2, None, None, 5, 1]
root = build_binary_tree(arr)

print(path_sum(root, 22))
# [[5, 4, 11, 2], [5, 8, 4, 5]]
```

Why Backtracking is neeeded?
When DFS reaches node 2:
    path = [5, 4, 11, 2]
    remaining = 0

It saves a copy, then removes 2:
result.append(path.copy())
path.pop()

As calls finish, each removes its own node. Eventually, the path returns to [5], ready to explore the right subtree.

| Line | Purpose |
|---|---|
| `result.append(path.copy())` | Saves an independent snapshot |
| `path.pop()` | Restores the path for other branches |

Without .copy(), every saved entry would reference the same mutable list, which becomes empty when traversal finishes.


Why don’t we restore remaining?
Each recursive call has its own local remaining, an immutable integer. Updating it does not change the caller’s value.

In contrast, all calls share the same mutable path list, so its changes must be undone.


Complexity
Let:
- n = number of nodes.
- h = maximum number of nodes in a root-to-leaf path.
- S = total number of values across all returned paths.

| Resource | Complexity | Reason |
|---|---:|---|
| Time | **O(n + S)** | Visit nodes once and copy successful paths |
| Auxiliary space, excluding output | **O(h)** | Recursion stack and current path |
| Output storage | **O(S)** | All returned path values |


If there are k matching paths, then S ≤ k × h, giving an alternative bound of O(n + kh) time.


Edge cases
Arrays below use our compact level-order format.

| Case | Tree | Target | Expected |
|---|---|---:|---|
| Empty tree | `[]` | 0 | `[]` |
| Matching single node | `[5]` | 5 | `[[5]]` |
| Nonmatching single node | `[5]` | 0 | `[]` |
| Zero-valued leaf | `[0]` | 0 | `[[0]]` |
| Target reached before a leaf | `[5, 2]` | 5 | `[]` |
| Negative value completes path | `[5, -3]` | 2 | `[[5, -3]]` |
| Zero extends a valid sum to a leaf | `[5, 0]` | 5 | `[[5, 0]]` |
| Different paths have identical values | `[1, 2, 2]` | 3 | `[[1, 2], [1, 2]]` |
| Skewed tree | `[1, None, 2, None, 3]` | 6 | `[[1, 2, 3]]` |

Problem 8 — Lowest Common Ancestor
Medium · 20 min

LeetCode 236 · DFS
Given the root of a binary tree and references to two distinct nodes p and q, find their lowest common ancestor. Assume both nodes exist in the tree. A node can be its own ancestor.
Tree: [3,5,1,6,2,0,8], p=5, q=1
Output: 3

Also asked as "Find the deepest shared ancestor of two nodes in a hierarchy."

Function: lowest_common_ancestor(root, p, q) -> Node
HintsIf the current node is p or q, return it. Search both subtrees. If both return a node, the current node is the LCA. Otherwise propagate whichever non-None result exists. Compare node identities, not just values.

The standard optimal solution is recursive DFS in O(n) time and O(h) auxiliary space. It finds where the paths to p and q first meet when working upward.

1. Naive approach: find and compare two root-to-node paths

Algorithm:
1. Use DFS to find the path from the root to p.
2. Use DFS again to find the path from the root to q.
3. Compare the paths from the beginning.
4. Return the last node shared by both paths before they diverge.

With backtracking and one saved path per target:
- Time: O(n)—two traversals are still O(n).
- Auxiliary space: O(h)—two paths plus the recursion stack.
This is a good solution. The improvement below avoids explicitly storing and comparing the paths; it does not improve the asymptotic complexity.

2. Efficient solution: let each subtree report what it found

At each node:
1. If it is None, return None.
2. If it is p or q, return that node.
3. Search the left and right subtrees.
4. If both sides return a node, the current node is the LCA.
5. Otherwise, return whichever side found something.


```python
def lowest_common_ancestor(root, p, q):
    if root is None:
        return None

    if root is p or root is q:
        return root

    left = lowest_common_ancestor(root.left, p, q)
    right = lowest_common_ancestor(root.right, p, q)

    if left is not None and right is not None:
        return root

    return left if left is not None else right

## UsAGE

root = build_binary_tree([3, 5, 1, 6, 2, 0, 8])

p = root.left   # Actual node containing 5
q = root.right  # Actual node containing 1

ancestor = lowest_common_ancestor(root, p, q)
print(ancestor.value)  # 3
```

Edge Cases 

| Case | `p`, `q` values | LCA value | Why |
|---|---|---:|---|
| Opposite sides of root | `5, 1` | 3 | Paths meet at root |
| Both in the same subtree | `6, 2` | 5 | Shared parent below root |
| One target is ancestor of the other | `5, 2` | 5 | A node can be its own ancestor |
| One target is the root | `3, 8` | 3 | Root is ancestor of every node |
| Both in the right subtree | `0, 8` | 1 | Shared parent below root |

Follow-up questions
| Question | Answer |
|---|---|
| What if one or both targets might be absent? | Add a presence check or return target counts along with the LCA candidate. Return `None` unless both are found. |
| What if the tree is a BST with unique values? | Move left if both targets are smaller, right if both are larger; otherwise return the current node. O(h) time and O(1) iterative space. |
| What if nodes already have parent pointers? | Align their depths, then move both upward until they meet. O(h) time, O(1) space. |
| What if `p` and `q` are the same node? | If allowed and present, that node is the answer; the recursive solution already handles it. |
| Can you find the distance between the nodes? | `depth(p) + depth(q) - 2 * depth(LCA)`, with depths measured in edges. |
| What if we have many LCA queries on a static tree? | Preprocess, for example with binary lifting: O(n log n) preprocessing/storage and O(log n) per query. |
| Why is the returned ancestor the *lowest*? | If both targets are in one subtree, its result propagates upward. A new ancestor is selected only where the targets split across the two sides. |

Problem :


Given a binary tree, determine whether the tree contains any duplicate subtree of size two or more. Two subtrees are considered duplicates if they have the same structure and identical node values.

Return true if such a duplicate subtree exists; otherwise, return false.

Note: Subtrees consisting of only a single leaf node are not considered duplicate subtrees.

root = [1, 2, 3, 4, 5, N, 2, N, N, N, N, 4, 5]

Follow up: Display the duplicate subtree may be as a list and also print at the levels they;re appearing

Approach: 

For every node, construct:

signature = (node.value, left_subtree_id, right_subtree_id)

```python
def has_duplicate_subtree(root):
    subtree_ids = {}
    found = False

    def identify(node):
        nonlocal found

        if node is None:
            return 0

        left_id = identify(node.left)
        right_id = identify(node.right)

        signature = (node.value, left_id, right_id)

        if signature in subtree_ids:
            # Repeated leaves do not count.
            if node.left is not None or node.right is not None:
                found = True
        else:
            subtree_ids[signature] = len(subtree_ids) + 1

        return subtree_ids[signature]

    identify(root)
    return found

## Usage 

arr = [
    1, 2, 3, 4, 5, None, 2,
    None, None, None, None, 4, 5
]

root = build_binary_tree(arr)

# print(has_duplicate_subtree_naive(root))  # True
print(has_duplicate_subtree(root))        # True
```

| Node being processed | Signature | Action |
|---|---|---|
| First leaf `4` | `(4, 0, 0)` | Assign ID `1` |
| First leaf `5` | `(5, 0, 0)` | Assign ID `2` |
| First parent `2` | `(2, 1, 2)` | Assign ID `3` |
| Second leaf `4` | `(4, 0, 0)` | Reuse ID `1`; ignore duplicate leaf |
| Second leaf `5` | `(5, 0, 0)` | Reuse ID `2`; ignore duplicate leaf |
| Second parent `2` | `(2, 1, 2)` | Signature repeats and node is not a leaf → **found!** |

```python
# Extended solution 

from collections import deque 

class Node: 
    def __init__(self, value):
        self.value = value 
        self.left = None 
        self.right = None

def subtree_as_list(root):
    """Seralize in level order, preserving missing child positions"""
    if root is None: 
        return []

    result = [] 

    queue = deque([root])

    while queue:
        node = queue.popleft() 

        if node is None:
            result.append(None)
            continue 

        result.append(node.value)
        queue.append(node.left)
        queue.append(node.right)

    # Trailing None entries are unnecessary 
    while result and result[-1] is None:
        result.pop() 
    return result

def find_duplicate_subtrees(root):
    signature_to_id = {}
    occurrences = {}
    representatives = {}

    def dfs(node, level):
        if node is None:
            return 0 

        left_id = dfs(node.left, level+1)
        right_id = dfs(node.right, level+1)

        signature = (node.value, left_id, right_id)

        if signature not in signature_to_id:
            signature_to_id[signature] = len(signature_to_id) + 1

        subtree_id = signature_to_id[signature]

        # At least one child means subtree size is >= 2.
        if node.left is not None or node.right is not None:
            if subtree_id not in occurrences:
                occurrences[subtree_id] = []
                representatives[subtree_id] = node

            occurrences[subtree_id].append(level)

        return subtree_id

    dfs(root, 1)

    duplicates = []

    for subtree_id, levels in occurrences.items():
        if len(levels) >= 2:
            duplicates.append({
                "subtree": subtree_as_list(
                    representatives[subtree_id]
                ),
                "levels": sorted(levels),
                "count": len(levels)
            })

    return bool(duplicates), duplicates

### USAGE 
# 
# 
arr = [
    1, 2, 3, 4, 5, None, 2,
    None, None, None, None, 4, 5
]

root = build_binary_tree(arr)

exists, duplicates = find_duplicate_subtrees(root)

print(exists)

for item in duplicates:
    print("Duplicate subtree:", item["subtree"])
    print("Root levels:", item["levels"])
    print("Occurrences:", item["count"])        


```
