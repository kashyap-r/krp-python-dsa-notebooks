For an Array, Subarray, Subsequence and Subsets 

For an array, nums = [10, 20, 30]

| Type | Must be contiguous? | Preserves original order? | Examples |
|---|---|---|---|
| **Subarray** | Yes | Yes | `[10, 20]`, `[20, 30]` |
| **Subsequence** | No | Yes | `[10, 30]`, `[10, 20]` |
| **Subset** | No | Order doesn’t matter | `{10, 30}` is the same subset as `{30, 10}` |


```python
# Generate all possible All subarrays

def all_subarrays(arr):
    n = len(arr)
    result = []

    for i in range(n):
        current = []
        for j in range(i, n):
            current.append(arr[i])
            result.append(current.copy())            
    return result 

arr = [10, 20, 30]
result = all_subarrays(arr)
print (result)
```

```python
def allsubarray_fixedsize(arr, k):
    n = len(arr)
    if not 1 <= k <= n:
        return []
    result = []
    for i in range(n-k+1):
        current = []
        for j in range(i, i+k):            
            current.append(arr[j])        
        result.append(current)
    return result 

# usage 
arr = [10, 20, 30]
k = 2 
result = allsubarray_fixedsize(arr, k)
print (result)
```

1. All Subarrays - Slicing 

Choose a Start and end, python excludes the end index. 

```python
# all sub arrays slice method
def subarrays_slice(arr):
    result = []
    for start in range(len(arr)):
        for end in range(start + 1, len(arr) + 1):
            result.append(arr[start:end])

    return result

result = subarrays_slice([1, 2, 3])
print (result)
```

```python
# All subarrays index method
# for each starting position, exten the current subarray one element at a time

def subarrays_index(arr):
    result = [] 

    for start in range(len(arr)):
        current = []
        for end in range(start, len(arr)):
            current.append(arr[end])
            result.append(current.copy())
    return result

"""
The copy is necessary: otherwise, every saved entry for that start 
would refer to the same growing current list.
"""
# usage
arr = [1, 2, 3]
n =len(arr)
result = subarrays_index(arr)
print (result)
print ()
# Below is TIme O(n^3)
for i in range(n):
    for j in range(i, n):
        for k in range(j, n):
            print (arr[k], end=' ')
        print()
    print ()
```

Subarrays of fixed length - Using Slicing

Once the start is know end is start + k 


```python
def fixed_subarrays_slice(arr, k):
    if not 1 <= k <= len(arr):
        return []

    return [
        arr[start:start + k]
        for start in range(len(arr) - k + 1)
    ]

# Usage
arr = [1, 2, 3, 4, 5] 
k = 3

res = fixed_subarrays_slice(arr, k)
print (res)
```

```python
def fixed_subarrays_index(arr, k):
    if not 1 <= k <= len(arr):
        return []

    return [
        [arr[i] for i in range(start, start + k)]
        for start in range(len(arr) - k + 1)
    ]
#usage 
arr = [10, 20, 30, 40]
k = 2 

result = fixed_subarrays_index(arr, k)
print (result)
```

All Subsequences - Slicing 

For each element, there are two choices: exclude it or include it.

Generate the subsequences of everthing after the first element. Then produce the versions without and withthe first element.

```python
def subsequences_slice(arr):
    if not arr:
        return [[]]

    rest = subsequences_slice(arr[1:])

    return rest + [arr[:1] + seq for seq in rest]

answer = subsequences_slice([10, 20, 30])
print (answer)
```

```python
def subsequences_index(arr):
    result = []
    def backtrack(start, path):
        result.append(path.copy())
        for i in range(start, len(arr)):
            path.append(arr[i])
            backtrack(i+1, path)
            path.pop() 

    backtrack(0, [])
    return result

subsequences_index([10, 20, 30])
```

Subsequences of fixed l;ength - k (using slicing)

```python
def fixed_subsequences_slice(arr, k):
    if k == 0:
        return [[]]

    if k < 0 or k > len(arr):
        return []

    without_first = fixed_subsequences_slice(arr[1:], k)
    with_first = [
        arr[:1] + seq
        for seq in fixed_subsequences_slice(arr[1:], k - 1)
    ]

    return without_first + with_first
```

```python
# Using index methpd

def fixed_subsequences_index(arr, k):
    if k < 0 or k > len(arr):
        return []

    result = []

    def backtrack(start, path):
        if len(path) == k:
            result.append(path.copy())
            return

        needed = k - len(path)

        # Only start where enough elements remain.
        for i in range(start, len(arr) - needed + 1):
            path.append(arr[i])
            backtrack(i + 1, path)
            path.pop()

    backtrack(0, [])
    return result

```

## Subsets 



```python
# All subsets — slicing
subsets = subsequences_slice([10, 20, 30])

# All subsets — index method
subsets = subsequences_index([10, 20, 30])

# Subsets of size 2 — slicing
subsets = fixed_subsequences_slice([10, 20, 30], 2)

# Subsets of size 2 — index method
subsets = fixed_subsequences_index([10, 20, 30], 2)

# Fixed-size selections:
# [10, 20], [10, 30], [20, 30]
# Output order varies by implementation.
```
