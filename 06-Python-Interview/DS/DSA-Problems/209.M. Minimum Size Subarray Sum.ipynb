```python
# naive length  

def min_subarray_len(s: int, arr: list[int]) -> int:
    n = len(arr) 
    min_length = float('inf')

    for start in range(n):
        current_sum = 0
        for end in range(start, n):
            current_sum += arr[end]
            if current_sum >= s:
                curr_length = end - start + 1 
                min_length = min(min_length, curr_length)
                break            
    return min_length if min_length != float('inf') else 0
 
arr = [2,3,1,2,4,3]
s = 7
print(min_subarray_len(s, arr))
```

```python
def min_subarray_len(s: int, arr: list[int]) -> int:
    n = len(arr) 
    # initialize with infinity
    min_length = float('inf')

    # try every possible starting point (from the begining)
    for start in range(n):
        curr_sum = 0 
        # expand the subarray from the starting point
        for end in range(start, n):
            curr_sum += arr[end]

            if curr_sum >= s:
                curr_length = end - start + 1
                min_length = min(min_length, curr_length)
                break
    return min_length if min_length != float('inf') else 0   
    
```

```python
# Alternative approach using sliding window technique
def min_subarray_len(s: int, arr: list[int]) -> int:
    n = len(arr)
    min_length = float('inf')
    curr_sum = 0
    start = 0

    for end in range(n):
        curr_sum += arr[end]

        while curr_sum >= s:
            min_length = min(min_length, end - start + 1)
            curr_sum -= arr[start]
            start += 1

    return min_length if min_length != float('inf') else 0  
```

```python
# alternative approach using prefix sum and binary search
def min_subarray_len(s: int, arr: list[int]) -> int:
    n = len(arr)
    prefix_sum = [0] * (n + 1)

    for i in range(n):
        prefix_sum[i + 1] = prefix_sum[i] + arr[i]

    min_length = float('inf')

    for i in range(n):
        target = s + prefix_sum[i]
        left, right = i + 1, n

        while left <= right:
            mid = (left + right) // 2
            if prefix_sum[mid] >= target:
                min_length = min(min_length, mid - i)
                right = mid - 1
            else:
                left = mid + 1

    return min_length if min_length != float('inf') else 0    
```
