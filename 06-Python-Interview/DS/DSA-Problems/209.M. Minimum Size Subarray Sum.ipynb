Find the smallest subarray with a sum greater than or equal to a given value S. 


Solution Approach:

1. (Re)state the problem 
Problem Statement: Find the smallest subarray with a sum greater than or equal to a given value S. 

2. Identify, clarify and confirm the boundaries / contraints
    - Whether the array contains only positive values ?
    - Alternatively, can I assume the array to contain only positive values

    - What should I return if no such sub array exists (0 or 1)

    Boundaries 
        - empty input array: Return 0 
3. naive Solution 

    for i in range (len (arr)-1): 
        for j in range (i,n):
            if the sum becomes greater than or equal to S, calculate the subarray length and update your minimum lebgth




```python
# Correct naive solution 31st Oct, 2026 

def min_subarray_len(arr, target):
    shortest = float("inf")

    for i in range(len(arr)):
        total = 0 
        for j in range(i, len(arr)):
            total += arr[j]

            if total >= target:
                shortest = min(shortest, j - i + 1)
                break
    return 0 if shortest == float("inf") else shortest

# usage 
arr = [2, 3, 1, 2, 4, 3]
target = 7 

result = min_subarray_len(arr, target)
print (result)

# Complexity: Time - O(N^2), Space - O(1)
```

```python
# to pront the array as well 

def min_subarray(arr, target):
    left = total = start = 0
    shortest = float("inf")

    for right, value in enumerate(arr):
        total += value

        while total >= target:
            if right - left + 1 < shortest:
                shortest = right - left + 1
                start = left

            total -= arr[left]
            left += 1

    return [] if shortest == float("inf") else arr[start:start + shortest]

arr= [2, 3, 1, 2, 4, 3]
print(min_subarray(arr, 7)) 
```

Time: O(n). Auxiliary space: O(1), plus O(k) for the returned slice of length k. If multiple shortest subarrays qualify, this returns the first one.

Interview trap
Why does the problem specify positive integers?

Because adding a positive number cannot decrease the sum, and removing one cannot increase it.

With negative numbers, that monotonic behavior disappears. The standard shrinking strategy may fail.

```python
# Alternative approach using sliding window technique
def min_subarray_len(s: int, arr: list[int]) -> int:
    n = len(arr)
    min_length = float('inf')
    curr_sum = 0
    start = 0

    for end in range(n):
        curr_sum += arr[end]

        while curr_sum >= s:
            min_length = min(min_length, end - start + 1)
            curr_sum -= arr[start]
            start += 1

    return min_length if min_length != float('inf') else 0  
```

```python
# alternative approach using prefix sum and binary search
def min_subarray_len(s: int, arr: list[int]) -> int:
    n = len(arr)
    prefix_sum = [0] * (n + 1)

    for i in range(n):
        prefix_sum[i + 1] = prefix_sum[i] + arr[i]

    min_length = float('inf')

    for i in range(n):
        target = s + prefix_sum[i]
        left, right = i + 1, n

        while left <= right:
            mid = (left + right) // 2
            if prefix_sum[mid] >= target:
                min_length = min(min_length, mid - i)
                right = mid - 1
            else:
                left = mid + 1

    return min_length if min_length != float('inf') else 0    
```
