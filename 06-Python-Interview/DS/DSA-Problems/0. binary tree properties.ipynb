## Binary Tree


What is the best method to find the dept / height of a binary tree?

Both BFS and DFs fidn the tree's height in O(n) time. 

**Choose based on the tree's shape, python's recursion limit, and whether you need maximum depth and minimum depth**

For interview's, recursive DFS is usually the simplest starting point. 

1. First: depth versus height

- Depth of a node: Number of edges from the root to that node. Root depth = 0.
- Height of a node: Number of edges on the longest downward path from that node to a leaf. Leaf height = 0.
- Height of the tree: Height of its root, equal to the maximum node depth.
However, many coding problems define maximum depth as the number of nodes/levels, not edges.

For the below tree, 
[3, 9, 20, None, None, 15, 7]

| Level | Nodes | Depth in edges |
|---|---|---|
| 1 | `3` | 0 |
| 2 | `9, 20` | 1 |
| 3 | `15, 7` | 2 |

So its height in edges is 2, and its maximum depth in levels is 3.

```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None 

class BinaryTree:
    def __init__(self):
        self.root = None 

    def build_tree_from_array(self, nums):
        if not nums or nums[0] is None: 
            return None 
        root = Node(nums[0])
        queue = deque([root])

        i = 1 # index of the next child entry
        while queue and i < len(nums):
            parent = queue.popleft()

            # next entry represents this parent's left child
            if nums[i] is not None:
                parent.left = Node(nums[i])
                queue.append(parent.left)

            i += 1

            # next entry represents this parent's right child 
            if nums[i] is not None:
                parent.right = Node(nums[i])
                queue.append(parent.right)

            i += 1
        return root 

    def print_levels(self, root):
        if root is None:
            return

        queue = deque([root])

        while queue:
            level = []

            # Process only the nodes currently in this level.
            for _ in range(len(queue)):
                node = queue.popleft()
                level.append(node.data)

                if node.left is not None:
                    queue.append(node.left)

                if node.right is not None:
                    queue.append(node.right)

            print(level)

    # post order DFS - to find the max depth from the root
    def max_depth_dfs(self, root):
        if root is None:
            return 0 

        left_dept = self.max_depth_dfs(root.left)
        right_depth = self.max_depth_dfs(root.right)

        return 1 + max(left_dept, right_depth)

    # BFS - Count the levels 
    def max_dept_bfs(self, root):
        if root is None:
            return 0 

        queue = deque ([root])
        max_depth = 0 
        while queue:
            level_size= len(queue)
            for _ in range(level_size):
                current = queue.popleft()
                if current.left:
                    queue.append(current.left)
                if current.right:
                    queue.append(current.right)
            max_depth += 1 
        return max_depth

# usage
nums = [3, 9, 20, None, None, 15, 7]

bt = BinaryTree()
root = bt.build_tree_from_array([3, 9, 20, None, None, 15, 7])
bt.print_levels(root)

print (bt.max_depth_dfs(root))
print (bt.max_dept_bfs(root))
```

Why capture level_size? The queue grows as we append children. We must process only the nodes belonging to the current level before incrementing depth.
- Time: O(n).
- Auxiliary space: O(w), where w is the maximum width.

| Situation | Prefer | Why |
|---|---|---|
| Height of a moderately deep tree | Recursive DFS | Short, natural recurrence |
| Balanced, very wide tree | DFS | `O(log n)` stack versus potentially `O(n)` BFS queue |
| Very deep, narrow tree | BFS or iterative DFS | Avoid Python’s recursion limit; BFS queue stays small |
| Need level averages, level sums, or level-order output too | BFS | Already processes nodes by level |
| Need subtree heights, balance checking, or diameter too | Postorder DFS | Parent calculations depend on child results |
| Need minimum depth to a leaf | BFS | Can stop at the first leaf encountered |

When you know the depth...

Iterative DFS is also an option 

```python
def max_depth_iterative(root):
    if root is None:
        return 0 

    stack = [(root, 1)]
    maximum = 0 

    while stack:
        node, depth = stack.pop()
        maximum = max(maximum, depth)

        if node.right is not None: 
            stack.append((node.right, depth + 1))

        if node.left is not None:
            stack.append((node.left, depth + 1))

    return maximum 
```

This takes O(n) time and O(h) auxiliary space as an upper bound, without using Python’s recursive call stack. For a simple chain, its explicit stack actually stays at size one.
