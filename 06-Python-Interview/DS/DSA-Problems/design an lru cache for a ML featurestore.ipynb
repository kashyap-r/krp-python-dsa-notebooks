Problem Statement:

    Imagine you are building a real-time fraud detection service. Each transaction requires features such as:
        - Customer's average transaction amoutn over 30 days
        - Number of transactions in the last 24 hours. 
        - Customer risk score 
        - Merchant risk category 

    These features are retrieved from an online feature store, but repated lookups introduce latency and cost. 

Requirement:
    Design an in-memory cache that stores frequently accessed features and evicts the least recently used entries when capacity is reached. 

Conceptual Architecture:    

    
    

We'll use a Hashmap + Doubly Linked list

Why a DLL ?

We'll need two structures working together:

Structure               Responsibility 
HashMap                 Find a feature in O(1) time 
Doubly Linked LIst      Maintain the recency order and move / remove known nodes in O(1) time
Head                    Most recently used entry 
Tail                    Least recently used entry 



```python
class Node:
    def __init__(self, key, value):
        self.key = key 
        self.value = value
        self.prev = None 
        self.next = None 

class LRUCache:
    def __init__(self, capacity):
        if capacity <= 0:
            raise ValueError('Capacity must be positive')

        self.capacity = capacity 
        self.cache = {}

        self.head = Node(None, None)
        self.tail = Node(None, None)

        self.head.next = self.tail
        self.tail.prev = self.head
    
    def _remove(self, node):
        # detach an existing node in O(1) 
        previous = node.prev 
        following = node.next

        previous.next = following 
        following.prev = previous 

        node.prev = None 
        node.next = None 

    def _insert_front(self, node):
        # insert immediately after head 
        first = self.head.next 

        node.prev = self.head 
        node.next = first

        self.head.next = node 
        first.prev = node 

    def get(self, key):
        pass 

    def put(self, key, value):
        pass 


    def display(self):
        pass 

cache = LRUCache() 

```
