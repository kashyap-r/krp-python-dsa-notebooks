| Traversal | Order | When to use |
|---|---|---|

| **Postorder** | Left → Right → Node | Process children before parents: height, balance, diameter |


1. Preorder — the default stack template

Pop a node, process it, then push its children. Push right first so left is visited first.

```python
from collections import deque

class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

class BinaryTree:
    def __init__(self):
        self.root = None 

def build_tree_from_array(nums):
    if not nums or nums[0] is None:
        return None 

# preorder traversal using iterative method
def preorder(root):

    if root is None: 
        return []

    result = []
    stack = [root]

    while stack:
        node = stack.pop() 
        result.append(node.value)

        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    return result

# Inorder - go left before processing the node 
# Keep pushing nodes while moving left. When you can’t go further, 
# pop and process a node, then explore its right subtree.
# Even when curr is None, the stack may contain ancestors waiting to be processed.
def inorder(root):
    result = []
    stack = []
    curr = root

    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left

        curr = stack.pop() 
        result.append(curr.value)
        curr = curr.right
    return result

# postorder - defer processing until the children finish
# use a flag to distinguish first encounteriung a node from 
# returning to process it after its children
# Because the stack is last-in, first-out, 
# pushing parent → right → left means processing left → right → parent.
def postorder(root):
    if root is None:
        return []

    result = [] 
    stack = [(root, False)]

    while stack:
        node, ready = stack.pop()

        if ready:
            result.append(node.value)
        else:                
            stack.append((node, True))

            if node.right:
                stack.append((node.right, False))
            if node.left:
                stack.append((node.left, False))
    return result

# now its level order
def levelorder_traversal(root):
    if not root:
        return 

    result = []
    queue = deque([root])

    while queue:
        levels = []
        for _ in range(len(queue)):
            current = queue.popleft()
            levels.append(current.value)

            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)
        result.append(levels)
    return result

## lets' Xcute 

root = Node(
    4,
    Node(2, Node(1), Node(3)),
    Node(7, Node(6), Node(9))
    )

print(preorder(root))   # [4, 2, 1, 3, 7, 6, 9]
print(inorder(root))    # [1, 2, 3, 4, 6, 7, 9]
print(postorder(root))  # [1, 3, 2, 6, 9, 7, 4]
print(levelorder_traversal(root))
```

All three take O(n) time and O(h) auxiliary stack space, where h is tree height, plus O(n) for the returned result. All handle an empty tree.

Count Nodes and Leaves 

Given a Binary tree, return a tuple containing the total number of nodes and the number of leaf nodes. 

A leaf node has no children 

Input: [1, 2, 3, 4, 5, None, 6]

Output: (6, 3)

```python
# Count of nodes and leaf nodes 

from collections import deque 
class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None 

def count_nodes_leaves(root):
    if not root:
        return None 

    queue = deque([root])
    node_count = 0 
    leaf_count = 0 

    while queue:
        current = queue.popleft()
        if not current.left and not current.right:
            leaf_count += 1 
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)
        node_count += 1
    return (node_count, leaf_count)
## Usage

nums = [1, 2, 3, 4, 5, None, 6]

root  = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

# root.right.left = None
root.right.right = TreeNode(6)
print (f"Count (Nodes, Leaf Nodes): {count_nodes_leaves(root)}")


#        1
#      /   \
#     2     3
#    / \     \
#   4   5     6

# Time: O(n)—every node must be visited.
# Auxiliary space: O(w)—queue storage, where w is maximum width.
```

```python
# Recursive 
class TreeNode:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None 

def count_nodes_leaves_dfs(root) -> tuple[int, int]:
    if root is None:
        return (0, 0)

    if root.left is None and root.right is None:
        return (1, 1)

    left_nodes, left_leaves = count_nodes_leaves_dfs(root.left)
    right_nodes, right_leaves = count_nodes_leaves_dfs(root.right)

    return (
        1 + left_nodes + right_nodes,
        left_leaves + right_leaves
    )
## Usage
root  = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

# root.right.left = None
root.right.right = TreeNode(6)

result = count_nodes_leaves_dfs(root)
print (f"Count (Nodes, Leaf Nodes): {result}")
```

The recurrence is:
Total nodes = left subtree nodes + right subtree nodes + current node.
Total leaves = left subtree leaves + right subtree leaves, unless the current node itself is a leaf.

Recursive DFS can hit Python’s recursion limit on a very deep tree.

Edge cases to test

These arrays use our compact level-order representation. Build the tree before passing its root to your function.

| Case | Input array | Expected result | What it checks |
|---|---|---|---|
| Empty tree | `[]` | `(0, 0)` | No root |
| Single node | `[1]` | `(1, 1)` | Root itself is a leaf |
| Only left child | `[1, 2]` | `(2, 1)` | One missing child does not make a leaf |
| Only right child | `[1, None, 2]` | `(2, 1)` | Handles right-only branches |
| Skewed chain | `[1, 2, None, 3]` | `(3, 1)` | Separates leaf count from level count |
| Full three-level tree | `[1, 2, 3, 4, 5, 6, 7]` | `(7, 4)` | Several leaves on the same level |
| Leaves at different levels | `[1, 2, 3, None, None, 4, 5]` | `(5, 3)` | Leaves are not restricted to the deepest level |
| Duplicate/zero values | `[0, 0, 0]` | `(3, 2)` | Counts objects regardless of their values |

Follow-up questions to prepare for 

| Interviewer question | Your answer |
|---|---|
| Why use `and` in the leaf check? | A leaf must have **both** children missing. Using `or` incorrectly includes nodes with one child. |
| Why remove the inner `for` loop? | Level boundaries do not matter for counting nodes or leaves. |
| Can you do it using DFS? | Yes—combine counts from the left and right subtrees, as above. |
| What if the tree is extremely deep? | Use iterative DFS or BFS to avoid recursive call depth limits. |
| What if the tree is extremely wide? | DFS can use much less memory than BFS, especially for a balanced tree. |
| Do you need a `visited` set? | Not for a valid tree traversed through child references: each node is reached once. Cycles or shared children require clarified semantics and potentially a visited set. |
| How many non-leaf nodes are there? | `node_count - leaf_count`. |
| How many edges are there? | `node_count - 1` for a nonempty tree; `0` for an empty tree. |
| Can you count faster than O(n)? | Not for an arbitrary tree without extra information. Special structure or stored subtree counts can allow faster answers. |
