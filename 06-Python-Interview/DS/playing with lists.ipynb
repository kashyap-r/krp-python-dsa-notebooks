### Difference between list.sort(reverse=True), list.reverse() and reversed(list)

The difference comes down to whether we're sorting the data alphabetically/ numerically. simply flipping the current positions, and whether you are modifying the original list or making something new. 

Here is the direct comparison:

1. list.sort(reverse=True)

What it does: It sorts the elements in descending order (highest to lowest, or Z to A)

Modifies the original list? Yes (in-place)

Return value: None 

2. list.reverse() 

what is does: It flips the order of items exactly as they are right now. It does not look at the values or sort them. 

Modifies the original list? Yes (In-place) 

Return Value: None 

3. reversed(list) 

What it does: It provides a way to read the list backwards without touching or changing the original list at all 

Modifies original list? No 

Return Value: An iterator object (a lazy generator). To see it as a normal list, you must cast it using list() 


```python
# 1 
arr = [3, 1, 4, 2]
print (arr)
arr.sort(reverse=True)
print (arr)
print() 

# 2
arr = [3, 1, 4, 2]
arr.reverse()
print (arr)
print()

# 3
arr = [3, 1, 2, 4]
backwrds_iterator = reversed(arr)
print (list(backwrds_iterator))
print (arr)
```

**How does Slicing work on Lists?**

Inside the bracket [], you can have a maximum of two colons (:).

This divides the slicing parameters into three distinct positions, following this exact layout:
    
    arr[start:stop:step]

The Three Components

1. start (Before the 1st colon): The index where your slice begins (inclusive). If left blank, it defaults to the beginning (0 for positive steps, or the very end for negative steps).

2. stop (Between the colons): The index where your slice ends (exclusive—it stops just before this index). If left blank, it goes all the way to the edge of the list.

3. step (After the 2nd colon): The increment value. It tells Python how many steps to skip. If left blank, it defaults to 1.

**Understandig the 4 "colon" variations**

1. Zero Colons: arr[index]
    • What it does: Grabs a single element.
    • Example: arr[2] returns the item at index 2.

2. One Colon: arr[start:stop]
    • What it does: Extracts a continuous window of elements. The step defaults to 1.
    • Example: arr[1:4] starts at index 1 and goes up to (but excludes) index 4.

3. Two Colons: arr[start:stop:step]
    • What it does: Extracts elements with a custom jump pattern.
    • Example: arr[0:6:2] starts at index 0, stops before index 6, and grabs every 2nd element.

4. Shorthand Cleanups: arr[::step]
    • What it does: If you leave start and stop completely blank but keep both colons, Python assumes you want to look at the entire list but apply a step pattern.
    • Example: arr[::-1] means "Look at the whole list from start to end, but step backward by 1" (reversing it).

```python
def find_duplicates(arr):
    seen = set()
    duplicates = set()
    
    for num in arr:
        if num in seen:
            duplicates.add(num)
            # return True
        else:
            seen.add(num)
            
    return list(duplicates)

# --- Example ---
# nums = [1, 2, 3, 2, 4, 5, 1]
nums = [1, 2, 3, 4, 4]
print(find_duplicates(nums))  # Output: [1, 2]
```

### Running Maximum / Minimum

Sometimes instead of cumulative sum, maintain cumulative maximum.

```python
arr = [3, 1, 5, 2, 7]

# Running max 
running = []
current = float("-inf")

for x in arr:
    current = max(current, x)
    running.append(current)

print (arr)
print (running)
```

This idea appears in:
