**The difference: sort() VS sorted()**

The main difference is that .sort() modifies the original list, while sorted() leaves the original list alone and returns a brand new list.

```python
numbers = [3, 7, 1, 9, 5]
print (numbers)
result = numbers.sort()
print (numbers)
print (result)

arr = [6, 2, 8, 4, 0]
print (arr)
new_list = sorted(arr)
print (arr)
print (new_list)
```

**The Difference: reverse() VS reversed()**

Similar to sorting, .reverse() flips the original list in-place, while reversed() leaves the original list alone and returns an iterator object.


```python
letters = ['a', 'b', 'c']
print (letters)
result = letters.reverse() 
print (letters)
print (result)

letters = ['a', 'b', 'c','d']
res= reversed(letters)
print (letters)
print (res)
print (list(res))
```

**How does Slicing work on Lists?**

Inside the bracket [], you can have a maximum of two colons (:).

This divides the slicing parameters into three distinct positions, following this exact layout:
    
    arr[start:stop:step]

The Three Components

1. start (Before the 1st colon): The index where your slice begins (inclusive). If left blank, it defaults to the beginning (0 for positive steps, or the very end for negative steps).

2. stop (Between the colons): The index where your slice ends (exclusive—it stops just before this index). If left blank, it goes all the way to the edge of the list.

3. step (After the 2nd colon): The increment value. It tells Python how many steps to skip. If left blank, it defaults to 1.

**Understandig the 4 "colon" variations**

1. Zero Colons: arr[index]
    • What it does: Grabs a single element.
    • Example: arr[2] returns the item at index 2.

2. One Colon: arr[start:stop]
    • What it does: Extracts a continuous window of elements. The step defaults to 1.
    • Example: arr[1:4] starts at index 1 and goes up to (but excludes) index 4.

3. Two Colons: arr[start:stop:step]
    • What it does: Extracts elements with a custom jump pattern.
    • Example: arr[0:6:2] starts at index 0, stops before index 6, and grabs every 2nd element.

4. Shorthand Cleanups: arr[::step]
    • What it does: If you leave start and stop completely blank but keep both colons, Python assumes you want to look at the entire list but apply a step pattern.
    • Example: arr[::-1] means "Look at the whole list from start to end, but step backward by 1" (reversing it).


```python
# Lists Slicig and Dicing - 1 
arr = [10, 20, 30, 40, 50]
# to-do: extract the firt three elements using a single slice 
print (arr[:3])

# extract the last two elements using negative slicing 
print (arr[-2:])

arr = [1, 2, 3, 4, 5, 6]
# grab all the odd-indexed elements (i.e. elements at positions 1, 3, 5)
n = len(arr)
result = [arr[i] for i in range(n) if i % 2 != 0]
print (result)

# pure slicing method 
print(arr[1::2]) 

# write the slice syntax that completely reverses a list 
a = ['a', 'b', 'c']
print (a[::-1])

# create a shallow copy of an array using only slicing symbol 
a = [100, 200, 300, 400]
b = a[:]
print (b)
```
