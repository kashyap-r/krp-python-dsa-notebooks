## **STACK**

A stack is a last-in, first-out (LIFO) data structure: the most recently added item is the first one removed.

In Python, you usually implement a stack using a list. For interviews, the important skill is recognizing when a problem requires you to remember unfinished work and process the most recent item first.

For a Senior/Staff AI Engineer, this shows up in tree and graph traversal, nested expressions, dependency exploration, backtracking, and monotonic-stack problems.

1. What is a stack?

Think of a stack of plates: you add and remove plates from the top.

| Operation | Meaning | Python |
|---|---|---|
| Push | Add an item to the top | `stack.append(item)` |
| Pop | Remove and return the top item | `stack.pop()` |
| Peek | Read the top without removing it | `stack[-1]` |
| Is empty? | Check whether anything remains | `not stack` |
| Size | Count stored items | `len(stack)` |


Compare Stack with a Queue

| Structure | Removal rule | Typical purpose |
|---|---|---|
| Stack — LIFO | Most recently added first | DFS, undo, nested structures |
| Queue — FIFO | Earliest added first | BFS, level-order processing |

Complexity of Operations on Stack 

| Operation | Time complexity |
|---|---|
| `append(x)` | O(1) amortized |
| `pop()` from the end | O(1) amortized |
| `stack[-1]` | O(1) |
| `len(stack)` | O(1) |
| `x in stack` | O(n) |

“Amortized” means that occasional resizing is expensive, but the average cost across many operations is constant.

Avoid using insert(0, x) and pop(0) for a list-based stack: they shift other elements and cost O(n).

Why and when should you use a stack?

| Problem clue | Why a stack fits | Example |
|---|---|---|
| “Most recent unresolved item” | Resolve work in reverse arrival order | Matching brackets |
| “Finish this branch before another” | Preserve other branches for later | DFS |
| “Return to the previous decision” | Track decisions and alternatives | Backtracking |
| “Undo the latest action” | Reverse the order of actions | Editor history |
| “Nested structures” | Close the innermost structure first | Nested expressions |
| “Next greater/smaller element” | Keep unresolved candidates in order | Monotonic stack |
| “Evaluate postfix notation” | The most recent operands are needed next | Reverse Polish notation |


A useful question to ask yourself is:
**Does the next input need to interact with the most recent unresolved input?**


```python
stack = []

stack.append("A")
stack.append("B")
stack.append("C")
stack.append("D")
stack.append("E")

# display Stack
print (stack)

# peek operation - Inspect without removing
top = stack[-1]
print(f"Topmost element of the Stack: {top}")

# delete - remove the newest item 
# Both pop() and stack[-1] raise an IndexError on an empty list
print (stack.pop())
removed = stack.pop()
print (removed)

print ("The Stack is Empty" if not stack else stack)

print (f"Size if the Stack: {len(stack)}")
```

#### Implementing STACK using class, when encapsulation matters.

```python
class stack: 
    def __init__(self):
        self._items=[]

    def push(self, value):
        self._items.append(value)

    def pop(self):
        if self.is_empty():
            raise IndexError("Cannot pop from an empty Stack")
        return self._items.pop()

    def peef(self):
        if self.is_empty():
            raise IndexError ("Cannot peek into an empty stack")
        return self._items[-1]

    def is_empty(self):
        return not self._items

    def __len__(self):
        return len(self._items)

## Usage 
if __name__=="__main__":
    pass
```

**Problem #1: Valid Paranthesis**

Given a string containing only ()[]{}, return whether every opening bracket is closed in the correct order. 

Input: 

Expected Output:

Edge cases / Constraints / Boundary Conditions: 


```python
def is_valid_parentheses(s):
    bracket_map = {
        ")":"(",
        "]": "[",
        "}": "{",
    }
    stack = []

    for char in s:
        if char in "([{":
            # remember an one / unresolved opening bracket
            stack.append(char)
        else:
            if not stack:
                # a closing bracket has no opening partner
                return False

            if stack.pop() != bracket_map[char]:
                # must match the most recent opening. 
                return False
            
    # no opening brackets may remain unresolved 
    return not stack

## Usage

print(is_valid_parentheses("([])"))  # True
print(is_valid_parentheses("([)]"))  # False
print(is_valid_parentheses("(("))   # False
```

Explanation:

Invariant: The stack contains the opening brackets that have not yet been matched, in encounter order.

Complexity: O(n) time and O(n) worst-case auxiliary space.

```python
def is_valid_parentheses(s: str) -> bool:
    # Map every closing bracket to its corresponding opening bracket
    bracket_map = {
        ")": "(", 
        "}": "{", 
        "]": "["}
    stack = []
    
    for char in s:
        # If the character is a closing bracket
        if char in bracket_map:
            # Pop the top element from the stack if it's not empty, else use a dummy value '#'
            top_element = stack.pop() if stack else '#'
            
            # If the mapping doesn't match the top element, it's invalid
            if bracket_map[char] != top_element:
                return False
        else:
            # If it's an opening bracket, push it onto the stack
            stack.append(char)
            
    # If the stack is empty, all brackets were matched correctly
    return len(stack) == 0

# Test cases
print(is_valid_parentheses("()[]{}")) # Output: True
print(is_valid_parentheses("([)]"))   # Output: False (Wrong order)
print(is_valid_parentheses("("))      # Output: False (Unclosed)
```

How stacks relate to Recursion ?

A recursive function uses the lanaguage's call stack to remember unfinished calls. 

Each call must remember where execution should resume when its child call returns.

An explicit stack lets you manage pending work yourself. This is especially useful for deep trees or graphs where recursive Python code may hit the recursion limit.

```python
import time
def countdown(n):
    if n==0:
        return 
    print (n) 
    time.sleep(1)
    countdown(n-1)

# Usage
countdown(5)
```

Another Example - Iterative preorder traversal of a binary tree


```python
def preorder(root):
    if root is None: 
        return []

    stack = [root]
    result = []

    while stack:
        node = stack.pop() 
        result.append(node.val)

        if node.right:
            stack.append(node.right)

        if node.left:
            stack.append(node.left)

    return result
```

Why push the right child first? Because the stack reverses removal order. To visit left before right, push right before left.

For a binary tree:
- Time: O(n).
- Auxiliary stack space: O(h) worst-case bound, where h is tree height.
- Returned result: O(n) additional space.
An explicit stack still consumes memory. It avoids recursive call-depth limits, but does not make arbitrarily large traversals free.

Also, converting recursion to iteration sometimes requires storing more than a node:

stack = [(root, False)]  # Example: node plus whether it is ready to process.

That extra state can represent work that recursion otherwise remembers automatically.

### Monotonic Stack 

A monotonic stack is a regular stack with one strict rule: its elements must always be strictly sorted (either completely increasing or completely decreasing)

To keep this order, when you push a new element onto the stack, you must first pop out any exsting elements that break the sorted rule. 

💡 The Real-World Analogy: A Concert Queue

Imagine a line of people waiting for a concert. A very tall person arrives. If they stand behind shorter people, they block their view. In a monotonic decreasing stack, the tall person "kicks out" everyone shorter than them standing ahead until they find someone even taller.

There are two types:

1. Monotonically Increasing Stack: Sorted from bottom to top (e.g., [1, 3, 6, 9]). New small elements pop larger elements.

2. Monotonically Decreasing Stack: Sorted from bottom to top (e.g., [9, 6, 3, 1]). New large elements pop smaller elements.


Monotonic Stack: An Example 

array = [5, 3, 2, 10]

1. Process 5: stack is empty, push 5 - Stack: [5]

2. Process 3: Is 3 smaller than top (5)? Yes. Rule is safe. Push 3 - Stack: [5, 3]

3. Process 2: Is 2 smaller than top (3)? Yes. Rule is safe. Push 2 - Stack: [5, 3, 2]

4. Process 10:
   Is 10 smaller than top (2)? NO! 10 breaks the decreasing order.
   -> Pop 2! (We now know 10 is the "Next Greater Element" for 2).
   -> Pop 3! (10 is also the "Next Greater Element" for 3).
   -> Pop 5! (10 is also the "Next Greater Element" for 5).
   Stack is now empty. Push 10.
   Stack: [10]


Why use it?

It optimizes problems from a slow O(n^2) nested loop down to an incredibly fast O(n) linear runtime. Even though there is a while loop inside the for loop to pop items, each element is pushed and popped exactly once.


```python
def next_greater_element(nums):
    # initialize theoutput array filler with -1 
    result= [-1] * len(nums)
    stack = []

    for i in range(len(nums)):
        # While the stack is not empty AND the current number
        # is greater than the number represented by the top index of the stack
        while stack and nums[i] > nums[stack[-1]]:
            # We found the next greater element for the index at the top of the stack
            popped_index = stack.pop()
            result[popped_index] = nums[i]

        # push the current index onto the stack 
        stack.append(i)

    return result

nums = [5, 3, 2, 10, 6, 8]
print(next_greater_element(nums))
# Output: [10, 10, 10, -1, 8, -1]
```
