## defaultdict

Return a new dictionary-like object. defaultdict is a subclass of the built-in dict class. It overrides one method and adds one writable instance variable. The remaining functionality is the same as for the dict class



### defaultdict examples

```python
# Using list as the default_factory attribute, it is easy to group a 
# sequence of key-value pairs into a dictionary of lists
s = [('yellow', 1), ('blue', 2), ('yellow', 3), ('blue', 4), ('red', 1)]
d = defaultdict(list)
for k, v in s:
    d[k].append(v)

sorted(d.items())
```

```python
# same functionality as above using normal python dictionary
d = {} 
for k, v in s:
    d.setdefault(k, []).append(v)

sorted(d.items())
```

```python
# setting the default_factory to int makes the defaultdict useful for counting (like a bag or multiset )

s = 'mississippi'
d = defaultdict(int)
for k in s:
    d[k] += 1
print (d)
```

```python
# setting the default_factory to set makes defaultdict useful for building a dictionary of sets

s = [('red', 1), ('blue', 2), ('red', 3), ('blue', 4), ('red', 1), ('blue', 4)]

d = defaultdict(set)
for k, v in s:
    d[k].add(v)

sorted(d.items())
```

Usage Pattern: Grouping

The problem it solves:
Suppose we want to group employees by department:

```python
employees = [
    ("Alice", "AI"), 
    ("Bob", "Admin"),
    ("Charlie", "AI"),
    ("David", "Platform"),
    ("Eva", "Data"),
    ("Mia", "Admin")
]

# To group this with a normal dictionary
groups = {}

for name, department in employees: 
    if department not in groups:
        groups[department] = []
    groups[department].append(name)

print (groups)

newGroups = defaultdict(list)
for name, department in employees:
    newGroups[department].append(name)

for key, value in newGroups.items():
    print (f"{key}: {value}")
```

```python
# How to initialize 
d = defaultdict(list)

d["AI"].append("Rob")
d["AI"].append("Max")
print (d)

c = defaultdict(int)

c["apple"] += 1
c["orange"] += 1 
c["apple"] += 1

print (c)

# Also, a Set
e = defaultdict(set)

e["Alice"].add("Python")
e["Alice"].add("AWS")
e["Alice"].add("Python")

print (e)
```

```python
# ex. 1 Group the below transactions by customer
# Pattern: Aggregation Grouping

from collections import defaultdict
transactions = [
    ("Alice", 100), 
    ("Charlie", 50),
    ("Alice", 2000),
    ("Bob", 75)
]
# Pattern: Grouping 
transactionGroup = defaultdict(list)
for customer, amount in transactions:
    transactionGroup[customer].append(amount)
print (transactionGroup)

# Pattern: Aggregation
# Calculate the total transaction value per customer
totals = defaultdict(int)
for customer, amount in transactions:
    totals[customer] += amount
print(totals)
```

```python
# Pattern: Graph Adjacency List 
# Suppose the edges are:
edges = [
    ("A", "B"), 
    ("A", "C"),
    ("B", "D"),
    ("C", "D")
]

graph = defaultdict(list)

for u, v in edges:
    graph[u].append(v)

for key, value in graph.items():
    print (key, value)

print ()
# for an undirected graph 
uGraph = defaultdict(list)

for u, v in edges:
    uGraph[u].append(v)
    uGraph[v].append(u)

for key, value in uGraph.items():
    print (key, value)

```
