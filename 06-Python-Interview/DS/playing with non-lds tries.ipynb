How the insert() method works ?

Here is the exact play-by-play when you insert a character that doesn't exist yet:

1. The Check: Python looks at the current node's children dictionary to see if the character ch exists as a key.

2. The Creation: If it's not there, Python executes TrieNode(). This allocates a brand-new, empty node object in memory (with its own empty {} children dictionary and is_end = False).

3. The Writing: Python takes that fresh, empty node object and assigns it to node.children[ch].
The crucial realization: The node object itself doesn't know it represents the letter 'C'. It is the parent's dictionary key 'C' that holds the reference to it.

### How does it handle a real-world multi-page text document?

Yes, in the real world, the insert() method must be called for every single word in that document.

However, you don't manually type it out. You use a file-reading loop to clean the text, break it into individual words, and feed them into the Trie one by one.


```python
import re

# 1. Initialize your Trie
my_trie = Trie()

# 2. Open and read the entire multi-page document
with open("large_document.txt", "r", encoding="utf-8") as file:
    for line in file:
        # Clean the line: remove punctuation and split it into a list of words
        # (e.g., "Cat, Can, Car!" becomes ["Cat", "Can", "Car"])
        words = re.findall(r'\b\w+\b', line)
        
        # 3. Call the insert method for EVERY word found
        for word in words:
            # Optional: convert to lowercase so "Cat" and "cat" are treated as the same word
            my_trie.insert(word.lower()) 
```


#### Why would we do this for millions of words? Is it efficient?

It sounds like a lot of work to call insert() millions of times, but Tries are incredibly efficient for two massive reasons:

1. The loop is bounded by word length, not document size: Inserting the word "apple" takes exactly 5 steps, whether your document has 10 words or 10 billion words. It never has to search through the rest of the document.

2. Massive Memory Savings: In a huge document, words like "the", "and", "this", and "that" appear thousands of times. The first time "the" is read, the nodes are created. The next 10,000 times "the" appears, the if ch not in node.children: check fails instantly, no new memory is allocated, and it just quickly traces the existing path to the end.

```python
class TrieNode: 
    def __init__(self):
        self.children = [None]*28 
        self.isEndOfWord = False 

# method to nsert a key into trie 
def insert(root, key):
    # Initialize the curr pointer with the root node 
    curr = root 

    # iterate across the length of the string
    for c in key: 
        # check if the node exists for the current character in the trie 
        index = ord(c) - ord('a')
        if curr.children[index] is None:

            # If node for current character does not exist then make a new node 
            new_node = TrieNode()

            # keep the reference for the newly created node
            curr.children[index] = new_node

        # Move the curr pointer to the newly created node
        curr = curr.children[index]

    # Mark the end of the word 
    curr.isEndOfWord = True
    
def search(root, key):
    # Initialize the curr pointer with the root node 
    curr = root 

    # Iterate across the length of the string 
    for c in key:

        # Check if the node exists for teh current character in the Trie 
        index = ord(c) - ord('a')
        if curr.children[index] is None:
            return False
        # move the curr pointer to the already existing node for current character
        curr = curr.children[index]

    # return true if the word exists and is marked as ending
    return curr.isEndOfWord

# Time Complexity: O(n), where n is the length of the word to search.
# Auxiliary Space: O(1)

# Prefix Searching in Tries  
def is_prefix(root, key):
    current = root 
    for c in key: 
        index = ord(c) - ord('a')

        # If character doesn't exist, return false 
        if current.children[index] is None:
            return False 
        current = current.children[index]
    return True
# O(n) Time and O(1) Space 

```

```python
class TrieNode:
    def __init__(self):
        self.children = [None] * 26
        self.isLeaf = False


class Trie:
    def __init__(self):
        self.root = TrieNode()

    # Method to insert a key into the Trie
    def insert(self, key):
        curr = self.root
        for c in key:
            index = ord(c) - ord('a')
            if curr.children[index] is None:
                curr.children[index] = TrieNode()
            curr = curr.children[index]
        curr.isLeaf = True

    # Method to search a key in the Trie
    def search(self, key):
        curr = self.root
        for c in key:
            index = ord(c) - ord('a')
            if curr.children[index] is None:
                return False
            curr = curr.children[index]
        return curr.isLeaf

    # Method to check if a prefix exists in the Trie
    def isPrefix(self, prefix):
        curr = self.root
        for c in prefix:
            index = ord(c) - ord('a')
            if curr.children[index] is None:
                return False
            curr = curr.children[index]
        return True


if __name__ == '__main__':
    trie = Trie()
    arr = ["and", "ant", "do", "dad"]
    for s in arr:
        trie.insert(s)
    searchKeys = ["do", "gee", "bat"]
    for s in searchKeys:
        if trie.search(s):
            print("true", end= " ")
        else:
            print("false", end=" ")
    
    print()
    prefixKeys = ["ge", "ba", "do", "de"]
    for s in prefixKeys:
        if trie.isPrefix(s):
            print("true", end = " ")
        else:
            print("false", end = " ")
```

### Trie | (Delete)

In Trie deletion, we delete the key from bottom up using recursion. The following are the possible conditions when deleting key. 

    1. Key may not be there in Trie. Delete operation should not modify trie. 

    2. Key present as unique key (no part of key contains another key (prefix), nor the key itself is prefix of another key in trie). Delete all the nodes

    3. Key is prefix key of another long key in trie. Unmark the leaf node. 

    4. Key is present in trie, having atleast on other key as prefix key. Delete nodes from end of key until first leaf node of longest prefix key.
