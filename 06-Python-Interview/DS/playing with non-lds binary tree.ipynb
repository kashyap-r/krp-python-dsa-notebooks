## Binary Tree Insertion 

```python
from collections import deque

class TreeNode:
    def __init__(self, data=0, left=None, right=None):
        self.data = data
        self.left: TreeNode | None = left
        self.right: TreeNode | None = right

class BinaryTree:
    def __init__(self):
        self.root = None

    # insert using level order (keep tree balanced)
    def insert(self, data):
        new_node = TreeNode(data)
        if not self.root:
            self.root = new_node
            return 
        queue = deque([self.root])

        while queue:
            current = queue.popleft()

            if not current.left:
                current.left = new_node 
                return 
            else:
                queue.append(current.left)
            if not current.right:
                current.right = new_node
                return 
            else:
                queue.append(current.right)
    
    # insert a node at a given position in the tree
        def insert_at_position(self, node, data, position):
            if not node:
                return TreeNode(data)
            if position == 1:
                new_node = TreeNode(data)
                new_node.left = node
                return new_node
            elif position == 2:
                new_node = TreeNode(data)
                new_node.right = node
                return new_node
            else:
                if position % 2 == 0:
                    node.left = self.insert_at_position(node.left, data, position // 2)
                else:
                    node.right = self.insert_at_position(node.right, data, position // 2)
                return node
```

### Binary Tree Deletion 

 - The task isto delete a given node from it by making sure the tree shrinks from the bottom. 
 - The deleted node is replaced by the **bottom-most and right-most node**.
 - Here we do not have any order amongelements, so we replace them with the last element.

     10
    /  \
   20  30

If we delete 10 from the above tree.. the tree will become 

      30
     /
   20 

As the bottom & rightmost node in the above binary tree is 30 , replace the key node ie. 10 with 30 and remove the bottom & rightmost node.

To perform the Deletion in a Binary Tree, the idea is to traverse the tree in level order manner. 
See the steps below:

- Starting at the root, find the deepest and the right-most node in the binary tree and the node which we want to delete. 
- Replace the deepese rightmost node's data with the node to be deleted.
- Then delete the deepest rightmost node.


```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left: TreeNode | None = None
        self.right: TreeNode | None = None

class BinaryTree:
    def __init__(self):
        self.root: TreeNode | None = None

    def delete_deepest(self, dNode: TreeNode) -> None:
        if self.root is None:
            return

        if self.root is dNode:
            self.root = None
            return

        queue = [self.root]

        while queue:
            curr = queue.pop(0)

            if curr.left is dNode:
                curr.left = None
                return
            if curr.left is not None:
                queue.append(curr.left)

            if curr.right is dNode:
                curr.right = None
                return
            if curr.right is not None:
                queue.append(curr.right)

    def delete(self, key: int) -> None:
        if self.root is None:
            return 

        # if the tree has only one node
        if self.root.left is None and self.root.right is None:
            if self.root.data == key:
                self.root = None
            return

        key_node = None
        queue = [self.root]
        curr = None 

        # level order traversal to find the node to delete and the deepest node
        while queue:
            curr = queue.pop(0)

            if curr.data == key:
                key_node = curr

            if curr.left is not None:
                queue.append(curr.left)
            if curr.right is not None:
                queue.append(curr.right)

        if key_node:
            key_node.data = curr.data  # Replace key_node's data with deepest node's data
            self.delete_deepest(curr)  # Delete the deepest node

        return
```

Time Complexity: O(n) where n is the number of nodes 
Space: O(n) 

Search for a node in Binary tree 

The idea is to use any of the tree traversal methods to traverse the tree and while traversing check if the current nide matches with the given node, return true if any node matches with the given node and stop traversing further. 
If the tree is completely traversed and none of the node matches with the given node then return False.

```python
# Search for a node in the binary tree
# uses a stack 
def search(self, key: int) -> bool:
    if self.root is None:
        return False
    queue = [self.root]
    while queue:
        curr = queue.pop(0)
        if curr.data == key:
            return True
        if curr.left is not None:
            queue.append(curr.left)
        if curr.right is not None:
            queue.append(curr.right)
    return False

# search using BFS
# uses a deque for efficient popping from the left
def search_tree(self, data):
    if not self.root:
        return False
    queue = deque([self.root])
    while queue:
        current = queue.popleft()
        if current.data == data:
            return True
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)
    return False

def recursive_search(self, node: TreeNode | None, key: int) -> bool:
    if node is None:
        return False
    if node.data == key:
        return True
    return self.recursive_search(node.left, key) or self.recursive_search(node.right, key)

# Time COmpelxity: O(n) where n is the number of nodes in the tree
# Auxiliary Space: O(h) where h is the height of the tree
```

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left: TreeNode | None = None
        self.right: TreeNode | None = None

class BinaryTree:
    def __init__(self):
        self.root: TreeNode | None = None

# BFS: level order traversal 
def level_order_traversal(self):
    if not self.root:
        return []
    result = []
    queue = deque([self.root])
    while queue:
        current = queue.popleft()
        result.append(current.data)
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)
    return result

# level order traversal with recursion
def level_order_traversal_with_recursion(self, node, level, result):
    if not node:
        return
    if len(result) <= level:
        result.append([])
    result[level].append(node.data)
    self.level_order_traversal_with_recursion(node.left, level + 1, result)
    self.level_order_traversal_with_recursion(node.right, level + 1, result)

# DFS: in-order traversal
def in_order_traversal(self, node):
    if node:
        self.in_order_traversal(node.left)
        print(node.data, end=' ')
        self.in_order_traversal(node.right)

# DFS: pre-order traversal
def pre_order_traversal(self, node):
    if node:
        print(node.data, end=' ')
        self.pre_order_traversal(node.left)
        self.pre_order_traversal(node.right)

# DFS: post-order traversal
def post_order_traversal(self, node):
    if node:
        self.post_order_traversal(node.left)
        self.post_order_traversal(node.right)
        print(node.data, end=' ')


# inorder traversal without recursion
def inorder_traversal_without_recursion(self):
    if not self.root:
        return []
    result = []
    stack = []
    current = self.root
    while stack or current:
        if current:
            stack.append(current)
            current = current.left
        else:
            current = stack.pop()
            result.append(current.data)
            current = current.right
    return result

# preorder traversal without recursion
def preorder_traversal_without_recursion(self):
    if not self.root:
        return []
    result = []
    stack = [self.root]
    while stack:
        current = stack.pop()
        result.append(current.data)
        if current.right:
            stack.append(current.right)
        if current.left:
            stack.append(current.left)
    return result

# postorder traversal without recursion
def postorder_traversal_without_recursion(self):
    if not self.root:
        return []
    result = []
    stack1 = [self.root]
    stack2 = []
    while stack1:
        current = stack1.pop()
        stack2.append(current)
        if current.left:
            stack1.append(current.left)
        if current.right:
            stack1.append(current.right)
    while stack2:
        current = stack2.pop()
        result.append(current.data)
    return result
    
# height of the tree
def height(self, node):
    if not node:
        return 0
    left_height = self.height(node.left)
    right_height = self.height(node.right)
    return max(left_height, right_height) + 1

# count the number of nodes in the tree
def count_nodes(self, node):
    if not node:
        return 0
    return 1 + self.count_nodes(node.left) + self.count_nodes(node.right)

# count the number of leaf nodes in the tree
def count_leaf_nodes(self, node):
    if not node:
        return 0
    if not node.left and not node.right:
        return 1
    return self.count_leaf_nodes(node.left) + self.count_leaf_nodes(node.right)

# count the number of non-leaf nodes in the tree
def count_non_leaf_nodes(self, node):
    if not node or (not node.left and not node.right):
        return 0
    return 1 + self.count_non_leaf_nodes(node.left) + self.count_non_leaf_nodes(node.right)

# count the number of full nodes in the tree
def count_full_nodes(self, node):
    if not node:
        return 0
    count = 0
    if node.left and node.right:
        count = 1
    return count + self.count_full_nodes(node.left) + self.count_full_nodes(node.right)

# count the number of half nodes in the tree
def count_half_nodes(self, node):
    if not node:
        return 0
    count = 0
    if (node.left and not node.right) or (not node.left and node.right):
        count = 1
    return count + self.count_half_nodes(node.left) + self.count_half_nodes(node.right)

# count the number of nodes at a given level
def count_nodes_at_level(self, node, level):
    if not node:
        return 0
    if level == 1:
        return 1
    return self.count_nodes_at_level(node.left, level - 1) + self.count_nodes_at_level(node.right, level - 1)

# count the number of nodes at a given depth
def count_nodes_at_depth(self, node, depth):
    if not node:
        return 0
    if depth == 0:
        return 1
    return self.count_nodes_at_depth(node.left, depth - 1) + self.count_nodes_at_depth(node.right, depth - 1)

# count the number of nodes at a given height
def count_nodes_at_height(self, node, height):
    if not node:
        return 0
    if height == 0:
        return 1
    return self.count_nodes_at_height(node.left, height - 1) + self.count_nodes_at_height(node.right, height - 1)   

# count the number of nodes at a given distance from the root
def count_nodes_at_distance(self, node, distance):
    if not node:
        return 0
    if distance == 0:
        return 1
    return self.count_nodes_at_distance(node.left, distance - 1) + self.count_nodes_at_distance(node.right, distance - 1)

# count the number of nodes at a given distance from a given node
def count_nodes_at_distance_from_node(self, node, distance):
    if not node:
        return 0
    if distance == 0:
        return 1
    return self.count_nodes_at_distance_from_node(node.left, distance - 1) + self.count_nodes_at_distance_from_node(node.right, distance - 1)



# height of the tree without recursion
def height_iterative(self):
    if not self.root:
        return 0
    queue = deque([self.root])
    height = 0
    while queue:
        height += 1
        for _ in range(len(queue)):
            current = queue.popleft()
            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)
    return height

# count the number of nodes without recursion
def count_nodes_without_recursion(self):
    if not self.root:
        return 0
    count = 0
    queue = deque([self.root])
    while queue:
        current = queue.popleft()
        count += 1
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)
    return count

def diameter(root):
    max_diameter = 0 

    def dfs(node):
        nonlocal max_diameter

        if not node:
            return 0 

        left = dfs(node.left)
        right = dfs(node.right)

        #update diameter 
        max_diameter = max(max_diameter, left + right)
        return max(left, right) + 1

    dfs(root)
    return max_diameter

# another way to calculate the diameter of a binary tree
def diameter_of_tree(self):
    if not self.root:
        return 0

    def diameter_helper(node):
        if not node:
            return 0, 0  # height, diameter

        left_height, left_diameter = diameter_helper(node.left)
        right_height, right_diameter = diameter_helper(node.right)

        current_height = max(left_height, right_height) + 1
        current_diameter = max(left_diameter, right_diameter, left_height + right_height + 1)

        return current_height, current_diameter

    _, diameter = diameter_helper(self.root)
    return diameter
```

#### InOrder Traversal 

Approach 1: Using recursion 
Algorithm for Postorder Traversal: Postorder(tree) 

● Traverse the left subtree, i.e., call Postorder(left->subtree) 

● Traverse the right subtree, i.e., call Postorder(right->subtree) 

● Visit the root 

Approach #2: without recursion using a Stack

    The idea is to implement recursion using a stack. Starting from root node, keep on pushing the node into a stack and move to left node. When node becomes null, pop a node from the stack, print its value and move to the right node.

Approach #3: Implementing without stack and without recursion (Morris Travel Method)
    
    * For each node, check if it has a left child 
    * If it does not have a left child, visit it and move to the right child
    * If it has a left child, find the inorder predecessor (rightmost node in the left subtree)
    Make the current node as the right child of its inorder predecessor (temporary link)
    Move to the left child 
    * When you encounter a temporary link again, it means the left subtree is fully visited:
        - Remove the temporary link
        - Visit the current node 
        - Move to the right child. 

Note: For Morris Travel Method

### Binary Tree Traversal

BFS - Breadth First Search / Level Order Traversal

Level Order Traversal technique is a method to traverse a Tree such that all nodes present in the same level are traversed completely before traversing the next level.

Two Approaches 
    * Approach #1: Using Recursion 
    * Using Queue (Iterative)

```python
# Approach #1 - Using Recursion 

class Node:
    def __init__(self, value):
        self.data = value 
        self.left = None 
        self.right = None

# class BinaryTree:
#     def __init__(self, root_value):
#         self.root_value = Node(root_value)

def levelOrderRecursion(root, level, result):
    # base case 
    if root is None: 
        return 

    # add a new level to the result if needed 
    if len(result) <= level:
        result.append([])

    # Add current node's data to its corresponding level
    result[level].append(root.data)

    # recur for the left and right childeren 
    levelOrderRecursion(root.left, level+1, result)
    levelOrderRecursion(root.right, level+1, result)

# Function to perform level order reversal 
def levelOrder(root):
    # Stores the result level by level 
    result = []
    levelOrderRecursion(root, 0, result)
    return result

def main():
    root = Node(5) 
    
    root.left = Node(12)
    root.right = Node(13)

    root.left.left = Node(7)
    root.left.left.left = Node(17)
    root.left.left.right = Node(14)

    root.left.right = Node(14)
    root.left.right.left = Node(27)
    root.left.right.right = Node(3)

    root.right.right = Node(2) 
    root.right.right.left = Node(8) 
    root.right.right.right = Node(11)

    result = levelOrder(root)

    for level in result:
        print (' '.join(map(str, level)))

    # in the form of a list
    print (result)

if __name__ == '__main__':
    main()
```

Boundary Traversal 

Given the root of the binary tree, Find its boundary traversal in anti-clockwise order, starting from the root. 

The boundary of a binary tree consists of the following parts:

1. Left Boundary: The nodes on the left edge of the tree, excluding the leaf nodes. 
2. Leaf Nodes: All the leaf nodes from left to right.
3. Right Boundary: The nodes on the right edge of the tree, excluding the leaf nodes, traversed in bottom-up order. 

Note:

- If the root does not have a left subtree or right subtree, then the root itself is considered part of the respective boundary.

- Each node in the boundary should appear only once in the traversal.

Count Nodes 

```python
def count_nodes(root):
    if not root:
        return 0

    return 1 + count_nodes(root.left) + count_nodes(root.right)
```

Count Edges 

For any tree: Edges = Nodes - 1

```python
def count_edges(root):
    nodes = count_nodes(root)
    return nodes - 1 if nodes > 0 else 0
```

Best Path between Two Nodes

It could mean: 
- Shortest path 
- path sum max/min
- path between two given nodes

Approach -> Use LCA (Lowest Common Ancestor)

Steps: 
1. FInd LCA of node1 and node2 
2. Get path: 
    - node1 -> LCA 
    - LCA -> node2 


```python
# Step1: Find LCA 
def lca(root, p, q):
    if not root:
        return None

    if root == p or root == q:
        return root

    left = lca(root.left, p, q)
    right = lca(root.right, p, q)

    if left and right:
        return root

    return left if left else right

# Step 2: Find distance from Root to Node 
def find_path(root, target, path):
    if not root:
        return False

    path.append(root)

    if root == target:
        return True

    if (find_path(root.left, target, path) or
        find_path(root.right, target, path)):
        return True

    path.pop()
    return False

# Step 3: Combine the above two steps to find distance between two nodes
def path_between_nodes(root, p, q):

    ancestor = lca(root, p, q)

    path1 = []
    path2 = []

    find_path(ancestor, p, path1)
    find_path(ancestor, q, path2)

    return path1[::-1] + path2[1:]

# Time Complexity: O(n)
# Space Complexity: O(n)

```
