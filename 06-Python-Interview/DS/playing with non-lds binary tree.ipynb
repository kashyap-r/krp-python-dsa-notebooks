### Binary Tree Traversal

DFS - Depth First Search algorithm or traversal 

- Explore as far down along the left branch of the treee as possible 
- Visit the node before moving on to the right side of the tree

Three Variants, 

- InOrder Traversal
- PreOrder Traversal
- Post Order Traversal



#### InOrder Traversal 

Approach 1: Using recursion 
Algorithm for Postorder Traversal: Postorder(tree) 

● Traverse the left subtree, i.e., call Postorder(left->subtree) 

● Traverse the right subtree, i.e., call Postorder(right->subtree) 

● Visit the root 

Approach #2: without recursion using a Stack

    The idea is to implement recursion using a stack. Starting from root node, keep on pushing the node into a stack and move to left node. When node becomes null, pop a node from the stack, print its value and move to the right node.

Approach #3: Implementing without stack and without recursion (Morris Travel Method)
    
    * For each node, check if it has a left child 
    * If it does not have a left child, visit it and move to the right child
    * If it has a left child, find the inorder predecessor (rightmost node in the left subtree)
    Make the current node as the right child of its inorder predecessor (temporary link)
    Move to the left child 
    * When you encounter a temporary link again, it means the left subtree is fully visited:
        - Remove the temporary link
        - Visit the current node 
        - Move to the right child. 

Note: For Morris Travel Method
Refer: https://www.geeksforgeeks.org/dsa/inorder-tree-traversal-without-recursion-and-without-stack/

     




```python
# InOrder Traversal Using recursion 

class Node: 
    def __init__(self, x):
        self.data = x 
        self.left = None
        self.right = None

def recursiveInOrder(root, result):
    if root is None: 
        return 

    recursiveInOrder(root.left, result)
    result.append(root.data)
    recursiveInOrder(root.right, result)

# iterative method using stack
def iterativeInOrder(root):
    ans = []
    stack = [] 
    curr = root 

    while curr is not None or len(stack)>0:
        while curr is not None:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop() 
        ans.append(curr.data)
        curr = curr.right
    return ans

# morris travel method 
def morristravelInorder(root):
    result = []
    curr = root 

    while curr is not None:
        if curr.left is None:
            result.append(curr.data)
            curr = curr.right 
        else: 
            # find the inorder predecessor or curr 
            pred = curr.left 
            while pred.right is not None and pred.right != curr:
                pred = pred.right 

            # Make curr the right child of its inorder predecessor
            if pred.right is None:
                pred.right = curr 
                curr = curr.left 
            else: 
                # revert the changes made in the tree structure
                pred.right = None 
                result.append(curr.data)
                curr = curr.right
    return result

def displayTree(root):
    print (" ".join(map(str, root)))

if __name__=="__main__":
    root = Node(1)
    root.left = Node(2)  # type: ignore[assignment]
    root.right = Node(3) 
    root.left.left = Node(4) 
    root.left.right = Node(5) 
    root.right.left = Node(6) 
    root.right.right = Node(7)

    result = [] 
    recursiveInOrder(root, result)

    for node in result:
        print (node, end=" ")
    print() 

    res = iterativeInOrder(root)
    displayTree(res)

    ans = morristravelInorder(root)
    displayTree(ans)
```


#### Time COmplexity and Auxiliary Space

Using Recursive Method:

Time Complexity: O(n)
Auxiliary Space: O(h), h is the height of the tree

    In the worst case, h can be the same as n (when the tree is a skewed tree)
    In the best case, h can be the same as log n (when the tree is a complete tree)

Key Properties:

    If applied to a Binary Search Tree (BST), it returns elements in sorted order.

    Ensures that nodes are processed in a hierarchical sequence, making it useful for expression trees and BSTs.

Using Stack (iterative method) - 
    O(n) Time and O(h) Space

Using Morris travel method:
    Time Complexity: O(n), if we take a closer look, we can notice that every edge of the tree is traversed at most three times.
    
    Auxiliary Space: O(1)

### Post Order Traversal of a Binary Tree 

```python
class Node:
    def __init__(self, x):
        self.data = x 
        self.left = None 
        self.right = None 

def recursivePostOrder(root, result):
    if root is None: 
        return result 

    recursivePostOrder(root.left, result)
    recursivePostOrder(root.right, result)
    result.append(root.data)

 
```

Key Properties:

    It is also useful for generating the postfix expression from an expression tree.

### Binary Tree Traversal

BFS - Breadth First Search / Level Order Traversal

Level Order Traversal technique is a method to traverse a Tree such that all nodes present in the same level are traversed completely before traversing the next level.

Two Approaches 
    * Approach #1: Using Recursion 
    * Using Queue (Iterative)

```python
# Approach #1 - Using Recursion 

class Node:
    def __init__(self, value):
        self.data = value 
        self.left = None 
        self.right = None

# class BinaryTree:
#     def __init__(self, root_value):
#         self.root_value = Node(root_value)

def levelOrderRecursion(root, level, result):
    # base case 
    if root is None: 
        return 

    # add a new level to the result if needed 
    if len(result) <= level:
        result.append([])

    # Add current node's data to its corresponding level
    result[level].append(root.data)

    # recur for the left and right childeren 
    levelOrderRecursion(root.left, level+1, result)
    levelOrderRecursion(root.right, level+1, result)

# Function to perform level order reversal 
def levelOrder(root):
    # Stores the result level by level 
    result = []
    levelOrderRecursion(root, 0, result)
    return result

def main():
    root = Node(5) 
    
    root.left = Node(12)
    root.right = Node(13)

    root.left.left = Node(7)
    root.left.left.left = Node(17)
    root.left.left.right = Node(14)

    root.left.right = Node(14)
    root.left.right.left = Node(27)
    root.left.right.right = Node(3)

    root.right.right = Node(2) 
    root.right.right.left = Node(8) 
    root.right.right.right = Node(11)

    result = levelOrder(root)

    for level in result:
        print (' '.join(map(str, level)))

    # in the form of a list
    print (result)

if __name__ == '__main__':
    main()
```

```python
# Expected approach - Using iterative method (using queues)
from collections import deque

class Node: 
    def __init__(self, value):
        self.data = value 
        self.left = None 
        self.right = None 

class BinaryTree:
    def __init__(self):
        self.root = None

    def levelOrder(self):
        if self.root is None:
            return []

        queue = deque([self.root])
        result = []

        while queue:
            level = []

            for _ in range(len(queue)):
                current = queue.popleft()
                level.append(current.data)

                if current.left:
                    queue.append(current.left)
                if current.right:
                    queue.append(current.right)

            result.append(level)

        return result

def main():

    bt = BinaryTree()
    bt.root = Node(5) 
    
    bt.root.left = Node(12)
    bt.root.right = Node(13)

    bt.root.left.left = Node(7)
    bt.root.left.left.left = Node(17)
    bt.root.left.left.right = Node(23)

    bt.root.left.right = Node(14)
    bt.root.left.right.left = Node(27)
    bt.root.left.right.right = Node(3)

    bt.root.right.right = Node(2) 
    bt.root.right.right.left = Node(8) 
    bt.root.right.right.right = Node(11)

    result = bt.levelOrder()

    for level in result:
        print (' '.join(map(str, level)))

    # in the form of a list
    print (result)

if __name__ == '__main__':
    main()
```
