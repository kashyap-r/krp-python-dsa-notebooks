Singly Linked List
    - is a chain of nodes
    - Each node has 
        - data 
        - pointer to the next node 

Advantages of a SLL 
1. Dynamic size (no pre-allocation)
    - Grows and shrinks at runtime 
    - No ersizing cost like python lists 

Why this matters
    - When data volume is unknown or highly variable 
    - Example
        Streaming logs, incremental ingestion pipelines 

2. O(1) insertion / deletion (when pointer is known)
    - Insert at head -> O(1)
    - Delete after a given node -> O(1)

📌 Example
Maintaining a queue of tasks where items are constantly added/removed

3. No shifting cost 
    - Arrays reqire shifting elements ->. O(n)
    - Linked LIst just relinks pointers 

    Example
        - Undo stacks, operation histories

Disadvantages of a Singly Linked List
    ❌ 1️⃣ No random access
    - list[i] → ❌ not possible
    - Must traverse from head → O(n)
    📌 Pain point
    Anything that smells like indexing or slicing

    ❌ 2️⃣ Poor cache locality
    - Nodes scattered in memory
    - CPU cache misses → slower in practice
    📌 Why Python lists often win despite worse big-O

    ❌ 3️⃣ Reverse traversal is impossible
    - No prev pointer
    - Reversing list costs O(n)

    ❌ 4️⃣ Extra memory per element
    - Pointer overhead per node
    - Python object overhead makes this worse

    ❌ 5️⃣ Complex to debug and maintain
    - Pointer bugs = nightmare
    - One bad link → memory leak / infinite loop

Real-Life Applications (Where SLLs Shine)
✅ 1️⃣ Implementing stacks and queues
    - Stack → push/pop at head
    - Queue → enqueue at tail, dequeue at head
    📌 Common in interview questions and low-level systems

✅ 2️⃣ Undo / Redo functionality (basic)
    - Each operation points to the next
    - Lightweight history tracking
    📌 Text editors, simple command history

✅ 3️⃣ Graph adjacency lists
    - Each node maintains a linked list of neighbors
    📌 Especially in C/C++ graph implementations

✅ 4️⃣ Memory-constrained embedded systems
    - Predictable pointer operations
    - No costly reallocation
    📌 Firmware, RTOS task lists

✅ 5️⃣ Streaming / sequential data processing
    - Process nodes one by one
    - No need for random access
    📌 Log processing, event streams

Real-Life Use-Cases Where Singly Linked Lists Fail Hard
❌ 1️⃣ Search-heavy systems
    - Databases
    - Search engines
    - Analytics queries
    Why?
    O(n) lookup kills performance.
📌 Use instead: hash tables, trees, indexes

❌ 2️⃣ ML / Data Engineering workloads
    - Feature vectors
    - Tensor operations
    - Batch processing
    📌 Reality check
    NumPy, Pandas, PyTorch → contiguous memory is king

❌ 3️⃣ UI rendering and timelines
    - Need fast back/forward traversal
    - Random jumps (scrolling)
    📌 Doubly linked lists or arrays win

❌ 4️⃣ High-frequency trading / low-latency systems
    - Pointer chasing = cache misses
    - Unpredictable memory access
    📌 Arrays + pre-allocation dominate here

❌ 5️⃣ Python production code (most of the time)
    - Python’s list is:
    - Cache-friendly
    - Amortized O(1) append
    - Faster in practice
    📌 That’s why Python doesn’t expose a native linked list
    

Interview Gold: When to Use vs Avoid

    ✔ Use Singly Linked List when:

        - Insert/delete is frequent

        - Access is sequential

        - Size is unknown

        - Memory fragmentation is acceptable

    ✘ Avoid when:

        - Random access is required

        - Performance is critical

        - Data is numeric or vectorized

        - You’re in Python and list works

```python
### this is the one to follow 

class Node: 
    def __init__(self, data, next=None):
        self.data = data 
        self.next = next 

class LinkedList:
    def __init__(self):
        self.head = None 

    # insert at beginning 
    def insert_at_beginning(self, data):
        node = Node(data)
        node.next = self.head
        self.head = node 

    def insert_at_end(self, data):
        node= Node(data)
        if self.head is None:
            self.head = node 
            return 
        curr = self.head 
        while curr.next:
            curr = curr.next
        curr.next = node

    def insert_at_position(self, position, data):
        if position < 0:
            return None
        if position == 0:
            self.insert_at_beginning(data)
            return 
        curr = self.head 
        for _ in range(position-1):
            if curr is None:
                raise IndexError("Position Out of range")
            curr = curr.next
        if curr is None:
            raise IndexError("Position Out of range")
        node = Node(data) 
        node.next = curr.next
        curr.next = node 

    def insert_after_value(self, target, data):
        curr = self.head 
        while curr and curr.data != target:
            curr = curr.next 
        if curr is None:
            raise IndexError("Target Not Found")
        node = Node(data)
        node.next = curr.next
        curr.next = node 

    # ---------- Deletion ----------
    def delete_at_beginning(self):
        if self.head is None:
            raise IndexError("List is empty")
        self.head = self.head.next
 
    def delete_at_end(self):
        if self.head is None:
            raise IndexError("List is empty")
        if self.head.next is None:
            self.head = None
            return
        cur = self.head
        while cur.next.next:
            cur = cur.next
        cur.next = None
 
    def delete_at_position(self, pos):
        if self.head is None:
            raise IndexError("List is empty")
        if pos == 0:
            self.head = self.head.next
            return
        cur = self.head
        for _ in range(pos - 1):
            if cur.next is None:
                raise IndexError("Position out of range")
            cur = cur.next
        if cur.next is None:
            raise IndexError("Position out of range")
        cur.next = cur.next.next
 
    def delete_by_value(self, value):
        if self.head is None:
            return False
        if self.head.data == value:
            self.head = self.head.next
            return True
        cur = self.head
        while cur.next and cur.next.data != value:
            cur = cur.next
        if cur.next is None:
            return False
        cur.next = cur.next.next
        return True
 
    # ---------- Search ----------
    def search(self, value):
        """Return index of first match, or -1."""
        cur, i = self.head, 0
        while cur:
            if cur.data == value:
                return i
            cur = cur.next
            i += 1
        return -1
 
    # ---------- Traversal / Display ----------
    def length(self):
        count, cur = 0, self.head
        while cur:
            count += 1
            cur = cur.next
        return count
 
    def display(self):
        parts, cur = [], self.head
        while cur:
            parts.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(parts) + " -> None")
 
 
if __name__ == "__main__":
    ll = LinkedList()
    ll.insert_at_end(10)
    ll.insert_at_end(20)
    ll.insert_at_beginning(5)
    ll.insert_at_position(2, 15)
    ll.insert_after_value(20, 25)
    ll.display()                    # 5 -> 10 -> 15 -> 20 -> 25 -> None
 
    print(ll.search(20))            # 3
    print(ll.search(99))            # -1
 
    ll.delete_at_beginning()
    ll.delete_at_end()
    ll.delete_at_position(1)
    ll.delete_by_value(10)
    ll.display()                    # 20 -> None
    print(ll.length())              # 1
```

### Classic Interview Problems in Singly Linked List

#### A. Reverse a. Singly LInked LIst

Intuition: 
- Flip arrows one by one 
- Maintain perv, curr, next 


```python
from typing import Optional

class Node:
    def __init__(self, value):
        self.value = value
        self.next: Optional["Node"] = None

def display_list(head):
    curr = head
    while curr:
        print(curr.value, end="")
        if curr.next:
            print(" -> ", end="")
        curr = curr.next
    print()

def reverse_list(head):
    curr = head 
    prev= None

    while curr is not None:
        nextNode = curr.next
        curr.next = prev
        prev = curr 
        curr = nextNode

    return prev 
    
# Usage
head = Node(10)

current = head
current.next = Node(20)
current = current.next
current.next = Node(30)
current = current.next
current.next = Node(40)
display_list(head)
head = reverse_list(head)
display_list(head)
```

####B. Find the middle of a linked list  (fast and slow)

Why this works? 
- Fast moves 2x 
- Slow moves 1x
- when fast hits end -> slow is in the middle


```python
from typing import Optional

class Node:
    def __init__(self, value):
        self.value = value
        self.next: Optional["Node"] = None

def display_list(head):
    curr = head
    while curr:
        print(curr.value, end="")
        if curr.next:
            print(" -> ", end="")
        curr = curr.next
    print()

def find_middle(head):
    slow = fast = head 

    while fast and fast.next:
        slow = slow.next 
        fast = fast.next.next 
    return slow.value

# Usage
head = Node(10)
current = head
current.next = Node(20)
current = current.next
current.next = Node(30)
# current = current.next
# current.next = Node(40)
# current = current.next
# current.next = Node(50)

display_list(head)
mid = find_middle(head)
print (mid)
```

#### C. Detect Cycle (Floyd's Algorithm)

Use-case:

    - Infinite loops 

    - Corrupted Pointers 

    - Memory Leaks 

```python
from typing import Optional

class Node:
    def __init__(self, value):
        self.value = value
        self.next: Optional["Node"] = None

def display_list(head):
    curr = head
    while curr:
        print(curr.value, end="")
        if curr.next:
            print(" -> ", end="")
        curr = curr.next
    print()

def has_cycle(head):
    slow = fast = head 

    while fast and fast.next:
        slow = slow.next 
        fast = fast.next.next 

        if slow is fast:
            return True
    return False
    
# Usage
head = Node(10)

current = head
current.next = Node(20)
current = current.next
current.next = Node(30)
current = current.next
current.next = Node(40)
current.next.next = head.next
cycle_check = has_cycle(head)
print (cycle_check)
# display_list(head)
```

Time: O(n) 
Space: O(1) 

```python
def build_list(values, cycle_at=-1):
    """Connect the last node to index cycle_at; -1 means no cycle."""
    nodes = [Node(value) for value in values]

    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]

    if nodes and cycle_at != -1:
        nodes[-1].next = nodes[cycle_at]

    return nodes[0] if nodes else None


# (values, cycle entry index, expected result)
tests = [
    ([], -1, False),                    # Empty list
    ([10], -1, False),                  # Single node, no cycle
    ([10], 0, True),                    # Single node points to itself
    ([10, 20], -1, False),              # Two nodes, no cycle
    ([10, 20], 0, True),                # Last node points to head
    ([10, 20, 30, 40], -1, False),      # Even-length list, no cycle
    ([10, 20, 30], -1, False),          # Odd-length list, no cycle
    ([10, 20, 30, 40], 0, True),        # 40 points to 10
    ([10, 20, 30, 40], 1, True),        # 40 points to 20
    ([10, 20, 30, 40], 3, True),        # 40 points to itself
    ([10, 10, 10], -1, False),          # Repeated values are NOT a cycle
]

for values, cycle_at, expected in tests:
    head = build_list(values, cycle_at)
    actual = has_cycle(head)

    assert actual == expected, (
        f"FAILED: values={values}, cycle_at={cycle_at}, "
        f"expected={expected}, got={actual}"
    )

    print(f"PASS: {values}, cycle_at={cycle_at} → {actual}")
```

#### D. Remove Nth Node from End

Trick
    - Move fast pointer n steps 
    - Move both until fast hits end


```python
from typing import Optional

class Node:
    def __init__(self, value):
        self.value = value
        self.next: Optional["Node"] = None

def display_list(head):
    curr = head
    while curr:
        print(curr.value, end="")
        if curr.next:
            print(" -> ", end="")
        curr = curr.next
    print()

def remove_nth_from_end(head, n):
    dummy = Node(0)
    dummy.next = head
    slow = fast = dummy
    
    for _ in range(n):
        fast = fast.next
    
    while fast.next:
        slow = slow.next
        fast = fast.next
    
    slow.next = slow.next.next
    head = dummy.next

# Usage
head = Node(10)

current = head
current.next = Node(20)
current = current.next
current.next = Node(30)
current = current.next
current.next = Node(40)
current = current.next
current.next = Node(40)
current = current.next
current.next = Node(50)
current = current.next
current.next = Node(60)
current = current.next
current.next = Node(70)
display_list(head)
remove_nth_from_end(head, 5)
display_list(head)

```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    # -------------------------------
    # Utility: Convert to Python list
    # -------------------------------
    def to_list(self):
        res = []
        curr = self.head
        while curr:
            res.append(curr.data)
            curr = curr.next
        return res

    # -------------------------------
    # Insert at beginning — O(1)
    # -------------------------------
    def insertAtBegin(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # -------------------------------
    # Insert at end — O(n)
    # -------------------------------
    def insertAtEnd(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return

        curr = self.head
        while curr.next:
            curr = curr.next

        curr.next = new_node

    # -------------------------------
    # Insert at position (1-based)
    # -------------------------------
    def insertAtPos(self, pos, data):
        if pos < 1:
            raise ValueError("Position must be >= 1")

        if pos == 1:
            self.insertAtBegin(data)
            return

        curr = self.head

        for _ in range(pos - 2):
            if curr is None:
                raise IndexError("Position out of bounds")
            curr = curr.next

        if curr is None:
            raise IndexError("Position out of bounds")

        new_node = Node(data)
        new_node.next = curr.next
        curr.next = new_node

    # -------------------------------
    # Insert after a given value
    # -------------------------------
    def insertAfterValue(self, target, data):
        curr = self.head

        while curr:
            if curr.data == target:
                new_node = Node(data)
                new_node.next = curr.next
                curr.next = new_node
                return
            curr = curr.next

        raise ValueError("Target value not found")

    # -------------------------------
    # Length — O(n)
    # -------------------------------
    def length(self):
        count = 0
        curr = self.head
        while curr:
            count += 1
            curr = curr.next
        return count

    # -------------------------------
    # Delete at beginning — O(1)
    # -------------------------------
    def deleteAtBegin(self):
        if not self.head:
            raise IndexError("List is empty")

        val = self.head.data
        self.head = self.head.next
        return val

    # -------------------------------
    # Delete at position (1-based)
    # -------------------------------
    def deleteAtPos(self, pos):
        if pos < 1:
            raise ValueError("Position must be >= 1")

        if not self.head:
            raise IndexError("List is empty")

        if pos == 1:
            return self.deleteAtBegin()

        curr = self.head

        for _ in range(pos - 2):
            if curr.next is None:
                raise IndexError("Position out of bounds")
            curr = curr.next

        if curr.next is None:
            raise IndexError("Position out of bounds")

        val = curr.next.data
        curr.next = curr.next.next
        return val

    # -------------------------------
    # Delete last node — O(n)
    # -------------------------------
    def deleteLastNode(self):
        if not self.head:
            raise IndexError("List is empty")

        if not self.head.next:
            val = self.head.data
            self.head = None
            return val

        curr = self.head

        while curr.next.next:
            curr = curr.next

        val = curr.next.data
        curr.next = None
        return val

    # -------------------------------
    # Traverse (safe, non-destructive)
    # -------------------------------
    def traverse(self):
        curr = self.head
        result = []

        while curr:
            result.append(str(curr.data))
            curr = curr.next

        return " -> ".join(result) if result else "EMPTY"

    # -------------------------------
    # Reverse list — O(n)
    # -------------------------------
    def reverse(self):
        prev = None
        curr = self.head

        while curr:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt

        self.head = prev
```
