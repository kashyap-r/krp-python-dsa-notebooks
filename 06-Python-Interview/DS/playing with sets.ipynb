### Sets 

- Doesn't take duplicates
- Can stire multiple data types - heterogneous


```python
s = {1, 2, 3, 3}
print (s)
print (type(s))

s1 = set(['a', 'b', 'c', 'c', 'd'])
print (s1) 
print (type(s1)) 
```

```python
# Set methods
s1 = set(['a', 'b', 'c', 'c', 'd'])

print (len(s1))
print (s1)
print (sorted(s1))

# add elements to the set
s1.add('f')
print (s1)
s1.remove('c')
print (s1)

# frozen sets are immutable
s2 = frozenset(s1)
print (type(s2))
print (s2)
# s2.add('g')
```

```python
# Set operations on sets 

s3 = set(['a', 'b', 'c', 'c', 'd'])
s4 = {'c', 'd', 'e', 'f', 'g'}
# union operation 
s5 = sorted(s3.union(s4))
s6 = sorted(s4.union(s3))
print (s5)
print (s6)
print ()
# intersection 
s7 = sorted(s3.intersection(s4))
s8 = sorted(s4.intersection(s3))
print (s7)
print (s8)
print ()
# difference 
s9 = sorted(s3.difference(s4))
s10 = sorted(s4.difference(s3))
print (s9)
print (s10)
print ()
# difference update
s11 = s3.difference_update(s4)
print (s11)
s12 = s4.difference_update(s3)
print (s12)
```

Operators	    Notes

key in s	    containment check

key not in s	non-containment check

s1 == s2	    s1 is equivalent to s2

s1 != s2	    s1 is not equivalent to s2

s1 <= s2	    s1 is subset of s2

s1 < s2	        s1 is proper subset of s2

s1 >= s2	    s1 is superset of s2

s1 > s2	        s1 is proper superset of s2

s1 | s2	        the union of s1 and s2

s1 & s2	        the intersection of s1 and s2

s1 - s2	        the set of elements in s1 but not s2

s1 ˆ s2	        the set of elements in precisely one of s1 or s2
