```python
# listof 1D integers
a = ['a', 'b', 'c', 'd', 'e']
d = dict()
print ("Using enumerate")
for index, element in enumerate(a):
    d[index] = element
print (d)
# using dictionary comprehension 
d2 = {element: index for index, element in enumerate(a)}
print (d2)
print ()
for key, values in d.items():
    print (key, values)

print ()
# One-liner shortcut
d1 = dict(enumerate(a))

for key, values in d1.items():
    print (key, values)

```
