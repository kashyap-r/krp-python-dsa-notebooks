In Python, *args and **kwargs allow a function to accept a variable number of arguments when you do not know in advance how many inputs will be passed

- *args (Non-keyword arguments).Collects any number of extra-positional arguements into a tuple 

- **kwargs (Keyword arguments). Collects any number of extra named (key=value) arguments into a dictionary

(Note: The names args and kwargs are just conventions—the single * and double ** operators are what actually control the behavior.)

How they work ?

1. Using *args

Use *args when you want to pass a flexible list of unnamed values. Inside the function, you can iterate over args just like a standard tuple 

```python
def add_numbers(*args):
    return sum(args)

print (add_numbers(5, 2, 3, 1, 4))
```

```python
def add_numbers(*args):
    l1 = list(args)
    t1 = args
    print (l1)
    print (t1)
    print (args)

add_numbers(1, 2, 3)
```

Using **kwargs 

Use **kwargs when you want to handle named arguements dynamically. Inside the function, kwargs behaves like a standard dictionary

```python
def print_profile(**kwargs):
    for key, value in kwargs.items():
        print (f"{key}: {value}")

print_profile(name="Alice", role="Developer", city="Bengaluru")
```

Combining Both & Parameter Order 

You can use standard parameters, *args, and **kwargs in the same functioin, provided they follow strict order:
standard_args -> *args -> **kwargs in the same function, provided they follow a strict order

standard_args -> *args -> **kwargs 


```python
def order_pizza(size, *toppings, **details):
    print (f"size: {size}")
    print (f"toppings: {toppings}")
    print (f"order details: {details}")

order_pizza("Large", "Mushrooms", "Olives", delivery=True, tip=50)
```


Unpacking When Calling Functions

The * and ** operators also work in reverse when calling a function, allowing you to unpack existing lists/tuples or dictionaries directly into function arguments:


```python
def greet(first, last):
    print(f"Hello, {first} {last}!")

# Unpacking a list with *
names = ["James", "Bond"]
greet(*names)  # Output: Hello, Jane Doe!
print () 
people = [["Jane", "Doe"], ["James", "Bond"]]
for person in people:
    greet(*person)
print ()
# Unpacking a dictionary with **
name_dict = {"first": "John", "last": "Smith"}
greet(**name_dict)  # Output: Hello, John Smith!
print () 
```

Flatten the list in a single call

If your function accepts any number of arguments via *args, flatten the nested lists with itertools.chain.from_iterable() before unpacking

```python
import itertools

def print_all(*args):
    print(args)  # Output: (1, 2, 3, 4)

nested = [[1, 2], [3, 4]]
print_all(*itertools.chain.from_iterable(nested))

```


2. Unpacking a List of Dictionaries

Using func(**list_of_dicts) raises a TypeError because the ** operator only works on a single mapping (dictionary), not a list.

• Call the function once per dictionary. Loop through the list and unpack each dictionary using **. As noted by users on Reddit, the dictionary keys must match the function's parameter names.

```python
def create_user(name, role):
    print(f"{name} is a {role}")

users = [{"name": "Anaya", "role": "Admin"}, {"name": "Dev", "role": "Editor"}]

for user in users:
    create_user(**user)
```

Merge dictionaries into a single call. If your function accepts **kwargs and the dictionaries have unique keys, combine them into one dictionary before unpacking

```python
def show_config(**kwargs):
    print(kwargs)  # Output: {'theme': 'dark', 'font': 14}

configs = [{"theme": "dark"}, {"font": 14}]
merged = {k: v for d in configs for k, v in d.items()}
print (merged)
print ()
show_config(**merged)



```
