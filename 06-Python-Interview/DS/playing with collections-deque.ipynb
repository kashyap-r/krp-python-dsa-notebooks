### deque operations

Deques are a generalization of stacks and queues (the name is pronounced “deck” and is short for “double-ended queue”). Deques support thread-safe, memory efficient appends and pops from either side of the deque with approximately the same O(1) performance in either direction.

Though list objects support similar operations, they are optimized for fast fixed-length operations and incur O(n) memory movement costs for pop(0) and insert(0, v) operations which change both the size and position of the underlying data representation.

### Deque obejcts support the following methods

append (item): adds the item to the right side of the deque 

appendleft(item): adds item to the left side of the deque 

clear(): remove all elements  from the deque

copy(): creates a shallow copy of the deque 

count(value): count the number of deque elements equal to the value 

extend(iterable): extend the right side by appending elements from the iterable object 

extendleft(iterable): extend the left sideo f the deque by appending elements from the iterable. 

index(value[, start[, stop]]): return the position of the value in the deque 

insert(index, value): insert value intot he deque at position index.

pop(): remove and return the elment from the right side of the deque 

popleft(): remove and return an element from the left side of the deque

remove(value): remove the first occurance of the value. If not found raises a ValueError

reverse(): reverse the lement of the deque in-place and then return None 

rotate(n=1): Rotate the deque n steps to the right. If n is negative, rotate to the left



```python
# deque method examples 

from collections import deque 

d = deque('ghi')
for elem in d:
    print (elem.upper(), end = ' ')

d.append('j')
d.appendleft('f')
print (d) 

d.pop() 
print (d) 
d.popleft() 
print (d) 

print (list(d))
print (d[0])
print (d[-1])

list(reversed(d))

d.extend('jkl')
print (d)

d.rotate(1) 
print (d) 

d.rotate(-1)
print (d)

deque(reversed(d))

d.clear() 
# d.pop()

d.extendleft('abc')
print (d)

d.extendleft('321')
print (d)
```

```python
# deque recipes - shows various appraoches working with deques

# bounded length deques provide functionality similar to tail in unix:
def tail(filename, n=10):
    'Return the last n lines of a file'
    with open(filename) as f:
        return deque(f, n)

```

```python
import itertools

# the below approach maintains a sequence of recently added elements by appending 
# to the right and popping to the left
def moving_average(iterable, n=3):
    # moving_average([40, 30, 50, 46, 39, 44]) --> 40.0 42.0 45.0 43.0
    # https://en.wikipedia.org/wiki/Moving_average
    it = iterable
    d = deque(itertools.islice(it, n-1))
    d.appendleft(0) 
    s = sum(d)
    for elem in it:
        s += elem - d.popleft()
        d.append(elem)
        yield s/n

```

```python
# pattern: BFS 

from collections import deque 

def bfs(graph, start):
    q = deque([start])
    visited = {start}

    while q:
        node = q.popleft()
        print (node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                q.append(neighbor)
```
