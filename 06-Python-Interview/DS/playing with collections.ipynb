Usage Pattern: Grouping

The problem it solves:
Suppose we want to group employees by department:


```python
employees = [
    ("Alice", "AI"), 
    ("Bob", "Admin"),
    ("Charlie", "AI"),
    ("David", "Platform"),
    ("Eva", "Data"),
    ("Mia", "Admin")
]

# To group this with a normal dictionary
groups = {}

for name, department in employees: 
    if department not in groups:
        groups[department] = []
    groups[department].append(name)

print (groups)

newGroups = defaultdict(list)
for name, department in employees:
    newGroups[department].append(name)

for key, value in newGroups.items():
    print (f"{key}: {value}")
```

```python
# How to initialize 
d = defaultdict(list)

d["AI"].append("Rob")
d["AI"].append("Max")
print (d)

c = defaultdict(int)

c["apple"] += 1
c["orange"] += 1 
c["apple"] += 1

print (c)

# Also, a Set
e = defaultdict(set)

e["Alice"].add("Python")
e["Alice"].add("AWS")
e["Alice"].add("Python")

print (e)
```

```python
# ex. 1 Group the below transactions by customer
# Pattern: Aggregation Grouping

from collections import defaultdict
transactions = [
    ("Alice", 100), 
    ("Charlie", 50),
    ("Alice", 2000),
    ("Bob", 75)
]
# Pattern: Grouping 
transactionGroup = defaultdict(list)
for customer, amount in transactions:
    transactionGroup[customer].append(amount)
print (transactionGroup)

# Pattern: Aggregation
# Calculate the total transaction value per customer
totals = defaultdict(int)
for customer, amount in transactions:
    totals[customer] += amount
print(totals)
```

```python
# Pattern: Graph Adjacency List 
# Suppose the edges are:
edges = [
    ("A", "B"), 
    ("A", "C"),
    ("B", "D"),
    ("C", "D")
]

graph = defaultdict(list)

for u, v in edges:
    graph[u].append(v)

for key, value in graph.items():
    print (key, value)

print ()
# for an undirected graph 
uGraph = defaultdict(list)

for u, v in edges:
    uGraph[u].append(v)
    uGraph[v].append(u)

for key, value in uGraph.items():
    print (key, value)

```

### Counter

Applicable for the problems in the context of:

```python
# Basic Counter implementation 

from collections import Counter

nums = [1, 2, 3, 4, 5, 2, 2, 4, 6, 7, 8]
freq = Counter(nums)
print (type(freq))

for i, v in freq.items():
    print (i, v)
```

### pattern: most_common

Ex. Top-K Elements


```python
words = [
    "ai", 
    "data",
    "ai", 
    "ml",
    "data",
    "ai",
    "hr",
    "admin"
]

counts = Counter(words)
# top 2 most common 
counts.most_common(2)
```

```python
# Pattern: Anagram 

s = "listen"
t = "silent"

# How do you compare ?
# traditionally 
print (sorted(s) == sorted(t))

# Using counter
print (Counter(s) == Counter(t))

# to check is_anagram 
# 

def is_anagram(s, t):
    return Counter(s) == Counter(t)

"""
Complexity
Time: O(n)
Space: O(K) where k is the number of distinc characters
"""
```

### deque 

Mentioning of deque -> think efficient queue + stack 


```python
# normal method 

queue = []

queue.append(10)
queue.append(20)
queue.append(30)
queue.append(40)

queue.pop(0)

""" 
Here, Complexity is O(n), because everything shifts left

With deque, the time complexity for the same operation would be O(1)
"""

q = deque() 

q.append(10)
q.append(20)
q.append(30)

q.popleft()
# its O(1) here
```

```python
# deque operations

Deques are a generalization of stacks and queues (the name is pronounced “deck” and is short for “double-ended queue”). Deques support thread-safe, memory efficient appends and pops from either side of the deque with approximately the same O(1) performance in either direction.

Though list objects support similar operations, they are optimized for fast fixed-length operations and incur O(n) memory movement costs for pop(0) and insert(0, v) operations which change both the size and position of the underlying data representation.
```

```python
# Queue pattern 

q = deque() 

q.append("task1")
q.append("task2")
q.append("task3")
q.append("task4")

print (q)

while q: 
    task = q.popleft() 
    print (task) 

# theq is empty now 
print (q)
```

```python
# pattern: BFS 

from collections import deque 

def bfs(graph, start):
    q = deque([start])
    visited = {start}

    while q:
        node = q.popleft()
        print (node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                q.append(neighbor)
```
