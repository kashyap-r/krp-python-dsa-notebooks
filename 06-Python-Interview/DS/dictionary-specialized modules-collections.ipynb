### Specialized Python Dictionary Types 

#### The Collections module extensions 

**defaultdict**

The defaultdict from the collections module eliminates repetitive key existence checks by automatically initializing missing keys with a default value. This is very useful for accumulation tasks like counting, grouping, or building nested structures


References: 
1. www.python.org
2. https://www.geeksforgeeks.org/python/python-collections-module/
3. Many other sources.. 

```python
from collections import defaultdict

# Standard way of finding. word count 
word_count = {}
text = "the quick brown fox jumps over the lazy dog".split() 

print(text)

for word in text:
    if word not in word_count:
        word_count[word] = 0
    word_count[word] += 1 

# using default dict 
word_count_auto = defaultdict(int) 

for word in text:
    word_count_auto[word] += 1

print(type(word_count_auto))
print (dict(word_count_auto))


```

```python
from collections import defaultdict

# Grouping with defaultdict(list)
transactions = [
    ("2024-01-15", "groceries", 45.50),
    ("2024-01-15", "gas", 60.00),
    ("2024-01-16", "groceries", 32.75),
    ("2024-01-16", "entertainment", 25.00),
    ("2024-01-17", "gas", 55.00)
]

by_date = defaultdict(list)
for date, category, amount in transactions:
    by_date[date].append((category, amount))

for date, items in by_date.items():
    print(f"{date}: {items}")
```

```python
# Nested defaultdict for complex structures
nested = defaultdict(lambda: defaultdict(int))
events = [
    ("2024-01", "login", 150),
    ("2024-01", "purchase", 45),
    ("2024-02", "login", 200),
    ("2024-02", "purchase", 60)
]

for month, event_type, count in events:
    nested[month][event_type] += count

print(dict(nested))
```

### Counter 

The Counter class is a specialized dictionary subclass designed for counting hashable objects and performing multiset operations. It's particularly powerful for statistical analysis and frequency distributions. Let’s see an example of how this type works:


```python
# Count occurances in a sequence 
from collections import Counter

tags = ["python", "data", "python", "ml", "data", "python", "statistics", "ml"]
tag_counts = Counter(tags)
print(tag_counts)
```
