Contracts: is like a promise that you will provide some specific behavior. In Classes this means that you 'promise' some functionality

One way to create a contract is through the concept of Interface. In Python, we use an abstract class with no implementation for the purpose.


```python
from abc import ABC, abstractmethod

# interface contract here - children will have to implment all of the abstract methods. In an interface, 
# methods have no implementation, so we use 'pass'

class MyInterface(ABC):
    @abstractmethod
    def my_method(self):
        pass 

class MyClass(MyInterface):
    def my_method(self):
        print ("my_method implmentation in MyClass")

class AnotherClass(MyInterface):
    def my_method(self):
        print ("my_method implmentation in AnotherClass")

my_obj = MyClass()
my_obj.my_method()

another_obj = AnotherClass()
another_obj.my_method()
```

Inheritance: Abstract Classes

In many cases, you have actual functionality and not just a signature i.e. actual functionality behind the signature that you would like to have all child class inherit.

Note: In interfaces (above) you just have the signature - not the actual implementation. So you're effectively saying that interfaces have a skeleton of how/ what the child class should look like or inherit.
Abstract classes go a step further in adding re-usable functionality that is inherited by all children

This is how/ where you would use an Abstract class.
    * You can add specific implemented methods 
    * specific constrants or variables 

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, color):
        self.color = color 

    @abstractmethod
    def area(self):
        pass 

    @abstractmethod
    def perimeter(self):
        pass 

    def description(self):
        print (f"{self.__class__.__name__} has the color: {self.color}")

"""
In the child / derived class the the area() and the perimeter() method have to be implmented 
and the description() does not have to be implemented unless it needs a specific additional 
functionality
"""

class Rectangle(Shape):
    def __init__(self, width, height, color):
        super().__init__(color)
        self.width = width 
        self.height = height 

    def area(self):
        return self.width * self.height

    def perimeter (self):
        return 2 * (self.width + self.height)

rectangle = Rectangle(4, 5, "Blue")
print (rectangle.area())
print (rectangle.perimeter())
rectangle.description()
```

```python
# Example - Abstract Classes: Contracts + Reusable Code. 
from abc import ABC, abstractmethod

class DatabaseConnector(ABC):
    # Specific constraint/variable inherited by all children
    DEFAULT_PORT = 5432 

    def __init__(self, host: str):
        self.host = host  # Reusable instance variable

    # A concrete method (Actual functionality inherited by all children)
    def log_connection(self) -> None:
        print(f"Connecting to host {self.host} on port {self.DEFAULT_PORT}...")

    # The abstract contract (Must be unique to each child)
    @abstractmethod
    def execute_query(self, query: str) -> list:
        pass

# --- Child Class ---
class PostgresConnector(DatabaseConnector):
    def execute_query(self, query: str) -> list:
        # Custom implementation for Postgres
        return [f"Results for '{query}' from Postgres"]

# --- Testing the Abstract Class ---
pg = PostgresConnector("localhost")
pg.log_connection()  # Output: Connecting to host localhost on port 5432...
print(pg.execute_query("SELECT * FROM users"))
```

Encapsulation in Python

Python does not have true access modifiers like Java or C++ to strictly lock variables down. Instead, it relies on naming conventions and a mechanism called name mangling.


1. Protected Attributes (_variable)

• What it means: A single leading underscore is a gentle warning to other programmers. It says: "This is an internal detail. Please do not modify or access this directly from outside the class, unless you are creating a subclass."
• Enforcement: None. Python will allow you to read or rewrite _account_number anywhere in your code without an error. It relies entirely on the honor system.

2. Private Attributes (__variable)

• What it means: A double leading underscore triggers Name Mangling. Python physically rewrites the variable name behind the scenes to include the class name as a prefix.
• Enforcement: Structural. If you try to access __balance from outside the class, Python throws an AttributeError. It makes accidental tampering much harder.


When to Use Which?

• Use Public (variable) for standard data that is safe to read and modify directly from anywhere (e.g., account.user_display_name).
• Use Protected (_variable) when building tools, frameworks, or parent classes where you expect child classes to inherit and use the variable, but you want to tell external users to keep their hands off.
• Use Private (__variable) sparingly. Use it when data corruption poses a serious risk to how the class functions, or when you absolutely must force external code to use specific methods (getters/setters) to interact with that data.




```python
class BankAccount:
    def __init__(self, account_number, balance):
        self._account_number = account_number # protected attribute (single underscore)
        self.__balance = balance # private attribute (double underscore - name mangling)

    # getter method for private attribute 
    def get_balance(self):
        return self.__balance

    # setter method for prvate attribute
    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else: 
            print ("Invalid balance")

    # Public method that uses the private attribute
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Invalid deposit amount")

    # public method that uses private attribute
    def withdrawl(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount 
        else: 
            print ("Invalid Withdrawl Amount!")

# Testing the encapsulation
account = BankAccount("123456", 1000)

# Accessing the protected attribute (not recommended, but possible)
print("Account number:", account._account_number)

# Accessing and modifying the private attribute through getter and setter methods
print("Initial balance:", account.get_balance())
account.set_balance(500)
print("Updated balance:", account.get_balance())

# Using public methods that internally use the private attribute
account.deposit(100)
print("Balance after deposit:", account.get_balance())
account.withdrawl(50)
print("Balance after withdrawal:", account.get_balance())
```

The Pythonic Way: Using the @property Decorator 

@property is the gold standard when you need to enforce validation while keeping your code clean and intuitive.

```python
# The pythonic way 

class BankAccount: 
    def __init__(self, account_number: str, balance: float ):
        self._account_number = account_number
        self._balance = balance

    # The pythonic getter 
    @property
    def balance(self) -> float:
        """Retrieves the balance cleanly cia: account.balance"""
        return self._balance

    # The pythonic setter
    @balance.setter
    def balance(self, value: float) -> None:
        """Updates the balance with validation via account.balance  = value"""
        if value >= 0:
            self._balance += value 
        else: 
            print ("Invalid Balance!")

    # Public Methods 
    def deposit(self, amount: float)-> None:
        """Deposits a Valid positive amount"""
        if amount > 0:
            self._balance += amount
        else: 
            print ("Invalid Deposit Amount")

    def withdrawl(self, amount: float) -> None:
        """Withdraws an amount if funds are available"""
        if 0 < amount <= self._balance:
            self._balance -= amount 
        else:
            print ("Invaalid Withdrawl Amount")

        
# Testing the encapsulation
account = BankAccount("123456", 100000)

# Accessing the protected attribute (not recommended, but possible)
print("Account number:", account._account_number)

# Accessing and modifying the private attribute through getter and setter methods
print("Initial balance:", account.balance)
account.balance = 50000
print("Updated balance:", account.balance)

# Using public methods that internally use the private attribute
account.deposit(20000)
print("Balance after deposit:", account.balance)

account.withdrawl(50000)
print("Balance after withdrawal:", account.balance)
```

Single Underscore Switch: The private backend variable was changed from __balance to _balance. When using properties, you do not need double underscores because the property name (balance) acts as the public interface, and the single underscore tells developers the actual data store is protected.


The Rules of Thumb for Python Variables

Because properties already handle the security of your data interface, Python developers follow this naming standard:

• Use self._balance (Single Underscore) with properties. It signals "this is internal/protected," plays nicely with your property decorators, and allows child classes to seamlessly inherit and extend your code.

• Use self.__balance (Double Underscore) only when you are terrified of naming collisions. This is usually reserved for deep framework development where you want to absolutely guarantee a child class cannot accidentally overwrite a parent variable.


```python
# A small extension of the above problem 

class BankAccount:
    def __init__(self, account_number: str, balance: float):
        self._account_number = account_number
        
        # We trigger our property setter right inside the constructor 
        # to ensure the initial balance is also validated!
        self.balance = balance                

    @property
    def balance(self) -> float:
        """Retrieves the balance cleanly."""
        return self._balance

    @balance.setter
    def balance(self, value: float) -> None:
        """Validates and updates the balance. Throws an error if invalid."""
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value

    def deposit(self, amount: float) -> None:
        """Deposits a valid positive amount."""
        if amount <= 0:
            raise ValueError("Deposit amount must be greater than zero.")
        self.balance += amount  # Reuses the setter logic safely

    def withdrawal(self, amount: float) -> None:
        """Withdraws an amount if funds are available."""
        if amount <= 0:
            raise ValueError("Withdrawal amount must be greater than zero.")
        if amount > self._balance:
            raise ValueError("Insufficient funds for this withdrawal.")
        self.balance -= amount  # Reuses the setter logic safely


# --- 1. Creating a Subclass ---
class SavingsAccount(BankAccount):
    def __init__(self, account_number: str, balance: float, interest_rate: float):
        # Initialize the parent class attributes
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate

    def apply_interest(self) -> None:
        """Calculates and updates balance using internal protected variable."""
        # Using self._balance works perfectly here because it has a single underscore.
        # If it had double underscores, this line would crash!
        interest = self._balance * self.interest_rate
        self.deposit(interest) 


# Testing the encapsulation
account = BankAccount("123456", 100000)

# Accessing the protected attribute (not recommended, but possible)
print("Account number:", account._account_number)

# Accessing and modifying the private attribute through getter and setter methods
print("Initial balance:", account.balance)
account.balance = 50000
print("Updated balance:", account.balance)

# Using public methods that internally use the private attribute
account.deposit(20000)
print("Balance after deposit:", account.balance)

account.withdrawal(50000)
print("Balance after withdrawal:", account.balance)
```

```python
import math 

class Fraction: 
    def __init__(self, numerator: int, denominator: int):
        self._numerator = numerator
        self._denominator = denominator

    @property
    def numerator(self) -> int:
        return self._numerator 

    @numerator.setter
    def numerator(self, value: int) -> None:
        self._numerator = value 

    @property 
    def denominator(self) -> int:
        return self._denominator

    @denominator.setter
    def denominator(self, value: int)-> None:
        if value == 0:
            raise ValueError("The denominator must be non-zero")
        self._denominator = value 
    
    # Arithmatic operation 
    
    def add(self, other: "Fraction") -> "Fraction":
        """Adds two fractions: (a/b) + (c/d) = (ad + bc) / bd"""
        new_num = (self._numerator * other.denominator) + (self._denominator * other.numerator)
        new_den = self._denominator * other.denominator
        return Fraction(new_num, new_den)
    
    def subtract(self, other: "Fraction") -> "Fraction":
        """Subtracts another fraction: (a/b) - (c/d) = (ad - bc) / bd"""
        new_num = (self._numerator * other.denominator) - (self._denominator * other.numerator)
        new_den = self._denominator * other.denominator
        return Fraction(new_num, new_den)

    def multiply(self, other: "Fraction") -> "Fraction":
        """Multiplies two fractions: (a/b) * (c/d) = (ac) / (bd)"""
        new_num = self._numerator * other.numerator
        new_den = self._denominator * other.denominator
        return Fraction(new_num, new_den)

    def divide(self, other: "Fraction") -> "Fraction":
        """Divides by another fraction: (a/b) / (c/d) = (ad) / (bc)"""
        if other.numerator == 0:
            raise ValueError("Cannot divide by a fraction with a zero numerator.")
        new_num = self._numerator * other.denominator
        new_den = self._denominator * other.numerator
        return Fraction(new_num, new_den)

    # --- Utility Methods ---
    def simplify(self) -> "Fraction":
        """Simplifies the fraction to its lowest terms using the Greatest Common Divisor (GCD)."""
        gcd = math.gcd(self._numerator, self._denominator)
        
        # Handle signs cleanly: if denominator is negative, move the sign to the numerator
        simplified_num = self._numerator // gcd
        simplified_den = self._denominator // gcd
        
        if simplified_den < 0:
            simplified_num = -simplified_num
            simplified_den = -simplified_den
            
        return Fraction(simplified_num, simplified_den)

    def __str__(self) -> str:
        return f"{self._numerator}/{self._denominator}"

# --- Demonstration of the Class in Action ---
try:
    # 1. Create two fractions
    f1 = Fraction(1, 2)
    f2 = Fraction(2, 3)

    print(f"Fraction 1: {f1}")  # Output: 1/2
    print(f"Fraction 2: {f2}")  # Output: 2/3

    # 2. Perform Arithmetic Operations
    print(f"Addition: {f1.add(f2)}")        # Output: 7/6
    print(f"Subtraction: {f1.subtract(f2)}")  # Output: -1/6
    print(f"Multiplication: {f1.multiply(f2)}") # Output: 2/6
    print(f"Division: {f1.divide(f2)}")       # Output: 3/4

    # 3. Simplify a fraction
    f3 = Fraction(4, 8)
    print(f"Original: {f3} -> Simplified: {f3.simplify()}")  # Output: 4/8 -> 1/2

    # 4. Testing the Contract Violations
    # Broken Fraction creation:
    # broken = Fraction(5, 0)  # Raises ValueError: The denominator must be non-zero.
    
    # Broken Division operation:
    # f1.divide(Fraction(0, 4))  # Raises ValueError: Cannot divide by a fraction with a zero numerator.

except ValueError as e:
    print(f"Error caught: {e}")

```
