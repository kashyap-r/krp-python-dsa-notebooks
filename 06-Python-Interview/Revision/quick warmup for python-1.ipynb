### Bitwise Operators 

Bitwise operators are used to perform operations on binary representations of numbers. These operators work on bits (0s and 1s) and perform bit-level operations.

    Bitwise AND (&): Compares corresponding bits and returns 1 if both bits are 1, otherwise 0.

    Bitwise OR (|): Compares corresponding bits and returns 1 if at least one of the bits is 1.

    Bitwise XOR (Exclusive OR) (^): Compares corresponding bits and returns 1 if the bits are different, otherwise 0.

    Bitwise Left Shift (<<): Shifts the bits of the number to the left by the specified number of positions, adding 0s at the right.

    Bitwise Right Shift (>>): Shifts the bits of the number to the right by the specified number of positions, discarding the rightmost bits.
    

```python
a = 10
b = 4 

print (a & b)
print ( a | b)
print (~a)
print (a^b)
print ( a >> 2 )
print ( a << 2 )
```

### Function Arguements 

3. Variable-length Arguements 

These allow you to pass a variable number of arguements to a function.
Use *args for non-keyword and **kwargs for keyword arguments. *args allows a function to accept any number of positional arguments as a tuple. **kwargs allows a function to accept any number of keyword arguments as a dictionary.

#### Non-Keyword Arguments (*args)

```python
def print_names(*names):
    for name in names:
        print(name)

print_names("James", "Avengers", "Charlie", "Bob")

def sum_numbers(*args):
    return (sum(args))

print (sum_numbers(1, 2, 3, 4, 5))
```

#### Keyword Arguments (**kwargs)

The **kwargs syntax allows a function to accept any number of named (keyword) arguments. Inside the function, these arguments are captured and stored in a dictionary.


```python
def showProfile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

showProfile(name="Alice", role="Developer")
```

#### Using them together

You can use standard arguments, *args, and **kwargs all in a single function definition. However, you must follow this specific order:

1. Standard positional arguments

2. *args

3. **kwargs

```python
def master_function(standar_arg, *args, **kwargs):
    print (f"Standard arguements: {standar_arg}")
    print (f"Args: {args}")
    print (f"kwargs: {kwargs}")

master_function("Hello", 1, 2, 3, active=True, theme='dark')
```

### Lambda Functions  

```python
# lambda function for basic operation (square of a number)
square = lambda x: x ** 2 
print (square(25))

# lambda with condition (check if number is even or odd)
even_odd = lambda x: "Even" if x % 2 == 0 else "Odd"
print (even_odd(4))

# Lambda with LIst comprehension (transform a list to its squares)
numbers = [1, 2, 3, 4]
squared_numbers = [lambda x: x ** 2 for x in numbers]  # List of lambda functions
squared_list = [func(num) for func, num in zip(squared_numbers, numbers)]
print(squared_list)  

# Lambda with filter() (filter numbers greater than 2)
filtered = filter(lambda x: x> 2, numbers)
print (list(filtered))

# Lambda with map() (double each number in the list)
doubled = map(lambda x: x * 2, numbers)
print(list(doubled))  

# Lambda with reduce() (find the product of a list of numbers)
from functools import reduce
product = reduce(lambda x, y: x * y, numbers)
print(product)
```

How Python Stores List Elements?

In Python, a list doesn’t store actual values directly. 

Instead, it stores references (pointers) to objects in memory. This means numbers, strings and booleans are separate objects in memory and the list just keeps their addresses.

That’s why modifying a mutable element (like another list or dictionary) can change the original object, while immutables remain unaffected.


## Python Dictionary 

Python dictionary is a data structure that stores the value in key: value pairs. 

Values in a dictionary can be of any data type and can be duplicated, whereas keys can't be repeated and must be immutable.

Keys are case sensitive which means same name but different cases of Key will be treated distinctly.

Keys must be immutable which means keys can be strings, numbers or tuples but not lists.

Duplicate keys are not allowed and any duplicate key will overwrite the previous value.

Internally uses hashing. Hence, operations like search, insert, delete can be performed in Constant Time.

From Python 3.7 Version onward, Python dictionary are Ordered.


```python
## Dictionaries
d1 = {'Name': 'Kash', 1: [1, 2, 3, 4]}
print (d1)
print (d1.items())
print (d1.keys())
print (d1.values())

# Accessing values in a dict
print (d1['Name'])

print (d1.get(1))

d2 = {x: x**2 for x in [1, 2, 3, 4, 5]}
print (d2)
```
