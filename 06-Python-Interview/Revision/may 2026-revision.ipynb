### Exercise 

Loop through numbers 1 - 20 
- for 4 and 13, print x is unlucky 
- for even print d is even 
- for odd, print x is odd 

```python
# Exercise 

for num in range(1, 21):
    if num ==4 or num == 13:
        print (f"{num} is unlucky")
    elif num%2 == 0:
        print (f"{num} is even")
    else:
        print (f"{num} is odd")
```

```python
# alternative solution (has less print statements)
for num in range(1, 21):
    if num ==4 or num == 13:
        state =  "unlucky"
    elif num%2 == 0:
        state = "even"
    else:
        state = "odd"
    print (f"{num} is {state}")
```
