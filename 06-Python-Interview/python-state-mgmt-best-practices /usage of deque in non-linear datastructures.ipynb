Why Use deque instead of a list in Trees 

Using the iterative approach you wrote is highly valued in coding interviews. Interviewers love it because it shows you understand space optimization—it uses O(1) auxiliary space, whereas the recursive approach uses O(h) space on the call stack (where h is the height of the tree).

In short, a deque (double-ended queue) is optimized for fetching items from the front, while a Python list is not.
For a Breadth-First Search (BFS) / Level-Order traversal, you need a FIFO (First-In, First-Out) structure. This means you must constantly pull elements from the beginning of your collection.

Why list.pop(0) is slow

A Python list is a dynamic array stored in contiguous memory. When you remove the first element using list.pop(0):
1. The first element is deleted.
2. Every single remaining element in the list must be shifted one index to the left to close the gap.
If your tree has thousands of nodes, shifting the array every time you process a node makes your traversal drastically slower—turning an efficient O(n) algorithm into a sluggish O(n²) algorithm.

Why deque.popleft() is fast

A deque is designed precisely for fast pushes and pops from both ends. Removing the leftmost element requires no shifting at all; it just adjusts a memory pointer. It will always take the same fraction of a microsecond, no matter how massive your tree gets.
