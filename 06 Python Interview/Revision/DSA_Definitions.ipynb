```python
# Leetcde 121


def maxProfit(prices):
    left, right = 0, 1
    maxProfit = 0 
    while right != len(prices):
        if prices[left] < prices[right]:
            profit = prices[right] - prices[left]
            maxProfit = max(maxProfit, profit)
        else:
            left = right 
        right += 1
    return maxProfit

prices = [7,1,5,3,6,4]
print ("max profit:",maxProfit(prices))
```

#### Breadth First Search or BFS for a Graph

Breadth First Search (BFS) is a graph traversal algorithm that starts from a source node and explores the graph level by level. First, it visits all nodes directly adjacent to the source. 

Then, it moves on to visit the adjacent nodes of those nodes, and this process continues until all reachable nodes are visited.



Input: adj[][] = [[1, 2], [0, 2], [0, 1, 3, 4], [2], [2]]



adj = [[2, 3], [2], [0, 1], [0], [5], [4]]



Output: [0, 2, 3, 1, 4, 5]

```python
# BFS for single connected component

from collections import deque
def bfs(adj):
    V = len(adj)
    visited = [False] * V
    result = []

    src = 0 
    q = deque()
    visited[src] = True 
    q.append(src)

    while q:
        curr = q.popleft() 
        result.append(curr)

        # visited all the unvisited neighbors of the current node.
        for x in adj[curr]:
            if not visited[x]:
                visited[x] = True 
                q.append(x)
    return result


adj = [[2, 3], [2], [0, 1], [0], [5], [4]]

result = bfs(adj)

print (result)
```

#### BFS of a Disconnected Undirected Graph 



Input: adj = [[2, 3], [2], [0, 1], [0], [5], [4]]

Output: [0, 2, 3, 1, 4, 5]


```python
from collections import deque

def bfsConnected(adj, src, visited, result):
    q = deque()
    visited[src] = True
    q.append(src)

    while q:
        curr = q.popleft()
        result.append(curr)

        for x in adj[curr]:
            if not visited[x]:
                visited[x] = True 
                q.append(x)

def bfs(adj):
    V = len(adj)
    visited = [False] * V
    res = []

    for i in range(V):
        if not visited[i]:
            bfsConnected(adj, i, visited, res)
    return res

adj = [[2, 3], [2], [0, 1], [0], [5], [4]]

result = bfs(adj)

print (result)
```
