### Accessing dictionary items

1. Using get() Method, specify the "key" 
2. Get all keys using keys() method, all values using values() method and both keys, values using dict.items() method
3. using the in operator 

```python
# Accessing Dictionary Items 
d = {"name": "Anaconda", 1: "Python", (1, 2): [1, 2, 3, 4]}
print (d["name"])
print (d.get("name"))
print (d[1])
print (d[(1, 2)])

# Adding and Updateing Dictionary items 
d= {1: "I'm ", 2: "learning", 3: "Python!" }
print("Dictionary before:")
print (d) 
print("Dictionary after changes:")
# adding a new key value Pair
d["name"] = "Bond"
d[4] = "in Anaconda"
d[3] = "Python in"
print (d)

print ("Using in Operator")
for key in d:
    print (key, d[key])
```

```python
# Removing Dictionary Items
# There are 4 built-in methods 
# del: removes item using a key
# pop: removes item with given key and return its value 
# clear(): removes all items in the dictionary 
# popitem(): removes and returns the last inserted key-value pair


```

```python
# Iteerating through a dictionary 

d= {1: "I'm ", 2: "learning", 3: "Python!" }

print ("iterate over keys ")
for key in d:
    print (key)

print("iterate over values") 
for value in d.items():
    print (value)

print ("Iterate over key-value pairs") 
for key, value in d.items():
    print (f"{key}: {value}")

```

```python
# Sort dictionary by key or value
# Sorting by value

d = {'watermelon': 4, 'apple': 1, 'cheeku': 3, 'banana': 2}
print ("Unsorted dict")
print (d)

# Using sorted() with lambda 
sortasc = {k: v for k, v in sorted(d.items(), key= lambda item: item[1])}
print ("Sorted dict")
print (sortasc) 

# Using ordered dict 
from collections import OrderedDict
d = {"Saturday": 7, "Monday": 1, "Wednesday": 3, "Tuesday": 3}
print (d)
result = OrderedDict(sorted(d.items(), key= lambda item: item[1]))
print (result)

# Using loop with sorted
```

```python
# Creating a nested Dictionary 
students = {}

students['student1'] = {'name': 'Drake', 'age':23, 'grade': 'A'}
students['student2'] = {'name': 'Brake', 'age':24, 'grade': 'B'}
students['student3'] = {'name': 'Krake', 'age':25, 'grade': 'A+'}

print ("Student Details")
print (students)
```

```python
# Adding elements to a nested dictionary

person = {'employee1': {'name': 'Pumpkin', 'age': 24}}
print (person)
# adding a new key-value pair to existing inner dictionary 
person['employee1']['department'] = 'HR'

# adding a new inner dictionaty 
person['employee2'] = {'name': 'Apple', 'age': 34, 'department': 'devices'}

print ("updated dictionary")
print (person)
```
