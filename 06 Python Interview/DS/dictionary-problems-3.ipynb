1️⃣ Sort dictionary by value DESC, break ties by key ASC

📌 Problem statement

Given a dictionary of products and their sales count, sort it such that:

    - Higher sales come first

    - If two products have the same sales, sort alphabetically by product name


```python
sales = {
    "laptop": 120,
    "mouse": 50,
    "keyboard": 50,
    "monitor": 200,
    "usb": 120
}

"""
Output 
{
    'monitor': 200,
    'laptop': 120,
    'usb': 120,
    'keyboard': 50,
    'mouse': 50
}
"""
```

```python
# naive Solution
```

```python
# optimized solution 

sorted_sales = { 
    k: v for k, v in sorted(
        sales.items(),
        key=lambda item: (-item[1], item[0])
    )
}

print (sorted_sales)
```

2️⃣ Sort dictionary by length of key, then by value
📌 Problem statement

Given a dictionary of usernames and scores:

    * Sort by length of username (shorter first)

    * If same length, sort by score descending
    

```python
# Input 

scores = {
    "alex": 90,
    "johnny": 85,
    "sam": 92,
    "christopher": 88,
    "bob": 90
}

# Output 
{
    'sam': 92,
    'bob': 90,
    'alex': 90,
    'johnny': 85,
    'christopher': 88
}
```

```python
# naive solution 
```

```python
# Optimized solution 

sorted_scores = {
    k: v for k, v in sorted(
        scores.items(),
        key=lambda item: (len(item[0]), -item[1])
    )
}

sorted_scores
```

4️⃣ Sort dictionary of lists by aggregate value
📌 Problem statement

Given students and their marks in multiple subjects, sort students by average score, highest first.

📌 Interview insight:
This tests lambda over nested structures.

```python
# Input 

students = {
    "alice": [85, 90, 95],
    "bob": [70, 75, 80],
    "charlie": [88, 92, 90],
    "dave": [90, 90, 90]
}
```

#### Type: Sort Dictionary by value DESC, break ties by key ASC

**Problem statement**

Given a dictionary of products ad their sales count, sort it such that: 
- higher sales come first
- if two products have the same sales, sort alphabetically by product name


# Input: 

sales = {
    "laptop": 1200,
    "mouse": 500, 
    "keyboard": 500,
    "monitor": 2000,
    "usb": 1200,
    "cpu": 2500
}

# Expected Output:


```python
# naive solution 

def naive_sort_products(products_dict):
    # Step 1: Convert dictionary to a list of pairs (tuples)
    products_list = list(products_dict.items())
    n = len(products_list)
    
    # Step 2: Implement a naive Bubble Sort
    for i in range(n):
        for j in range(0, n - i - 1):
            # Extract names and sales for comparison
            name1, sales1 = products_list[j]
            name2, sales2 = products_list[j + 1]
            
            # Condition 1: Higher sales should come first
            # Condition 2: If sales are equal, sort alphabetically by name
            if (sales1 < sales2) or (sales1 == sales2 and name1 > name2):
                # Swap the elements
                products_list[j], products_list[j + 1] = products_list[j + 1], products_list[j]
                
    # Step 3: Convert the sorted list back into a dictionary
    return dict(products_list)

# Example Usage:
data = {"banana": 10, "apple": 50, "cherry": 50, "date": 20}
sorted_data = naive_sort_products(data)
print(sorted_data)
# Output: {'apple': 50, 'cherry': 50, 'date': 20, 'banana': 10}
```

```python

def optimized_sort_products(products_dict):
    # Sort by sales descending (-x[1]) and then by name alphabetically (x[0])
    sorted_items = sorted(products_dict.items(), key=lambda x: (-x[1], x[0]))
    
    # Convert the sorted list of tuples back into a dictionary
    return dict(sorted_items)

# Example Usage:
data = {"banana": 10, "apple": 50, "cherry": 50, "date": 20}
print(optimized_sort_products(data))
# Output: {'apple': 50, 'cherry': 50, 'date': 20, 'banana': 10}
```

### Explanation 

sorted_sales = {
    k: v for k, v in sorted(
        sales.items(),
        key = lambda item: (-item[1], item[0])
    )
}

print (sorted_sales)

sales.items(): Converts the dictionary into a list of key-value pairs (tuples), like [('laptop', 1200), ('mouse', 500)...]

key = lambda item: ...: Tells python exactly how to compare these pairs. 
Each item represents a single (key, value) tuple. 

    * item[0] is the key (ex. prodcut name)
    * item[1] is the value (ex. sales number)

(-item[1], item[0]): This is a tuple containing the sorting rules in order of priority:

Primary Sort (-item[1]): By making the value negative, Python flips the default ascending sort into a descending sort for numbers. A sales value of 50 becomes -50, and 10 becomes -10. Since -50 is smaller than -10, 50 gets placed first.

Secondary Sort (item[0]): If two items have the exact same sales value, Python looks at the second element in the tuple to break the tie. It sorts the keys in standard ascending order (A to Z or smallest to largest).



```python

sales = {
    "laptop": 1200,
    "mouse": 500, 
    "keyboard": 500,
    "monitor": 2000,
    "usb": 1200,
    "cpu": 2500
}

# sorted_sales = {
#     k: v for k, v in sorted(
#         sales.items(),
#         key = lambda item: (-item[1], item[0])
#     )
# }

# print (sorted_sales)

sorted_sales = {
    k: v for k, v in sorted(
        sales.items(),
        key = lambda item: (-item[1], item[0])
    )
}

print (sorted_sales)
```

### Using OrderedDict: 

This method sorts a dictionary by values and stores the result in an OrderedDict, which preserves insertion order.
