```python
"""
Sum of array 

Constraints: 
    1. Size of the array 
    2. Range of the array 
""" 

arr = [1, 2, 3, 4]
total = 0 
for item in arr:
    total += item
print (total)
```

```python
numbers = [1, 2, 3, 4, 5]
total = sum(numbers)
print(total)  # Output: 15
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
total = reduce(lambda x, y: x + y, numbers)
print(total)  # Output: 15
```
