A tuple is a built-in data type used to store an ordered, immutable collection of items

Immutable: Once created, you cannot add, remove, or modify its elements.

Ordered: The items have a defined sequence that will not change.

Heterogeneous: A single tuple can hold elements of different data types (e.g., strings, integers, lists, or other tuples).

Allows Duplicates: The same value can appear multiple times.




```python
# Creating tuples 

coordinates = (5, 7)

user_profile = ("Tony Stark", 35, "Engineer")

colors = "Red", "Green", "Blue"

empty_tuple = ()

print (type(coordinates))
print( type(user_profile))
print (type(colors))
print (empty_tuple)

# string
single = ("apple")
print (type(single))

# tuple 
single = ('apple',)
print (type(single))
```

```python
# Accessing element in a tuple

fruits = ("apple", "banana", "cherry", "date", "elaichi", "funnyfruit", "gooseberry")
print(fruits[0])
print (fruits[-1])
# slicing [start:stop] stop index is excluded
print (fruits[2:])
print(fruits[1:3])
print (fruits[::-1])
print()
for i in range(len(fruits)):
    print (fruits[i])

```

### Tuple operations 

```python
tup = (1, 2, 3, 3, 4) 

# tuples are indexed
print (tup[1])
print (tup[4])
# cannot update
# tup[2] = 100
print (tup)
# accessing tuple using negative index 
print (tup[-1])
print (tup[-len(tup)])
```

#### Operations related to tuples 

```python
tup = (1, 2, 3, 4, 5)

for x in tup:
    print (x, end=' ')
print()
# concantenation of python tuples 
tup1 = (0, 1, 2, 3)
tup2 = ('python', 'learn')
print (tup1+tup2)

# create nested tuples 
tup3 = (tup1, tup2)
print (tup3)

# initializing tuples with constants
tup4 = ('Pyt') * 3
print (tup4)
print ()

tup5 = ('Pyt',) * 3
print (tup5)
```

```python
# Slicing tuples 

tup = (0, 1, 2, 3)
print (tup[1:])
print (tup[::-1])
print (tup[2:4])
```

### Tuple unpacking with Asterisk (*)

    - It is about extracting tuple values back into distinct variables 


```python
# standard unpacking 
point = (10, 20)
x, y = point 

# unpacking with the asterisk (*) operator to collect remaining items
scores = (98, 85, 76, 64, 50)
highest, second, *rest = scores

print (highest)
print (second)
print (rest)
```

```python
tup = (1, 2, 3, 4, 5)
a, *b, c = tup 
print (a)
print (b) 
print (c) 
```

### Tuple Built-ins Cont'd ... 

Built-in Function	Description
all()	Returns true if all element are true or if tuple is empty
any()	return true if any element of the tuple is true. if tuple is empty, return false
len()	Returns length of the tuple or size of the tuple
enumerate()	Returns enumerate object of tuple
max()	return maximum element of given tuple
min()	return minimum element of given tuple
sum()	Sums up the numbers in the tuple
sorted()	input elements in the tuple and return a new sorted list
tuple()	Convert an iterable to a tuple.
