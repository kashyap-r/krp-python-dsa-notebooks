# Core Dictionary architecutre and hashing

At a structural level, a Python dictionary implements a hash table. This architectural choice is what gives the dictionary its speed and versatility. When you define a dictionary, you are essentially creating a sparse array, often called a bucket array.

When you insert a key-value pair, Python passes the key through a hash function. This function calculates a unique integer (the hash) that determines the specific index in the bucket array where the value will be stored.

Because of this design:

    Keys must be hashable, which usually means they must be of an immutable type (e.g., str, int, tuple)

    Values can be mutable, including lists, other dictionaries, or custom objects

    Lookups, insertions, and deletions run in average amortized O(1) time

This structure is important for performance. While searching for an item in a list requires iterating through elements one by one, an O(n) operation, retrieving a value from a dictionary is an O(1) operation on average.

### Dictionary Comprehension

- important for 
    * unverting a dictionary 
    * filtering out null values from a dataset


```python
# classic usecase 
num = [1, 2, 3, 4, 5, 6]
squares = {x: x**2 for x in num}

print (squares)
print() 

# filtering data during creation 
raw_data = {"a": 10, "b": 20, "c": 30, "d": 40, "e": None}

clean_data = {k: v for k, v in raw_data.items() if v is not None}
print (clean_data)
```

```python
# dict.fromkeys() 
# creates a new dictionary with specified keys and a single value

categories = ["electronics", "clothing", "food", "books"]
inventory = dict.fromkeys(categories, 0)
print (inventory)

# Initialize with None for OPtional fields 
user_fields = ["email", "phone", "address", "company"]
user_profile = dict.fromkeys(user_fields)
print (user_profile)
```

```python
categories = ["A", "B", "C"]
wrong_way = dict.fromkeys(categories, [])
print (wrong_way)
wrong_way["A"].append(1)
print (wrong_way)

"""
Wrong Way

* dict.fromkeys(categories, []) creates a dictionary where each key in categories shares the same list object as the value.

* When you modify one list (e.g., wrong_way["A"].append(1)), it affects all keys because they all reference the same list.

Output: {'A': [1], 'B': [1], 'C': [1]}
"""
```

```python
right_way = {cat: [] for cat in categories}
print(right_way)

right_way["A"].append(1)
print(right_way)

"""
Right Way:

* {cat: [] for cat in categories} uses dictionary comprehension to create a new, independent list for each key.

* Modifying one list (e.g., right_way["A"].append(1)) does not affect others.

Output: {'A': [1], 'B': [], 'C': []}
"""
```

### Python Dictionary methods dfor Access and Modification 

```python
# Bracket notation 

product = {
    "name": "Macbook Pro M5 Max",
    "price": 7999.99,
    "stock": 2,
    "category": "Electronics"
}

# Direct access with brackets 
print (product["name"])
print (product["price"])

# attempt any non-existent key returns an error 
try:
    print (product["location"])
except KeyError as e:
    print (f"Key not found: {e}")
```

```python
# get() method
# used to retrieve values, returns None instead of an exception if key doesn't exist 

product = {
    "name": "Macbook Pro M5 Max",
    "price": 7999.99,
    "stock": 2,
    "category": "Electronics"
}

manufacturer = product.get("manufacturer")
warranty = product.get("warranty", "No Warranry Information")
print (manufacturer)
print (warranty)

# .get() is especially useful in data pipelines
customer_data = {"name": "John Doe", "email": "john@example.com"}

phone = customer_data.get("phone", "Not provided")
print(phone)

address = customer_data.get("address", "Not provided")
print(address)
```

### .setdefault() method 
Combines retrieval and insertion in a single operation 

It retrieves a value if the key exists, or inserts a default value and returns it if the key is missing, which is perfect for accumulation patterns. 


Page Visits Counter:

    page_visits = {} initializes an empty dictionary to track visits to different pages.

    page_visits.setdefault("home", 0) checks if the key "home" exists. If not, it sets "home" to 0 and returns 0. If it exists, it returns the current value.

    page_visits["home"] += 1 increments the visit count for "home".

    The second call to setdefault("home", 0) returns the updated count without changing it.

```python
# .setdefault() method

page_visits = {} 

# Frst visit to "home" - inserts 0 and returns it 
count = page_visits.setdefault("home", 0)
print (count)
page_visits["home"] += 1 

# subsequent call returns existing value
count = page_visits.setdefault("home", 0)
print (count)
```

Grouping Items:

inventory is a list of tuples, each containing an item and its category.

grouped = {} initializes an empty dictionary to group items by category.

The loop iterates over inventory, using setdefault(category, []) to ensure each category has a list. It appends the item to the appropriate category list.

The final print(grouped) shows items grouped by their categories, resulting in {'fruit': ['apple', 'banana'], 'vegetable': ['carrot', 'broccoli']}.

Note: Remember "key" in a dictionary immutable



```python
# Practical example: grouping items
inventory = [
    ("apple", "fruit"),
    ("carrot", "vegetable"),
    ("banana", "fruit"),
    ("broccoli", "vegetable")
]

grouped = {}
for item, category in inventory:
    grouped.setdefault(category, []).append(item)

print(grouped)
```
