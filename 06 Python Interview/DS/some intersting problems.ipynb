✅ 50 Coding Problems (Structured, Realistic, Varying Difficulty)
🟢 EASY–MEDIUM (1–15) — Foundations but with real-world framing
1. Top K Frequent Transactions

Description:
You are given a stream of financial transactions (user_id, amount). Design a function to return the top K users with the highest number of transactions in a given time window. The system should support both batch input and streaming updates. Consider how you will efficiently maintain counts as new transactions arrive. The solution should scale to millions of users and support frequent queries.
Hints: HashMap + Min Heap, approximate counting for scale

2. Deduplicate Events

Description:
Given a list of events with unique event_ids but possible duplicates due to retries, remove duplicates while preserving order. The system must handle out-of-order arrivals and large-scale ingestion. Consider memory constraints and whether you can use TTL-based deduplication.
Hints: HashSet + sliding window / Bloom filter

3. Moving Average of Payments

Description:
Design a function that calculates the moving average of transaction amounts over the last N transactions. The function should support real-time updates and constant-time computation per update. Consider how to efficiently handle sliding windows.
Hints: Queue + running sum

4. Group Transactions by User

Description:
Given a large dataset of transactions, group them by user_id and return aggregated statistics (count, total spend, average). Assume the dataset may not fit into memory.
Hints: External sorting / MapReduce-style grouping

5. First Unique Transaction

Description:
Given a stream of transactions, return the first transaction that has not been repeated. The system should support continuous ingestion and queries.
Hints: HashMap + Queue

6. Validate Balanced Logs

Description:
Given a sequence of log entries representing system start/stop events, validate whether all processes are correctly opened and closed.
Hints: Stack

7. Longest Increasing Spend Streak

Description:
Given daily spend values, find the longest strictly increasing sequence. Handle missing days and irregular timestamps.
Hints: DP / sliding window

8. Merge K Sorted Transaction Streams

Description:
Merge multiple sorted transaction streams into one sorted stream by timestamp. Each stream represents a shard.
Hints: Min Heap

9. Detect Duplicate Within K Distance

Description:
Check if any transaction ID appears twice within a window of size K.
Hints: Sliding window + HashSet

10. Reverse Log Blocks

Description:
Reverse log entries in chunks of size K to simulate batch reprocessing.
Hints: Linked list / array chunking

11. Two Sum on Transaction Amounts

Description:
Find if two transactions sum to a target value. Extend to streaming case.
Hints: HashMap

12. Find Missing Transaction IDs

Description:
Transaction IDs are sequential but some are missing. Find missing IDs efficiently.
Hints: XOR / HashSet

13. Rate Limiter (Basic)

Description:
Implement a rate limiter that allows N requests per minute per user.
Hints: Sliding window / token bucket

14. Intersection of Two User Lists

Description:
Find common users between two datasets efficiently.
Hints: HashSet

15. Sort by Frequency

Description:
Sort transactions based on frequency of occurrence.
Hints: Heap / bucket sort
