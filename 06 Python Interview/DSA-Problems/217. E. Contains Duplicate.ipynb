217. Contains Duplicate 

Given an interger array nums, return True if any value appears twice in the array, and return false if every element is distinct.

Example 1:
nums = [1, 2, 3, 1]

Output = True 

Input nums = [1, 2, 3, 4]
Output False 


```python
# latest Submission


def containsDuplicate(nums):    
    return len(nums) != len(set(nums))

nums = [1, 2, 3, 4]
containsDuplicate(nums)
```

```python
## Old Submission 

from typing import List
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        seen = set()
        for num in nums:
            if num in seen:
                return True
            seen.add(num)
        return False
    
```
