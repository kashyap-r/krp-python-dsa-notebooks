### Below is the Optimal Solution 

```python
def countPairs(deliciousness):
    MOD = 10**9 + 7
    max_val = max(deliciousness)
    max_sum = max_val * 2
    
    # Precompute powers of 2 up to max_sum
    powers = []
    p = 1
    while p <= max_sum:
        powers.append(p)
        p *= 2
    
    freq = {}
    result = 0
    
    for x in deliciousness:
        for p in powers:
            complement = p - x
            if complement in freq:
                result = (result + freq[complement]) % MOD
        freq[x] = freq.get(x, 0) + 1
    
    return result


print(countPairs([1,3,5,7,9]))  # Output: 2
```
