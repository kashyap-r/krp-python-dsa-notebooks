### String Manipulation


```python
s1 = "Learning Python"

s2 = "Learning TypeScript"

print (s1) 
print (s2)

print (s1[10])
print(s2[-1])
```

```python
# String slicing 
s1 = "Learning Python"
print (s1[1:4])
print (s1[2:])
print (s1[:4])
# print (s1[::-1])

s = s1 + 'T'
print (s)
```

```python
# String Methods 

s1 = "  Learning Python.  "

print (f"The String: {s1}")
print (f"Length of the string: {len(s1)}")
print (f"String in Upper case: {s1.upper()}")
print (f"string in Lowercase: {s1.lower()}")
print (f"{s1.strip()}")
s1 = s1.replace("Python", "TypeScript")
print (s1)
```
