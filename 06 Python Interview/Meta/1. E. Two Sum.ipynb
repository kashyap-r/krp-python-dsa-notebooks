###  1. Two Sum

Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]

Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]


Constraints: 
* 2 <= nums.length <= 10^4  (Array length constraint)    
    - array to have atleast 2 elements and maximum of 10000 elements  
* -10^9 <= nums[i] <= 10^9 (Value range COnstraint)
    - each number in the array can be very large and both positive or negative
    - This means you can’t rely on small integer tricks (like using array indices directly as hash keys without a proper map). 
    You need a data structure like a hash map/dictionary to handle the wide range.
* -10^9 <= target <= 10^9 (Target Constraint)
    - The target sum can also be very large or very small, so your algorithm must handle extreme values without overflow.
    - In Python, integers are unbounded, but in languages like C++/Java, you need to be mindful of integer limits.
* Valid Answer Constraint
    - Only one valid answer exists.
    - This guarantees uniqueness: you don’t have to worry about multiple pairs summing to the target.
    - It simplifies the problem because once you find a valid pair, you can return immediately.


```python
# Approach 1 - Naive Brute Force Method

nums = [2,7,11,15]
target = 9

def check2Sum(arr, target):
    n = len(arr)
    for i in range(n-1):                
        for j in range(i+1, n):
            if arr[i] + arr[j] == target:
                return True
    return False

check2Sum(nums, target)

# Time O(n^2)
# Space O(1) 
```

#### Approach 2 - Using Sorting and Binary Search - Time O(n * log n) and O(1) space

To check if a pair with a given sum exists in the array, we first sort the array. 
Then for each element, we compute the required complement (i.e. target - arr[i]) and perform binary search on the remaining sub-array (from index i+1 to the end) to find that complement

1. Sort the array in ascending order (non-decreasing order) 
2. Loop through each element arr[i] from index 0 to n-2 
3. for each arr[i], calculate complement = target - arr[i]
4. perform binary search for the complement in the subarray from index i+1 to n-1 
5. If the complement is found, return True 
6. is the llop completes without finding any pair, return False.  


```python
# Approach #2 

def binarySearch(arr, left, right, target):
    # left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2 
        if arr[mid] == target:
            return mid 
        elif arr[mid] < target:
            left = mid + 1 
        else:
            right = mid - 1
    return -1

def twoSum(arr, target):
    arr.sort() 
    for i in range(len(arr)):
        complement = target - arr[i]

        # use binary Search to find the complement
        if binarySearch(arr, i+1, len(arr)-1, complement):
            return (i+1, len(arr)-1)
        
    # If no pair is found
    return False

if __name__ == "__main__":
    arr = [0, -1, 2, -3, 1]
    target = -2
    result = []
    result = twoSum(arr, target)
    print (result)
    if result:
        print("true")
    else:
        print("false")

```

```python
## Another way of doing the binary search method 

def twoSum2Pointers(nums, target):
    # pair each number with its index 
    indexed_nums = list (enumerate(nums))
    # sort by value
    indexed_nums.sort(key=lambda x: x[1])

    left, right = 0, len(nums) - 1 
    while left < right:
        total = indexed_nums[left][1] + indexed_nums[right][1]
        if total == target:
            return [indexed_nums[left][0], indexed_nums[right][0]]
        elif total < target:
            left += 1 
        else:
            right -= 1 
    return []

if __name__ == "__main__":
    arr = [0, -1, 2, -3, 1]
    target = -2
    result = []
    result = twoSum2Pointers(arr, target)
    print (result)

"""
Time Complexity: 𝑂(𝑛log⁡𝑛) (due to sorting)

Space Complexity: 𝑂(𝑛) (for index tracking)
"""
```

#### [Better Approach 2] Sorting and Two-Pointer Technique - O(n × log(n)) time and O(1) space

    The idea is to use the two-pointer technique but for using the two-pointer technique, the array must be sorted. Once the array is sorted then we can use this approach by keeping one pointer at the beginning (left) and another at the end (right) of the array. 

Check the sum of the elements at these two pointers:

    If the sum equals the target, we’ve found the pair.
    If the sum is less than the target, move the left pointer to the right to increase the sum.
    If the sum is greater than the target, move the right pointer to the left to decrease the sum.



```python
def twoSum(arr, target):
    arr.sort
    left, right = 0, len(arr)-1

    # iterate while the left pointer is < the right 
    while left < right:
        sum = arr[left] + arr[right]

        if sum == target:
            return True 
        elif sum < target:
            left += 1 
        else:
            right -= 1 
    return False

if __name__ == "__main__":
    arr = [0, -1, 2, -3, 1]
    target = -2
    result = []
    if twoSum(arr, target):
        print ("true")
    else:
        print ("false")
```

```python
def twoSum(arr, target):
    arr.sort()
    left, right = 0, len(arr) - 1

    # Iterate while left pointer is less than right
    while left < right:
        sum = arr[left] + arr[right]

        # Check if the sum matches the target
        if sum == target:
            return True
        elif sum < target:
            
            # Move left pointer to the right
            left += 1
        else:
            
            # Move right pointer to the left
            right -= 1

    # If no pair is found
    return False

if __name__ == "__main__":
    arr = [0, -1, 2, -3, 1]
    target = -2
    if twoSum(arr, target):
        print("true")
    else:
        print("false")
```

```python
## Using Hashset : O(n) Time and O(n) space 


def TwoSum(arr, target):
    hashSet = set()
    for i in range(len(arr)):
        complement = target - arr[i]

        if not complement in hashSet:
            hashSet.add(arr[i])
        else:
            return True 
    
    return False

if __name__ == "__main__":
    arr = [0, -1, 2, -3, 1]
    target = -2
    if TwoSum(arr, target):
        print("true")
    else:
        print("false")

    
```
