### Prefix Sum 

#### Core Idea 

A Prefix Sum array stores cumulative sums so that range queries can be answered in O(1) time.

Given:
arr = [a0, a1, a2, a3, ... , an-1]

We build:
prefix[i] = a0 + a1 + a2 + ... + ai

So:
prefix[i] = prefix[i-1] + arr[i]

Why is this Powerful?

Without prefix sum:
    Range sum query → O(n)

With prefix sum:
    Range sum query → O(1)

sum(L, R) = prefix[R] - prefix[L-1]

This is a preprocessing for fast queries pattern.

### Basic Example 

arr = [2, 4, 1, 6, 3]

The idea is to create an array prefixSum[] of size n, and for each index i in range 1 
to n - 1, set prefixSum[i] = prefixSum[i - 1] + arr[i].

Build Prefix: 

prefix[0] = 2
prefix[1] = 2 + 4 = 6
prefix[2] = 6 + 1 = 7 
prefix[3] = 7 + 6 = 13
prefix[4] = 13 + 3 = 16 

Query Example:
Sum from index 1 to 3 :
= prefix[3] - prefix[0]
= 13 - 2 
= 11 

Check from the array:
= 4 + 1 + 6
= 11 


### When should you use Prefix Sum?

✔️ Multiple range queries
    - Sum of subarrays
    - Count of elements in range
    - Frequency in intervals

✔️ Subarray sum problems
    - Count subarrays = K
    - Longest subarray with sum = K
    - Subarray divisible by K

✔️ 2D grid problems
    - Matrix sum queries
    - Image smoothing
    - Heatmap aggregation

✔️ Difference array (range updates)

✔️ Probability & ML use cases
    - Cumulative distribution
    - Histogram prefixing
    - Rolling window metrics
    - Feature aggregation by time bucket

 At Meta E6:
Expect it in:
    - Stream aggregation
    - Large-scale log analysis
    - Time-series window analytics
    - Feature engineering in ML pipelines

### Worked Example #1.E. Range Sum Query  

Problem:
    Given an array and multiple queries (L, R), return Sum.

Context (Meta like)    
    Log volume per minute. Query total logs between time windows.

Approach: 
    1. Build Prefix array 
    2. For each query:
        sum = prefix[R]  - prefix[L-1]

   

```python
class NumArray:
    def __init__ (self, nums):
        self.prefix = [0] * len(nums)
        self.prefix[0] = nums[0]

        for i in range(1, len(nums)):
            self.prefix[i] = self.prefix[i-1] + nums[i]

    def sumRange(self, left, right):
        if left == 0:
            return self.prefix[right]
        return self.prefix[right] - self.prefix[left-1]

if __name__ == "__main__":
    arr = [2, 4, 1, 6, 3]
    l = 1
    r =  3

    sr = NumArray(arr)
    print (sr.sumRange(l, r))
    print (sr.sumRange(2, 4))
    # print (sr.sumRange(3, 5))
```
