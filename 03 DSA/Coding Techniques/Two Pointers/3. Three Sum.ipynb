### Better Approach - Using Hash Set - O(n^2) Time and O(n) space

Traverse every element arr[i] in a for loop. For every element in the array, use the hashing based solution of 2 sum problem to check if there is a pair with equal sum to given sum - arr[i]

Steps: 
1. Iterate through the array, fixing the first element (arr[i]) for the triplet 
2. For each arr[i], use a Hash Set to store potential second elements and run another 
loop inside it for j from i+1 to n-1.
3. Inside a nested loop, check if given sum - arr[i] - arr[j] is present in the hash set. 
If yes, then print the triplet.
4. If no triplet is found in the entire array, the function returns false.


```python
def findTriplets(arr, target):
    n = len(arr)

    # fix the first element as arr[i]
    for i in range(n-2):
        # hash set to store potential second elements 
        st =set()

        # fix the third element as arr[j]
        for j in range(i+1, n):
            second = target - arr[i] - arr[j]
            if second in st:
                return (second, arr[i], arr[j])
            st.add(arr[j])
    return False

if __name__ == "__main__":
    arr = [1, 4, 45, 6, 10, 8]
    target = 16
    print (findTriplets(arr, target))
    # if findTriplets(arr, target):
    #     print("true")
    # else:
    #     print("false")
```

### Expected Approach
#### Approach #3 - Sorting and Two POinter method 

We first sort the array. After sorting, we traverse every element arr[i] in a loop. For every arr[i], use the Two Pointer Technique based solution of 2 Sum Problem to check if there is a pair with sum equal to given sum - arr[i].

```python
def findTripletSums(arr, target):
    n = len(arr)
    arr.sort()

    for i in range(n-2):
        l = i+1 
        r = n-1 

        reqSum = target - arr[i]

        while l < r:
            if arr[l] + arr[r] == reqSum:
                return (arr[i], arr[l], arr[r]) 
            if arr[l] + arr[r] < reqSum:
                l += 1
            else:
                r -= 1
    return False


if __name__ == "__main__":
    arr = [1, 4, 45, 6, 10, 8]
    target = 17
    print (findTripletSums(arr, target))
    # if findTriplets(arr, target):
    #     print("true")
    # else:
    #     print("false")
```
