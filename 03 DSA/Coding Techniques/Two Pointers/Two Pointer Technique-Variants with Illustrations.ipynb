### Pattern 1: Two Sum 

Finding two numbers in a sorted array that sum to a target.

Example: 

arr = [1, 2, 3, 6, 10]
target = 8

Key Invariant
    Everything left of left pointer is too small.
    Everything right of right pointer is too big.


```python
def twoSum(arr, target):
    left, right = 0, len(arr) - 1

    while left < right:
        curr_sum = arr[left] + arr[right]
        if curr_sum == target:
            return (arr[left], arr[right])
        elif curr_sum < target:
            left+= 1
        else:
            right -= 1 
    return None  

arr = [1, 2, 3, 4, 5, 6]
target = 5
print (twoSum(arr,target))
```
