Two Sum  - Pair Sum in a Sorted array 

Given a 1-based indexed integer array arr[] that is sorted in non-decreasing order, along with an integer target. find two elements in the array such that their sum is equal to target. 

If such a pair exists, return the indices of the two elements in increasing order. If no such pair exists, return [-1, -1].

Example: 
arr = [2, 7, 11, 15]
target = 9


```python
# naive approach

def twoSumNaive(arr):
    n = len(arr) 
    for i in range(n-1):    
        for j in range(i+1, n):
            if (arr[i] + arr[j] == target):
                print (f"{arr[i]}+{arr[j]} = {target} same as target!")
                
arr = [2, 7, 11, 15]
target = 9

twoSumNaive(arr)
```

In the above code Time us O(n^2)
Space O(1)

```python
# Two pointer approach

def twoSum2P(arr):
    left, right = 0, len(arr)-1

    while (left < right):

        sum = arr[left] + arr[right]
        if sum == target:
            return (arr[left], arr[right])
        elif sum < target: 
            left += 1
        else:
            right -= 1
    return None

arr = [2, 7, 11, 15]
target = 9

print (f"Sum of ({twoSum2P(arr)}) == {target} !!")
```

Time O(n)
Space O(1)
