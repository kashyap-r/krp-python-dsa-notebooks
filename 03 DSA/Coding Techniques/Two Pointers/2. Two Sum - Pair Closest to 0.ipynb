Two Sum - Pair Closest to 0

Given an integer array arr[], find the sum of any two elements whose sum is closest to zero.

Note: In case if we have two ways to form sum closest to zero, return the maximum sum among them.

### Write a basic algorithm (Naive Approach)

Check the sum of every possible pair of elements in the array and leets track of the pair with the minimum absolute sum. 

- loop through the array i = 1 to n-1 
- loop through the array j = i+1 to n
- calculate sum = arr[i] + a[j]
- 

```python
def minSumPair(arr):
    # first initialize the result with sum of first two elements
    res = arr[0] + arr[1]
    # print (f"1. initial sum={res}")
    n = len(arr)

    for i in range(n-1):
        # print (f"ith iteration={i}")
        for j in range(i+1, n):
            sum = arr[i] + arr[j]
            # print (f"jth iteration={j}")
            # print (f"sum of {arr[i]} + {arr[j]} = {sum} ")
            if abs(sum) < abs(res):
                res = sum 
                # print (f" sum < res (inital value): res = {res}")                
            elif abs(sum) == abs(res):
                res = max(res, sum)
                # print(f"Now res = max(res, sum) = {res} ")
    return res
            
if __name__ == "__main__":
    arr = [0, -8, -6, -3]
    print (minSumPair(arr))
```

### Approach #2: Using Sorting + Binary Search (Better approach)

O(n log n) Time
O(1) space

The idea is to find the pair whose sum is closest to zero by sorting the array and using binary search for efficient look up. 

For each element, we search for its closest complement using binary search, updating the clsoest sum found so far. 

If an exact zero sum is found, we return immediately. 

Steps:
1. Sort the array to enable binary search 
2. Iterate through each element and fix it as the first element  of the pair 
3. Use Binary Search to find the closest second element .
4. Update the closest sum if the current pair gives a smaller absolute sum 
5. return 0 if an exact zero sum pair is found 
6. return the closest sum after checking all the pairs  



```python
def minAbsSumPair(arr):
    arr.sort()
    # [-8, -6, 0, 3]
    n = len(arr)
    # initialise a floating point variable of lowest possible value 
    # cant use 0 here as it will result in checkig only for sum == 0
    # and will defeat the purpose and we looking for sum closest to 0 including 0 
    res = float('-inf')

    for i in range(n):
        # initialise the first element of the array as one of the pair
        # and find the other element using binary search 
        init_val = arr[i]        
        left, right = i+1, n-1
        while left <= right:
            mid = (left+right)//2
            
            # next sum is sum of mid point of the array and initial arr value  
            curr = arr[mid] + init_val
            # if exact pair is found 
            if curr == 0:
                return 0
            # update res if the current sum is closer
            if abs(curr) < abs(res):
                res = curr
            # you need a tie breaker when curr has the same value on either side 
            # of the zero like +2 and -2. I'm randomly choosing +2 
            # to test this condition use input as [5, -2, 2, 6]
            elif abs(curr) == abs(res):
                res = max(res, curr) 
            # if the current is smaller than 0, go to the left side else to 
            # the right side of the array
            if curr < 0:
                left = mid + 1
            else:
                right = mid - 1
    return res

if __name__ == "__main__":
    # arr = [0, -8, -6, 3]
    arr = [5, -2, 2, 6]
    print (minAbsSumPair(arr))
```

### Approach #3 - Using Sorting + Two Pointer method 
O(n. log(n)) Time 
O(1) Space 

The idea is to sort the array first and use two pointers, one at the leftmost (smallest) and the other at the rightmost (largest) element. We calculate their sum and check if it's closer to zero than our current best sum. If we find a sum of zero, we return immediately. Otherwise, we adjust the pointers based on whether the sum is positive (move right pointer left) or negative (move left pointer right).

Steps:
- Sort the array in ascendin order
- Initialize two pointers: one at the start and one at the end
- Initialize variables to store the closest sum and its absolute difference 
- Iterate while the left pointer is less than right pointer 
- update the closest sum if the current sum is closer to zero 
- move the left pointer right if the sum is negaive; else move the right pointer left


```python
# Python Code for Two Sum - Closest Pair
# Using Two Pointer
def minAbsSumPair(arr):
  
    # Sorting the array in ascending order
    arr.sort()

    i, j = 0, len(arr) - 1

    # Initializing sum with the first and last elements
    sum_val = arr[i] + arr[j]

    # Initializing the result with the absolute value
    # of the initial sum
    diff = abs(sum_val)

    while i < j:
      
        # If we have zero sum, there's no result
        # better. Hence, we return 0.
        if arr[i] + arr[j] == 0:
            return 0

        # If we get a better result, we update the
        # difference
        if abs(arr[i] + arr[j]) < abs(diff):
            diff = arr[i] + arr[j]
            sum_val = arr[i] + arr[j]
        elif abs(arr[i] + arr[j]) == abs(diff):
          
            # If there are multiple pairs with the same
            # minimum absolute difference, return the
            # pair with the larger sum
            sum_val = max(sum_val, arr[i] + arr[j])

        # If the current sum is greater than zero, we
        # search for a smaller sum
        if arr[i] + arr[j] > 0:
            j -= 1
        # Else, we search for a larger sum
        else:
            i += 1

    return sum_val


# Driver Code
arr = [0, -8, -6, 3]
print(minAbsSumPair(arr))
```
