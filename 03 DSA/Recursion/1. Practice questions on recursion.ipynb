```python
# Question 8

def fun(a, n):
    if(n == 1):
        return a[0]
    else:
        x = fun(a, n - 1)
    if(x > a[n - 1]):
        return x
    else:
        return a[n - 1]

# Driver code
arr = [12, 10, 30, 50, 100]
print(fun(arr, 5))
```

This function finds the maximum element in the array.

Step‑by‑Step Execution
Call 1: fun(arr, 5)
- n=5, not 1 → go into x = fun(arr, 4)
Call 2: fun(arr, 4)
- n=4, not 1 → go into x = fun(arr, 3)
Call 3: fun(arr, 3)
- n=3, not 1 → go into x = fun(arr, 2)
Call 4: fun(arr, 2)
- n=2, not 1 → go into x = fun(arr, 1)
Call 5: fun(arr, 1)
- n=1 → base case → return a[0] = 12

Now the recursion unwinds
- Back to Call 4 (n=2)
    - x = 12
    - Compare x (12) vs a[1] (10) → return 12
- Back to Call 3 (n=3)
    - x = 12
    - Compare x (12) vs a[2] (30) → return 30
- Back to Call 2 (n=4)
    - x = 30
    - Compare x (30) vs a[3] (50) → return 50
- Back to Call 1 (n=5)
    - x = 50
    - Compare x (50) vs a[4] (100) → return 100

Key Insight
- The line x = fun(a, n-1) is executed once per recursive call until the base case (n=1).
- The if (x > a[n-1]) block executes every time the recursion unwinds, i.e., at each level after the recursive call returns.
- So:
    - Recursive descent: only the x = fun(a, n-1) line runs.
    - Recursive ascent (unwinding): the if (x > a[n-1]) comparison runs at each level.

✅ Final Result: The function returns 100, the maximum element.

Time complexity: O(n)

Auxiliary Space: O(n)

```python
def fun(a, b):
    if (b == 0):
        return 0 
    if (b % 2 == 0):
        return fun(a + a, b//2) 
    
    return fun(a + a, b//2) + a 

# Driver code

print(fun(4, 3))
```

It calculates a*b (a multiplied b).

Time Complexity: O(log2(b))
Auxiliary Space:  O(log2(b))

```python
def fun(count):
    print(count)
    if(count < 3):
        count+=1
        fun(fun(fun(count))) 
    
    return count 
 

if __name__=="__main__":  
    
    fun(1)
```

The main() function calls fun(1). fun(1) prints "1" and calls fun(fun(fun(2))). fun(2) prints "2" and calls fun(fun(fun(3))). So the function call sequence becomes fun(fun(fun(fun(fun(3))))). fun(3) prints "3" and returns 3 (note that the count is not incremented and no more functions are called as if the condition is not true for count 3). So the function call sequence reduces to fun(fun(fun(fun(3)))). fun(3) again prints "3" and returns 3. So the function call again reduces to fun(fun(fun(3))) which again prints "3" and reduces it to fun(fun(3)). This continues and we get "3" printed 5 times on the screen

```python
fp = 15


def fun(n):
    global fp
    if (n <= 2):
        fp = 1
        return 1

    t = fun(n - 1)
    f = t + fp
    fp = t
    return f

# Driver code


print(fun(5))
```

The program calculates n-th Fibonacci Number. The statement t = fun ( n-1, fp ) gives the (n-1)th Fibonacci number and *fp is used to store the (n-2)th Fibonacci Number. The initial value of *fp (which is 15 in the above program) doesn't matter. The following recursion tree shows all steps from 1 to 10, for the execution of fun(5, &x).
