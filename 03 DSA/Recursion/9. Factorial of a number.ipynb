```python
# Iterative solution 

def factorial(n):
    fact = 1
    i = 2 
    while i <= n :
        fact *= i
        i += 1
    return fact

factorial(5)
```

```python
# using recursion 
def factorial(n):
    if n == 0:
        return 1
    return n * factorial (n-1)

factorial(5)
```

```python
# using for loop 
def factorial(n):
    fact = 1 
    if n == 0:
        return fact     
    for i in range(n, 0, -1):
        fact *= i
    return fact

factorial(5)
```
