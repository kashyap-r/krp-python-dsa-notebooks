### Climbing staairs to reach the top 

Given n stairs, and a person standing at the ground wants to climb stairs to reach the top. 

The person can climb wither 1 stair or 2 stairs at a time, count the number of ways the person can reach the top. 

Example: 

Input: 1 
Output: There is only onw way to climb the stairs 

Input: n = 4 
Output: 5 
Explanation: There are five ways to reach 4th stair: {1, 1, 1, 1}, {1, 1, 2}, {2, 1, 1}, {1, 2, 1} and {2, 2}. 



### Naive Approach: Using recursion
- O(2^n) Time and O(n) space

In this approach, we observe that to reach the nth stair, one can come either from the (n−1)th or (n−2)th stair. Similarly, to reach the (n−1)th stair, we must know the ways to reach the (n−2)th and (n−3)rd stairs, and so on. This way we can observe that this can be done using recursion.



```python
def countways(n):
    # base case 
    if n == 0 or n == 1:
        return 1 
    return countways(n-1) + countways(n-2)



```
