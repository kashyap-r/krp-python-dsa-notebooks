### HashMap, HashSet and Hashtable

In Python, the concepts of HashMap, HashSet, and HashTable are implemented using built-in data structures.

1. **HashMap in Python → dict**
    
    - A HashMap stores key-value pairs with unique keys.

    - Python’s built-in dict is a hash map.

✅ Features: Fast lookup, insertion, deletion (average O(1)).


```python
# HashMap example using dict 

hash_map = {
    "name": "Kash",
    "age": 25,
    "city": "Bengaluru"
}

# Adding a new key-value pair
hash_map["country"] = "India"

# Accessing values safely
print("Name:", hash_map.get("name", "Not Found"))

# Updating a value
hash_map["age"] = 26

# Removing a key
removed_value = hash_map.pop("city", None)

# Iterating over key-value pairs
for key, value in hash_map.items():
    print(f"{key}: {value}")

# hash_map
```

2. **HashSet in Python → set**

    - A HashSet stores unique, unordered elements.
    
    -Python’s built-in set is a hash set.

Example:

```python
hash_set = {1, 2, 3, 4}

hash_set.add(5)
hash_set.add(3)
print(hash_set)

hash_set.discard(2)
print(hash_set)

print (4 in hash_set)

for item in hash_set:
    print (item)
```

### HashTable in Python  → dict or collections.OrderedDict

A HashTable is conceptually the same as a HashMap — key-value storage using hashing.

In Python, dict is the default hash table, but for thread-safe operations, use collections.defaultdict or threading.Lock.


```python
import threading

hash_table = {}
lock = threading.Lock()

def safe_insert(key, value):
    with lock:
        hash_table[key] = value

# Example usage
safe_insert("id", 101)
safe_insert("status", "active")

print(hash_table)
```

✅ Features: Same as HashMap, but can be made thread-safe with locks.

HashMap and HashTable in Python - Another perspective 

In Python, HashMap and HashTable are data structures used to store key-value pairs. While Python does not explicitly have a HashMap or HashTable class, these concepts are implemented using the built-in dict data type. Here's a breakdown of their functionality and differences:

#### HashMap in Python

A HashMap is a data structure that maps keys to values using a hash function. In Python, dictionaries (dict) serve as HashMaps. They are unsynchronized, meaning they are not thread-safe but are faster in single-threaded environments.

Example of a HashMap:


```python
# Creating a dictionary (HashMap)
hash_map = {'Alice': 25, 'Bob': 30, 'Charlie': 35}
# Accessing a value
print(hash_map['Alice']) # Output: 25
# Adding a new key-value pair
hash_map['David'] = 40
# Updating a value
hash_map['Alice'] = 26
print(hash_map)

# Deleting a key-value pair
del hash_map['Bob']
print(hash_map) # Output: {'Alice': 26, 'Charlie': 35, 'David': 40}
```

Key Features:

    Non-synchronized: Not thread-safe.

    Allows null values: Keys must be hashable, but values can be None.

    Fast operations: Average time complexity for search, insert, and delete is O(1).
    

#### HashTable in Python

A HashTable is similar to a HashMap but is synchronized, meaning it is thread-safe. Python does not have a built-in HashTable class, but you can use collections.defaultdict or implement one manually. Synchronization can also be achieved using threading locks.

Example of a HashTable:


```python
# Implementing a basic HashTable
class HashTable:
   def __init__(self, size):
       self.size = size
       self.table = [[] for _ in range(size)]
   def _hash(self, key):
       return hash(key) % self.size
   def set(self, key, value):
       index = self._hash(key)
       for pair in self.table[index]:
           if pair[0] == key:
               pair[1] = value
               return
       self.table[index].append([key, value])
   def get(self, key):
       index = self._hash(key)
       for pair in self.table[index]:
           if pair[0] == key:
               return pair[1]
       return None
   def delete(self, key):
       index = self._hash(key)
       self.table[index] = [pair for pair in self.table[index] if pair[0] != key]
# Example usage
hash_table = HashTable(10)
hash_table.set('Alice', 25)
hash_table.set('Bob', 30)
print(hash_table.get('Alice')) # Output: 25
hash_table.delete('Alice')
print(hash_table.get('Alice')) # Output: None
```

Key Features:

    Synchronized: Thread-safe but slower due to locking mechanisms.

    No null keys: Keys and values cannot be None.

    Collision handling: Uses techniques like chaining or open addressing.

Differences Between HashMap and HashTable

    Thread Safety: HashMap is not synchronized, while HashTable is synchronized.

    Performance: HashMap is faster due to the lack of synchronization overhead.

    Null Handling: HashMap allows null values, while HashTable does not.

Applications

    HashMap is ideal for single-threaded applications where performance is critical.

    HashTable is suitable for multi-threaded environments where thread safety is required.

Both structures are widely used in scenarios like caching, indexing, and implementing associative arrays. Python's dict is a powerful and flexible implementation of these concepts.
