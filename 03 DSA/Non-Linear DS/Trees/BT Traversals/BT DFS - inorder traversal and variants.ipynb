Inorder Tree Traversal 

Approach #1: Using Recursion 

                - O(n) Time and O(h) Space 

Approach #2: Using Stack [Naive Approach] 
                
                - O(n) Time and O(h) Space 

Approach #3: Using Morris Traversal Algorithm [Expected Approach]
                
                - O(n) Time and O(1) Space 

Where h is the height of the tree

```python
# Approach #1 

class Node:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None 

def inOrderUsingRec(root):
    result = []
    if not root:
        return 
    inOrderUsingRec(root.left)
    print (root.data, end=' ')
    inOrderUsingRec(root.right)
    return result

def inOrderUsingStack(root):
    result  = []
    stack = []
    curr = root 

    while curr is not None or len(stack) > 0:
        # reach the left most node of the current node
        while curr is not None:
            # place pointer to a tree node on the stack 
            # before traversing the node's left subtree
            stack.append(curr)
            curr = curr.left 

        # current must be none at this point 
        curr = stack.pop()
        result.append(curr.data)

        # we've visited the node and its left subtree. 
        # now its the right subtree's turn 
        curr = curr.right 

    return result

    

def printTree(v):
    print(" ".join(map(str, v)))

if __name__ == "__main__":
    
    # Constructed binary tree is
    #          1
    #        /   \
    #      2      3
    #    /  \
    #  4     5
    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)

    res = inOrderUsingStack(root)
    printTree(res)

    res = inOrderUsingRec(root)
    printTree(res)
```

### Approach #3 - MOrris Traversal for Inorder

Approach: We want to traverse the tree and come back to the root after finishing the  left subtree, but without a stack. Morries travel achieves thisby temporarily modifying the tree 

    - For each node check if it has a left child 
    - If it does not have a left child, visit it and move to the right child 
    - If it has a left child, find the inorder predecessor (righmost node in the left subtree)
      Make the current node as the right child of its inorder predecessor (temporary link) 
      Move to the left child 
    - When you encounter a temporary link again, it means the left subtree is fully visited:
        Remove the temporary link 
        Visit the current node
        Move to the right child

Time O(n)

Space O(1)


```python
class Node: 
    def __init__(self, x):
        self.data = x
        self.left = None 
        self.right = None 

def inOrder(root):
    result = []
    curr = root 

    while curr is not None:
        if curr.left is None:
            # if no left child, visit this node and go right 
            result.append(curr.data)
            curr = curr.right
        else:
            # Find the inorder predecessor of curr 
            prev = curr.left 
            while prev.right is not None and prev.right != curr:
                prev = prev.right 
            # Make current the right child of inorder predecessor
            if prev.right is None:
                prev.right = curr
                curr = curr.left 
            else:
                # revert the chsnges madei n the tree structure 
                prev.right = None 
                result.append(curr.data)
                curr = curr.right 
    return result 


if __name__ == "__main__":
  
    # Representation of input binary tree:
    #           1
    #          / \
    #         2   3
    #        / \  
    #       4   5
    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)

    res = inOrder(root)
    
    for data in res:
        print(data, end=" ")
```
