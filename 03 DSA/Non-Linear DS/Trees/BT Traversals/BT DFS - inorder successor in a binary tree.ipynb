### Inorder successor in Binary Tree

Given a bonary tree and a node, the task is to find the inorder sucecssor of this node.

Inorder Successor of a node in the binary tree is the next node in the Inorder traversal of the Binary Tree. Inorder Successor is NULL for the last node in Inorder traversal.
