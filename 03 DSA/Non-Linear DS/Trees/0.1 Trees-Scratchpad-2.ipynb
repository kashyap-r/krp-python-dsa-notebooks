```python
# Binary Tree
from collections import deque 

class Node:
    def __init__(self, x):
        self.key = x 
        self.left = None 
        self.right = None 

class BinaryTree:
    def __init__(self):
        self.root = None 
    

    def insert(self, key):
        if not self.root:
            self.root = Node(key)
            return 
        
        q = deque()
        q.append(self.root)

        while q:

            curr = q.popleft()

            if curr.left is not None:
                q.append(curr.left)
            else:
                curr.left = Node(key)
                


        
```
