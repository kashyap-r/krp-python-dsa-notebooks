### Binary Tree - Level order traversal, Using the height of the tree
By Recursion - another way 
Design choice:

    printlevelOrder() separates the logic: compute height, then print level by level.

    This avoids storing nodes in a list of lists, but directly prints them.

Call order:

    - Yes, printCurrentLevel(root.left, level-1) is fully executed before moving to printCurrentLevel(root.right, level-1).

    - This is depth-first recursion.

    - The level is not reset — it decreases as recursion goes deeper. When recursion unwinds back to the root, the right subtree call begins at the same level.

Time complexity: 𝑂(𝑛2) in worst case (because for each level, you traverse the tree again).

Space complexity: 𝑂(ℎ) recursion stack, where ℎ = height of tree.

```python
class Node:
    def __init__(self, value):
        self.data = value 
        self.left = None 
        self.right = None 

# Recursively computes the height of the tree
def height(node):
    if node is None:
        return 0 
    else:
        lheight = height(node.left)
        rheight = height(node.right)
        if lheight >= rheight:
            return lheight + 1
        else:
            return rheight + 1

# first computes the tree height, theniterates from level 1 to h
# For each level calls printCurrentLevel() to proint nodes at that depth
def printlevelOrder(root):
    # height of the tree 
    h = height(root)

    for i in range(1, h+1):
        printCurrentLevel(root, i)

# This method ensures nodes at the given depth are printed left to right
def printCurrentLevel(root, level):
    if root is None:
        return 0
    if level == 1:
        print (root.data, end=' ')
    elif level > 1:
        printCurrentLevel(root.left, level - 1)
        printCurrentLevel(root.right, level - 1)

if __name__ == "__main__":
    # Representation of the input tree:
    #     12
    #    /  \
    #   8   18
    #  / \
    # 5   11
    root = Node(12)
    root.left = Node(8)
    root.right = Node(18)
    root.left.left = Node(5)
    root.left.right = Node(11)

    print(printlevelOrder(root))
```

### Comparision of Time and Space Complexities for the three recursive approaches to BFS 

1️⃣ Height + Print Current Level (your last program)
Logic: Compute height, then for each level call a recursive DFS to print nodes at that depth.

Time Complexity:
    Height computation = 𝑂(𝑛).
    For each level, you traverse parts of the tree again.
    Worst case (skewed tree): 𝑂(𝑛^2).
    Best case (balanced tree): closer to 𝑂(𝑛.ℎ), but still can degrade.

Space Complexity:
    Recursion stack = 𝑂(ℎ), where ℎ = height of tree.
    No extra data structures beyond recursion.

2️⃣ Recursive Helper with level parameter (your earlier levelOrderRec)
Logic: Pass level down the recursion, directly appending values into res[level].

Time Complexity:
    Each node is visited exactly once.
    𝑂(𝑛).

Space Complexity:
    Recursion stack = 𝑂(ℎ).
    Result list = 𝑂(𝑛).
    Much more efficient than method 1.

3️⃣ Merged Recursive Function (single function with default args)
Logic: Same as #2, but cleaner API (no helper).

Time Complexity:
    Same as #2: 𝑂(𝑛).

Space Complexity:
    Same as #2: 𝑂(ℎ) recursion + 𝑂(𝑛)  result list.
Difference: Just syntactic elegance, not complexity.




### Binary Tree - Level Order Traversal - [Expected] [Iterative] approach - Using Queue

O(n) Time 
O(n) Space 

How it works:

1. Use a queue to store nodes level by level.

2. At each iteration:

    * Measure level_size (number of nodes at this level).

    * Process all nodes at that level.

    * Enqueue their children for the next level.

3. Append the collected nodes into res.


⚖️ Complexity
Time Complexity: 𝑂(𝑛) — each node is visited once.

Space Complexity:

    - Queue can hold up to 𝑂(𝑛)  nodes in the worst case (last level of a complete tree).

    - Result list also 𝑂(𝑛).


🎯 Why Iterative BFS Is Preferred

    Efficient: avoids repeated traversals (unlike recursive height‑based BFS).

    Scalable: handles very deep trees without recursion depth limits.

    Industry standard: used in graph algorithms, tree serialization, and large‑scale systems.

```python
from collections import deque

class Node: 
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None

def levelOrderIterative(root):
    if root is None:
        return []
    
    # Create an empty queue    
    res = []
    queue = deque([root])

    while queue:
        # number of nodes at current level 
        levelSize = len(queue)
        levelNodes = []

        for _ in range(levelSize):
            node = queue.popleft()
            levelNodes.append(node.data)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

        res.append(levelNodes)
    return res

# Example usage
if __name__ == "__main__":
    root = Node(12)
    root.left = Node(8)
    root.right = Node(18)
    root.left.left = Node(5)
    root.left.right = Node(11)

    print(levelOrderIterative(root))   # [[12], [8, 18], [5, 11]]
```

### Binary Tree Traversal - Depth First Search (Recrsive Approach)

#### PreOrder Traversal
    node -> left -> right

#### InOrder Traversal
    left -> node -> right

#### PostOrder Traversal
    left -> right -> node


```python
class Node:
    def __init__(self, data):
        self.data = data 
        self.left = None 
        self.right = None 

def preOrder(node, res):
    if not node:
        return 
        
    res.append(node.data)
    preOrder(node.left, res)
    preOrder(node.right, res)

def inOrder(node, result):
    if not node:
        return
    inOrder(node.left, result)
    result.append(node.data)
    inOrder(node.right, result)

def postOrder(node, result):
    if not node:
        return
    postOrder(node.left, result)
    postOrder(node.right, result)
    result.append(node.data)
    
    

if __name__ == '__main__':
    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)
    result = []
    preOrder(root, result)
    # # print (res) 
    # # print(' '.join(map(str, res)))
    print(*result)
    result = []
    inOrder(root, result)    
    print(' -> '.join(map(str, result)))
    result = []
    postOrder(root, result)    
    print(' -> '.join(map(str, result)))
```

🧭 Traversal Insights
Preorder (Root → Left → Right)

    Successor: usually the left child, else right child, else climb up.

    Predecessor: parent if you’re a left child, or deepest node in left subtree if you’re a right child.

Inorder (Left → Root → Right)

    Successor: smallest node in right subtree.

    Predecessor: largest node in left subtree.

    Most important in BST operations (deletion, search).

Postorder (Left → Right → Root)

    Successor: parent if you’re a right child, else deepest node in right subtree.

    Predecessor: deepest descendant if you have children, else climb up.

### Preorder Successor

#### Preorder traversal = Root → Left → Right.

**Successor of a node is:**

    If it has a left child → that child.

    Else if it has a right child → that child.

    Else → climb up until you find a node whose right child exists, then that right child is the successor.

### Postorder Successor

#### Postorder traversal = Left → Right → Root.

**Successor of a node is:**

    If it’s the right child → parent.

    If it’s the left child and parent has a right subtree → leftmost node of that right subtree in postorder.

    Otherwise → parent.

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
        self.parent = None

# Preorder successor
def preorder_successor(node):
    if node.left:
        return node.left
    if node.right:
        return node.right
    curr = node
    while curr.parent:
        if curr == curr.parent.left and curr.parent.right:
            return curr.parent.right
        curr = curr.parent
    return None

# Postorder successor
def postorder_successor(node):
    if not node.parent:
        return None
    parent = node.parent
    if parent.right and node == parent.left:
        # go deep into right subtree
        curr = parent.right
        while curr.left or curr.right:
            curr = curr.left if curr.left else curr.right
        return curr
    return parent
```

## Predecessor Algorithms

### Preorder Predecessor
Preorder = Root → Left → Right.

Predecessor of a node is:

If it has a parent and the node is the right child, then predecessor is:

If parent has a left subtree → go deep into that left subtree’s rightmost node in preorder.

Else → parent.

If it’s the left child, predecessor is simply the parent.

If it’s the root → no predecessor.

### Postorder Predecessor
Postorder = Left → Right → Root.

Predecessor of a node is:

If it has a right child → that right child’s deepest descendant (postorder last).

Else if it has a left child → that left child’s deepest descendant.

Else → climb up until you find a node where current is a right child, then predecessor is the left subtree’s deepest node.

If it’s the very first node in traversal → no predecessor.


```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
        self.parent = None

# Preorder predecessor
def preorder_predecessor(node):
    if not node.parent:
        return None
    parent = node.parent
    if node == parent.left:
        return parent
    if node == parent.right:
        if parent.left:
            curr = parent.left
            while curr.right or curr.left:
                curr = curr.right if curr.right else curr.left
            return curr
        return parent
    return None

# Postorder predecessor
def postorder_predecessor(node):
    # If node has children, predecessor is deepest descendant
    if node.right:
        curr = node.right
        while curr.left or curr.right:
            curr = curr.right if curr.right else curr.left
        return curr
    if node.left:
        curr = node.left
        while curr.left or curr.right:
            curr = curr.right if curr.right else curr.left
        return curr
    # Otherwise climb up
    curr = node
    while curr.parent:
        if curr == curr.parent.right and curr.parent.left:
            # deepest node in left subtree
            temp = curr.parent.left
            while temp.left or temp.right:
                temp = temp.right if temp.right else temp.left
            return temp
        curr = curr.parent
    return None
```
