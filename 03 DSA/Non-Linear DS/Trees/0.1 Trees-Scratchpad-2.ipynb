```python
# Binary Tree
from collections import deque 

class Node:
    def __init__(self, x):
        self.key = x 
        self.left = None 
        self.right = None 

class BinaryTree:
    def __init__(self):
        self.root = None 
    

    def insert(self, key):

        if not self.root:
            self.root = Node(key)
            return 
        
        q = deque()
        q.append(self.root)

        while q:
            curr = q.popleft()

            if curr.left is not None:
                q.append(curr.left)
            else:
                curr.left = Node(key)
                return self.root

            if curr.right is not None:
                q.append(curr.right)
            else:
                curr.right = Node(key)
                return self.root

    def inorder(self, node):
        if node is None:
            return 
        self.inorder(node.left)
        print (node.key, end= ' ')
        self.inorder(node.right)


if __name__ == "__main__":
    bt = BinaryTree()
    bt.insert(10)
    bt.inorder(bt.root)



        
```
