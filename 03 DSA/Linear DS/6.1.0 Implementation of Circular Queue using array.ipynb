### Implementation of Circular Queue Using Array

A circular queue is a linear data structure that overcomes the limitations of a simple queue. In a normal array implementation, dequeue() can be O(n) or we may waste space. 

Using a circular array, both enqueue() and dequeue() can be done in O(1).

In this implementation, arr is used to store elements. some variables are maintained:

    arr[] : array to store elements.
    capacity : maximum size of the queue.
    front : index of the front element.
    size : current number of elements in the queue.

Operations On Circular Queue:
enqueue(x) :

Purpose: Insert an element x at the rear of the circular queue.

    Check for full queue: If size == capacity, the queue is full print message or return.
    Compute rear index: rear = (front + size) % capacity ensures circular behavior.
    Insert element: arr[rear] = x.
    Update size: Increment size by 1.
    Time Complexity: O(1) Space Complexity: O(1) for the array

dequeue() :

Purpose: Remove and return the front element from the circular queue.

    Check for empty queue: If size == 0, the queue is empty print message or return -1.
    Retrieve front element: res = arr[front].
    Move front forward: front = (front + 1) % capacity circular movement.
    Update size: Decrement size by 1.
    Return element: Return res.
    Time Complexity: O(1) Space Complexity: O(1)

getRear() :

Purpose: Return the element at the rear of the circular queue.

    Check for empty queue: If size == 0, the queue is empty → return -1.
    Compute rear index: rear = (front + size - 1) % capacity.
    Return element: Return arr[rear].
    Time Complexity: O(1) Space Complexity: O(1)

getFront() :

Purpose: Return the element at the front of the circular queue.

    Check for empty queue: If size == 0, the queue is empty → return -1.
    Return element: arr[front] is the front element.
    Time Complexity: O(1) Space Complexity: O(1)
    
                    

```python
class myQueue:
    def __init__(self, cap):
         # fixed-size array
        self.arr = [0]*cap    
           # index of front element
        self.front = 0 
          # current number of elements
        self.size = 0      
         # maximum capacity
        self.capacity = cap    

    # Insert an element at the rear
    def enqueue(self, x):
        if self.size == self.capacity:
            print("Queue is full!")
            return
        
        rear = (self.front + self.size) % self.capacity
        # print (f"front={self.front} + size = {self.size} % capacity={self.capacity} = rear i.e {rear}")

        self.arr[rear] = x
        self.size += 1

    # Remove an element from the front
    def dequeue(self):
        if self.size == 0:
            print("Queue is empty!")
            return -1
        res = self.arr[self.front]
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return res

    # Get the front element
    def getFront(self):
        if self.size == 0:
            return -1
        return self.arr[self.front]

    # Get the rear element
    def getRear(self):
        if self.size == 0:
            return -1
        rear = (self.front + self.size - 1) % self.capacity
        return self.arr[rear]

    def printQueue(self):
        #check if the queue is empty 
        if self.size == 0 :
            print ("Queue is empty")
            return 
        
        for i in range(self.size):
            # index = i,why cant you directly print arr[i] .. see below explanation 
            index = (self.front + i) % self.capacity
            print (self.arr[index], end = ' ')        

    # another method to print the contents of the Queue
    def __str__(self):
        if self.size == 0:
            return "Queue is Empty"
        elements = []
        for i in range(self.size):
            index = (self.front + i) % self.capacity
            elements.append(str(self.arr[index]))
        return "\nQueue Contents: "+ " ".join(elements)
    
    # Iterator Support 
    """
    def __iter__(self):
        self._iter_index = 0
        return self
    """

    def __next__(self):
        if self._iter_index >= self.size:
            raise StopIteration
        index = (self.front + self._iter_index) % self.capacity
        value = self.arr[index]
        self._iter_index += 1
        return value
    
    # Generator 
    def __iter__(self):
        for i in range(self.size):
            index = (self.front + i) % self.capacity
            yield self.arr[index]   


if __name__ == "__main__":
    q = myQueue(5)
    q.enqueue(10)
    q.enqueue(20)
    q.enqueue(30)
    q.enqueue(40)
    q.printQueue()

    print (q)

    for item in q:
        print("Iterated:", item)


    # print(q.getFront(), q.getRear())
    # q.dequeue()
    # print(q.getFront(), q.getRear())
    # q.enqueue(40)
    # print(q.getFront(), q.getRear())
```

Why we still need index = (self.front + i) % self.capacity in the printQueue() method 
- The queue is stored in a fixed-size array (self.arr) that doesn’t shift elements when you enqueue or dequeue.
- Instead, it uses circular indexing: the front pointer moves forward, and the rear wraps around when it reaches the end.
- So the logical order of elements in the queue may not match their physical order in the array.
The calculation (self.front + i) % self.capacity maps the logical position (i-th element from the front) to the actual index in the array.

The Key takeaway:
The range(self.size) ensures we only loop over the right number of elements.
The (self.front + i) % self.capacity ensures we access them in the correct logical order, regardless of where they sit in the circular buffer.

Now that this is clear, shall I show you how to add a __str__ method so you can just do print(q) and it will automatically use this same logic to display the queue neatly?




```python
class myQueue:
    def __init__(self, cap):
        self.arr = [0]*cap
        self.front = 0
        self.size = 0
        self.capacity = cap

    def enqueue(self, x):
        if self.size == self.capacity:
            print("Queue is full!")
            return
        rear = (self.front + self.size) % self.capacity
        self.arr[rear] = x
        self.size += 1

    def dequeue(self):
        if self.size == 0:
            print("Queue is empty!")
            return -1
        res = self.arr[self.front]
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return res

    def getFront(self):
        if self.size == 0:
            return -1
        return self.arr[self.front]

    def getRear(self):
        if self.size == 0:
            return -1
        rear = (self.front + self.size - 1) % self.capacity
        return self.arr[rear]

    def __str__(self):
        if self.size == 0:
            return "Queue is empty!"
        elements = []
        for i in range(self.size):
            index = (self.front + i) % self.capacity
            elements.append(str(self.arr[index]))
        return "Queue contents: " + " ".join(elements)

    def __iter__(self):
        for i in range(self.size):
            index = (self.front + i) % self.capacity
            yield self.arr[index]

    # 👇 Indexing and slicing
    def __getitem__(self, key):
        if isinstance(key, int):
            # Handle negative indices
            if key < 0:
                key += self.size
            if key < 0 or key >= self.size:
                raise IndexError("Queue index out of range")
            index = (self.front + key) % self.capacity
            return self.arr[index]

        elif isinstance(key, slice):
            start, stop, step = key.indices(self.size)
            result = []
            for i in range(start, stop, step):
                index = (self.front + i) % self.capacity
                result.append(self.arr[index])
            return result

        else:
            raise TypeError("Queue indices must be integers or slices")    

    def __len__(self):
        return self.size

    # Truthiness support
    def __bool__(self):
        return self.size > 0
    
    # reset the queue    
    def clear(self):
        # Reset the queue to its initial empty state
        self.arr = [0] * self.capacity
        self.front = 0
        self.size = 0

    # check for membership
    def __contains__(self, item):
        # iterate through the queue and check for memership 
        for i in range (self.size):
            index = (self.front + i) % self.capacity 
            if self.arr[index] == item:
                return True 
        return False
 
        
if __name__ == "__main__":
    q = myQueue(5)
    q.enqueue(10)
    q.enqueue(20)
    q.enqueue(30)
    q.enqueue(40)

    print(q)        # Queue contents: 10 20 30 40
    print(q[0])     # 10
    print(q[-1])    # 40
    print(q[1:3])   # [20, 30]
    print(10 in q)

    print (len(q))

    q.clear()
    print(q)        # Queue is empty!
```
