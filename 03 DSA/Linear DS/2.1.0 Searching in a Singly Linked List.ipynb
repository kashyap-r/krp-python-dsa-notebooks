# Search for an element in a linked list 
 Given the head of a linked list and a key,find the key by traversing through the 
 nodes sequentially

```python
class Node:
    def __init__(self, value):
        self.value = value 
        self.next = None 

class SinglyLinkedList:
    def __init__(self):
        self.head = None 

    def insertAtStart(self, val):
        newNode = Node(val)
        newNode.next =  self.head 
        self.head = newNode
        return self.head 

    def insertAtEnd(self, val):
        if self.head is None:
            return self.insertAtStart(val)
        lastNode = self.head 
        while lastNode.next is not None:        
            lastNode = lastNode.next

        newNode = Node(val)
        lastNode.next = newNode 
    
    def traversal(self):
        curr = self.head 
        while curr:
            print (curr.value, end=" ")
            curr = curr.next 

    # Searching based on a key - iterative approach 
    # O(n) Time and O(1) space because you need jsut the one variable
    def iterSearch(self, key):
        curr = self.head

        while curr is not None:
            if curr.value == key:
                return True 
            curr = curr.next 

        return False 
    
    # searching based on a key - recursive approach 
    # O(n) time and O(n) space for the recursive stack
    def search(self, key):
        return self._recSearch(self.head, key)

    def _recSearch(self, node, key):
        if node is None:
            return False
        if node.value == key:
            return True
        return self._recSearch(node.next, key)
    
    def recSearch(self, node, key):
        if node is None:
            return False
        if node.value == key:
            return True 
        
        return self.recSearch(node.next, key)

    

ll = SinglyLinkedList()
ll.insertAtStart(10)
ll.insertAtStart(5)
ll.insertAtEnd(20)
ll.insertAtEnd(30)
ll.insertAtEnd(40)
ll.insertAtEnd(50)

ll.traversal()
ll.iterSearch(40)
ll.iterSearch(45)

print (ll.search(10))
ll.search(55)
```
