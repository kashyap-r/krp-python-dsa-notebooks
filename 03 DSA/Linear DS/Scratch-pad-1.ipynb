# Singly Linked List 

```python
class Node:
    def __init__(self, x):
         self.data = x 
         self.next = None 

class SinglyLinkedList:
    def __init__(self):
        self.head = None 

    # convert linked list to Python list
    def toList(self):
        res = []
        curr = self.head 
        while curr:
            res.append(curr.data)
            curr = curr.next 
        return res 

    # insert at beginning - O(1)
    def insertAtBegin(self, x):
        newNode = Node(x)         
        newNode.next = self.head 
        self.head = newNode        
    
    # inser at end - O(n) 
    def insertAtEnd(self, x):
        # if the list is empty, 
        # create the list with a head 
        if not self.head:
            self.insertAtBegin(x)
            return

        curr = self.head 
        # check if the current node points to the next element 
        # till the end of the list, as last node has next pointer as null
        while curr.next: 
            curr = curr.next 
        newNode = Node(x)
        curr.next = newNode
        return self.head 

    # insert at position (1-based) 
    def insertAtPos(self, pos, x):
        if pos < 1:
            raise ValueError("Position must be >= 1")
        
        if pos == 1:
            self.insertAtBegin(x)
            return
        
        curr = self.head
        for i in range (pos - 2):
            if curr is None:
                raise IndexError("Position out of bounds")
            curr = curr.next 
        
        # if position is greater than number of nodes
        if curr is None:
            raise IndexError("Position out of bounds") 
               
        newNode = Node(x)
        newNode.next = curr.next
        curr.next = newNode

    def insertAtPos2(self, pos, data):
        if pos == 0:
            self.insertAtBegin(data)
        
        curr = self.head
        count = 0 
        while curr is not None and count < pos - 1:
            curr = curr.next 
            count += 1 
        
        newNode = Node(data)
        newNode.next = curr.next 
        curr.next = newNode 

    # Insert at a position after a given value
    def insertAtPosVal(self, target, data):
        curr = self.head 
        while curr: 
            if curr.data == target:
                newNode = Node(data)
                newNode.next = curr.next
                curr.next = newNode
                return 
            curr = curr.next 

    # length of the linked list - O(n) 
    def length(self):
        count = 0 
        curr = self.head 
        while curr:
            count+=1 
            curr = curr.next 
        return count 

    # Delete at beginning - O(1)
    def deleteAtBegin(self):
        if self.head is None:
            raise IndexError("List s empty")
         
        delval = self.head.data 
        self.head = self.head.next 
        return delval

    # Delete at Position (1-based)
    def deleteAtPos(self, pos):
        # check if the position is < 1 
        if pos < 1:
            raise ValueError("Position must be >= 1")
        
        # if the list is empty 
        if not self.head:
            raise IndexError("Empty List !!") 

        # if pos ==1 i.e. head is to be deleted
        if pos == 1:
            return self.deleteAtBegin()
        
        curr = self.head

        for _ in range(pos - 2):
            if curr.next is None:
                raise IndexError("Position is Out of bounds")
            curr = curr.next

        if curr is None or curr.next is None:
            raise IndexError("Position is out of bounds")
        
        delval = curr.next.data
        curr.next = curr.next.next
        return delval 

    def deleteAtPos2(self, pos):
        # check if the position is < 1 
        if pos < 1:
            raise ValueError("Position must be >= 1")
        
        # if the list is empty 
        if not self.head:
            raise IndexError("Empty List !!") 

        # if pos ==1 i.e. head is to be deleted
        if pos == 1:
            return self.deleteAtBegin()
        
        curr = self.head
        # traverse to the node before the one to be deleted
        prev = None
        for i in range(pos):
            prev = curr 
            curr = curr.next

        # delete the node at the position
        prev.next = curr.next 
        return self.head 
    
    def deleteLastNode(self):
        if self.head is None:
            raise IndexError("List is empty") 
        # if the list has only one node 
        if self.head.next is None:
            delval = self.head.data
            self.head = None 
            return delval
        # Now find the second last node
        curr = self.head 
        while curr.next.next:
            curr = curr.next

        delval = curr.next.data
        curr.next = None
        return delval
    
    # Traversing a Linked LIst - Recursive approach 
    def traverseListRec(self):
        # base condition 
        if self.head is None:
            print() 
            return 
        print (self.head.data, end="")

        if self.head.next is not None:
            print (" -> ", end="")

        self.traverseListRec(self.head.next) 

    # Traversing a Linked List - Iterative approach 
    def traverseList(self):
        while self.head is not None:
            print (self.head.data, end=" ") 
            if self.head.next is not None:
                print (" -> ", end = "")
            self.head = self.head.next 
        print() 

        # -------------------------------
    # Traverse (safe, non-destructive)
    # -------------------------------
    def traverse(self):
        curr = self.head
        result = []

        while curr:
            result.append(str(curr.data))
            curr = curr.next

        return " -> ".join(result) if result else "EMPTY"
    
    # -------------------------------
    # Reverse list — O(n)
    # -------------------------------
    def reverse(self):
        prev = None
        curr = self.head

        while curr:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt

        self.head = prev


    def __iter__(self):
        curr = self.head
        while curr:
            yield curr.data
            curr = curr.next

    def __str__(self):
        return self.traverse()

```

```python
def print_list(ll):
    """Helper to print list without modifying it"""
    curr = ll.head
    res = []
    while curr:
        res.append(str(curr.data))
        curr = curr.next
    return " -> ".join(res) if res else "EMPTY"


def run_tests():
    print("\n================ LINKED LIST TEST HARNESS ================\n")

    # -------------------------------------------------------
    # TEST 1: Insert at beginning on empty list
    # -------------------------------------------------------
    print("TEST 1: Insert at beginning (empty list)")
    ll = SinglyLinkedList()
    ll.insertAtBegin(10)
    print("Expected: 10")
    print("Actual  :", print_list(ll))
    print()

    # -------------------------------------------------------
    # TEST 2: Insert multiple at beginning
    # -------------------------------------------------------
    print("TEST 2: Insert multiple at beginning")
    ll.insertAtBegin(20)
    ll.insertAtBegin(30)
    print("Expected: 30 -> 20 -> 10")
    print("Actual  :", print_list(ll))
    print()

    # -------------------------------------------------------
    # TEST 3: Insert at position (middle)
    # -------------------------------------------------------
    print("TEST 3: Insert at position (pos=2)")
    ll.insertAtPos(2, 25)
    print("Expected: 30 -> 25 -> 20 -> 10")
    print("Actual  :", print_list(ll))
    print("NOTE: This will likely FAIL due to curr.head bug\n")

    # -------------------------------------------------------
    # TEST 4: Insert at position beyond length
    # -------------------------------------------------------
    print("TEST 4: Insert at position beyond length")
    ll.insertAtPos(100, 99)
    print("Expected: No change")
    print("Actual  :", print_list(ll))
    print()

    # -------------------------------------------------------
    # TEST 5: Insert using insertAtPos2
    # -------------------------------------------------------
    print("TEST 5: insertAtPos2 (pos=2)")
    try:
        ll.insertAtPos2(2, 77)
        print("Expected: insertion at pos=2")
        print("Actual  :", print_list(ll))
    except Exception as e:
        print("FAILED with exception:", e)
    print("NOTE: Likely fails if curr becomes None\n")

    # -------------------------------------------------------
    # TEST 6: Insert at end
    # -------------------------------------------------------
    print("TEST 6: Insert at end")
    try:
        ll.insertAtEnd(999)
        print("Expected: last element = 999")
        print("Actual  :", print_list(ll))
    except Exception as e:
        print("FAILED with exception:", e)
    print("NOTE: This WILL FAIL due to while loop bug\n")

    # -------------------------------------------------------
    # TEST 7: Length check
    # -------------------------------------------------------
    print("TEST 7: Length")
    print("Expected: >=1")
    print("Actual  :", ll.length())
    print()

    # -------------------------------------------------------
    # TEST 8: Delete at beginning
    # -------------------------------------------------------
    print("TEST 8: Delete at beginning")
    val = ll.deleteAtBegin()
    print("Deleted:", val)
    print("List   :", print_list(ll))
    print()

    # -------------------------------------------------------
    # TEST 9: Delete at position (middle)
    # -------------------------------------------------------
    print("TEST 9: Delete at position (pos=2)")
    try:
        ll.deleteAtPos(2)
        print("Expected: node removed at pos 2")
        print("Actual  :", print_list(ll))
    except Exception as e:
        print("FAILED with exception:", e)
    print("NOTE: Possible off-by-one bug\n")

    # -------------------------------------------------------
    # TEST 10: Delete last node
    # -------------------------------------------------------
    print("TEST 10: Delete last node")
    val = ll.deleteLastNode()
    print("Deleted:", val)
    print("List   :", print_list(ll))
    print()

    # -------------------------------------------------------
    # TEST 11: Delete from empty list
    # -------------------------------------------------------
    print("TEST 11: Delete from empty list")
    empty_ll = SinglyLinkedList()
    val = empty_ll.deleteAtBegin()
    print("Expected: -1 or message")
    print("Actual  :", val)
    print()

    # -------------------------------------------------------
    # TEST 12: Traverse iterative (BUG: destroys list)
    # -------------------------------------------------------
    print("TEST 12: Traverse iterative (destructive)")
    ll2 = SinglyLinkedList()
    ll2.insertAtBegin(1)
    ll2.insertAtBegin(2)
    ll2.insertAtBegin(3)

    print("Before traverse:", print_list(ll2))
    ll2.traverseList()
    print("After traverse :", print_list(ll2))
    print("NOTE: This WILL EMPTY the list (bug)\n")

    # -------------------------------------------------------
    # TEST 13: Traverse recursive (BUG: signature mismatch)
    # -------------------------------------------------------
    print("TEST 13: Traverse recursive")
    try:
        ll2.traverseListRec()
    except Exception as e:
        print("FAILED with exception:", e)
        print("NOTE: Method definition is incorrect\n")

    print("\n================ END OF TESTS ================\n")


# Run tests
if __name__ == "__main__":
    run_tests()
```
