```python
a = [10, 20, 30, 40]

# accessing a list using index 
print (a[3])
print (a[-1])

# accessing using for loop
for item in a:
    print (item, end=" ")
print ()

for i in range(len(a)):
    print (a[i], end= ' ')
print() 

# Using List slicing 
print (a[0:])
print (a[::-1])
print (a[3:4])
print (a[2:])
```

#### List comprehension 

```python

a = [10, 20, 30, 40, 50]

b = [item for item in a if item > 20]
print (a) 
print (b)
```

```python
# square the elements of the array
a = [1, 2, 3, 4, 5]
res = [val **2 for val in a]
print (res)
```

#### Conditional Statements in List Comprehension 

```python
a = [1, 2, 3, 4, 5, 6, 7, 8, 9, 0]
res = [val for val in a if val % 2 == 0]
print (res)
```

```python
# create a list of numbers > 20 from the list of given numbers 
a = [1, 2, 3, 4, 21, 22, 25, 56, 75]
res = [val for val in a if val > 20]
print (res)
print() 

# Create a list from range 
a = [i for i in range(10)]
print (a)
print()
```

```python
# using nested loops

c = [(x, y, z) for x in range(3) for y in range(3) for z in range(3)]
print (c)
len(c)
```

```python
mat = [[1, 2, 3], [4, 5, 6], [6, 7, 8]]
print (mat)

res = [val for row in mat for val in row]
print (res)
```

The statement  
res = [val for row in mat for val in row]

is same as 

res = []

for row in mat:         # 1. Outer loop: Grab one row at a time
    for val in row:     # 2. Inner loop: Grab each value inside that row
        res.append(val) # 3. Action: Add the value to the new list

```python
# list comprehension using if-else 

a = [1, 2, 3, 4, 5]
res = ["Even" if n % 2 == 0 else "Odd" for n in a]
print (res)
```

```python
# list comprehension using if 

a = [1, 2, 3, 4, 5]
res = [n for n in a if n % 2 == 0]
print (res)
```

```python
# nested if-else 
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

res = ['Divisible by 2' if n % 2 == 0 else 'Divisible by 3' if n%3==0 else 'Other' for n in a]
print (res) 

```

```python
# Updating / Changing list 
a = [1, 2, 3, 4, 5, 6]
# a[1] = 11
print (a)

# Using Slicing for multiple items
a[2:3] = [21, 31]
# a[1:3] = [21, 31]
print (a)
print () 
# Using list comprehension 
# Doubling even numbers in the array 
res = [x * 2 if x % 2 == 0 else x for x in a]
print (res)

# Using a loop with enumerate
# Adding 5 to odd numbers in the array
for i, x in enumerate(a): 
    if x % 2 != 0:
        a[i] += 5
print (a)

# Using map() 
a = [10, 20, 30]
a = list(map(lambda x: x +1, a))
print (a)
```

```python
# Replace values in a List in Python 
"""
Given a list of values and a value to replace, the task is to update one or 
more matching values with a new value
"""
# Using List indexing 
a = [1, 2, 3, 4, 5]
a[2] = 99
print (a)
print ()

# using list comprehension
a = [1, 2, 3, 4, 5]
res = [99 if x ==3 else x for x in a]
print (res)

# Using map() with lambda
a = [1, 2, 3, 4, 5]
res = list(map(lambda x: 99 if x == 5 else x, a))
print (res)

# using for loop 
a = [1, 2, 3, 4, 5]
for i in range(len(a)):
    if a[i] == 4:
        a[i] = 99
print (a)
```

```python
# Python list append() method

a = [2, 3, 5, 6, 7]
a.append(9)
print (a)
a.append([11, 13])
print (a)

for i in range(15, 23, 2):
    a.append(i)
print (a)
```

```python
# list insert() method
a = ['banana','Cherry', 'date', 'elaichi', 'funfruit']
a.insert(0, 'apple')
print (a)

a = [1, 2, 3, 4, 5]
i = a.index(4) 
a.insert(i, 30)
print(a)
```

```python
from itertools import chain

# list extend() method
a = [1, 2, 3, 4]
n = [5, 6, 7, 8]
a.extend(n)

print(a)

# Using the += Operator
a = [1, 2, 3]
n = [4, 5, 6]

a += n
print (a)

# using Slicing 
a = [1, 2, 3, 4]
b = [5, 6, 7, 8]
a[len(a):] = b
print (a)

# Using itertools.chain() 
a = [1, 3, 5, 7]
b = [9, 11, 13, 15]

a.extend(chain(b))
print(a)

# remove the element from a list 
a = [10, 20, 30, 20, 40, 50]
a.remove(20)
print(a)

# using pop() 
val = a.pop(4)
print (a)
print (val)
val = a.pop() 
print (a)

# using del 
del a[1]
print (a)

a = [10, 20, 30, 20, 40, 50]
del a[1:4]
print(a)
a.clear() 
print (a)
```
